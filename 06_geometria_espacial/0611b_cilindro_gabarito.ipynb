## Básico

**Exercício 1.** Verdadeiro ou falso: (a) a seção meridiana de um cilindro reto é um retângulo; (b) no cilindro equilátero, a altura é igual ao raio; (c) num cilindro oblíquo, a altura é menor que a geratriz; (d) um cilindro reto e um oblíquo, com o mesmo raio e a mesma altura, têm o mesmo volume; (e) a planificação da superfície lateral de um cilindro reto é um retângulo.

```python
respostas_ex1 = {
    "a": True,  # o retângulo 2r × h
    "b": False,  # no equilátero, a altura é igual ao DIÂMETRO: h = 2r
    "c": True,  # h = g sen θ, e sen θ < 1 quando a geratriz é inclinada
    "d": True,  # pelo Cavalieri, as seções têm a mesma área em cada altura
    "e": True,  # as medidas do retângulo são 2πr e h
}

print(respostas_ex1)
```

**Exercício 2.** Cilindro reto de raio 3 e altura 10: (a) área lateral; (b) área total; (c) volume.

```python
raio, altura = 3, 10

area_lateral_ex2 = 2 * math.pi * raio * altura  # (a) 60π
area_total_ex2 = 2 * math.pi * raio * (altura + raio)  # (b) 60π + 2 · 9π = 78π
volume_ex2 = math.pi * raio**2 * altura  # (c) 90π

print(area_lateral_ex2, area_total_ex2, volume_ex2)

# Conferência com o utils_3d: um prisma regular de 2000 lados quase não difere do
# cilindro.
solido = u3d.prisma(u3d.poligono_regular(2000, raio=raio), (0, 0, altura))
print(f"Conferência: área total {u3d.medir_area_total(*solido):.2f}")
print(f"Conferência: volume {u3d.medir_volume(*solido):.2f}")
```

**Exercício 3.** Cilindro equilátero de raio 2: (a) altura; (b) área total; (c) volume.

```python
raio = 2

altura_ex3 = 2 * raio  # (a) no cilindro equilátero, h = 2r = 4
area_total_ex3 = 6 * math.pi * raio**2  # (b) S_t = 6πr² = 24π
volume_ex3 = 2 * math.pi * raio**3  # (c) volume = 2πr³ = 16π

print(altura_ex3, area_total_ex3, volume_ex3)
```

## Intermediário

**Exercício 4.** Caixa d'água cilíndrica de 1,2 m de diâmetro e 1,5 m de altura: capacidade em litros.

```python
raio = 1.2 / 2  # o raio é metade do diâmetro: 0,6 m
volume_m3 = math.pi * raio**2 * 1.5  # em m³, porque as medidas estão em metros

litros_ex4 = volume_m3 * 1000  # 1 m³ = 1000 L

print(litros_ex4)
```

**Exercício 5.** Lata de raio 3,3 cm, com um rótulo que cobre uma faixa de 10 cm de altura: área do rótulo.

```python
# O rótulo é a área lateral de um cilindro de raio 3,3 e altura 10: a altura da lata não
# importa, só a altura da faixa.
area_rotulo_ex5 = 2 * math.pi * 3.3 * 10

print(area_rotulo_ex5)
```

**Exercício 6.** A seção meridiana de um cilindro reto tem área 48 cm² e a altura é 8 cm: (a) raio; (b) volume.

```python
# (a) A seção meridiana é um retângulo 2r × h, então 2r · 8 = 48.
raio_ex6 = 48 / (2 * 8)

# (b) Volume = πr²h = π · 9 · 8 = 72π.
volume_ex6 = math.pi * raio_ex6**2 * 8

print(raio_ex6, volume_ex6)
```

## Desafio

**Exercício 7.** Para latas de 350 cm³, testando os raios de 1,50 cm a 8,00 cm, de 0,01 em 0,01: a razão $\frac{h}{r}$ do raio de menor área total.

```python
melhor_raio = None
menor_area = float("inf")  # qualquer área calculada será menor que "infinito"

for k in range(651):
    raio = 1.5 + k * 0.01
    altura = 350 / (math.pi * raio**2)
    area_total = 2 * math.pi * raio * altura + 2 * math.pi * raio**2
    # Sempre que aparece uma área menor, ela e o seu raio passam a ser os "campeões".
    if area_total < menor_area:
        menor_area = area_total
        melhor_raio = raio

razao_ex7 = (350 / (math.pi * melhor_raio**2)) / melhor_raio

print(melhor_raio, razao_ex7)
print(f"Menor área total: {menor_area:.2f} cm²")
```

O resultado não depende do volume escolhido: para qualquer volume, a lata de menor área tem $h = 2r$. Provar isso exige uma ferramenta do Cálculo (a derivada), que você verá no módulo 11.
