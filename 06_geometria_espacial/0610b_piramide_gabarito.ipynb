**Exercício 2.** Pirâmide quadrangular regular de aresta da base 6 e altura 4: (a) apótema da base; (b) apótema da pirâmide; (c) aresta lateral; (d) área total; (e) volume.

```python
lado, altura = 6, 4

apotema_da_base_ex2 = lado / 2  # (a) na base quadrada, m = ℓ / 2 = 3
apotema_da_piramide_ex2 = math.sqrt(altura**2 + apotema_da_base_ex2**2)  # (b) VOM: g = 5

# (c) Triângulo VMA: a² = g² + (ℓ/2)² = 25 + 9 = 34.
aresta_lateral_ex2 = math.sqrt(apotema_da_piramide_ex2**2 + (lado / 2) ** 2)

# (d) S_ℓ = p · g, com p = 4 · 6 / 2 = 12; a base tem área 36.
semiperimetro = 4 * lado / 2
area_total_ex2 = semiperimetro * apotema_da_piramide_ex2 + lado**2

volume_ex2 = lado**2 * altura / 3  # (e) volume = (1/3) · 36 · 4 = 48

print(
    apotema_da_base_ex2, apotema_da_piramide_ex2, aresta_lateral_ex2, area_total_ex2,
    volume_ex2,
)

# Conferência com o utils_3d.
solido = u3d.piramide_regular(4, lado, altura)
print(f"Conferência: área total {u3d.medir_area_total(*solido):.4f}")
print(f"Conferência: volume {u3d.medir_volume(*solido):.4f}")
```

**Exercício 6.** Pirâmide do Louvre (base de 35 m, altura de 21,6 m): (a) o apótema da pirâmide; (b) a área lateral.

```python
lado, altura = 35, 21.6

# (a) Na base quadrada, m = 35 / 2 = 17,5. No triângulo VOM: g² = h² + m².
m = lado / 2
apotema_ex6 = math.sqrt(altura**2 + m**2)

# (b) S_ℓ = p · g, com p = 4 · 35 / 2 = 70. Sem a base: o chão não leva vidro!
p = 4 * lado / 2
area_lateral_ex6 = p * apotema_ex6

print(apotema_ex6, area_lateral_ex6)
```

**Exercício 7.** No cubo $ABCDEFGH$ de aresta 2, as pirâmides de vértice $G$ e bases $ABCD$, $ABFE$ e $ADHE$: (a) o volume de cada uma; (b) são congruentes? (c) preenchem o cubo?

```python
# (a) A base de cada pirâmide é uma face do cubo (área 4). A altura é a distância de G
#     à face da base, que é a aresta do cubo (2).
volume_ex7 = (1 / 3) * 4 * 2  # 8/3 ≈ 2,667

# (b) Girando o cubo em torno da diagonal AG, cada pirâmide ocupa o lugar da seguinte.
congruentes_ex7 = True

# (c) As três bases são as três faces que NÃO contêm G. Todo ponto do cubo está num
#     segmento que sai de G e termina numa dessas faces, então nada fica de fora.
#     A soma dos volumes confirma: 3 · 8/3 = 8, o volume do cubo.
preenchem_ex7 = True

print(volume_ex7, congruentes_ex7, preenchem_ex7)

# As três pirâmides afastadas e, depois, juntas.
cubo = u3d.vertices_do_cubo()
pecas = [
    u3d.piramide([cubo[letra] for letra in base], cubo["G"])
    for base in ("ABCD", "ABFE", "ADHE")
]
print("Conferência:", [round(u3d.medir_volume(*peca), 4) for peca in pecas])
cores = [
    u3d.CORES["plano"], u3d.CORES["plano_secundario"], u3d.CORES["plano_terciario"],
]
for titulo, distancia in (("Afastadas", 1.5), ("Juntas: um cubo", 0)):
    fig = u3d.criar_figura(titulo=titulo)
    for peca, cor in zip(u3d.afastar_pecas(pecas, distancia), cores, strict=True):
        fig = u3d.desenhar_poliedro(fig, *peca, cor=cor, opacidade=0.6)
    u3d.mostrar(fig)
```

Esta é outra forma de "ver" o $\frac{1}{3}$: três pirâmides iguais formam um cubo.

**Exercício 8.** Para a pirâmide de base com área 1 e altura 1 (volume exato $\frac{1}{3}$), o menor $N$ para o qual a soma das $N$ placas difere de $\frac{1}{3}$ em menos de $0{,}001$.

```python
def soma_das_placas(N: int) -> float:
    """Soma os volumes das N placas que aproximam a pirâmide."""
    soma = 0
    for k in range(N):
        soma = soma + (1 - k / N) ** 2 * (1 / N)
    return soma


# Enquanto a diferença não for menor que 0,001, tentamos o N seguinte. O laço para no
# primeiro N que serve, que é justamente o menor.
N = 1
while soma_das_placas(N) - 1 / 3 >= 0.001:
    N = N + 1
menor_N_ex8 = N

print(menor_N_ex8)
print(f"Diferença com N = 500: {soma_das_placas(500) - 1 / 3:.7f}")
print(f"Diferença com N = 501: {soma_das_placas(501) - 1 / 3:.7f}")
```
