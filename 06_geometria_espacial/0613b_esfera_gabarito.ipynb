## Básico

**Exercício 1.** Verdadeiro ou falso: (a) toda seção plana de uma esfera é um círculo (ou um ponto, se o plano for tangente); (b) o equador é um círculo máximo; (c) todo paralelo é um círculo máximo; (d) o plano tangente a uma esfera é perpendicular ao raio no ponto de tangência; (e) dobrando o raio de uma esfera, o volume dobra.

```python
respostas_ex1 = {
    "a": True,  # r = √(R² − d²): um círculo se d < R, um ponto se d = R
    "b": True,  # o plano do equador passa pelo centro: o raio da seção é R
    "c": False,  # só o equador; os outros paralelos estão em planos fora do centro
    "d": True,  # é a propriedade do plano tangente (a bola na mesa)
    "e": False,  # o volume tem R³: dobrando o raio, o volume fica 2³ = 8 vezes maior
}

print(respostas_ex1)
```

**Exercício 2.** Esfera de raio 3: (a) área; (b) volume; (c) raio da seção feita a 2 do centro.

```python
raio = 3

area_ex2 = 4 * math.pi * raio**2  # (a) 4π · 9 = 36π
volume_ex2 = 4 / 3 * math.pi * raio**3  # (b) (4/3)π · 27 = 36π
raio_secao_ex2 = math.sqrt(raio**2 - 2**2)  # (c) √(9 − 4) = √5

print(area_ex2, volume_ex2, raio_secao_ex2)

# Conferência com o utils_3d: a seção pelo plano a 2 do centro e um domo geodésico
# bem fino (5 subdivisões) inscrito na esfera.
_, raio_conferido = u3d.secao_da_esfera((0, 0, 0), raio, ((0, 0, 2), (0, 0, 1)))
print(f"Conferência: raio da seção {raio_conferido:.4f}")
domo = u3d.poliedro_aproximando_esfera(raio, 5)
print(f"Conferência: área do domo {u3d.medir_area_total(*domo):.2f}")
print(f"Conferência: volume do domo {u3d.medir_volume(*domo):.2f}")
```

**Exercício 3.** Esfera de raio 6, ângulo de 30°: (a) área do fuso; (b) volume da cunha.

```python
raio, angulo = 6, 30
fracao = angulo / 360  # 30° é 1/12 da volta

area_fuso_ex3 = fracao * 4 * math.pi * raio**2  # (a) (1/12) · 144π = 12π
volume_cunha_ex3 = fracao * 4 / 3 * math.pi * raio**3  # (b) (1/12) · 288π = 24π

print(area_fuso_ex3, volume_cunha_ex3)
```

## Intermediário

**Exercício 4.** A Terra (raio 6371 km): (a) comprimento do equador; (b) km por grau de longitude no equador; (c) comprimento do paralelo de latitude 60°.

```python
R_terra = 6371

equador_ex4 = 2 * math.pi * R_terra  # (a) ≈ 40 030 km
km_por_grau_ex4 = equador_ex4 / 360  # (b) ≈ 111 km
# (c) O raio do paralelo é R·cos(60°) = R/2: o paralelo tem metade do equador.
paralelo_60_ex4 = 2 * math.pi * R_terra * math.cos(math.radians(60))

print(equador_ex4, km_por_grau_ex4, paralelo_60_ex4)
```

**Exercício 5.** Bola de basquete de 24 cm de diâmetro: (a) volume; (b) área da superfície.

```python
raio = 24 / 2  # o problema dá o diâmetro: o raio é a metade, 12 cm

volume_ex5 = 4 / 3 * math.pi * raio**3  # (a) (4/3)π · 1728 = 2304π ≈ 7238 cm³
area_ex5 = 4 * math.pi * raio**2  # (b) 4π · 144 = 576π ≈ 1810 cm²

print(volume_ex5, area_ex5)
```

**Exercício 6.** Esfera dentro do cilindro equilátero que a envolve: (a) razão entre os volumes; (b) razão entre a área da esfera e a área total do cilindro.

```python
R = 1  # as razões não dependem do raio; com R = 1 as contas ficam simples

volume_esfera = 4 / 3 * math.pi * R**3
volume_cilindro = math.pi * R**2 * (2 * R)  # base πR², altura 2R
razao_volumes_ex6 = volume_esfera / volume_cilindro  # (a) (4/3)π / 2π = 2/3

area_esfera = 4 * math.pi * R**2
# Área total do cilindro: lateral 2πR · 2R = 4πR² mais duas bases πR²: 6πR².
area_total_cilindro = 2 * math.pi * R * (2 * R) + 2 * math.pi * R**2
razao_areas_ex6 = area_esfera / area_total_cilindro  # (b) 4π / 6π = 2/3

print(razao_volumes_ex6, razao_areas_ex6)
```

**Exercício 8.** O menor número de subdivisões $k$ para o qual a área do poliedro de `u3d.poliedro_aproximando_esfera(1, k)` fica a menos de 1% de $4\pi$.

```python
menor_k_ex8 = None

for k in range(5):
    vertices, faces = u3d.poliedro_aproximando_esfera(1, k)
    area = u3d.medir_area_total(vertices, faces)
    erro = (4 * math.pi - area) / (4 * math.pi)  # quanto falta, em fração de 4π
    print(f"k = {k}: {len(faces)} faces, área {area:.4f}, erro {erro:.2%}")
    if erro < 0.01:
        menor_k_ex8 = k
        break  # achamos o menor: não precisamos testar os outros

print(menor_k_ex8)
```

Com 3 subdivisões o domo já tem 1280 faces, e a área fica a menos de 0,5% da área da esfera. Os domos geodésicos, como o famoso domo do Epcot, nos Estados Unidos, usam exatamente essa ideia: muitos triângulos quase planos formando uma superfície quase esférica.
