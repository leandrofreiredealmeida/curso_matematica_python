**Exercício 4.** Telhado de duas águas com inclinação de 30° sobre uma planta de 8 m × 10 m: área do telhado e número de telhas (16 por m²).

```python
largura_da_planta = 8
comprimento_da_planta = 10
inclinacao = 30
telhas_por_m2 = 16

# A planta é a projeção do telhado no chão: S' = S · cos θ, logo S = S' / cos θ.
area_planta = largura_da_planta * comprimento_da_planta
area_telhado_ex4 = area_planta / math.cos(math.radians(inclinacao))
# Não dá para comprar um pedaço de telha: arredondamos para cima.
telhas_ex4 = math.ceil(area_telhado_ex4 * telhas_por_m2)

print(area_telhado_ex4, telhas_ex4)  # 92.376... 1479
```

**Exercício 5.** Medida do diedro de aresta $BD$ formado pela base (lado de $A$) e pelo plano $BDE$.

```python
aresta = 2

# Seja O o ponto médio de BD (o centro da base). Na base, AO ⊥ BD (as diagonais do
# quadrado são perpendiculares). Pelas três perpendiculares (notebook 03), como
# AE ⊥ base e AO ⊥ BD, temos EO ⊥ BD. Então o plano AOE é perpendicular à aresta
# BD, e o ângulo AÔE é a seção reta do diedro.
ao = math.sqrt(aresta**2 + aresta**2) / 2  # metade da diagonal da base: √2
ae = aresta

# No triângulo AOE, retângulo em A: tan θ = AE / AO = 2 / √2 = √2.
diedro_ex5 = math.degrees(math.atan(ae / ao))

print(diedro_ex5)  # 54.7356...

# Conferência com o utils_3d (que usa vetores, assunto do módulo 07).
print(round(u3d.medir_angulo_do_diedro(B, D, A, E), 4))
```

**Exercício 7.** Os planos $ABGH$ e $ADGF$ contêm a diagonal $AG$: (a) a medida do diedro de aresta $AG$ com faces por $B$ e por $D$; (b) o ângulo entre os planos $ABGH$ e $ADGF$.

```python
# (a) No notebook 03 vimos que AG ⊥ BDE, e AG fura BDE num ponto P. Como o plano
# BDE é perpendicular à aresta AG e B, D estão nas faces, o ângulo BPD é a seção
# reta do diedro.
# O triângulo BDE é equilátero: BD, DE e EB são diagonais de faces (2√2 cada).
# A é equidistante de B, D e E (AB = AD = AE = 2), e G também (GB = GD = GE = 2√2):
# a reta AG é o lugar dos pontos equidistantes de B, D e E (notebook 04), então P
# é o circuncentro do triângulo equilátero, que coincide com o seu centro.
# No centro de um triângulo equilátero, os segmentos até os vértices dividem a
# volta completa em três ângulos iguais: BPD = 360° / 3 = 120°.
diedro_ex7 = 360 / 3

# (b) Os planos ABGH e ADGF formam diedros de 120° e de 180° − 120° = 60°. O ângulo
# entre eles é o menor: 60°.
angulo_planos_ex7 = min(diedro_ex7, 180 - diedro_ex7)

print(diedro_ex7, angulo_planos_ex7)  # 120.0 60.0

# Conferência com o utils_3d (que usa vetores, assunto do módulo 07).
P = u3d.intersecao_reta_plano(A, G, plano_pelo_nome("BDE"))
print("P =", tuple(round(c, 4) for c in P))
print(f"Ângulo BPD: {u3d.medir_angulo(P, B, D):.1f}°")
print(f"Diedro de aresta AG: {u3d.medir_angulo_do_diedro(A, G, B, D):.1f}°")
angulo = u3d.medir_angulo_entre_planos(plano_pelo_nome("ABGH"), plano_pelo_nome("ADGF"))
print(f"Ângulo entre ABGH e ADGF: {angulo:.1f}°")

fig = u3d.criar_figura(titulo="A seção reta BPD do diedro de aresta AG")
fig = u3d.desenhar_cubo(fig)
# Dentro do cubo, os dois semiplanos de origem AG aparecem como os triângulos ABG e
# ADG.
fig = u3d.desenhar_poligono(fig, [A, B, G], cor=u3d.CORES["plano"], opacidade=0.35)
fig = u3d.desenhar_poligono(
    fig, [A, D, G], cor=u3d.CORES["plano_secundario"], opacidade=0.35
)
fig = u3d.desenhar_segmento(fig, A, G, cor=u3d.CORES["destaque"], espessura=6)
fig = u3d.desenhar_poligono(
    fig, [B, D, E], cor=u3d.CORES["plano_terciario"], opacidade=0.25
)
fig = u3d.desenhar_segmento(fig, P, B, cor=u3d.CORES["reta"], espessura=5)
fig = u3d.desenhar_segmento(fig, P, D, cor=u3d.CORES["reta"], espessura=5)
fig = u3d.desenhar_angulo(fig, P, B, D, rotulo="120°", raio=0.4)
fig = u3d.desenhar_ponto(fig, P, rotulo="P", cor=u3d.CORES["destaque"])
u3d.mostrar(fig)

# O plano ACGE também contém AG. Os três planos (ABGH, ADGF e ACGE) dividem o
# espaço em volta de AG em seis diedros de 60°.
for nome1, nome2 in [("ABGH", "ACGE"), ("ACGE", "ADGF")]:
    plano1, plano2 = plano_pelo_nome(nome1), plano_pelo_nome(nome2)
    angulo = u3d.medir_angulo_entre_planos(plano1, plano2)
    print(f"Ângulo entre {nome1} e {nome2}: {angulo:.1f}°")
```

**Exercício 8.** Para um diedro reto, qual inclinação $\varphi$ dá uma seção igualmente inclinada de 60°? Encontrar com um laço de 1° a 90°.

```python
diedro = u3d.vertices_do_diedro(90)
a, b = diedro["aresta_inicio"], diedro["aresta_fim"]
p1, p2 = diedro["ponto_face1"], diedro["ponto_face2"]
vertice = (0, 0, 0)  # um ponto da aresta

melhor_inclinacao_ex8 = None
menor_diferenca = 1000  # começa bem grande: qualquer diferença real será menor

for inclinacao in range(1, 91):
    secao = u3d.secao_igualmente_inclinada(a, b, p1, p2, vertice, inclinacao)
    angulo = u3d.medir_angulo(*secao)
    # Guardamos a inclinação sempre que encontramos uma seção mais perto de 60°.
    if abs(angulo - 60) < menor_diferenca:
        menor_diferenca = abs(angulo - 60)
        melhor_inclinacao_ex8 = inclinacao

print(melhor_inclinacao_ex8, menor_diferenca)  # 45 e uma diferença quase zero

# De onde vem o 45°? Para um diedro de abertura θ e inclinação φ, a seção σ
# satisfaz cos σ = cos²φ + sen²φ · cos θ. Com θ = 90°, cos σ = cos²φ; para σ = 60°,
# cos²φ = 1/2, ou seja, cos φ = √2/2 e φ = 45°.
phi = math.radians(45)
print(math.degrees(math.acos(math.cos(phi) ** 2)))  # 60.0
```
