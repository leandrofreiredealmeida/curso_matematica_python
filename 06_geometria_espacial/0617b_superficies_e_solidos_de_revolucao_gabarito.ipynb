O lado que fica no eixo é a altura do cilindro; o outro lado é o raio.

```python
volume_lado_maior_ex2 = math.pi * 2**2 * 5   # raio 2, altura 5: 20π
volume_lado_menor_ex2 = math.pi * 5**2 * 2   # raio 5, altura 2: 50π

print(volume_lado_maior_ex2, volume_lado_menor_ex2)

# Conferência por Pappus: a área é 10 nos dois casos, mas o centro do retângulo fica
# a 1 do eixo no item (a) e a 2,5 do eixo no item (b).
print(10 * 2 * math.pi * 1, 10 * 2 * math.pi * 2.5)
```

**Exercício 3.** O toro com $a = 1$ e $b = 3$: (a) volume e (b) área.

Por Pappus: a região é o círculo (área $\pi a^2$) e a curva é a circunferência (comprimento $2\pi a$); os dois centroides são o centro, que percorre $2\pi b$.

```python
a, b = 1, 3
volume_toro_ex3 = math.pi * a**2 * 2 * math.pi * b   # = 2π²a²b = 6π² ≈ 59,22
area_toro_ex3 = 2 * math.pi * a * 2 * math.pi * b    # = 4π²ab = 12π² ≈ 118,44

print(volume_toro_ex3, area_toro_ex3)
```

## Intermediário

**Exercício 4.** O pião: o triângulo de catetos 3 e 4 girando em torno da hipotenusa, (a) pela soma dos dois cones e (b) por Pappus.

Relações métricas (módulo 04): hipotenusa $c = 5$, altura relativa à hipotenusa $h = \frac{3 \cdot 4}{5} = 2{,}4$ (o raio comum) e projeções $m = \frac{9}{5} = 1{,}8$ e $n = \frac{16}{5} = 3{,}2$ (as alturas dos dois cones).

```python
a, b = 3, 4
c = math.sqrt(a**2 + b**2)   # 5
h = a * b / c                # 2,4: o raio da base comum
m = a**2 / c                 # 1,8: altura do primeiro cone
n = b**2 / c                 # 3,2: altura do segundo cone (m + n = c)

# (a) Dois cones de mesmo raio h: (1/3)πh²m + (1/3)πh²n = (1/3)πh²(m + n).
volume_cones_ex4 = math.pi * h**2 * m / 3 + math.pi * h**2 * n / 3

# (b) Pappus: área do triângulo = 6; o baricentro fica a 1/3 da altura relativa à
# hipotenusa, medida a partir dela. Como a hipotenusa está no eixo, d = h / 3 = 0,8.
area = a * b / 2
d = h / 3
volume_pappus_ex4 = area * 2 * math.pi * d

print(volume_cones_ex4, volume_pappus_ex4, 9.6 * math.pi)   # os três valem 9,6π
```

**Exercício 6.** O centroide de um semicírculo de raio 3, por Pappus ao contrário.

O semicírculo, girando em torno do diâmetro, gera a esfera de raio $R$: $\frac{4}{3}\pi R^3 = \frac{\pi R^2}{2} \cdot 2\pi d$, então $d = \frac{4R}{3\pi}$.

```python
R = 3
volume_esfera = 4 / 3 * math.pi * R**3
area_semicirculo = math.pi * R**2 / 2
d_ex6 = volume_esfera / (area_semicirculo * 2 * math.pi)   # = 4R / (3π) = 4/π

print(d_ex6, 4 / math.pi)   # ≈ 1,273

# Conferência com o utils_3d (o semicírculo como um polígono de muitos lados):
semicirculo = [(R * math.cos(t), R * math.sin(t)) for t in np.linspace(0, math.pi, 2000)]
print(u3d.medir_area_e_centroide_2d(semicirculo)[1][1])
```

## Desafio

**Exercício 7.** O menor número de discos que aproxima o volume da esfera de raio 1 com erro relativo abaixo de 1%.

Na altura $z$, o raio do disco é $\sqrt{1 - z^2}$ (Pitágoras na seção meridiana).

```python
exato = 4 * math.pi / 3
N = 0
erro_relativo = 1
while erro_relativo >= 0.01 and N < 1000:
    N = N + 1
    dz = 2 / N
    soma = 0
    for i in range(N):
        z_meio = -1 + (i + 0.5) * dz            # o meio da fatia i
        soma = soma + math.pi * (1 - z_meio**2) * dz   # π r² dz, com r² = 1 − z²
    erro_relativo = (soma - exato) / exato

menor_N_ex7 = N

print(menor_N_ex7, f"erro relativo = {100 * erro_relativo:.2f}%")

# Com N = 7, o erro ainda passa de 1%; com N = 8, fica em cerca de 0,78%.
```

**Exercício 8.** O volume do vaso de contorno $r(z) = 2 + \operatorname{sen} z$, com $z$ de $0$ a $2\pi$, usando 1000 discos.

```python
numero_de_discos = 1000
z_inicial, z_final = 0, 2 * math.pi
dz = (z_final - z_inicial) / numero_de_discos
soma = 0
for i in range(numero_de_discos):
    z_meio = z_inicial + (i + 0.5) * dz
    raio = 2 + math.sin(z_meio)
    soma = soma + math.pi * raio**2 * dz

volume_ex8 = soma

print(volume_ex8, 9 * math.pi**2)   # ≈ 88,83 cm³: o valor exato é 9π²
```

Os dois desafios usam a mesma ideia: **somar muitas fatias finas**. No primeiro, você viu que o erro cai depressa quando o número de discos aumenta; no segundo, calculou um volume sem nenhuma fórmula pronta. No módulo 11 (Limite, Integral e Derivada), você vai ver o que acontece quando o número de discos cresce sem parar, e como obter o valor exato $9\pi^2$ do vaso com uma conta curta.
