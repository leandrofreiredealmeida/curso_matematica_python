**Exercício 2.** Num cubo de aresta 3: (a) a diagonal de uma face; (b) a diagonal do cubo; (c) a área total.

```python
a = 3
diagonal_face_ex2 = a * math.sqrt(2)  # (a) Pitágoras na face: √(a² + a²) = a√2 ≈ 4,243
diagonal_cubo_ex2 = a * math.sqrt(3)  # (b) √(a² + a² + a²) = a√3 ≈ 5,196
area_ex2 = 6 * a**2  # (c) seis quadrados de lado 3: 54

print(diagonal_face_ex2, diagonal_cubo_ex2, area_ex2)
```

**Exercício 3.** Paralelepípedo retângulo de dimensões 3, 4 e 12: (a) a diagonal; (b) a área total.

```python
a, b, c = 3, 4, 12
diagonal_ex3 = math.sqrt(a**2 + b**2 + c**2)  # (a) √(9 + 16 + 144) = √169 = 13
area_ex3 = 2 * (a * b + a * c + b * c)  # (b) 2(12 + 36 + 48) = 192

print(diagonal_ex3, area_ex3)

# Conferência com o utils_3d.
v = u3d.vertices_do_paralelepipedo(3, 4, 12)
print(u3d.medir_distancia(v["A"], v["G"]), u3d.medir_area_total(*u3d.paralelepipedo_retangulo(3, 4, 12)))
```

## Intermediário

**Exercício 4.** Caixa de 1,0 m × 0,8 m × 0,6 m: (a) a diagonal; (b) cabe uma vassoura de 1,5 m? (c) e uma de 1,4 m?

```python
# (a) d = √(1,0² + 0,8² + 0,6²) = √(1 + 0,64 + 0,36) = √2 ≈ 1,414 m.
diagonal_ex4 = math.sqrt(1.0**2 + 0.8**2 + 0.6**2)
# (b) e (c): a vassoura cabe quando não é maior que a diagonal, o maior segmento
#     que cabe dentro da caixa.
cabe_15_ex4 = 1.5 <= diagonal_ex4  # False: 1,5 > 1,414
cabe_14_ex4 = 1.4 <= diagonal_ex4  # True: 1,4 < 1,414

print(diagonal_ex4, cabe_15_ex4, cabe_14_ex4)
```

**Exercício 5.** Prisma hexagonal regular de aresta da base 2 cm e altura 5 cm: (a) área lateral; (b) área de uma base; (c) área total.

```python
lado, altura = 2, 5
# (a) Seis retângulos 2 × 5, ou 2p · h com 2p = 6 · 2 = 12: 12 · 5 = 60.
area_lateral_ex5 = 6 * lado * altura
# (b) O hexágono regular é formado por 6 triângulos equiláteros de lado 2, cada um
#     com área ℓ²√3/4 = √3: a base tem área 6√3 ≈ 10,392.
area_base_ex5 = 6 * lado**2 * math.sqrt(3) / 4
# (c) S_t = S_ℓ + 2 S_b = 60 + 12√3 ≈ 80,785.
area_total_ex5 = area_lateral_ex5 + 2 * area_base_ex5

print(area_lateral_ex5, area_base_ex5, area_total_ex5)

# Conferência com o utils_3d.
prisma = u3d.prisma(u3d.poligono_regular(6, lado=2), (0, 0, 5))
print(u3d.medir_area_total(*prisma))
```

**Exercício 6.** Paralelepípedo retângulo com $a + b + c = 12$ e diagonal $\sqrt{50}$: a área total.

```python
# (a + b + c)² = d² + S, então S = (a + b + c)² − d² = 144 − 50 = 94.
soma = 12
diagonal = math.sqrt(50)
area_ex6 = soma**2 - diagonal**2

print(area_ex6)

# Um paralelepípedo assim: 3, 4 e 5 (3 + 4 + 5 = 12 e 9 + 16 + 25 = 50).
print(2 * (3 * 4 + 3 * 5 + 4 * 5))
```

**Exercício 7.** Cubo de aresta 2 cortado por um plano perpendicular a uma diagonal, pelo centro: (a) o lado do hexágono regular; (b) a sua área.

```python
# (a) Dois pontos médios vizinhos estão em arestas que saem de um mesmo vértice, a 1
#     unidade dele. Essas arestas são perpendiculares, então o lado do hexágono é a
#     hipotenusa de um triângulo retângulo de catetos 1 e 1: √2 ≈ 1,414.
lado_ex7 = math.sqrt(1**2 + 1**2)
# (b) O hexágono regular é formado por 6 triângulos equiláteros de lado √2, cada um
#     com área (√2)²√3/4 = √3/2. Total: 6 · √3/2 = 3√3 ≈ 5,196.
area_ex7 = 6 * lado_ex7**2 * math.sqrt(3) / 4

print(lado_ex7, area_ex7)

# Conferência com o utils_3d.
cubo = u3d.paralelepipedo_retangulo(2, 2, 2)
hexagono = u3d.secao_plana(*cubo, ((1, 1, 1), (1, 1, 1)))
print(u3d.medir_distancia(hexagono[0], hexagono[1]), u3d.medir_area_do_poligono(hexagono))
```

**Exercício 8.** Menor caminho sobre a superfície da caixa $3 \times 4 \times 12$, de $A$ até $G$.

```python
# Abrindo duas faces vizinhas, elas formam um retângulo, e o caminho mais curto é a
# diagonal dele. As três escolhas de faces dão retângulos (3 + 4) × 12,
# (3 + 12) × 4 e (4 + 12) × 3:
candidatos = [
    math.sqrt((3 + 4) ** 2 + 12**2),  # √193 ≈ 13,89
    math.sqrt((3 + 12) ** 2 + 4**2),  # √241 ≈ 15,52
    math.sqrt((4 + 12) ** 2 + 3**2),  # √265 ≈ 16,28
]
caminho_ex8 = min(candidatos)

print(candidatos)
print(caminho_ex8)
```

Por dentro (voando), o caminho seria a diagonal da caixa, 13. Por fora (andando), o menor caminho é $\sqrt{193} \approx 13{,}89$: ele cruza a aresta $BF$ a $\frac{36}{7} \approx 5{,}14$ de $B$, juntando as faces $ABFE$ e $BCGF$.
