## Básico

**Exercício 1.** Conversões: (a) 1 m³ em litros; (b) 2,5 L em cm³; (c) 1 cm³ em mm³; (d) 3000 L em m³.

```python
# Em volume, cada passo de unidade multiplica (ou divide) por 1000 = 10 × 10 × 10.
litros_ex1 = 1 * 1000  # (a) 1 m³ = 1000 dm³ = 1000 L
cm3_ex1 = 2.5 * 1000  # (b) 1 L = 1 dm³ = 1000 cm³, então 2,5 L = 2500 cm³
mm3_ex1 = 1 * 1000  # (c) 1 cm³ = 10 mm × 10 mm × 10 mm = 1000 mm³
m3_ex1 = 3000 / 1000  # (d) 1000 L = 1 m³, então 3000 L = 3 m³

print(litros_ex1, cm3_ex1, mm3_ex1, m3_ex1)
```

**Exercício 2.** Volumes: (a) cubo de aresta 3; (b) paralelepípedo retângulo $3 \times 4 \times 12$; (c) prisma hexagonal regular de aresta da base 2 e altura 5.

```python
volume_cubo_ex2 = 3**3  # (a) V = a³ = 27
volume_paralelepipedo_ex2 = 3 * 4 * 12  # (b) V = abc = 144

# (c) A base é um hexágono regular de lado 2: 6 triângulos equiláteros de lado 2,
#     cada um com área ℓ²√3 / 4.
area_base = 6 * 2**2 * math.sqrt(3) / 4  # = 6√3
volume_prisma_ex2 = area_base * 5  # V = S_b · h = 30√3 ≈ 51,96

print(volume_cubo_ex2, volume_paralelepipedo_ex2, volume_prisma_ex2)

# Conferência do item (c) com o utils_3d.
prisma_hexagonal = u3d.prisma(u3d.poligono_regular(6, lado=2), (0, 0, 5))
print(f"Conferência: {u3d.medir_volume(*prisma_hexagonal):.4f}")
```

**Exercício 3.** Verdadeiro ou falso: (a) dobrar a aresta de um cubo dobra o volume; (b) prismas com bases de mesma área e mesma altura têm o mesmo volume; (c) o prisma oblíquo tem volume menor que o reto de mesma base e mesma aresta lateral; (d) sólidos de mesmo volume têm seções de mesma área em todas as alturas.

```python
respostas_ex3 = {
    # (a) Falso: dobrando a aresta, as três dimensões dobram e o volume fica
    #     multiplicado por 2 × 2 × 2 = 8 (seção 2).
    "a": False,
    # (b) Verdadeiro: V = S_b · h vale para qualquer prisma, reto ou oblíquo, de
    #     qualquer base (seção 4).
    "b": True,
    # (c) Verdadeiro: no oblíquo, h = ℓ · sen θ < ℓ, que é a altura do reto
    #     (seção 4).
    "c": True,
    # (d) Falso: o Princípio de Cavalieri não vale a volta. Por exemplo, um cubo
    #     2 × 2 × 2 e uma caixa 4 × 2 × 1 deitada têm volume 8, mas as seções
    #     horizontais têm áreas 4 e 8 (seção 3).
    "d": False,
}
print(respostas_ex3)
```

## Intermediário

**Exercício 4.** Caixa d'água de $1{,}5 \text{ m} \times 1{,}2 \text{ m} \times 1{,}0 \text{ m}$: (a) a capacidade em litros; (b) quantos dias ela dura para 4 pessoas que gastam 150 L por dia cada.

```python
volume_m3 = 1.5 * 1.2 * 1.0  # 1,8 m³
litros_ex4 = volume_m3 * 1000  # (a) 1 m³ = 1000 L: 1800 L

gasto_por_dia = 4 * 150  # 600 L por dia
dias_ex4 = litros_ex4 / gasto_por_dia  # (b) 1800 / 600 = 3 dias

print(litros_ex4, dias_ex4)
```

**Exercício 5.** Aquário de fundo $60 \text{ cm} \times 30 \text{ cm}$; a pedra faz o nível subir 2 cm. Volume da pedra (a) em cm³ e (b) em litros.

```python
# A água que subiu forma um paralelepípedo com a base do fundo e a altura da subida:
# o volume dele é o volume da pedra.
volume_cm3_ex5 = 60 * 30 * 2  # (a) 3600 cm³
volume_litros_ex5 = volume_cm3_ex5 / 1000  # (b) 1 L = 1000 cm³: 3,6 L

print(volume_cm3_ex5, volume_litros_ex5)
```

**Exercício 6.** Casa de planta $8 \text{ m} \times 10 \text{ m}$, telhado com 30° de inclinação, chuva de 25 mm: quantos litros o telhado capta?

```python
# A chuva cai na vertical: o que importa é a área da planta (a projeção do telhado
# no chão), e não a área do telhado. A inclinação de 30° não entra na conta!
area_da_planta = 8 * 10  # 80 m²
litros_ex6 = area_da_planta * 25  # 1 mm sobre 1 m² = 1 L: 80 × 25 = 2000 L

print(litros_ex6)
```

**Exercício 7.** A aresta de um cubo aumenta 10%. Em quantos por cento aumenta o volume?

```python
# A nova aresta é 1,1a, então o novo volume é (1,1a)³ = 1,331a³: o volume ficou
# multiplicado por 1,331, isto é, aumentou 33,1% (e não 30%, nem 10%).
fator = 1.1**3
aumento_ex7 = (fator - 1) * 100

print(fator, aumento_ex7)
```

**Exercício 8.** Três faces de um paralelepípedo retângulo, com um vértice em comum, têm áreas 12 cm², 15 cm² e 20 cm². Qual é o volume?

```python
# Com dimensões a, b, c, as três faces têm áreas ab, ac e bc. Multiplicando:
#     (ab) · (ac) · (bc) = a² b² c² = (abc)² = V²
# Então V = √(12 · 15 · 20) = √3600 = 60, sem precisar descobrir a, b e c.
volume_ex8 = math.sqrt(12 * 15 * 20)

print(volume_ex8)

# Conferência com o utils_3d, usando as dimensões 3, 4 e 5.
print(f"Conferência: {u3d.medir_volume(*u3d.paralelepipedo_retangulo(3, 4, 5)):.4f}")
```

As dimensões são 3, 4 e 5. Confira: $3 \cdot 4 = 12$, $3 \cdot 5 = 15$, $4 \cdot 5 = 20$, e $3 \cdot 4 \cdot 5 = 60$.
