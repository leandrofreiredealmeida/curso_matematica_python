**Exercício 2.** Cone de raio 3 e altura 4: raios das esferas (a) inscrita e (b) circunscrita.

Seção meridiana: um triângulo isósceles de base 6, altura 4 e lados iguais a $g = 5$. A esfera inscrita vira a circunferência inscrita; a circunscrita, a circunferência circunscrita.

```python
R, H = 3, 4
g = math.sqrt(H**2 + R**2)  # 5
# (a) r = área / semiperímetro = (R · H) / (R + g) = 12 / 8
r_ex2 = R * H / (R + g)  # 1,5
# (b) (H − Rₑ)² + R² = Rₑ²  ⟹  Rₑ = g² / (2H) = 25 / 8
R_ex2 = g**2 / (2 * H)  # 3,125

print(r_ex2, R_ex2)

# Conferência com o utils_3d, no triângulo da seção meridiana.
print(u3d.medir_circunferencia_inscrita((-3, 0), (3, 0), (0, 4)))
print(u3d.medir_circunferencia_circunscrita((-3, 0), (3, 0), (0, 4)))
```

**Exercício 3.** Esfera de raio 2: volumes (a) da esfera, (b) do cilindro equilátero e (c) do cone equilátero circunscritos.

Seção meridiana: o círculo de raio 2 inscrito num quadrado de lado 4 (cilindro) e num triângulo equilátero de altura $3R = 6$ (cone).

```python
R = 2
volume_esfera_ex3 = 4 / 3 * math.pi * R**3  # (a) 32π/3 ≈ 33,51
volume_cilindro_ex3 = math.pi * R**2 * (2 * R)  # (b) raio R e altura 2R: 16π ≈ 50,27
# (c) No triângulo equilátero, o raio inscrito é 1/3 da altura: altura 3R = 6, raio R√3.
raio_do_cone = R * math.sqrt(3)
volume_cone_ex3 = math.pi * raio_do_cone**2 * (3 * R) / 3  # 24π ≈ 75,40

print(volume_esfera_ex3, volume_cilindro_ex3, volume_cone_ex3)
print("Razões × 4 / esfera:", [v / volume_esfera_ex3 * 4 for v in
      (volume_esfera_ex3, volume_cilindro_ex3, volume_cone_ex3)])  # 4, 6, 9
```

## Intermediário

**Exercício 4.** Prisma hexagonal regular de aresta da base 2 e altura 5: volumes dos cilindros (a) inscrito e (b) circunscrito.

Aqui não há seção meridiana: olhamos para a **base**. O hexágono regular de lado 2 tem a circunferência inscrita de raio igual ao apótema e a circunscrita de raio igual ao lado.

```python
lado, altura = 2, 5
m = lado * math.sqrt(3) / 2  # apótema do hexágono: altura de um triângulo equilátero, √3
R_base = lado  # no hexágono regular, o raio da circunscrita é igual ao lado

volume_inscrito_ex4 = math.pi * m**2 * altura  # (a) π · 3 · 5 = 15π ≈ 47,12
volume_circunscrito_ex4 = math.pi * R_base**2 * altura  # (b) π · 4 · 5 = 20π ≈ 62,83

print(volume_inscrito_ex4, volume_circunscrito_ex4)
```

**Exercício 5.** Cilindro de altura 6 inscrito numa esfera de raio 5: (a) raio e (b) volume.

Seção meridiana: um retângulo $2r \times 6$ inscrito num círculo de raio 5; a diagonal do retângulo é um diâmetro.

```python
R, h = 5, 6
r_ex5 = math.sqrt(R**2 - (h / 2) ** 2)  # (a) r² + 3² = 5², então r = 4
volume_ex5 = math.pi * r_ex5**2 * h  # (b) π · 16 · 6 = 96π ≈ 301,59

print(r_ex5, volume_ex5)
```

**Exercício 6.** Tronco de cone circunscrito a uma esfera, com raios das bases 9 e 4: (a) raio da esfera, (b) altura e (c) volume do tronco.

Seção meridiana: uma circunferência inscrita num trapézio isósceles de bases 18 e 8. Como o trapézio é circunscritível, $g = R + r = 13$.

```python
R, r = 9, 4
rho_ex6 = math.sqrt(R * r)  # (a) média geométrica dos raios: √36 = 6
altura_ex6 = 2 * rho_ex6  # (b) a altura é o diâmetro da esfera: 12
# (c) Volume do tronco de cone (notebook 14).
volume_ex6 = math.pi * altura_ex6 / 3 * (R**2 + R * r + r**2)  # 4π · 133 = 532π ≈ 1671,33

# Conferência: g = R + r = 13 e √(h² + (R − r)²) = √(144 + 25) = 13.
print(R + r, math.sqrt(altura_ex6**2 + (R - r) ** 2))
print(rho_ex6, altura_ex6, volume_ex6)
```

## Desafio

**Exercício 7.** O maior cilindro inscrito num cone de raio 3 e altura 6: (a) a altura de maior volume e (b) a razão entre o volume desse cilindro e o do cone.

Seção meridiana: um retângulo dentro de um triângulo; a semelhança dá $r = 3\left(1 - \frac{h}{6}\right)$.

```python
R_cone, H_cone = 3, 6
alturas = [k / 100 for k in range(1, 600)]   # 0,01; 0,02; ...; 5,99
volumes = []
for h in alturas:
    r = R_cone * (1 - h / H_cone)  # semelhança: r / R = (H − h) / H
    volumes.append(math.pi * r**2 * h)

maior_volume = max(volumes)
melhor_altura_ex7 = alturas[volumes.index(maior_volume)]  # (a) 2, um terço da altura
volume_cone = math.pi * R_cone**2 * H_cone / 3  # 18π
razao_ex7 = maior_volume / volume_cone  # (b) 8π / 18π = 4/9 ≈ 0,444

print(melhor_altura_ex7, razao_ex7)

# O gráfico do volume em função da altura, com o máximo marcado.
grafico = u3d.criar_grafico(
    titulo="Cilindro inscrito no cone de raio 3 e altura 6",
    rotulo_x="altura h do cilindro", rotulo_y="volume do cilindro",
)
grafico = u3d.desenhar_curva(grafico, alturas, volumes)
grafico = u3d.marcar_ponto_no_grafico(grafico, melhor_altura_ex7, maior_volume, rotulo="máximo")
u3d.mostrar(grafico)
```

**Exercício 8.** O maior cilindro inscrito numa esfera de raio 1.

Seção meridiana: um retângulo inscrito num círculo; $r^2 = 1 - \left(\frac{h}{2}\right)^2$.

```python
alturas = [k / 1000 for k in range(1, 2000)]   # 0,001; 0,002; ...; 1,999
volumes = []
for h in alturas:
    r_quadrado = 1 - (h / 2) ** 2  # Pitágoras na seção meridiana
    volumes.append(math.pi * r_quadrado * h)

melhor_altura_ex8 = alturas[volumes.index(max(volumes))]  # ≈ 1,155

print(melhor_altura_ex8, 2 / math.sqrt(3))  # o valor exato é 2/√3

# Nesse cilindro, r = √(1 − 1/3) = √(2/3), e h = r√2 (não 2r: ele não é equilátero).
r_otimo = math.sqrt(1 - (melhor_altura_ex8 / 2) ** 2)
print(f"r = {r_otimo:.4f};  h / r = {melhor_altura_ex8 / r_otimo:.4f} (√2 ≈ {math.sqrt(2):.4f})")
```

Os dois desafios são **problemas de máximo**: procurar, entre infinitas possibilidades, a que dá o maior valor. Aqui fizemos isso "na força bruta", testando milhares de alturas.
