**Exercício 2.** Razões de semelhança: (a) áreas e (b) volumes com $k = 3$; (c) áreas e (d) volumes numa maquete 1:100.

```python
k = 3
razao_areas_ex2 = k**2  # (a) 9
razao_volumes_ex2 = k**3  # (b) 27

k_maquete = 100  # cada comprimento real é 100 vezes o da maquete
razao_areas_maquete_ex2 = k_maquete**2  # (c) 10 000
razao_volumes_maquete_ex2 = k_maquete**3  # (d) 1 000 000

print(razao_areas_ex2, razao_volumes_ex2, razao_areas_maquete_ex2, razao_volumes_maquete_ex2)
```

**Exercício 3.** Pirâmide de altura 12 e volume 216, cortada a 4 do vértice: (a) volume da pirâmide pequena; (b) volume do tronco.

```python
k = 4 / 12  # razão de semelhança da pirâmide pequena: d / H = 1/3

volume_pequena_ex3 = 216 * k**3  # (a) 216 / 27 = 8
volume_tronco_ex3 = 216 - volume_pequena_ex3  # (b) 216 − 8 = 208

print(volume_pequena_ex3, volume_tronco_ex3)
```

## Intermediário

**Exercício 4.** Tronco de pirâmide quadrangular regular com bases de lados 6 e 3 e altura 4: (a) volume; (b) apótema do tronco; (c) área lateral; (d) área total.

```python
lado_maior, lado_menor, h = 6, 3, 4
B = lado_maior**2  # 36
b = lado_menor**2  # 9

volume_ex4 = h / 3 * (B + math.sqrt(B * b) + b)  # (a) 4/3 · (36 + 18 + 9) = 84

# (b) Os apótemas das bases quadradas são metade dos lados: m = 3 e m' = 1,5.
m, m_linha = lado_maior / 2, lado_menor / 2
apotema_ex4 = math.sqrt(h**2 + (m - m_linha) ** 2)  # √(16 + 2,25) = √18,25 ≈ 4,272

# (c) Semiperímetros: p = 4 · 6 / 2 = 12 e p' = 4 · 3 / 2 = 6.
p, p_linha = 4 * lado_maior / 2, 4 * lado_menor / 2
area_lateral_ex4 = (p + p_linha) * apotema_ex4  # 18 · 4,272 ≈ 76,90

area_total_ex4 = area_lateral_ex4 + B + b  # (d) 76,90 + 45 ≈ 121,90

print(volume_ex4, apotema_ex4, area_lateral_ex4, area_total_ex4)

# Conferência com o utils_3d.
tronco = u3d.tronco_de_piramide_regular(4, 6, 3, 4)
print(f"Conferência: volume {u3d.medir_volume(*tronco):.4f}, "
      f"área total {u3d.medir_area_total(*tronco):.4f}")
```

**Exercício 5.** Copo em forma de tronco de cone (raios 4 cm e 3 cm, altura 10 cm): (a) volume; (b) "volume errado" com a média das áreas das bases.

```python
R, r, h = 4, 3, 10

volume_ex5 = math.pi * h / 3 * (R**2 + R * r + r**2)  # (a) 10π/3 · 37 = 370π/3 ≈ 387,46 mL
# (b) A média das bases é (16π + 9π) / 2 = 12,5π; vezes 10 dá 125π ≈ 392,70.
volume_errado_ex5 = (math.pi * R**2 + math.pi * r**2) / 2 * h

print(volume_ex5, volume_errado_ex5)
print(f"O erro seria de {volume_errado_ex5 - volume_ex5:.2f} mL a mais.")
```

**Exercício 6.** Cano de raio 2 cm cortado em bisel, com geratrizes de 5 cm e 9 cm: (a) volume; (b) área lateral.

```python
raio, g_m, g_M = 2, 5, 9
media = (g_M + g_m) / 2  # o eixo do tronco mede 7 cm

volume_ex6 = math.pi * raio**2 * media  # (a) 4π · 7 = 28π ≈ 87,96 cm³
area_lateral_ex6 = 2 * math.pi * raio * media  # (b) 4π · 7 = 28π ≈ 87,96 cm²

print(volume_ex6, area_lateral_ex6)
```

**Exercício 8.** Baldes de pipoca semelhantes: o pequeno (altura 10 cm, raios 4 cm e 3 cm) custa R\$ 10,00, e o grande (altura 15 cm) custa R\$ 25,00. (a) Volume do balde grande; (b) qual é o mais vantajoso.

```python
volume_pequeno = math.pi * 10 / 3 * (4**2 + 4 * 3 + 3**2)  # 370π/3 ≈ 387,46 cm³

k = 15 / 10  # os baldes são semelhantes com razão 1,5
volume_grande_ex8 = volume_pequeno * k**3  # (a) 387,46 · 3,375 ≈ 1307,68 cm³

# (b) Preço por cm³ de cada balde: quanto menor, mais vantajoso.
preco_por_cm3_pequeno = 10 / volume_pequeno
preco_por_cm3_grande = 25 / volume_grande_ex8
print(f"Pequeno: R$ {preco_por_cm3_pequeno:.4f} por cm³")
print(f"Grande: R$ {preco_por_cm3_grande:.4f} por cm³")
if preco_por_cm3_grande < preco_por_cm3_pequeno:
    mais_vantajoso_ex8 = "grande"
else:
    mais_vantajoso_ex8 = "pequeno"

print(volume_grande_ex8, mais_vantajoso_ex8)
```

O balde grande custa 2,5 vezes mais, mas tem $1{,}5^3 = 3{,}375$ vezes mais pipoca! É o mesmo raciocínio dos coelhos de chocolate do início do notebook: quem compara só a altura (ou o "tamanho") subestima muito a diferença de volume.
