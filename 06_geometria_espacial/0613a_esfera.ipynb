## Por que as bolhas de sabão são redondas?

Sopre uma bolha de sabão: ela é sempre redonda, mesmo que o aro tenha formato de estrela. Gotas de chuva pequenas e gotas de óleo na água também ficam redondas. Por quê?

A película de sabão age como uma membrana elástica que tenta **encolher** o máximo possível. Para guardar uma certa quantidade de ar, ela procura o formato com a **menor área**. E esse formato é a esfera.

No notebook 11, você descobriu que, entre os cilindros, o equilátero é o que gasta menos material. Vamos comparar agora um **cubo**, um **cilindro equilátero** e uma **esfera** com o **mesmo volume**. (As fórmulas da esfera serão explicadas nas seções 3 e 4. Por enquanto, confie nelas!)

🕹️ Escolha o volume no widget abaixo.

```python
def comparar_solidos(volume: int) -> None:
    """Compara as áreas de um cubo, de um cilindro equilátero e de uma esfera com o
    mesmo volume."""
    aresta_do_cubo = volume ** (1 / 3)
    # Cilindro equilátero: volume = 2πr³.
    raio_do_cilindro = (volume / (2 * math.pi)) ** (1 / 3)
    # Esfera: volume = (4/3)πR³.
    raio_da_esfera = (3 * volume / (4 * math.pi)) ** (1 / 3)

    area_cubo = 6 * aresta_do_cubo**2
    area_cilindro = 6 * math.pi * raio_do_cilindro**2
    area_esfera = 4 * math.pi * raio_da_esfera**2

    # Os três sólidos lado a lado, apoiados no mesmo plano, com 2 cm entre eles.
    meio = aresta_do_cubo / 2
    x_do_cilindro = aresta_do_cubo + 2 + raio_do_cilindro
    x_da_esfera = x_do_cilindro + raio_do_cilindro + 2 + raio_da_esfera

    fig = u3d.criar_figura(titulo="Mesmo volume, áreas diferentes")
    cubo = u3d.paralelepipedo_retangulo(aresta_do_cubo, aresta_do_cubo, aresta_do_cubo)
    fig = u3d.desenhar_poliedro(fig, *cubo, cor=u3d.CORES["plano"], opacidade=0.5)
    fig = u3d.desenhar_cilindro(
        fig, (x_do_cilindro, meio, 0), raio_do_cilindro, (0, 0, 2 * raio_do_cilindro),
        cor=u3d.CORES["plano_secundario"], opacidade=0.5,
    )
    fig = u3d.desenhar_esfera(
        fig, (x_da_esfera, meio, raio_da_esfera), raio_da_esfera,
        cor=u3d.CORES["plano_terciario"], opacidade=0.5,
    )
    # Os nomes ficam acima de cada sólido.
    fig = u3d.adicionar_rotulo(fig, (meio, meio, aresta_do_cubo + 1.5), "cubo")
    fig = u3d.adicionar_rotulo(
        fig, (x_do_cilindro, meio, 2 * raio_do_cilindro + 1.5), "cilindro"
    )
    fig = u3d.adicionar_rotulo(
        fig, (x_da_esfera, meio, 2 * raio_da_esfera + 1.5), "esfera"
    )
    fig = u3d.definir_camera(fig, olho=(0.3, -2.0, 0.6))
    u3d.mostrar(fig)

    print(f"{'Sólido':<22}{'Área (cm²)':>12}{'% do cubo':>12}")
    solidos = [
        ("Cubo", area_cubo),
        ("Cilindro equilátero", area_cilindro),
        ("Esfera", area_esfera),
    ]
    for nome, area in solidos:
        print(f"{nome:<22}{area:>12.1f}{area / area_cubo:>12.1%}")
    print()
    print("Para o mesmo volume, a esfera é sempre a que tem a menor área.")


interact(
    comparar_solidos,
    volume=widgets.IntSlider(
        min=100, max=2000, step=100, value=1000, description="volume (cm³)",
        continuous_update=False,
    ),
);
```

A esfera é a campeã da economia: nenhum outro sólido guarda o mesmo volume com menos área. É por isso que bolhas e gotas são redondas, e é por isso que tantos animais se enrolam "em bola" para se proteger do frio (menos área exposta, menos calor perdido).

## 1. Superfície esférica e esfera

> 📌 **Definição.**
> - A **superfície esférica** de centro $O$ e raio $R$ é o conjunto dos pontos do espaço que estão a uma distância **igual** a $R$ de $O$ (é o lugar geométrico do notebook 04).
> - A **esfera** é o sólido formado pelos pontos a uma distância **menor ou igual** a $R$ de $O$.

A superfície esférica é a "casca"; a esfera é a "bola cheia". Uma bola de futebol murcha lembra uma superfície esférica; uma bola de gude ou uma laranja lembram esferas.

### Elementos

Escolha uma reta que passa pelo centro: ela é o **eixo**. Na Terra, o eixo é a reta em torno da qual ela gira.
- **Polos:** os dois pontos onde o eixo fura a superfície (polo norte e polo sul).
- **Equador:** a seção feita pelo plano **perpendicular ao eixo** que passa pelo **centro**.
- **Paralelos:** as seções feitas por planos **paralelos ao equador**.
- **Meridianos:** as semicircunferências que vão de um polo ao outro, contidas em planos que contêm o eixo.
- **Círculo máximo:** qualquer seção feita por um plano que passa pelo **centro**. Seu raio é $R$, o maior possível. O equador é um círculo máximo; os paralelos (exceto o equador) não são.

> 🕵️ **Curiosidade histórica.** Por volta de 240 a.C., o grego **Eratóstenes** mediu o tamanho da Terra sem sair do Egito. Ele sabia que, ao meio-dia do solstício de verão, o sol iluminava o fundo de um poço na cidade de Siena (estava a pino), enquanto em Alexandria, mais ao norte, um bastão vertical fazia sombra com um ângulo de cerca de 7,2°, ou $\frac{1}{50}$ da volta. Como as duas cidades ficam aproximadamente no mesmo meridiano, a distância entre elas devia ser $\frac{1}{50}$ do comprimento de um meridiano completo. O resultado ficou surpreendentemente próximo do valor atual.

🕹️ Escolha um paralelo e um meridiano no widget.

```python
def explorar_elementos(latitude: int, longitude: int, mostrar_grade: bool) -> None:
    """Mostra o eixo, os polos, o equador e o paralelo e o meridiano escolhidos."""
    raio = 2
    fig = u3d.criar_figura(titulo="Os elementos da esfera")
    fig = u3d.desenhar_esfera(fig, (0, 0, 0), raio, opacidade=0.2)

    # O eixo passa pelo centro e fura a superfície nos dois polos.
    fig = u3d.desenhar_segmento(
        fig, (0, 0, -1.3 * raio), (0, 0, 1.3 * raio), cor=u3d.CORES["apoio"],
        espessura=3, tracejado=True,
    )
    fig = u3d.desenhar_ponto(fig, u3d.ponto_na_esfera(90, 0, raio), rotulo="N")
    fig = u3d.desenhar_ponto(fig, u3d.ponto_na_esfera(-90, 0, raio), rotulo="S")

    if mostrar_grade:
        # Uma grade discreta: paralelos e meridianos de 30° em 30°.
        for lat in [-60, -30, 30, 60]:
            fig = u3d.desenhar_paralelo(fig, lat, raio, espessura=1)
        for lon in range(0, 360, 30):
            fig = u3d.desenhar_meridiano(fig, lon, raio, espessura=1)

    fig = u3d.desenhar_paralelo(fig, 0, raio, cor=u3d.CORES["plano"], espessura=5)
    fig = u3d.adicionar_rotulo(
        fig, u3d.ponto_na_esfera(0, -60, 1.2 * raio), "equador", cor=u3d.CORES["plano"]
    )
    fig = u3d.desenhar_paralelo(fig, latitude, raio, cor=u3d.CORES["reta"], espessura=5)
    fig = u3d.desenhar_meridiano(
        fig, longitude, raio, cor=u3d.CORES["reta_secundaria"], espessura=5
    )
    # O paralelo e o meridiano escolhidos se encontram num único ponto.
    encontro = u3d.ponto_na_esfera(latitude, longitude, raio)
    fig = u3d.desenhar_ponto(fig, encontro, cor=u3d.CORES["destaque"], tamanho=9)
    u3d.mostrar(fig)

    if latitude > 0:
        lado = "ao norte do equador"
    elif latitude < 0:
        lado = "ao sul do equador"
    else:
        lado = "sobre o equador"
    print(f"Paralelo (azul): latitude {abs(latitude)}°, {lado}")
    print(f"Meridiano (laranja): longitude {longitude}°")
    print("O ponto vermelho é o encontro dos dois.")


interact(
    explorar_elementos,
    latitude=widgets.IntSlider(
        min=-80, max=80, step=10, value=40, description="latitude (°)",
        continuous_update=False,
    ),
    longitude=widgets.IntSlider(
        min=-180, max=180, step=15, value=30, description="longitude (°)",
        continuous_update=False,
    ),
    mostrar_grade=widgets.Checkbox(value=True, description="Mostrar a grade completa"),
);
```

## 2. Seções da esfera e posições relativas

### O raio de uma seção

Corte uma esfera de raio $R$ por um plano a uma distância $d$ do centro ($d < R$). Seja $O'$ o pé da perpendicular de $O$ ao plano e $P$ um ponto da borda da seção. O segmento $OO'$ é perpendicular ao plano (notebook 03), então o triângulo $OO'P$ é retângulo em $O'$, com hipotenusa $OP = R$:
$$\boxed{r^2 = R^2 - d^2}$$

Toda seção plana de uma esfera é um **círculo** de centro $O'$ e raio $r = \sqrt{R^2 - d^2}$. Quanto mais longe do centro, menor a seção.

### Posições relativas entre plano e esfera

| Distância do centro ao plano | Posição | Interseção |
|---|---|---|
| $d < R$ | **secante** | um círculo |
| $d = R$ | **tangente** | um único ponto |
| $d > R$ | **exterior** | vazia |

> 📌 **Propriedade.** O plano tangente a uma esfera é **perpendicular ao raio** no ponto de tangência. (Uma bola apoiada numa mesa toca a mesa num único ponto, bem embaixo do centro.)

Da mesma forma, uma **reta** pode ser secante (dois pontos em comum), tangente (um ponto) ou exterior à esfera, conforme a distância do centro à reta seja menor, igual ou maior que $R$.

**No dia a dia:** fatias de uma laranja ou de um tomate cortados em rodelas (as do meio são maiores), a bola de bilhar apoiada na mesa.

🕹️ Mova o plano de corte.

```python
def cortar_esfera(raio_da_esfera: float, distancia: float, direcao_do_plano: str) -> None:
    """Corta a esfera por um plano a uma distância dada do centro O."""
    if distancia < raio_da_esfera:
        raio_da_secao = math.sqrt(raio_da_esfera**2 - distancia**2)
        posicao = "secante"
    elif distancia == raio_da_esfera:
        raio_da_secao = 0
        posicao = "tangente"
    else:
        raio_da_secao = None
        posicao = "exterior"

    # O vetor normal (perpendicular ao plano) tem comprimento 1. Andando `distancia`
    # nessa direção a partir do centro O, chegamos ao pé da perpendicular O'.
    if direcao_do_plano == "Horizontal":
        normal = (0, 0, 1)
    else:
        normal = (math.sqrt(2) / 2, 0, math.sqrt(2) / 2)  # inclinado 45°
    centro_da_secao = tuple(distancia * n for n in normal)
    plano = (centro_da_secao, normal)

    fig = u3d.criar_figura(titulo=f"Plano {posicao}")
    fig = u3d.desenhar_esfera(fig, (0, 0, 0), raio_da_esfera, opacidade=0.2)
    fig = u3d.desenhar_plano(
        fig, centro_da_secao, normal, opacidade=0.15, tamanho=2 * raio_da_esfera + 2
    )
    fig = u3d.desenhar_ponto(fig, (0, 0, 0), rotulo="O")
    if posicao == "secante":
        fig = u3d.desenhar_circunferencia(
            fig, centro_da_secao, raio_da_secao, normal=normal,
            cor=u3d.CORES["destaque"], espessura=6,
        )
        if distancia > 0:
            # Um ponto P da borda da seção (o primeiro vértice de um quadrado
            # inscrito nela) e o triângulo retângulo OO'P.
            ponto_p = u3d.poligono_regular(
                4, raio=raio_da_secao, centro=centro_da_secao, normal=normal
            )[0]
            fig = u3d.desenhar_poligono(
                fig, [(0, 0, 0), centro_da_secao, ponto_p],
                cor=u3d.CORES["plano_secundario"], opacidade=0.4,
            )
            fig = u3d.desenhar_angulo_reto(
                fig, centro_da_secao, (0, 0, 0), ponto_p,
                tamanho=0.2 * min(distancia, raio_da_secao),
            )
            fig = u3d.desenhar_ponto(fig, centro_da_secao, rotulo="O'")
            fig = u3d.desenhar_ponto(fig, ponto_p, rotulo="P")
    elif posicao == "tangente":
        # O raio até o ponto de tangência T é perpendicular ao plano.
        fig = u3d.desenhar_segmento(
            fig, (0, 0, 0), centro_da_secao, cor=u3d.CORES["plano_secundario"]
        )
        fig = u3d.desenhar_ponto(
            fig, centro_da_secao, rotulo="T", cor=u3d.CORES["destaque"], tamanho=9
        )
    u3d.mostrar(fig)

    print(f"R = {raio_da_esfera}, d = {distancia}: plano {posicao}")
    if posicao == "secante":
        print(f"Raio da seção: r = √({raio_da_esfera}² − {distancia}²) = {raio_da_secao:.4f}")
    elif posicao == "tangente":
        print("A seção é um único ponto (raio 0).")
    else:
        print("O plano não toca a esfera: não há seção.")
    conferencia = u3d.secao_da_esfera((0, 0, 0), raio_da_esfera, plano)
    if conferencia is None:
        print("Conferência com o utils_3d: nenhuma seção")
    else:
        print(f"Conferência com o utils_3d: raio da seção {conferencia[1]:.4f}")


interact(
    cortar_esfera,
    raio_da_esfera=widgets.FloatSlider(
        min=1, max=4, step=0.5, value=3, description="raio R",
        continuous_update=False,
    ),
    distancia=widgets.FloatSlider(
        min=0, max=5, step=0.25, value=2, description="distância d",
        continuous_update=False,
    ),
    direcao_do_plano=widgets.Dropdown(
        options=["Horizontal", "Inclinado"], value="Horizontal", description="plano",
    ),
);
```

## 3. O volume da esfera: a anticlépsidra

Para usar o Princípio de Cavalieri, precisamos de um sólido cujo volume já conhecemos e cujas seções tenham a mesma área que as da esfera. Esse sólido é a **anticlépsidra**.

> 📌 **Definição.** A **anticlépsidra** é o sólido que sobra quando, de um cilindro equilátero de raio $R$ (altura $2R$), retiramos dois cones de raio $R$ e altura $R$, com vértices no centro do cilindro. (O nome vem de "clépsidra", o relógio de água em forma de dois cones unidos pelo vértice.)

**As seções.** Corte a esfera e a anticlépsidra por um plano a uma distância $d$ do centro:
- **esfera:** a seção é um círculo de raio $\sqrt{R^2 - d^2}$, com área $\pi(R^2 - d^2)$;
- **anticlépsidra:** a seção é um **anel** (uma "coroa circular"): o círculo do cilindro (raio $R$) menos o círculo do cone. Como o cone tem altura $R$ e raio $R$, a uma distância $d$ do vértice o raio do cone é $d$. Área: $\pi R^2 - \pi d^2 = \pi(R^2 - d^2)$.

As áreas são **iguais** em todas as alturas! Pelo Princípio de Cavalieri:
$$\text{volume da esfera} = \text{volume do cilindro} - 2 \cdot \text{volume do cone} = \pi R^2 \cdot 2R - 2 \cdot \frac{1}{3}\pi R^2 \cdot R = 2\pi R^3 - \frac{2}{3}\pi R^3$$
$$\boxed{\text{volume da esfera} = \frac{4}{3}\pi R^3}$$

> 🕵️ **Curiosidade histórica.** **Arquimedes** (século III a.C.) descobriu que a esfera tem exatamente $\frac{2}{3}$ do volume do cilindro que a envolve, e também $\frac{2}{3}$ da sua área total. Ele se orgulhava tanto desse resultado que pediu que uma esfera dentro de um cilindro fosse gravada no seu túmulo. Mais de um século depois, o romano **Cícero** contou ter encontrado o túmulo, abandonado, justamente por causa desse desenho.

🕹️ Compare as seções da semiesfera e da semianticlépsidra.

```python
def comparar_com_anticlepsidra(raio: float, altura_do_corte: float) -> None:
    """Corta a semiesfera e a semianticlépsidra pelo mesmo plano horizontal."""
    d = altura_do_corte * raio
    area_secao_esfera = math.pi * (raio**2 - d**2)
    area_secao_anel = math.pi * raio**2 - math.pi * d**2
    volume_esfera = 4 / 3 * math.pi * raio**3
    volume_cilindro = math.pi * raio**2 * (2 * raio)

    # À esquerda, a semiesfera de cima; à direita, a metade de cima da anticlépsidra:
    # um cilindro de raio R e altura R do qual sai um cone com o vértice no centro da
    # base de baixo e a base na tampa.
    x_da_anticlepsidra = 2 * raio + 1.5
    centro_de_baixo = (x_da_anticlepsidra, 0, 0)
    fig = u3d.criar_figura(titulo="Semiesfera e semianticlépsidra")
    fig = u3d.desenhar_esfera(fig, (0, 0, 0), raio, opacidade=0.3, latitude_minima=0)
    fig = u3d.desenhar_paralelo(fig, 0, raio, cor=u3d.CORES["solido"])
    fig = u3d.desenhar_cilindro(fig, centro_de_baixo, raio, (0, 0, raio), opacidade=0.2)
    fig = u3d.desenhar_cone(
        fig, centro_de_baixo, (x_da_anticlepsidra, 0, raio), raio,
        cor=u3d.CORES["destaque"], opacidade=0.3,
    )

    # O plano de corte, largo o bastante para alcançar os dois sólidos.
    fig = u3d.desenhar_plano(
        fig, (x_da_anticlepsidra / 2, 0, d), (0, 0, 1), opacidade=0.1,
        tamanho=x_da_anticlepsidra + 2 * raio + 1,
    )
    # Seção da semiesfera: um círculo de raio √(R² − d²).
    raio_da_secao = math.sqrt(raio**2 - d**2)
    if raio_da_secao > 0:
        circulo = u3d.poligono_regular(60, raio=raio_da_secao, centro=(0, 0, d))
        fig = u3d.desenhar_poligono(fig, circulo, cor=u3d.CORES["reta"], opacidade=0.7)
    else:
        fig = u3d.desenhar_ponto(fig, (0, 0, d), cor=u3d.CORES["reta"], tamanho=8)
    # Seção da semianticlépsidra: o anel entre o cone (raio d) e o cilindro (raio R).
    fig = u3d.desenhar_coroa_circular(
        fig, (x_da_anticlepsidra, 0, d), raio, d, cor=u3d.CORES["reta"], opacidade=0.7
    )
    fig = u3d.definir_camera(fig, olho=(0.4, -1.8, 0.9))
    u3d.mostrar(fig)

    print(f"Distância do corte ao centro: d = {d:.2f}")
    print(f"Seção da semiesfera: π(R² − d²) = {area_secao_esfera:.4f}")
    print(f"Seção da semianticlépsidra (anel): πR² − πd² = {area_secao_anel:.4f}")
    print(f"Volume da esfera: (4/3)πR³ = {volume_esfera:.4f}")
    print(f"Volume do cilindro que envolve a esfera: πR² · 2R = {volume_cilindro:.4f}")
    print(f"Razão entre os volumes: {volume_esfera / volume_cilindro:.4f} (dois terços)")


interact(
    comparar_com_anticlepsidra,
    raio=widgets.FloatSlider(
        min=1, max=3, step=0.5, value=2, description="raio R",
        continuous_update=False,
    ),
    altura_do_corte=widgets.FloatSlider(
        min=0, max=1, step=0.1, value=0.5, description="corte (× R)",
        continuous_update=False,
    ),
);
```

## 4. A área da superfície esférica

### A ideia das pirâmides

Imagine a superfície da esfera dividida em muitas regiõezinhas minúsculas. Ligue cada uma delas ao centro: a esfera fica dividida em muitas "pirâmides" finíssimas, todas com vértice no centro e altura praticamente igual a $R$. Somando os volumes dessas pirâmides ($\frac{1}{3} \cdot \text{base} \cdot \text{altura}$, notebook 10):
$$\text{volume da esfera} \approx \frac{1}{3} \cdot (\text{soma das bases}) \cdot R = \frac{1}{3} \cdot S \cdot R$$

Quanto menores as regiõezinhas, melhor a aproximação. Igualando ao volume da seção 3:
$$\frac{1}{3} S R = \frac{4}{3}\pi R^3 \qquad\Longrightarrow\qquad \boxed{S = 4\pi R^2}$$

**Três formas de lembrar**
- A área da esfera é igual à área de **quatro círculos máximos**.
- A área da esfera é igual à **área lateral do cilindro** que a envolve: $2\pi R \cdot 2R = 4\pi R^2$ (outro resultado de Arquimedes).
- A área da esfera é $\frac{2}{3}$ da área total desse cilindro: $\frac{4\pi R^2}{6\pi R^2} = \frac{2}{3}$.

> 💡 **Dica.** Experimente em casa: desenhe no papel alguns círculos com o mesmo raio de uma laranja. Descasque a laranja e cubra os círculos com os pedaços da casca. A casca toda cobre cerca de **quatro** círculos!

> ⚠️ **Erro comum.**
> - Confundir as fórmulas: a **área** tem $R^2$ (é medida em cm², m²...) e o **volume** tem $R^3$ (cm³, m³...).
> - Usar o diâmetro no lugar do raio. Se o problema der o diâmetro, divida por 2 antes.

### Vendo a área aparecer

O widget abaixo aproxima a esfera de raio 1 por um poliedro de faces triangulares (como um **domo geodésico**), cada vez mais fino. Para cada face, calculamos o volume da "pirâmide" que ela forma com o centro.

🕹️ Aumente o número de subdivisões.

```python
def aproximar_esfera(subdivisoes: int, mostrar_piramides: bool) -> None:
    """Aproxima a esfera de raio 1 por um poliedro e soma as pirâmides com vértice no
    centro."""
    vertices, faces = u3d.poliedro_aproximando_esfera(1, subdivisoes)

    # A soma das pirâmides: cada face é a base de uma pirâmide com vértice no centro,
    # e a altura é a distância do centro ao plano da face.
    area_total = 0
    volume_pelas_piramides = 0
    for face in faces:
        pontos = [vertices[i] for i in face]
        area = u3d.medir_area_do_poligono(pontos)
        plano_da_face = u3d.plano_por_tres_pontos(*pontos)
        altura = u3d.medir_distancia_ponto_plano((0, 0, 0), plano_da_face)
        area_total = area_total + area
        volume_pelas_piramides = volume_pelas_piramides + area * altura / 3

    fig = u3d.criar_figura(titulo=f"Um domo com {len(faces)} faces triangulares")
    opacidade = 0.25 if mostrar_piramides else 0.6
    fig = u3d.desenhar_poliedro(
        fig, vertices, faces, cor=u3d.CORES["plano"], opacidade=opacidade,
        traco_unico=True,
    )
    if mostrar_piramides:
        cores = [
            u3d.CORES["destaque"], u3d.CORES["plano_secundario"],
            u3d.CORES["plano_terciario"], u3d.CORES["reta_secundaria"],
            u3d.CORES["reta"],
        ]
        # Cinco faces espalhadas pelo poliedro, cada uma ligada ao centro.
        for k, cor in enumerate(cores):
            face = faces[k * len(faces) // 5]
            piramide = u3d.piramide([vertices[i] for i in face], (0, 0, 0))
            fig = u3d.desenhar_poliedro(fig, *piramide, cor=cor, opacidade=0.8)
    u3d.mostrar(fig)

    print(f"Número de faces: {len(faces)}")
    print(
        f"Área total do poliedro: {area_total:.4f}   "
        f"(4π = {4 * math.pi:.4f}; {area_total / (4 * math.pi):.2%} dela)"
    )
    print(
        f"Volume pela soma das pirâmides: {volume_pelas_piramides:.4f}   "
        f"((4/3)π = {4 / 3 * math.pi:.4f}; "
        f"{volume_pelas_piramides / (4 / 3 * math.pi):.2%} dele)"
    )
    print(f"Um terço da área vezes o raio: {area_total * 1 / 3:.4f}")


interact(
    aproximar_esfera,
    subdivisoes=widgets.IntSlider(
        min=0, max=4, step=1, value=1, description="subdivisões",
        continuous_update=False,
    ),
    mostrar_piramides=widgets.Checkbox(
        value=False, description="Mostrar as pirâmides de algumas faces"
    ),
);

print("Subdivisões | faces | área do poliedro")
for k in range(5):
    vertices, faces = u3d.poliedro_aproximando_esfera(1, k)
    print(f"{k:>11} | {len(faces):>5} | {u3d.medir_area_total(vertices, faces):.4f}")
print(f"     esfera |       | {4 * math.pi:.4f}  (4π)")
```

## 5. Fuso e cunha esféricos

Corte uma laranja em gomos, com cortes que passam pelo eixo: cada gomo é uma **cunha esférica**, e a casca de cada gomo é um **fuso esférico**.

> 📌 **Definição.** Considere dois semiplanos com origem no eixo da esfera, formando um **diedro** de medida $\alpha$ (notebook 05).
> - O **fuso esférico** de ângulo $\alpha$ é a parte da **superfície** esférica entre os dois semiplanos.
> - A **cunha esférica** de ângulo $\alpha$ é a parte da **esfera** (o sólido) entre os dois semiplanos.

**As fórmulas, por proporção.** A volta inteira (360°) corresponde à esfera inteira. Um ângulo $\alpha$ corresponde à fração $\frac{\alpha}{360°}$:
$$S_{\text{fuso}} = \frac{\alpha}{360°} \cdot 4\pi R^2 = \frac{\pi R^2 \alpha}{90°} \qquad\qquad \text{volume}_{\text{cunha}} = \frac{\alpha}{360°} \cdot \frac{4}{3}\pi R^3 = \frac{\pi R^3 \alpha}{270°}$$

**Aplicações**
- **Gomos** de laranja (cunhas) e **gomos** de uma bola de praia ou de um guarda-chuva (fusos).
- **Fusos horários:** a Terra gira 360° em 24 horas, ou seja, 15° por hora. Por isso, o planeta é dividido em 24 fusos horários de aproximadamente 15° cada (na prática, as fronteiras seguem também divisas de países e estados).

🕹️ Abra o gomo no widget.

```python
def abrir_gomo(raio: float, angulo: int, mostrar: str) -> None:
    """Desenha o fuso ou a cunha de ângulo dado e calcula a área e o volume."""
    fracao = angulo / 360
    area_do_fuso = fracao * 4 * math.pi * raio**2
    volume_da_cunha = fracao * 4 / 3 * math.pi * raio**3

    fig = u3d.criar_figura(titulo=f"{mostrar} de {angulo}°")
    fig = u3d.desenhar_esfera(fig, (0, 0, 0), raio, opacidade=0.1)
    # O eixo da esfera: os dois semiplanos que limitam o gomo saem dele.
    fig = u3d.desenhar_segmento(
        fig, (0, 0, -1.3 * raio), (0, 0, 1.3 * raio), cor=u3d.CORES["apoio"],
        espessura=3, tracejado=True,
    )
    if mostrar == "Fuso":
        fig = u3d.desenhar_fuso(fig, 0, angulo, raio)
    else:
        fig = u3d.desenhar_cunha(fig, 0, angulo, raio)
    u3d.mostrar(fig)

    print(f"Fração da esfera: {angulo}/360 = {fracao:.4f}")
    print(f"Área do fuso: {fracao:.4f} · 4π · {raio}² = {area_do_fuso:.2f}")
    print(f"Volume da cunha: {fracao:.4f} · (4/3)π · {raio}³ = {volume_da_cunha:.2f}")
    if angulo == 15:
        print("Este é o tamanho aproximado de um fuso horário!")


interact(
    abrir_gomo,
    raio=widgets.FloatSlider(
        min=1, max=3, step=0.5, value=2, description="raio R",
        continuous_update=False,
    ),
    angulo=widgets.IntSlider(
        min=15, max=360, step=15, value=60, description="ângulo α (°)",
        continuous_update=False,
    ),
    mostrar=widgets.RadioButtons(options=["Fuso", "Cunha"], value="Fuso"),
);
```

## 6. Latitude e longitude

Para localizar um ponto na Terra (pensando nela como uma esfera), usamos dois ângulos:

> 📌 **Definição.**
> - A **latitude** $\varphi$ de um ponto é o ângulo entre o raio que vai até ele e o plano do **equador**. Vai de 0° (no equador) até 90° norte (N) ou sul (S), nos polos.
> - A **longitude** $\lambda$ é o ângulo entre o meridiano do ponto e um meridiano de referência, o de **Greenwich** (em Londres). Vai de 0° até 180° leste (L) ou oeste (O). A longitude é um **diedro** entre semiplanos que contêm o eixo da Terra.

Pontos com a mesma latitude formam um **paralelo**; pontos com a mesma longitude, um **meridiano**. A Linha do Equador passa, por exemplo, pela cidade de Macapá; o Trópico de Capricórnio (latitude de cerca de 23,4° S) atravessa o Brasil.

### O tamanho dos paralelos

O paralelo de latitude $\varphi$ é uma seção da esfera feita a uma distância $d = R\sin\varphi$ do centro. Pela seção 2:
$$r^2 = R^2 - R^2\sin^2\varphi = R^2\cos^2\varphi \qquad\Longrightarrow\qquad \boxed{r = R\cos\varphi}$$

Então o comprimento do paralelo é $2\pi R\cos\varphi$: os paralelos encolhem à medida que nos aproximamos dos polos.

### Distâncias

- **Ao longo de um meridiano** (ou do equador), cada grau corresponde a $\frac{2\pi R}{360°}$. Na Terra ($R \approx 6371$ km), **1° ≈ 111 km**.
- **Ao longo de um paralelo** de latitude $\varphi$, cada grau de longitude vale $\frac{2\pi R\cos\varphi}{360°}$: em 60° de latitude, metade do que vale no equador.

> 💡 **Dica.** O caminho mais curto entre dois pontos de uma esfera vai sobre um **círculo máximo**. É por isso que, num mapa plano, as rotas dos aviões em voos longos parecem "curvas": elas seguem círculos máximos da Terra.

🕹️ Escolha um lugar no globo. (Latitudes ao sul e longitudes a oeste de Greenwich aparecem negativas nos controles.)

```python
R_terra = 6371  # km, o raio médio da Terra


def formatar_posicao(latitude: int, longitude: int) -> str:
    """Escreve a posição como no mapa, por exemplo "25° S, 50° O"."""
    norte_ou_sul = "N" if latitude > 0 else "S" if latitude < 0 else ""
    leste_ou_oeste = "L" if longitude > 0 else "O" if longitude < 0 else ""
    return (
        f"{abs(latitude)}° {norte_ou_sul}".strip()
        + ", "
        + f"{abs(longitude)}° {leste_ou_oeste}".strip()
    )


def localizar_no_globo(latitude: int, longitude: int) -> None:
    """Marca o ponto escolhido num globo e calcula medidas na Terra real."""
    raio_do_paralelo = R_terra * math.cos(math.radians(latitude))
    comprimento_do_paralelo = 2 * math.pi * raio_do_paralelo
    km_por_grau_no_paralelo = comprimento_do_paralelo / 360
    # Ao longo do meridiano, cada grau de latitude vale 2πR/360.
    distancia_ao_equador = abs(latitude) * 2 * math.pi * R_terra / 360

    # O desenho usa um "globo" de raio 1.
    ponto = u3d.ponto_na_esfera(latitude, longitude)
    fig = u3d.criar_figura(titulo=formatar_posicao(latitude, longitude))
    fig = u3d.desenhar_esfera(fig, (0, 0, 0), 1, opacidade=0.25)
    for lat in [-60, -30, 30, 60]:
        fig = u3d.desenhar_paralelo(fig, lat, espessura=1)
    for lon in range(30, 360, 30):
        fig = u3d.desenhar_meridiano(fig, lon, espessura=1)
    fig = u3d.desenhar_paralelo(fig, 0, cor=u3d.CORES["plano"], espessura=4)
    fig = u3d.desenhar_meridiano(fig, 0, cor=u3d.CORES["plano_secundario"], espessura=4)
    fig = u3d.adicionar_rotulo(
        fig, u3d.ponto_na_esfera(45, 0, 1.2), "Greenwich",
        cor=u3d.CORES["plano_secundario"],
    )

    # O paralelo e o meridiano do ponto escolhido.
    fig = u3d.desenhar_paralelo(fig, latitude, cor=u3d.CORES["reta"], espessura=4)
    fig = u3d.desenhar_meridiano(
        fig, longitude, cor=u3d.CORES["reta_secundaria"], espessura=4
    )
    # O raio do centro até o ponto e, se o ponto não for um polo, o raio do paralelo,
    # que vai do eixo até o ponto (o centro do paralelo fica no eixo, na altura do
    # ponto).
    fig = u3d.desenhar_segmento(fig, (0, 0, 0), ponto, cor=u3d.CORES["apoio"])
    if abs(latitude) < 90:
        centro_do_paralelo = (0, 0, ponto[2])
        fig = u3d.desenhar_segmento(
            fig, centro_do_paralelo, ponto, cor=u3d.CORES["reta"], tracejado=True
        )
    # A latitude φ: o ângulo entre o raio e o plano do equador.
    if latitude != 0:
        no_equador = u3d.ponto_na_esfera(0, longitude)
        fig = u3d.desenhar_segmento(
            fig, (0, 0, 0), no_equador, cor=u3d.CORES["apoio"], tracejado=True
        )
        fig = u3d.desenhar_angulo(
            fig, (0, 0, 0), no_equador, ponto, rotulo="φ", raio=0.35
        )
    fig = u3d.desenhar_ponto(fig, ponto, cor=u3d.CORES["destaque"], tamanho=9)
    # A câmera gira junto com a longitude, para o ponto ficar sempre à vista.
    fig = u3d.definir_camera(fig, olho=u3d.ponto_na_esfera(20, longitude, 2.2))
    u3d.mostrar(fig)

    print(f"Posição: {formatar_posicao(latitude, longitude)}")
    print(f"Raio do paralelo: {raio_do_paralelo:.0f} km")
    print(f"Comprimento do paralelo: {comprimento_do_paralelo:.0f} km")
    print(f"1° de longitude neste paralelo: {km_por_grau_no_paralelo:.0f} km")
    print(f"Distância ao equador, ao longo do meridiano: {distancia_ao_equador:.0f} km")


interact(
    localizar_no_globo,
    latitude=widgets.IntSlider(
        min=-90, max=90, step=5, value=-25, description="latitude (°)",
        continuous_update=False,
    ),
    longitude=widgets.IntSlider(
        min=-180, max=180, step=5, value=-50, description="longitude (°)",
        continuous_update=False,
    ),
);

print("Latitude | comprimento do paralelo (km)")
for latitude in [0, 23.4, 45, 60, 90]:
    comprimento = 2 * math.pi * R_terra * math.cos(math.radians(latitude))
    print(f"{latitude:>7}° | {comprimento:>10.0f}")
```

## Resumo

**Superfície esférica:** pontos a distância $R$ do centro. **Esfera:** pontos a distância $\le R$.

**Elementos:** eixo, polos, equador, paralelos, meridianos. **Círculo máximo:** seção por um plano que passa pelo centro.

**Seção a distância $d$ do centro:** círculo de raio $r = \sqrt{R^2 - d^2}$.
Plano e esfera: secante ($d < R$), tangente ($d = R$, plano ⟂ raio), exterior ($d > R$).

**Volume** (Cavalieri e anticlépsidra): $\dfrac{4}{3}\pi R^3$, ou $\dfrac{2}{3}$ do cilindro circunscrito.

**Área** (pirâmides com vértice no centro): $4\pi R^2$ = quatro círculos máximos = área lateral do cilindro circunscrito.

**Fuso** de ângulo $\alpha$: $\dfrac{\alpha}{360°} \cdot 4\pi R^2$. **Cunha**: $\dfrac{\alpha}{360°} \cdot \dfrac{4}{3}\pi R^3$.

**Latitude e longitude:** raio do paralelo $= R\cos\varphi$. Na Terra ($R \approx 6371$ km): equador ≈ 40 030 km; 1° de meridiano ≈ 111 km; 15° de longitude = 1 fuso horário.

**A esfera é o sólido de menor área para um volume dado.**
