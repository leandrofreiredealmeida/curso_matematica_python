**Exercício 2.** Existe triedro com faces (a) 90°, 90°, 90°; (b) 60°, 60°, 60°; (c) 120°, 120°, 120°; (d) 50°, 60°, 100°; (e) 50°, 60°, 120°?

```python
# Basta conferir as quatro condições: cada face menor que a soma das outras duas
# e a soma das três menor que 360°.
# (a) soma 270°, e 90 < 180: existe (é o triedro trirretângulo).
# (b) soma 180°, e 60 < 120: existe (é o vértice do tetraedro regular).
# (c) soma 360°: NÃO existe (as faces ficam planas, sem ponta).
# (d) soma 210°, e a maior face 100 < 50 + 60 = 110: existe.
# (e) soma 230°, mas a maior face 120 não é menor que 50 + 60 = 110: NÃO existe.
casos_ex2 = {
    "a": (90, 90, 90),
    "b": (60, 60, 60),
    "c": (120, 120, 120),
    "d": (50, 60, 100),
    "e": (50, 60, 120),
}
existe_ex2 = {item: existe_triedro(*faces) for item, faces in casos_ex2.items()}
print(existe_ex2)
```

**Exercício 4.** Diedros do triedro de faces $f_a = 90°$, $f_b = 90°$ e $f_c = 60°$ (o canto de uma barra de chocolate triangular).

```python
# A aresta c forma 90° com a (face f_b = 90°) e 90° com b (face f_a = 90°). Pelo
# teorema fundamental da perpendicularidade (notebook 03), c é perpendicular ao
# plano da face f_c, que contém a e b.
#
# d_c: o plano da face f_c é perpendicular à aresta c, então a própria face f_c é
#      uma seção reta do diedro d_c. Logo d_c = f_c = 60°.
# d_a: a face f_b contém c, que é perpendicular ao plano da face f_c. Então os
#      planos das faces f_b e f_c são perpendiculares (notebook 03), e o diedro de
#      aresta a entre eles é reto: d_a = 90°.
# d_b: pelo mesmo motivo (a face f_a também contém c), d_b = 90°.
diedros_ex4 = {"a": 90, "b": 90, "c": 60}
print(diedros_ex4)

# Conferência com o utils_3d (que usa vetores, assunto do módulo 07).
t = u3d.vertices_do_triedro(90, 90, 60)
print({x: round(medida, 1) for x, medida in elementos(t)["diedros"].items()})
```

**Exercício 5.** Duas faces de um triedro medem 70° e 50°: entre que valores pode estar a terceira face?

```python
# Seja x a terceira face. As condições da seção 2 são:
#   x < 70 + 50           →  x < 120
#   70 < 50 + x           →  x > 20
#   50 < 70 + x           →  vale sempre (x > −20)
#   70 + 50 + x < 360     →  x < 240 (já garantido por x < 120)
minimo = 70 - 50
maximo = 70 + 50
intervalo_ex5 = (minimo, maximo)
print(intervalo_ex5)  # (20, 120)

# Conferência: testamos as faces inteiras de 1° a 179° com as condições da seção 2.
possiveis = [x for x in range(1, 180) if existe_triedro(70, 50, x)]
print(f"De {possiveis[0]}° a {possiveis[-1]}° (os extremos 20° e 120° não valem).")
```

**Exercício 6.** Número máximo de (a) triângulos equiláteros, (b) quadrados, (c) pentágonos regulares e (d) hexágonos regulares que formam, sozinhos, um ângulo poliédrico convexo.

```python
# Com k faces iguais em volta do vértice, a soma das faces é k · (ângulo interno),
# e ela precisa ser menor que 360°. Além disso, k ≥ 3.
maximo_ex6 = {}
for item, lados in zip("abcd", [3, 4, 5, 6], strict=True):
    angulo_interno = 180 * (lados - 2) / lados
    k = 2
    while (k + 1) * angulo_interno < 360:
        k += 1
    # Se nem 3 faces cabem, o laço não avança e k fica 2: não há ângulo poliédrico.
    maximo_ex6[item] = k if k >= 3 else 0

# (a) 5 · 60° = 300° < 360°, mas 6 · 60° = 360°: no máximo 5.
# (b) 3 · 90° = 270° < 360°, mas 4 · 90° = 360°: no máximo 3.
# (c) 3 · 108° = 324° < 360°, mas 4 · 108° = 432°: no máximo 3.
# (d) 3 · 120° = 360°: nem 3 hexágonos cabem; resposta 0.
print(maximo_ex6)  # {'a': 5, 'b': 3, 'c': 3, 'd': 0}
```

**Exercício 8.** Diedro de um triedro com as três faces de 60° (o vértice de um tetraedro regular).

```python
# No tetraedro regular de aresta 1, seja M o ponto médio de uma aresta. Os
# segmentos de M até os dois vértices opostos são alturas de triângulos
# equiláteros de lado 1: h = √3/2. Eles são perpendiculares à aresta, então formam
# a seção reta do diedro. O terceiro lado desse triângulo é uma aresta (1).
aresta = 1
h = math.sqrt(3) / 2 * aresta

# Lei dos cossenos: 1² = h² + h² − 2 · h · h · cos θ.
cosseno = (h**2 + h**2 - aresta**2) / (2 * h * h)
diedro_ex8 = math.degrees(math.acos(cosseno))
print(cosseno, diedro_ex8)  # 0.333... 70.528...  (cos θ = 1/3)

# Conferência com o utils_3d (que usa vetores, assunto do módulo 07).
t = u3d.vertices_do_triedro(60, 60, 60)
print(f"Diedro medido: {elementos(t)['diedros']['a']:.4f}°")
```
