## Básico

**Exercício 1.** Verdadeiro ou falso: (a) 90° com duas retas concorrentes de um plano ⟹ perpendicular ao plano; (b) perpendicular a uma reta de um plano ⟹ perpendicular ao plano; (c) duas retas perpendiculares a um mesmo plano são paralelas; (d) duas retas perpendiculares a uma mesma reta são paralelas; (e) se dois planos são perpendiculares, toda reta de um é perpendicular ao outro.

```python
respostas_ex1 = {
    # (a) Verdadeiro: é o teorema fundamental da perpendicularidade (seção 2).
    "a": True,
    # (b) Falso: um esquadro só não basta. A diagonal AH forma 90° com AB, que está
    #     na base, mas AH é oblíqua à base.
    "b": False,
    # (c) Verdadeiro: é a propriedade 3 (as colunas de um prédio).
    "c": True,
    # (d) Falso: no espaço, elas podem ser paralelas, concorrentes ou reversas
    #     (erro comum da seção 3). Ex.: AD e BF formam 90° com AB e são reversas.
    "d": False,
    # (e) Falso: só as retas de um plano perpendiculares à interseção são
    #     perpendiculares ao outro (erro comum da seção 5). Ex.: a diagonal AF da
    #     face ABFE não é perpendicular à base, embora ABFE ⊥ ABCD.
    "e": False,
}
print(respostas_ex1)

# Conferência dos contraexemplos com o utils_3d (que usa vetores, assunto do módulo 07).
base = plano_pelo_nome("ABCD")
print("AH ⊥ base?", u3d.reta_perpendicular_ao_plano(A, H, base))
print("AD e BF:", u3d.classificar_posicao_retas(A, D, B, F))
print("AF ⊥ base?", u3d.reta_perpendicular_ao_plano(A, F, base))
```

**Exercício 2.** A reta é perpendicular ao plano? $AE$ e $ABCD$, $AB$ e $BCGF$, $AG$ e $ABCD$, $BF$ e $EFGH$, $AC$ e $BDHF$.

```python
perpendiculares_ex2 = {
    # AE forma 90° com AB e com AD, concorrentes em A, ambas na base.
    "AE e ABCD": True,
    # AB forma 90° com BC e com BF, concorrentes em B, ambas na face BCGF.
    "AB e BCGF": True,
    # AG é a diagonal do cubo: forma 90° com BD (AG está no plano ACGE, que é
    # perpendicular a BD), mas não forma 90° com AB (o ângulo é cerca de 54,7°).
    "AG e ABCD": False,
    # BF forma 90° com EF e com FG, concorrentes em F, ambas na face de cima.
    "BF e EFGH": True,
    # AC ⊥ BD (diagonais do quadrado) e AC é ortogonal a BF (BF ⊥ base). BD e BF
    # são concorrentes em B, logo AC ⊥ BDHF (Pare e pense da seção 2).
    "AC e BDHF": True,
}

# Conferência com a função do utils_3d (que usa vetores, assunto do módulo 07).
for par, resposta in perpendiculares_ex2.items():
    reta, plano = par.split(" e ")
    conferida = u3d.reta_perpendicular_ao_plano(
        vertices[reta[0]], vertices[reta[1]], plano_pelo_nome(plano)
    )
    print(f"{par}: {resposta} (conferido: {conferida})")
```

**Exercício 3.** Os planos são perpendiculares? $ABCD$ e $ABFE$, $ABCD$ e $EFGH$, $ACGE$ e $BDHF$, $ABFE$ e $BCGF$, $BDE$ e $ABCD$.

```python
perpendiculares_ex3 = {
    # ABFE contém AE, que é perpendicular à base.
    "ABCD e ABFE": True,
    # As bases são paralelas, não perpendiculares.
    "ABCD e EFGH": False,
    # ACGE contém AC, que é perpendicular a BDHF (exercício 2).
    "ACGE e BDHF": True,
    # ABFE contém AB, que é perpendicular a BCGF (exercício 2).
    "ABFE e BCGF": True,
    # Os planos se cortam na reta BD. Em BDE, a reta perpendicular a BD pelo centro
    # O da base é OE, que é oblíqua à base (E não fica acima de O). Se os planos
    # fossem perpendiculares, OE seria perpendicular à base (teorema da seção 5).
    "BDE e ABCD": False,
}

# Conferência com a função do utils_3d (que usa vetores, assunto do módulo 07).
for par, resposta in perpendiculares_ex3.items():
    nome_1, nome_2 = par.split(" e ")
    conferida = u3d.planos_sao_perpendiculares(
        plano_pelo_nome(nome_1), plano_pelo_nome(nome_2)
    )
    print(f"{par}: {resposta} (conferido: {conferida})")
```

## Intermediário

**Exercício 4.** Poste de 6 m a 8 m do meio-fio: comprimento do menor cabo do topo do poste até o meio-fio.

```python
altura_do_poste = 6
distancia_ao_meio_fio = 8

# Pelas três perpendiculares, o menor cabo é PM, preso no pé M da perpendicular de
# O ao meio-fio. O triângulo POM é retângulo em O: PM² = PO² + OM².
cabo_ex4 = (altura_do_poste**2 + distancia_ao_meio_fio**2) ** 0.5

print(cabo_ex4)  # 10.0
```

**Exercício 5.** Cubo de aresta 3: distância do vértice $G$ à reta $AB$.

```python
aresta = 3

# Três perpendiculares com r = GC (perpendicular à base em C), s = AB e M = B,
# pois CB ⊥ AB. Então GB ⊥ AB, e a distância de G até AB é GB.
# Pitágoras no triângulo GCB, retângulo em C: GB² = GC² + CB².
distancia_ex5 = (aresta**2 + aresta**2) ** 0.5  # 3·√2

print(distancia_ex5)  # 4.2426...

# Conferência com o utils_3d (que usa vetores, assunto do módulo 07).
v3 = u3d.vertices_do_cubo(aresta=3)
pe = u3d.pe_da_perpendicular_a_reta(v3["G"], v3["A"], v3["B"])
print("Pé da perpendicular de G até AB:", pe, "(é o vértice B:", v3["B"], ")")
print(f"Distância medida: {u3d.medir_distancia(v3['G'], pe):.4f}")
```

**Exercício 6.** Retas que formam 90° com $AB$: posição relativa de $AE$ e $BF$, $AD$ e $AE$, $AE$ e $BC$.

```python
posicoes_ex6 = {
    # As duas são perpendiculares à base (propriedade 3): paralelas.
    "AE e BF": "paralelas",
    # As duas passam por A.
    "AD e AE": "concorrentes",
    # BC está na base e não passa por A, onde AE fura a base (critério do notebook 02).
    "AE e BC": "reversas",
}

# Conferência com a função do utils_3d (que usa vetores, assunto do módulo 07).
for par, resposta in posicoes_ex6.items():
    r, s = par.split(" e ")
    conferida = u3d.classificar_posicao_retas(
        vertices[r[0]], vertices[r[1]], vertices[s[0]], vertices[s[1]]
    )
    print(f"{par}: {resposta} (conferido: {conferida})")
```

## Desafio

**Exercício 7.** (a) A diagonal $AG$ é perpendicular ao plano $BDE$? (b) Os planos $ACGE$ e $BDE$ são perpendiculares?

```python
# (a) Vamos achar duas retas concorrentes de BDE que formam 90° com AG.
#
# 1. BD ⊥ AC (diagonais do quadrado da base) e BD ⊥ AE (AE é perpendicular à base,
#    então forma 90° com todas as retas dela). AC e AE são concorrentes em A, logo,
#    pelo teorema fundamental, BD ⊥ ACGE. Como AG está em ACGE, BD forma 90° com AG.
# 2. BE ⊥ AF (diagonais do quadrado ABFE) e BE ⊥ AD (AD é perpendicular à face
#    ABFE). AF e AD são concorrentes em A, logo BE ⊥ ADGF. Como AG está em ADGF,
#    BE forma 90° com AG.
# 3. BD e BE são concorrentes em B e estão em BDE. Pelo teorema fundamental, AG ⊥ BDE.
ag_perpendicular_a_bde_ex7 = True

# (b) AG ⊥ BDE e AG ⊂ ACGE. Pela definição, ACGE ⊥ BDE.
acge_perpendicular_a_bde_ex7 = True

print(ag_perpendicular_a_bde_ex7, acge_perpendicular_a_bde_ex7)

# Conferência com as funções do utils_3d (que usam vetores, assunto do módulo 07).
plano_bde = plano_pelo_nome("BDE")
print("BD ⊥ ACGE?", u3d.reta_perpendicular_ao_plano(B, D, plano_pelo_nome("ACGE")))
print("BE ⊥ ADGF?", u3d.reta_perpendicular_ao_plano(B, E, plano_pelo_nome("ADGF")))
print("AG ⊥ BDE?", u3d.reta_perpendicular_ao_plano(A, G, plano_bde))
plano_acge = plano_pelo_nome("ACGE")
print("ACGE ⊥ BDE?", u3d.planos_sao_perpendiculares(plano_acge, plano_bde))

fig = u3d.criar_figura(titulo="AG ⊥ BDE e ACGE ⊥ BDE")
fig = u3d.desenhar_cubo(fig)
fig = u3d.desenhar_plano_por_tres_pontos(
    fig, B, D, E, rotulo="BDE", opacidade=0.3, tamanho=1
)
fig = u3d.desenhar_plano_por_tres_pontos(
    fig, A, C, G, rotulo="ACGE", cor=u3d.CORES["plano_secundario"], opacidade=0.2,
    tamanho=1,
)
fig = u3d.desenhar_segmento(fig, A, G, cor=u3d.CORES["reta"], espessura=6)
fig = u3d.desenhar_segmento(fig, B, D, cor=u3d.CORES["reta_secundaria"], espessura=5)
fig = u3d.desenhar_segmento(fig, B, E, cor=u3d.CORES["reta_secundaria"], espessura=5)
fig = u3d.desenhar_intersecao_reta_plano(fig, A, G, plano_bde, rotulo="P")
u3d.mostrar(fig)
```

**Exercício 8.** Em que fração da diagonal $AG$ fica o ponto $P$ onde ela fura o plano $BDE$?
