## Essa rampa está dentro da norma?

Uma escola vai construir uma rampa de acessibilidade para vencer um degrau. No Brasil, a norma de acessibilidade (ABNT NBR 9050) limita, na maioria dos casos, a inclinação das rampas a **8,33%**: no máximo 1 metro de subida para cada 12 metros na horizontal. Rampas mais íngremes são perigosas e cansativas para quem usa cadeira de rodas.

Para projetar a rampa, o engenheiro precisa responder a várias perguntas:
- Quanto espaço a rampa ocupa **no chão**? (a **projeção** da rampa sobre o chão)
- Qual é a **altura** a vencer? (a **distância** do topo da rampa até o chão)
- Qual é o **ângulo** da rampa com o chão? E a inclinação em porcentagem?

🕹️ Mexa na altura e no comprimento horizontal da rampa e veja se ela atende à norma.

```python
import math

import numpy as np

# As coordenadas (x, y, z) abaixo servem só para posicionar o desenho.
LARGURA_DA_RAMPA = 1.5  # em metros
LIMITE_DA_NORMA = 8.33  # inclinação máxima, em %


def desenhar_rampa(altura: float, comprimento_horizontal: float) -> None:
    """Desenha a rampa, a sua projeção no chão, a altura e o ângulo com o chão."""
    c, h = comprimento_horizontal, altura
    # O pedaço de chão desenhado acompanha o tamanho da rampa.
    chao = ((c / 2, LARGURA_DA_RAMPA / 2, 0), (0, 0, 1))
    rampa = [(0, 0, 0), (c, 0, h), (c, LARGURA_DA_RAMPA, h), (0, LARGURA_DA_RAMPA, 0)]

    fig = u3d.criar_figura(titulo="A rampa e a sua projeção no chão")
    fig = u3d.desenhar_plano(fig, *chao, rotulo="chão", opacidade=0.2, tamanho=c + 4)
    fig = u3d.desenhar_poligono(fig, rampa, rotulo="rampa", cor=u3d.CORES["solido"])
    fig = u3d.desenhar_projecao_ortogonal(
        fig, rampa, chao, fechar=True, preencher=True, mostrar_projetantes=True
    )
    # A altura: o segmento vertical do topo da rampa até o chão.
    fig = u3d.desenhar_segmento(
        fig, (c, 0, 0), (c, 0, h), rotulo="altura", cor=u3d.CORES["destaque"],
        espessura=6,
    )
    fig = u3d.desenhar_angulo_reto(
        fig, (c, 0, 0), (0, 0, 0), (c, 0, h), tamanho=min(0.3, h / 2)
    )
    fig = u3d.desenhar_angulo(fig, (0, 0, 0), (c, 0, 0), (c, 0, h), raio=1)
    fig = u3d.definir_camera(fig, olho=(0.4, -1.9, 0.6))
    u3d.mostrar(fig)

    # Pitágoras
    comprimento_rampa = math.sqrt(comprimento_horizontal**2 + altura**2)
    # Subida para cada 100 de horizontal
    inclinacao_percentual = altura / comprimento_horizontal * 100
    # Tangente = cateto oposto / cateto adjacente
    angulo = math.degrees(math.atan(altura / comprimento_horizontal))

    print(f"Espaço ocupado no chão (projeção): {comprimento_horizontal:.2f} m")
    print(f"Comprimento da rampa: {comprimento_rampa:.2f} m")
    print(f"Inclinação: {inclinacao_percentual:.2f}%")
    print(f"Ângulo com o chão: {angulo:.2f}°")
    if inclinacao_percentual <= LIMITE_DA_NORMA:
        print("✔ A rampa atende ao limite de 8,33%.")
    else:
        print("✘ A rampa é íngreme demais para o limite de 8,33%.")


interact(
    desenhar_rampa,
    altura=widgets.FloatSlider(
        min=0.1, max=1.5, step=0.1, value=0.5, description="altura (m)",
        continuous_update=False,
    ),
    comprimento_horizontal=widgets.FloatSlider(
        min=2, max=15, step=1, value=6, description="horizontal (m)",
        continuous_update=False,
    ),
);
```

No fim, você vai ver também por que quem sobe a rampa em linha reta enfrenta a maior inclinação possível dela.

## 1. Projeção ortogonal

> 📌 **Definição.**
> - A **projeção ortogonal** de um ponto $P$ sobre um plano $\alpha$ é o pé $P'$ da perpendicular a $\alpha$ traçada por $P$. A reta $PP'$ se chama **projetante**.
> - A **projeção ortogonal de uma figura** sobre $\alpha$ é o conjunto das projeções de todos os pontos da figura.

Se $P$ já está em $\alpha$, sua projeção é ele mesmo.

> 📎 **Conexão.** O pé da perpendicular é o mesmo do notebook 03 ([`0603a_perpendicularidade.ipynb`](0603a_perpendicularidade.ipynb)). No `utils_3d.py`, a projeção de um ponto é calculada pela mesma função: `u3d.pe_da_perpendicular_ao_plano`.

**Projeção de uma reta**

| Posição da reta $r$ | Projeção sobre $\alpha$ |
|---|---|
| $r \perp \alpha$ | um **ponto** (o pé da perpendicular) |
| $r$ não perpendicular a $\alpha$ | uma **reta** $r'$ |

No segundo caso, $r'$ é a interseção de $\alpha$ com o único plano que contém $r$ e é perpendicular a $\alpha$ (propriedade 3 da seção 5 do notebook 03).

**Projeção de um segmento $AB$:** é um ponto (se $AB \perp \alpha$) ou um segmento $A'B'$. Se $AB$ é **paralelo** a $\alpha$, a projeção tem o **mesmo comprimento**: $A'B' = AB$. Nos outros casos, a projeção é menor que o segmento.

🎯 **Aplicações**
- **Sombra ao meio-dia.** Nas regiões entre os trópicos, em certos dias do ano o sol do meio-dia fica exatamente a pino. Nesse momento, a sombra de um objeto no chão é praticamente a sua projeção ortogonal sobre o chão.
- **Plantas de arquitetura.** A planta baixa de uma casa é a projeção ortogonal da casa sobre o plano do chão; as **vistas** de frente e de lado do desenho técnico são projeções sobre planos verticais.

> 🕵️ **Curiosidade histórica.** No fim do século XVIII, o francês **Gaspard Monge** criou a **Geometria Descritiva**: um método para representar objetos 3D com precisão usando projeções ortogonais em dois ou mais planos. O método foi considerado tão útil para projetos de fortificações militares que, por um tempo, foi tratado como segredo. Até hoje, ele é a base do desenho técnico de engenharia e arquitetura.

🕹️ Mexa no segmento $AB$ abaixo e observe a projeção $A'B'$.

```python
def desenhar_projecao_de_segmento(inclinacao: int, giro: int) -> None:
    """Desenha o segmento AB, o plano α e a projeção A'B'."""
    alfa = ((0, 0, 0), (0, 0, 1))  # o plano α, horizontal
    A = (0, 0, 1)

    # Seno e cosseno (módulo 05) servem só para inclinar e girar o desenho de AB:
    # "inclinacao" é o ângulo de AB com α e "giro" diz para que lado AB aponta.
    i, g = math.radians(inclinacao), math.radians(giro)
    B = (
        A[0] + 3 * math.cos(i) * math.cos(g),
        A[1] + 3 * math.cos(i) * math.sin(g),
        A[2] + 3 * math.sin(i),
    )
    A_ = u3d.pe_da_perpendicular_ao_plano(A, alfa)  # o ponto A'
    B_ = u3d.pe_da_perpendicular_ao_plano(B, alfa)  # o ponto B'

    # Quando A' e B' coincidem, escrevemos um rótulo só.
    rotulos = ["A' = B'", ""] if inclinacao == 90 else ["A'", "B'"]

    fig = u3d.criar_figura(titulo="O segmento AB e a sua projeção A'B'")
    fig = u3d.desenhar_plano(fig, *alfa, rotulo="α", tamanho=8, opacidade=0.25)
    fig = u3d.desenhar_segmento(fig, A, B, cor=u3d.CORES["reta"], espessura=6)
    fig = u3d.desenhar_pontos(fig, [A, B], rotulos=["A", "B"])
    fig = u3d.desenhar_projecao_ortogonal(fig, [A, B], alfa, rotulos=rotulos)
    u3d.mostrar(fig)

    print(f"AB = {u3d.medir_distancia(A, B):.3f}")
    print(f"A'B' = {u3d.medir_distancia(A_, B_):.3f}")
    if inclinacao == 0:
        print(
            "AB é paralelo a α: a projeção tem o mesmo comprimento "
            "(verdadeira grandeza)."
        )
    elif inclinacao == 90:
        print("AB é perpendicular a α: a projeção é um único ponto.")
    else:
        print("A projeção é menor que o segmento.")


interact(
    desenhar_projecao_de_segmento,
    inclinacao=widgets.IntSlider(
        min=0, max=90, step=15, value=30, description="inclinação (°)",
        continuous_update=False,
    ),
    giro=widgets.IntSlider(
        min=0, max=180, step=30, value=0, description="giro (°)",
        continuous_update=False,
    ),
);
```

### As vistas de uma casinha

Um arquiteto representa uma casa no papel pelas suas projeções ortogonais em três planos: o **chão** (planta baixa ou vista de cima), a **parede do fundo** (vista de frente) e uma **parede lateral** (vista de lado). Quando a projeção de uma figura tem o mesmo tamanho e forma da figura, dizemos que ela aparece em **verdadeira grandeza**.

🕹️ Ligue e desligue cada vista e gire a figura.

```python
# A casinha: um bloco de 2 × 3 × 2 com telhado de duas águas (cumeeira 1 acima das
# paredes, ao longo do comprimento). Como na Geometria Descritiva, ela fica afastada
# dos três planos de projeção, para que as projetantes apareçam.
CASA = {
    "A": (1, 1, 1), "B": (3, 1, 1), "C": (3, 4, 1), "D": (1, 4, 1),  # chão da casa
    "E": (1, 1, 3), "F": (3, 1, 3), "G": (3, 4, 3), "H": (1, 4, 3),  # topo das paredes
    "M": (2, 1, 4), "N": (2, 4, 4),  # as pontas da cumeeira
}
ARESTAS_DA_CASA = [
    "AB", "BC", "CD", "DA", "EF", "FG", "GH", "HE", "AE", "BF", "CG", "DH",
    "MN", "EM", "FM", "HN", "GN",
]
PAREDES = ["ABFME", "DCGNH", "ADHE", "BCGF"]  # a frente e o fundo são pentágonos
TELHADO = ["EMNH", "FMNG"]

# Os três planos de projeção, como (ponto, normal), e a cor de cada vista.
VISTAS = {
    "de cima": (((2, 2.5, 0), (0, 0, 1)), u3d.CORES["plano"]),
    "de frente": (((2, 6, 2.5), (0, 1, 0)), u3d.CORES["plano_secundario"]),
    "de lado": (((-1, 2.5, 2.5), (1, 0, 0)), u3d.CORES["plano_terciario"]),
}


def desenhar_vistas(
    vista_de_cima: bool, vista_de_frente: bool, vista_de_lado: bool,
    mostrar_projetantes: bool,
) -> None:
    """Desenha a casinha e as suas projeções nos planos das vistas escolhidas."""
    fig = u3d.criar_figura(titulo="As três vistas de uma casinha", altura=600)
    for nome in PAREDES:
        fig = u3d.desenhar_poligono(
            fig, [CASA[letra] for letra in nome], opacidade=0.3, mostrar_borda=False
        )
    for nome in TELHADO:
        fig = u3d.desenhar_poligono(
            fig, [CASA[letra] for letra in nome], cor=u3d.CORES["reta_secundaria"],
            opacidade=0.3, mostrar_borda=False,
        )
    for aresta in ARESTAS_DA_CASA:
        fig = u3d.desenhar_segmento(
            fig, CASA[aresta[0]], CASA[aresta[1]], cor=u3d.CORES["apoio"]
        )

    escolhidas = {
        "de cima": vista_de_cima, "de frente": vista_de_frente, "de lado": vista_de_lado
    }
    for nome, (plano, cor) in VISTAS.items():
        fig = u3d.desenhar_plano(
            fig, *plano, rotulo=f"vista {nome}", cor=cor, opacidade=0.1, tamanho=7
        )
        if not escolhidas[nome]:
            continue
        # Cada aresta da casinha é projetada no plano da vista: dois pontos por vez.
        for aresta in ARESTAS_DA_CASA:
            fig = u3d.desenhar_projecao_ortogonal(
                fig, [CASA[aresta[0]], CASA[aresta[1]]], plano, cor=cor,
                mostrar_projetantes=False, tamanho=2,
            )
        if mostrar_projetantes:
            # As projetantes de todos os vértices, uma vez só para cada vértice.
            fig = u3d.desenhar_projecao_ortogonal(
                fig, list(CASA.values()), plano, cor=cor, ligar_pontos=False, tamanho=2
            )

    fig = u3d.definir_camera(fig, olho=(1.7, -1.5, 1.0))
    u3d.mostrar(fig)
    # As três vistas juntas bastam para reconstruir o objeto: é assim que funciona o
    # desenho técnico (cada vista mostra duas das três dimensões da casa).
    print("Vista de cima: comprimento e largura (a planta baixa).")
    print("Vista de frente: largura e altura (o formato do telhado aparece aqui).")
    print("Vista de lado: comprimento e altura.")


interact(
    desenhar_vistas,
    vista_de_cima=widgets.Checkbox(value=True, description="Vista de cima"),
    vista_de_frente=widgets.Checkbox(value=True, description="Vista de frente"),
    vista_de_lado=widgets.Checkbox(value=True, description="Vista de lado"),
    mostrar_projetantes=widgets.Checkbox(
        value=False, description="Mostrar projetantes"
    ),
);
```

## 2. Segmento perpendicular e segmentos oblíquos

Seja $P$ um ponto fora de um plano $\alpha$, e $O$ o pé da perpendicular por $P$. O segmento $PO$ é o **segmento perpendicular**. Qualquer outro segmento $PA$, com $A$ em $\alpha$, é um **segmento oblíquo**; o segmento $OA$ é a **projeção** de $PA$ sobre $\alpha$.

Como o triângulo $POA$ é retângulo em $O$, Pitágoras dá $PA^2 = PO^2 + OA^2$. Daí saem três propriedades:

1. **O perpendicular é o menor.** $PO < PA$ para qualquer oblíquo $PA$.
2. **Projeções iguais, oblíquos iguais.** Se $OA = OB$, então $PA = PB$. E vale a volta: oblíquos iguais têm projeções iguais.
3. **Projeção maior, oblíquo maior.** Se $OA < OB$, então $PA < PB$.

Uma consequência bonita da propriedade 2: os pés de todos os oblíquos de **mesmo comprimento** formam uma **circunferência** de centro $O$.

🎯 **Aplicação: os cabos de uma torre.** Torres de antena e postes altos são presos por **cabos de estai**. Para que a torre fique equilibrada, os cabos costumam ter o mesmo comprimento e ficam ancorados no chão sobre uma circunferência centrada no pé da torre. A propriedade 2 garante que isso funciona.

🕹️ Mexa na altura da torre, no raio da circunferência e no número de cabos.

```python
def desenhar_cabos(altura_da_torre: int, raio: int, numero_de_cabos: int) -> None:
    """Desenha a torre, a circunferência das âncoras, os cabos e as suas projeções."""
    chao = ((0, 0, 0), (0, 0, 1))
    O = (0, 0, 0)  # o pé da torre
    P = (0, 0, altura_da_torre)  # o topo da torre

    # Seno e cosseno (módulo 05) servem só para espalhar as âncoras igualmente sobre
    # a circunferência: uma a cada 360°/n.
    ancoras = []
    for k in range(numero_de_cabos):
        t = 2 * math.pi * k / numero_de_cabos
        ancoras.append((raio * math.cos(t), raio * math.sin(t), 0))

    fig = u3d.criar_figura(titulo="Cabos de estai de mesmo comprimento", altura=550)
    fig = u3d.desenhar_plano(
        fig, *chao, rotulo="chão", tamanho=2 * raio + 4, opacidade=0.2
    )
    fig = u3d.desenhar_circunferencia(
        fig, O, raio, cor=u3d.CORES["apoio"], tracejado=True
    )
    fig = u3d.desenhar_segmento(fig, O, P, cor=u3d.CORES["apoio"], espessura=10)
    fig = u3d.desenhar_angulo_reto(fig, O, ancoras[0], P, tamanho=0.12 * raio)
    for ancora in ancoras:
        fig = u3d.desenhar_segmento(fig, P, ancora, cor=u3d.CORES["reta"], espessura=5)
        # A projeção do cabo PA no chão é o segmento OA.
        fig = u3d.desenhar_projecao_ortogonal(
            fig, [P, ancora], chao, cor=u3d.CORES["apoio"], mostrar_projetantes=False,
            tamanho=3,
        )
    fig = u3d.desenhar_pontos(fig, [O, P], rotulos=["O", "P"])
    u3d.mostrar(fig)

    # Pitágoras no triângulo POA, retângulo em O: PA² = PO² + OA².
    comprimento_do_cabo = math.sqrt(altura_da_torre**2 + raio**2)
    total = numero_de_cabos * comprimento_do_cabo

    print(f"Pitágoras: √({altura_da_torre}² + {raio}²) = {comprimento_do_cabo:.2f} m")
    # Conferência com a trena do utils_3d: mede os n cabos de uma vez.
    medidos = u3d.medir_distancia(np.array(ancoras), P)
    for numero, medida in enumerate(medidos, start=1):
        print(f"  cabo {numero}: {medida:.2f} m")
    print(f"Total de cabo necessário: {total:.2f} m")


interact(
    desenhar_cabos,
    altura_da_torre=widgets.IntSlider(
        min=4, max=20, step=1, value=12, description="altura (m)",
        continuous_update=False,
    ),
    raio=widgets.IntSlider(
        min=2, max=12, step=1, value=5, description="raio (m)", continuous_update=False
    ),
    numero_de_cabos=widgets.IntSlider(
        min=3, max=8, step=1, value=4, description="nº de cabos",
        continuous_update=False,
    ),
);
```

## 3. Distâncias geométricas

> 📌 **Definição.** A **distância** entre duas figuras é o comprimento do **menor** segmento que liga um ponto de uma a um ponto da outra.

No espaço, as distâncias mais importantes são:

| Entre | Como medir | Onde aparece |
|---|---|---|
| dois pontos | o segmento que os une | — |
| ponto e reta | o segmento perpendicular do ponto à reta | notebook 03 (três perpendiculares) |
| **ponto e plano** | o segmento perpendicular $PO$ | seção 2 (propriedade 1) |
| **reta e plano paralelos** | a distância de **qualquer** ponto da reta ao plano | abaixo |
| **planos paralelos** | a distância de **qualquer** ponto de um deles ao outro | abaixo |
| **retas reversas** | o segmento da **perpendicular comum** | abaixo |

**Reta e plano paralelos; planos paralelos.** Todos os pontos de uma reta paralela a um plano estão à mesma distância dele (os segmentos perpendiculares são lados opostos de retângulos). O mesmo vale para dois planos paralelos. É essa distância que chamamos de **altura** de um prisma (notebook 08) e de **distância entre andares** de um prédio.

**Exemplo no cubo.** No notebook 03 (Desafio 8), você descobriu que a diagonal $AG$ fura o plano $BDE$ num ponto $P$ com $AP = \frac{1}{3} AG$. Como $AG \perp BDE$, o segmento $AP$ é o segmento perpendicular de $A$ ao plano. Num cubo de aresta 2: $d(A, BDE) = \frac{1}{3} \cdot 2\sqrt{3} = \frac{2\sqrt{3}}{3} \approx 1{,}155$.

### Distância entre retas reversas

**Teorema**
> Dadas duas retas reversas $r$ e $s$, existe **uma única** reta que é concorrente com as duas e perpendicular a ambas: a **perpendicular comum**. A distância entre $r$ e $s$ é o comprimento do segmento $AB$ da perpendicular comum, com $A \in r$ e $B \in s$.

**Construção**
1. Trace o plano $\alpha$ que contém $r$ e é paralelo a $s$ (notebook 02).
2. Projete $s$ ortogonalmente sobre $\alpha$, obtendo a reta $s'$. Ela é paralela a $s$ e corta $r$ num ponto $A$.
3. Trace por $A$ a perpendicular ao plano $\alpha$. Ela está no plano que projeta $s$ sobre $\alpha$ e corta $s$ num ponto $B$.
4. O segmento $AB$ é perpendicular a $\alpha$ (logo forma 90° com $r$) e perpendicular a $s'$, portanto a $s$.

Além disso, $AB$ é a distância entre o plano $\alpha$ e o plano paralelo a $\alpha$ que contém $s$: duas retas reversas sempre estão em planos paralelos, e a distância entre elas é a distância entre esses planos.

🎯 **Aplicação.** Em indústrias e prédios, canos de gás, água e cabos elétricos passam uns por cima dos outros. As normas de segurança exigem uma **distância mínima** entre eles, e é a perpendicular comum que mede essa distância.

🕹️ Avance os passos da construção.

```python
# As retas reversas do notebook 02 (as coordenadas servem só para o desenho).
r1, r2 = (-2, 0, 0), (2, 0, 0)  # dois pontos de r
s1, s2 = (0, -2, 2), (0, 2, 3)  # dois pontos de s
ALFA_RS = u3d.plano_paralelo_a_reta(r1, r2, s1, s2)  # contém r e é paralelo a s
PONTO_A, PONTO_B = u3d.perpendicular_comum(r1, r2, s1, s2)

FRASES = {
    1: "Passo 1: as retas r e s são reversas.",
    2: "Passo 2: o plano α contém r e é paralelo a s (construção do notebook 02).",
    3: "Passo 3: a projeção s' de s sobre α é paralela a s e corta r no ponto A.",
    4: "Passo 4: a perpendicular a α traçada por A corta s no ponto B.",
    5: "Passo 5: AB forma 90° com r e com s. É a perpendicular comum!",
}


def desenhar_construcao(passo: int) -> None:
    """Desenha a construção da perpendicular comum, até o passo escolhido."""
    fig = u3d.criar_figura(titulo=f"A perpendicular comum: passo {passo}", altura=550)
    fig = u3d.desenhar_reta_por_dois_pontos(
        fig, r1, r2, rotulo="r", cor=u3d.CORES["reta"], comprimento=7, espessura=6
    )
    fig = u3d.desenhar_reta_por_dois_pontos(
        fig, s1, s2, rotulo="s", cor=u3d.CORES["reta_secundaria"], comprimento=7,
        espessura=6,
    )
    if passo >= 2:
        fig = u3d.desenhar_plano_paralelo_a_reta(
            fig, r1, r2, s1, s2, rotulo="α", tamanho=7, opacidade=0.25
        )
    if passo >= 3:
        fig = u3d.desenhar_projecao_ortogonal(
            fig, [s1, s2], ALFA_RS, rotulos=None, cor=u3d.CORES["reta_secundaria"],
            mostrar_projetantes=True,
        )
        s1_ = u3d.pe_da_perpendicular_ao_plano(s1, ALFA_RS)
        s2_ = u3d.pe_da_perpendicular_ao_plano(s2, ALFA_RS)
        fig = u3d.desenhar_reta_por_dois_pontos(
            fig, s1_, s2_, rotulo="s'", cor=u3d.CORES["reta_secundaria"],
            comprimento=7, tracejado=True,
        )
        fig = u3d.desenhar_ponto(fig, PONTO_A, rotulo="A")
    if passo >= 4:
        fig = u3d.desenhar_perpendicular_ao_plano(
            fig, ALFA_RS, PONTO_A, cor=u3d.CORES["apoio"], comprimento=6, espessura=3,
            tracejado=True,
        )
        fig = u3d.desenhar_ponto(fig, PONTO_B, rotulo="B")
    if passo >= 5:
        fig = u3d.desenhar_perpendicular_comum(fig, r1, r2, s1, s2, rotulo="AB")
    u3d.mostrar(fig)

    print(FRASES[passo])
    if passo >= 5:
        distancia = u3d.medir_distancia_entre_retas(r1, r2, s1, s2)
        print(f"Distância entre r e s: {distancia:.3f}")


interact(
    desenhar_construcao,
    passo=widgets.IntSlider(
        min=1, max=5, step=1, value=1, description="passo", continuous_update=False
    ),
);
```

## 4. Ângulo de uma reta com um plano

> 📌 **Definição.** O **ângulo entre uma reta $r$ e um plano $\alpha$** é:
> - $90°$, se $r \perp \alpha$;
> - $0°$, se $r$ é paralela a $\alpha$ ou está contida nele;
> - o ângulo entre $r$ e sua projeção ortogonal $r'$ sobre $\alpha$, se $r$ é oblíqua.

**Por que a projeção?** Porque, entre todas as retas do plano, a projeção é a que forma o **menor** ângulo com $r$. É a medida "honesta" da inclinação: nenhuma reta do chão é mais "parecida" com a rampa do que a sombra dela.

**Calculando com trigonometria.** Se $PA$ é um segmento oblíquo, com $P$ a uma altura $h = PO$ do plano e projeção $OA$, o triângulo $POA$ é retângulo em $O$ e o ângulo $\theta$ fica no vértice $A$:
$$\sin\theta = \frac{PO}{PA}, \qquad \cos\theta = \frac{OA}{PA}, \qquad \tan\theta = \frac{PO}{OA}$$

Em particular, **o comprimento da projeção** de um segmento é $A'B' = AB \cdot \cos\theta$.

**Inclinação em porcentagem.** Em rampas, ruas e telhados, a inclinação costuma ser dada em porcentagem: é a tangente vezes 100.
$$\text{inclinação} = \tan\theta \cdot 100\%$$

> ⚠️ **Erro comum.** Achar que 100% de inclinação é uma parede vertical. Na verdade, 100% significa $\tan\theta = 1$, ou seja, $\theta = 45°$: sobe 1 metro para cada metro na horizontal. Uma parede vertical teria inclinação "infinita".

🕹️ Gire a reta $t$ dentro do plano, em volta do ponto $A$ onde $r$ fura o plano, e acompanhe o ângulo entre $r$ e $t$. Onde ele é menor?

```python
ALFA_H = ((0, 0, 0), (0, 0, 1))  # o plano α, horizontal
PONTO_A_R = (0, 0, 0)  # onde r fura α
PONTO_P_R = (2, 0, 2)  # outro ponto de r


def direcao_de_t(giro: int) -> tuple[float, float, float]:
    """Um ponto da reta t de α, que passa por A e está girada do ângulo dado."""
    # Seno e cosseno (módulo 05) servem só para girar o desenho de t dentro de α.
    theta = math.radians(giro)
    return (math.cos(theta), math.sin(theta), 0)


def desenhar_angulo_minimo(giro_de_t: int) -> None:
    """Desenha r, a sua projeção r' e uma reta t de α, com o ângulo entre r e t."""
    t2 = direcao_de_t(giro_de_t)
    P_ = u3d.pe_da_perpendicular_ao_plano(PONTO_P_R, ALFA_H)  # projeção de P

    fig = u3d.criar_figura(titulo="O ângulo entre r e as retas de α")
    fig = u3d.desenhar_plano(fig, *ALFA_H, rotulo="α", tamanho=6, opacidade=0.25)
    fig = u3d.desenhar_reta_por_dois_pontos(
        fig, PONTO_A_R, PONTO_P_R, rotulo="r", cor=u3d.CORES["reta"], comprimento=6,
        espessura=6,
    )
    fig = u3d.desenhar_projecao_ortogonal(
        fig, [PONTO_P_R], ALFA_H, rotulos=["P'"], cor=u3d.CORES["apoio"]
    )
    fig = u3d.desenhar_reta_por_dois_pontos(
        fig, PONTO_A_R, P_, rotulo="r'", cor=u3d.CORES["apoio"], comprimento=6,
        tracejado=True,
    )
    fig = u3d.desenhar_reta_por_dois_pontos(
        fig, PONTO_A_R, t2, rotulo="t", cor=u3d.CORES["reta_secundaria"],
        comprimento=6, espessura=5,
    )
    fig = u3d.desenhar_angulo_entre_retas(fig, PONTO_A_R, PONTO_P_R, PONTO_A_R, t2)
    fig = u3d.desenhar_pontos(fig, [PONTO_A_R, PONTO_P_R], rotulos=["A", "P"])
    u3d.mostrar(fig)

    angulo_r_t = u3d.medir_angulo_entre_retas(PONTO_A_R, PONTO_P_R, PONTO_A_R, t2)
    angulo_r_alfa = u3d.medir_angulo_reta_plano(PONTO_A_R, PONTO_P_R, ALFA_H)
    print(f"Ângulo entre r e t: {angulo_r_t:.1f}°")
    print(f"Ângulo entre r e o plano: {angulo_r_alfa:.1f}°")
    if giro_de_t in (0, 180):
        print("t coincide com a projeção r': este é o menor ângulo possível!")

    # Uma pequena tabela: o ângulo entre r e t para vários giros de t.
    print()
    print(f"{'giro de t':^12}{'ângulo entre r e t':^22}")
    tabela = {
        giro: u3d.medir_angulo_entre_retas(
            PONTO_A_R, PONTO_P_R, PONTO_A_R, direcao_de_t(giro)
        )
        for giro in range(0, 181, 30)
    }
    menor = min(tabela.values())
    for giro, angulo in tabela.items():
        marca = "  ← mínimo (t = r')" if abs(angulo - menor) < 1e-9 else ""
        print(f"{f'{giro}°':^12}{f'{angulo:.1f}°':^22}{marca}")


interact(
    desenhar_angulo_minimo,
    giro_de_t=widgets.IntSlider(
        min=0, max=180, step=15, value=90, description="giro de t (°)",
        continuous_update=False,
    ),
);
```

## 5. Reta de maior declive

Quando chove, a água escorre pelo telhado sempre pelo caminho mais "inclinado". Uma bola solta numa rampa desce em linha reta, perpendicular à borda da rampa. Esse caminho tem nome.

> 📌 **Definição.** Sejam $\alpha$ e $\beta$ dois planos secantes, com reta de interseção $i$. Uma **reta de maior declive** de $\alpha$ em relação a $\beta$ é uma reta de $\alpha$ **perpendicular a $i$**.

**Teorema**
> Entre todas as retas de $\alpha$, as retas de maior declive são as que formam o **maior ângulo** com $\beta$. Esse ângulo mede a **inclinação** do plano $\alpha$ em relação a $\beta$.

A ideia da justificativa: tome um ponto $P$ de $\alpha$ a uma altura $h$ de $\beta$. Para qualquer reta de $\alpha$ por $P$ que encontra $i$ num ponto $Q$, o seno do ângulo com $\beta$ é $\frac{h}{PQ}$. Esse seno é **máximo** quando $PQ$ é o **menor** possível, ou seja, quando $PQ$ é perpendicular a $i$.

🎯 **Aplicações**
- **Telhados e calhas.** A água da chuva escorre pelas retas de maior declive, perpendiculares ao beiral; é por isso que a calha fica ao longo do beiral.
- **Rampas.** Quem sobe a rampa do gancho em linha reta, de frente, segue a reta de maior declive e enfrenta a inclinação máxima.
- **Estradas de serra.** Subir uma montanha em linha reta seria íngreme demais para carros. Por isso as estradas fazem **zigue-zague**: um caminho que forma um ângulo com a reta de maior declive é bem menos inclinado.

🕹️ Incline o plano $\alpha$ e gire a direção do caminho dentro dele. Compare a inclinação do caminho com a inclinação da reta de maior declive.

```python
BETA = ((1, 0, 0), (0, 0, 1))  # o plano β, horizontal
ORIGEM = (0, 0, 0)  # um ponto da reta i


def inclinacao_do_caminho(
    inclinacao_do_plano: int, direcao_do_caminho: int
) -> tuple[float, float]:
    """Devolve a inclinação do caminho, em graus e em porcentagem."""
    # Seno e cosseno (módulo 05) servem só para inclinar o desenho de α e para
    # girar o caminho dentro dele: u sobe pelo maior declive e v corre ao longo de i.
    phi, delta = math.radians(inclinacao_do_plano), math.radians(direcao_do_caminho)
    u = (math.cos(phi), 0, math.sin(phi))
    v = (0, 1, 0)
    caminho = tuple(math.cos(delta) * a + math.sin(delta) * b for a, b in zip(u, v))
    graus = u3d.medir_angulo_reta_plano(ORIGEM, caminho, BETA)
    porcentagem = math.tan(math.radians(graus)) * 100  # tangente × 100
    return graus, porcentagem


def desenhar_maior_declive(inclinacao_do_plano: int, direcao_do_caminho: int) -> None:
    """Desenha α, β, a reta i, a reta de maior declive por P e o caminho por P."""
    # Seno e cosseno (módulo 05) servem só para posicionar o desenho (veja acima).
    phi, delta = math.radians(inclinacao_do_plano), math.radians(direcao_do_caminho)
    u = (math.cos(phi), 0, math.sin(phi))
    v = (0, 1, 0)
    alfa = u3d.plano_por_tres_pontos(ORIGEM, u, v)
    P = tuple(2 * coordenada for coordenada in u)  # no meio do pedaço de α desenhado
    caminho = tuple(math.cos(delta) * a + math.sin(delta) * b for a, b in zip(u, v))

    fig = u3d.criar_figura(titulo="A reta de maior declive", altura=550)
    fig = u3d.desenhar_plano(fig, *BETA, rotulo="β", tamanho=6, opacidade=0.2)
    fig = u3d.desenhar_plano(
        fig, P, alfa[1], rotulo="α", cor=u3d.CORES["plano_secundario"], tamanho=4,
        opacidade=0.3,
    )
    fig = u3d.desenhar_intersecao_de_planos(
        fig, alfa, BETA, rotulo="i", cor=u3d.CORES["apoio"], comprimento=6
    )
    fig = u3d.desenhar_reta(
        fig, P, u, rotulo="maior declive", cor=u3d.CORES["destaque"], comprimento=4,
        espessura=6,
    )
    fig = u3d.desenhar_angulo_reto(fig, ORIGEM, u, v)
    fig = u3d.desenhar_reta(
        fig, P, caminho, rotulo="caminho", cor=u3d.CORES["reta"], comprimento=4,
        espessura=6,
    )
    fig = u3d.desenhar_ponto(fig, P, rotulo="P")
    fig = u3d.definir_camera(fig, olho=(1.2, -1.8, 0.9))
    u3d.mostrar(fig)

    # Inclinação da reta de maior declive: em graus e em porcentagem (tangente × 100).
    declive = u3d.medir_angulo_reta_plano(ORIGEM, u, BETA)
    declive_percentual = math.tan(math.radians(declive)) * 100
    graus, porcentagem = inclinacao_do_caminho(inclinacao_do_plano, direcao_do_caminho)
    print(f"Reta de maior declive: {declive:.1f}° ({declive_percentual:.1f}%)")
    print(f"Caminho escolhido: {graus:.1f}° ({porcentagem:.1f}%)")

    print()
    print(f"Plano α inclinado {inclinacao_do_plano}°:")
    print(f"{'direção do caminho':^20}{'inclinação (°)':^16}{'inclinação (%)':^16}")
    for direcao in range(0, 91, 15):
        graus, porcentagem = inclinacao_do_caminho(inclinacao_do_plano, direcao)
        print(f"{f'{direcao}°':^20}{graus:^16.1f}{porcentagem:^16.1f}")


interact(
    desenhar_maior_declive,
    inclinacao_do_plano=widgets.IntSlider(
        min=10, max=60, step=5, value=30, description="inclinação de α (°)",
        continuous_update=False,
    ),
    direcao_do_caminho=widgets.IntSlider(
        min=0, max=90, step=15, value=45, description="direção do caminho (°)",
        continuous_update=False,
    ),
);
```

## 6. Lugares geométricos no espaço

> 📌 **Definição.** Um **lugar geométrico** é o conjunto de **todos** os pontos que têm uma certa propriedade, e **somente** eles.

Na Geometria Plana você conheceu a mediatriz (pontos equidistantes de dois pontos), a bissetriz (pontos equidistantes dos lados de um ângulo) e a circunferência (pontos a uma distância fixa de um centro). No espaço, essas ideias ganham uma dimensão a mais:

| Propriedade dos pontos | Lugar geométrico no espaço | No plano era... |
|---|---|---|
| equidistantes de dois pontos $A$ e $B$ | o **plano mediador** de $AB$: perpendicular a $AB$ no ponto médio | a mediatriz |
| equidistantes de dois planos secantes | dois **planos bissetores**, perpendiculares entre si, que contêm a reta de interseção | as bissetrizes |
| a uma distância $R$ de um ponto $C$ | a **superfície esférica** de centro $C$ e raio $R$ | a circunferência |
| a uma distância $d$ de um plano $\alpha$ | **dois planos** paralelos a $\alpha$, um de cada lado | duas retas paralelas |
| equidistantes de dois planos paralelos | o plano paralelo a eles, **no meio** | a reta paralela no meio |
| equidistantes de três pontos não colineares $A$, $B$, $C$ | a **reta perpendicular** ao plano $ABC$ pelo **circuncentro** do triângulo | um único ponto (o circuncentro) |

**Por que o plano mediador funciona?** Se $M$ é o ponto médio de $AB$ e $P$ está no plano perpendicular a $AB$ por $M$, os triângulos $PMA$ e $PMB$ são retângulos em $M$, com $MA = MB$ e o cateto $PM$ comum; logo $PA = PB$. E se $P$ estiver fora desse plano, um raciocínio parecido mostra que $PA \neq PB$.

**Vendo um lugar geométrico surgir.** O widget abaixo sorteia milhares de pontos aleatórios no espaço e guarda **só** os que têm (aproximadamente) a propriedade escolhida. Veja a forma que aparece e depois compare com o lugar geométrico exato.

🕹️ Escolha a propriedade e marque a caixa para ver o lugar geométrico exato.

```python
LUGARES = [
    "Equidistantes de A e B",
    "Equidistantes de dois planos secantes",
    "A uma distância 2 de C",
    "Equidistantes de A, B e C",
]


def desenhar_lugar_geometrico(lugar: str, mostrar_exato: bool) -> None:
    """Sorteia pontos, guarda os que têm a propriedade e desenha a nuvem."""
    # Sorteio: cada ponto tem x, y e z escolhidos ao acaso entre -3 e 3, dentro de
    # um cubo de aresta 6. A "semente" 2026 faz o sorteio sair sempre igual.
    rng = np.random.default_rng(2026)
    ultimo = lugar == "Equidistantes de A, B e C"
    # O último lugar é uma reta, muito "fina": sorteamos mais pontos e aceitamos uma
    # tolerância um pouco maior para que algum ponto caia perto dela.
    N = 300_000 if ultimo else 30_000
    tol = 0.08 if ultimo else 0.05
    pontos = rng.uniform(-3, 3, size=(N, 3))

    fig = u3d.criar_figura(titulo=lugar, altura=550)
    if lugar == "Equidistantes de A e B":
        A, B = (-1.5, -0.5, 0), (1.5, 0.5, 0.5)
        # Guardamos os pontos com PA e PB quase iguais: |PA - PB| < tolerância.
        diferenca = u3d.medir_distancia(pontos, A) - u3d.medir_distancia(pontos, B)
        guardados = pontos[np.abs(diferenca) < tol]
        fig = u3d.desenhar_segmento(fig, A, B, cor=u3d.CORES["destaque"], espessura=5)
        fig = u3d.desenhar_pontos(fig, [A, B], rotulos=["A", "B"])
        if mostrar_exato:
            fig = u3d.desenhar_plano(
                fig, *u3d.plano_mediador(A, B), rotulo="plano mediador",
                cor=u3d.CORES["plano_secundario"], opacidade=0.3,
            )

    elif lugar == "Equidistantes de dois planos secantes":
        alfa = ((0, 0, 0), (0, 0, 1))  # horizontal
        # β contém o eixo do desenho (a reta de interseção) e está inclinado de 60°.
        # O seno e o cosseno servem só para inclinar o desenho.
        sessenta = math.radians(60)
        beta = u3d.plano_por_tres_pontos(
            (0, 0, 0), (1, 0, 0), (0, math.cos(sessenta), math.sin(sessenta))
        )
        # Guardamos os pontos com d(P, α) e d(P, β) quase iguais.
        diferenca = u3d.medir_distancia_ponto_plano(
            pontos, alfa
        ) - u3d.medir_distancia_ponto_plano(pontos, beta)
        guardados = pontos[np.abs(diferenca) < tol]
        fig = u3d.desenhar_plano(fig, *alfa, rotulo="α", opacidade=0.1)
        fig = u3d.desenhar_plano(
            fig, *beta, rotulo="β", cor=u3d.CORES["plano_terciario"], opacidade=0.1
        )
        fig = u3d.desenhar_intersecao_de_planos(fig, alfa, beta, comprimento=6)
        if mostrar_exato:
            for bissetor in u3d.planos_bissetores(alfa, beta):
                fig = u3d.desenhar_plano(
                    fig, *bissetor, cor=u3d.CORES["plano_secundario"], opacidade=0.3
                )

    elif lugar == "A uma distância 2 de C":
        C = (0, 0, 0)
        # Guardamos os pontos com PC quase igual a 2.
        guardados = pontos[np.abs(u3d.medir_distancia(pontos, C) - 2) < tol]
        fig = u3d.desenhar_ponto(fig, C, rotulo="C")
        if mostrar_exato:
            fig = u3d.desenhar_esfera(fig, C, 2, cor=u3d.CORES["plano_secundario"])

    else:
        A, B, C = (2, 0, 0), (-1, 1.5, 0), (-1, -1.5, 0)
        # Guardamos os pontos com PA = PB e PA = PC (aproximadamente).
        pa = u3d.medir_distancia(pontos, A)
        pb = u3d.medir_distancia(pontos, B)
        pc = u3d.medir_distancia(pontos, C)
        guardados = pontos[(np.abs(pa - pb) < tol) & (np.abs(pa - pc) < tol)]
        fig = u3d.desenhar_poligono(fig, [A, B, C], cor=u3d.CORES["solido"])
        fig = u3d.desenhar_pontos(fig, [A, B, C], rotulos=["A", "B", "C"])
        if mostrar_exato:
            # A reta comum aos planos mediadores de AB e de AC.
            fig = u3d.desenhar_intersecao_de_planos(
                fig, u3d.plano_mediador(A, B), u3d.plano_mediador(A, C),
                cor=u3d.CORES["plano_secundario"], comprimento=6, espessura=8,
            )

    fig = u3d.desenhar_nuvem_de_pontos(fig, guardados, cor=u3d.CORES["reta"], tamanho=2)
    u3d.mostrar(fig)
    quantidade = len(guardados)
    # Escrevemos os números no jeito brasileiro: 30.000 e 0,05.
    n_texto = f"{N:,}".replace(",", ".")
    tol_texto = f"{tol}".replace(".", ",")
    print(
        f"Dos {n_texto} pontos sorteados, {quantidade} têm a propriedade "
        f"(com tolerância de {tol_texto})."
    )


interact(
    desenhar_lugar_geometrico,
    lugar=widgets.Dropdown(options=LUGARES, value=LUGARES[0], description="Lugar"),
    mostrar_exato=widgets.Checkbox(
        value=False, description="Mostrar o lugar geométrico exato"
    ),
);
```

## Resumo

**Projeção ortogonal:** de um ponto é o pé da perpendicular; de uma reta é um ponto (se $r \perp \alpha$) ou uma reta; de um segmento de comprimento $L$ que forma ângulo $\theta$ com o plano: $L' = L \cos\theta$. Segmento paralelo ao plano ⟹ projeção em verdadeira grandeza.

**Perpendicular e oblíquos (de um ponto $P$ a um plano):** $PA^2 = PO^2 + OA^2$.
- O perpendicular é o menor.
- Projeções iguais ⟺ oblíquos iguais (pés numa circunferência).
- Projeção maior ⟺ oblíquo maior.

**Distâncias:** sempre o **menor** segmento.
- Ponto–plano: o segmento perpendicular.
- Reta–plano paralelos e planos paralelos: de qualquer ponto.
- Retas reversas: a **perpendicular comum** (única), igual à distância entre os planos paralelos que as contêm.

**Ângulo reta–plano:** $90°$ se perpendicular; $0°$ se paralela ou contida; senão, o ângulo com a projeção (o **menor** ângulo com retas do plano).
$\sin\theta = \dfrac{\text{altura}}{\text{oblíquo}}$, $\cos\theta = \dfrac{\text{projeção}}{\text{oblíquo}}$, $\tan\theta = \dfrac{\text{altura}}{\text{projeção}}$. Inclinação $= \tan\theta \cdot 100\%$ (100% = 45°).

**Reta de maior declive:** reta de $\alpha$ perpendicular a $i = \alpha \cap \beta$; forma o **maior** ângulo com $\beta$.

**Lugares geométricos:** plano mediador · planos bissetores · superfície esférica · dois planos paralelos a distância $d$ · plano paralelo no meio · reta perpendicular pelo circuncentro.
