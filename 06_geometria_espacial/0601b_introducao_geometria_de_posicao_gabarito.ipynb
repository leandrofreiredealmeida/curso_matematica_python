```python
def contar_retas_e_planos(n: int) -> tuple[int, int]:
    """Conta as retas e os planos determinados por n pontos do espaço
    (reaproveitada de 0601a_introducao_geometria_de_posicao).

    Supõe que não há três pontos alinhados nem quatro pontos no mesmo plano.
    Cada par de pontos (i < j) determina uma reta; cada trio (i < j < k), um plano.
    """
    retas = 0
    for i in range(n):
        for j in range(i + 1, n):  # j começa depois de i: cada par conta uma vez só
            retas += 1

    planos = 0
    for i in range(n):
        for j in range(i + 1, n):
            for k in range(j + 1, n):  # i < j < k: cada trio conta uma vez só
                planos += 1

    return retas, planos
```

**Exercício 1.** Verdadeiro ou falso: (a) por três pontos quaisquer passa um único plano; (b) dois pontos distintos determinam uma única reta; (c) duas retas sem ponto comum são paralelas; (d) dois pontos distintos de uma reta num plano ⟹ a reta está contida nele; (e) dois planos distintos podem ter um único ponto comum.

```python
respostas_ex1 = {
    # (a) Falso: se os três pontos forem colineares, passam infinitos planos por eles.
    "a": False,
    # (b) Verdadeiro: é o postulado da determinação da reta.
    "b": True,
    # (c) Falso: no espaço elas também podem ser reversas (não coplanares).
    "c": False,
    # (d) Verdadeiro: é o postulado da inclusão.
    "d": True,
    # (e) Falso: se dois planos distintos têm um ponto comum, têm uma reta inteira
    #     em comum (postulado da interseção).
    "e": False,
}
print(respostas_ex1)
```

**Exercício 2.** Classifique as retas que contêm os pares de arestas do cubo: $AB$ e $EF$, $AB$ e $AD$, $AB$ e $GH$, $AE$ e $CD$, $BF$ e $DH$.

```python
posicoes_ex2 = {
    # AB e EF: arestas opostas da face ABFE (uma em baixo, outra em cima).
    "AB e EF": "paralelas",
    # AB e AD: as duas saem do vértice A, na base ABCD.
    "AB e AD": "concorrentes",
    # AB e GH: estão no plano diagonal ABGH e não se encontram.
    "AB e GH": "paralelas",
    # AE e CD: AE é vertical, CD é horizontal e fica em outra face; nenhum plano
    # contém as duas.
    "AE e CD": "reversas",
    # BF e DH: duas arestas verticais, ambas no plano diagonal BDHF.
    "BF e DH": "paralelas",
}

# Conferência com a função do utils_3d (que usa vetores, assunto do módulo 07).
vertices = u3d.vertices_do_cubo()
for par, resposta in posicoes_ex2.items():
    r, s = par.split(" e ")
    conferida = u3d.classificar_posicao_retas(
        vertices[r[0]], vertices[r[1]], vertices[s[0]], vertices[s[1]]
    )
    print(f"{par}: {resposta} (conferido: {conferida})")
```

## Desafio

**Exercício 7.** Quantos pares de arestas reversas existem no cubo? Percorra todos os pares com laços aninhados e use `u3d.classificar_posicao_retas`.

```python
arestas = ["AB", "BC", "CD", "DA", "EF", "FG", "GH", "HE", "AE", "BF", "CG", "DH"]

pares_reversos_ex7 = 0
for i in range(len(arestas)):
    # j começa em i + 1: assim cada par aparece uma vez só (AB-CG, mas não CG-AB).
    for j in range(i + 1, len(arestas)):
        r = arestas[i]
        s = arestas[j]
        posicao = u3d.classificar_posicao_retas(
            vertices[r[0]], vertices[r[1]], vertices[s[0]], vertices[s[1]]
        )
        if posicao == "reversas":
            pares_reversos_ex7 += 1

# Conferência pelo raciocínio: cada uma das 12 arestas tem 4 reversas a ela, o que
# dá 12 * 4 = 48; mas assim cada par foi contado duas vezes, então são 48 / 2 = 24.
print(pares_reversos_ex7)
```

**Exercício 8.** Qual é o maior número de regiões em que três planos podem dividir o espaço?

```python
# O máximo é 8. Dois planos secantes (como duas paredes prolongadas) dividem o espaço
# em 4 regiões. Um terceiro plano que corta os dois (como o chão prolongado) divide
# cada uma dessas 4 regiões em duas: uma em cima e outra embaixo, 4 * 2 = 8.
# São os 8 "cantos de sala" em volta do ponto O.
maximo_de_regioes_ex8 = 8
print(maximo_de_regioes_ex8)

# Os três planos do canto da sala, agora prolongados para todos os lados
# (pedaços centrados no canto O).
chao = ((0, 0, 0), (0, 0, 1))
parede_1 = ((0, 0, 0), (1, 0, 0))
parede_2 = ((0, 0, 0), (0, 1, 0))

fig = u3d.criar_figura(titulo="Três planos dividindo o espaço em 8 regiões")
fig = u3d.desenhar_plano(fig, *chao, rotulo="α", tamanho=4, opacidade=0.25)
fig = u3d.desenhar_plano(
    fig, *parede_1, rotulo="β", cor=u3d.CORES["plano_secundario"], tamanho=4,
    opacidade=0.25,
)
fig = u3d.desenhar_plano(
    fig, *parede_2, rotulo="γ", cor=u3d.CORES["plano_terciario"], tamanho=4,
    opacidade=0.25,
)
for plano_1, plano_2 in [(chao, parede_1), (chao, parede_2), (parede_1, parede_2)]:
    fig = u3d.desenhar_intersecao_de_planos(fig, plano_1, plano_2, comprimento=4)
fig = u3d.desenhar_ponto(fig, (0, 0, 0), rotulo="O")
u3d.mostrar(fig)
```
