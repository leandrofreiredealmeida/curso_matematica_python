## Básico

**Exercício 1.** Verdadeiro ou falso: (a) por um ponto fora de uma reta passa uma única paralela a ela; (b) uma reta paralela a um plano é paralela a todas as retas dele; (c) se dois planos são paralelos, toda reta de um é paralela ao outro; (d) duas retas paralelas a uma terceira são paralelas entre si; (e) se um plano contém duas retas paralelas a outro plano, os planos são paralelos.

```python
respostas_ex1 = {
    # (a) Verdadeiro: é o postulado das paralelas, que vale também no espaço.
    "a": True,
    # (b) Falso: ela é paralela só às retas do plano com a mesma direção dela; com
    #     as outras, é reversa (seção 2).
    "b": False,
    # (c) Verdadeiro: é a propriedade 3 dos planos paralelos.
    "c": True,
    # (d) Verdadeiro: é a transitividade do paralelismo, demonstrada na seção 1.
    "d": True,
    # (e) Falso: as duas retas precisam ser CONCORRENTES. Duas retas paralelas entre
    #     si podem estar num plano inclinado (como o piso inclinado só numa direção).
    "e": False,
}
print(respostas_ex1)

# Contraexemplo do item (e) no cubo: o plano diagonal ABGH contém a reta HG e a reta
# que liga os pontos médios de AH e BG. As duas são paralelas entre si e paralelas
# à base ABCD (ambas são paralelas a AB). Mesmo assim, ABGH corta a base na reta AB.
plano_ABGH = u3d.plano_por_tres_pontos(A, B, G)
plano_base = u3d.plano_por_tres_pontos(A, B, C)
print("HG e a base:", u3d.classificar_posicao_reta_plano(H, G, plano_base))
print("ABGH e a base:", u3d.classificar_posicao_planos(plano_ABGH, plano_base))
```

**Exercício 2.** Classifique a posição de cada reta em relação a cada plano do cubo: $EF$ e $ABCD$, $AE$ e $ABCD$, $CD$ e $ABCD$, $BG$ e $ADHE$, $AG$ e $EFGH$.

```python
posicoes_ex2 = {
    # EF não está na base e é paralela a AB, que está na base: condição de paralelismo.
    "EF e ABCD": "paralela",
    # AE sai do vértice A (que está na base) e sobe: fura a base em A.
    "AE e ABCD": "concorrente",
    # CD é uma aresta da própria base.
    "CD e ABCD": "contida",
    # BG não está na face ADHE e é paralela a AH, que está nela (BG e AH são lados
    # opostos do retângulo ABGH).
    "BG e ADHE": "paralela",
    # AG é a diagonal do cubo: sai de A, embaixo, e fura a face de cima em G.
    "AG e EFGH": "concorrente",
}

# Conferência com a função do utils_3d (que usa vetores, assunto do módulo 07).
for par, resposta in posicoes_ex2.items():
    reta, plano = par.split(" e ")
    plano_dado = u3d.plano_por_tres_pontos(*(vertices[letra] for letra in plano[:3]))
    conferida = u3d.classificar_posicao_reta_plano(
        vertices[reta[0]], vertices[reta[1]], plano_dado
    )
    print(f"{par}: {resposta} (conferido: {conferida})")
```

**Exercício 3.** Ângulo entre as retas $AB$ e $CG$, $AB$ e $DC$, $AB$ e $CH$, $AE$ e $BG$; e a classificação de $AB$ e $CG$.

```python
angulos_ex3 = {
    # AB e CG são reversas. Trocamos CG pela paralela BF, que encontra AB em B:
    # o ângulo ABF é reto.
    "AB e CG": 90,
    # AB e DC são paralelas (lados opostos da base): o ângulo é 0°.
    "AB e DC": 0,
    # AB e CH são reversas. Trocamos AB pela paralela DC, que encontra CH em C:
    # o ângulo DCH é o ângulo entre um lado e a diagonal do quadrado CDHG, 45°.
    "AB e CH": 45,
    # AE e BG são reversas. Trocamos AE pela paralela BF, que encontra BG em B:
    # o ângulo FBG é o ângulo entre um lado e a diagonal do quadrado BCGF, 45°.
    "AE e BG": 45,
}

# (e) AB e CG formam 90°, mas são reversas (não se cruzam): são ortogonais.
classificacao_ex3 = "ortogonais"

print(angulos_ex3, classificacao_ex3)

# Conferência com a função do utils_3d (que usa vetores, assunto do módulo 07).
for par in angulos_ex3:
    r, s = par.split(" e ")
    angulo = u3d.medir_angulo_entre_retas(
        vertices[r[0]], vertices[r[1]], vertices[s[0]], vertices[s[1]]
    )
    print(f"{par}: {angulo:.1f}°")
```

## Intermediário

**Exercício 4.** Três planos paralelos cortados por $r$ e $s$: $AB = 4$, $BC = 6$, $A'B' = 5$. Calcule $B'C'$ e $A'C'$.

```python
ab = 4
bc = 6
ab_linha = 5

# Tales no espaço: AB / BC = A'B' / B'C'. Isolando B'C':
# B'C' = A'B' · BC / AB.
bc_linha_ex4 = ab_linha * bc / ab

# A'C' é a soma dos dois segmentos da transversal s.
ac_linha_ex4 = ab_linha + bc_linha_ex4

print(bc_linha_ex4, ac_linha_ex4)  # 7.5 e 12.5
```

**Exercício 5.** Cabo de 45 m do chão até a laje do 10º andar de um prédio com andares de mesma altura: comprimento entre dois andares consecutivos e do chão até a laje do 4º andar.

```python
comprimento_cabo = 45

# Os pisos são planos paralelos igualmente espaçados. Pelo Tales no espaço, a razão
# entre os pedaços do cabo é a mesma razão entre as alturas dos vãos: todos os 10
# pedaços são iguais, qualquer que seja a inclinação do cabo.
trecho_por_andar_ex5 = comprimento_cabo / 10

# Do chão até a laje do 4º andar há 4 vãos.
trecho_ate_4o_andar_ex5 = 4 * trecho_por_andar_ex5

print(trecho_por_andar_ex5, trecho_ate_4o_andar_ex5)  # 4.5 e 18.0
```

**Exercício 6.** Posição dos planos do cubo: $ABCD$ e $EFGH$, $ABCD$ e $ABFE$, $ACGE$ e $BDHF$, $BDE$ e $CFH$.

```python
posicoes_ex6 = {
    # Bases opostas do cubo: não têm ponto em comum.
    "ABCD e EFGH": "paralelos",
    # A base e uma face lateral se cortam na aresta AB.
    "ABCD e ABFE": "secantes",
    # Os dois planos diagonais se cortam na reta vertical que passa pelos centros
    # das bases (o "eixo" do cubo).
    "ACGE e BDHF": "secantes",
    # BD ∥ FH e BE ∥ CH. Como BD e BE são concorrentes em B e cada uma é paralela a
    # uma reta de CFH, a condição de paralelismo garante que os planos são paralelos.
    "BDE e CFH": "paralelos",
}

# Conferência com a função do utils_3d (que usa vetores, assunto do módulo 07).
for par, resposta in posicoes_ex6.items():
    nome_1, nome_2 = par.split(" e ")
    plano_1 = u3d.plano_por_tres_pontos(*(vertices[letra] for letra in nome_1[:3]))
    plano_2 = u3d.plano_por_tres_pontos(*(vertices[letra] for letra in nome_2[:3]))
    conferida = u3d.classificar_posicao_planos(plano_1, plano_2)
    print(f"{par}: {resposta} (conferido: {conferida})")
```

## Desafio

**Exercício 7.** No cubo, $r = AE$, $s = BC$ e $t = GH$ são reversas duas a duas. Qual aresta é paralela a $t$ e se apoia em $r$ e em $s$?

```python
# Construção da seção 5:
# 1. O plano α por r = AE, paralelo a t = GH. Pela paralela a GH traçada por A,
#    que é a aresta AB, α é o plano de AE e AB: a face ABFE.
# 2. O plano β por s = BC, paralelo a t = GH. Pela paralela a GH traçada por B,
#    que é de novo a aresta AB, β é o plano de BC e AB: a base ABCD.
# 3. A reta procurada é α ∩ β: a aresta AB.
# De fato: AB ∥ GH (faces opostas do retângulo ABGH), AB encontra AE em A e
# encontra BC em B.
aresta_ex7 = "AB"

x1, x2 = vertices[aresta_ex7[0]], vertices[aresta_ex7[1]]
posicao_x_e_t_ex7 = u3d.classificar_posicao_retas(x1, x2, G, H)
posicao_x_e_r_ex7 = u3d.classificar_posicao_retas(x1, x2, A, E)
posicao_x_e_s_ex7 = u3d.classificar_posicao_retas(x1, x2, B, C)
print(aresta_ex7, posicao_x_e_t_ex7, posicao_x_e_r_ex7, posicao_x_e_s_ex7)

# A mesma construção feita pelo utils_3d: os planos α e β e a reta x = α ∩ β.
alfa = u3d.plano_paralelo_a_reta(A, E, G, H)
beta = u3d.plano_paralelo_a_reta(B, C, G, H)
print("AB contida em α:", u3d.classificar_posicao_reta_plano(A, B, alfa))
print("AB contida em β:", u3d.classificar_posicao_reta_plano(A, B, beta))

fig = u3d.criar_figura(titulo="A barra de sustentação: x = α ∩ β")
fig = u3d.desenhar_cubo(fig)
fig = u3d.desenhar_plano_paralelo_a_reta(
    fig, A, E, G, H, rotulo="α", opacidade=0.25, tamanho=4
)
fig = u3d.desenhar_plano_paralelo_a_reta(
    fig, B, C, G, H, rotulo="β", cor=u3d.CORES["plano_secundario"], opacidade=0.25,
    tamanho=4,
)
fig = u3d.desenhar_reta_por_dois_pontos(
    fig, A, E, rotulo="r", cor=u3d.CORES["reta"], comprimento=6, espessura=6
)
fig = u3d.desenhar_reta_por_dois_pontos(
    fig, B, C, rotulo="s", cor=u3d.CORES["reta_secundaria"], comprimento=6, espessura=6
)
fig = u3d.desenhar_reta_por_dois_pontos(
    fig, G, H, rotulo="t", cor=u3d.CORES["plano_terciario"], comprimento=6, espessura=6
)
fig = u3d.desenhar_intersecao_de_planos(fig, alfa, beta, rotulo="x", comprimento=6)
u3d.mostrar(fig)
```

**Exercício 8.** Ângulo entre a diagonal $AG$ do cubo e a aresta $AB$, em graus, com 1 casa decimal.

```python
a = 1

# Pitágoras na face BCGF (triângulo BCG, retângulo em C): BG² = a² + a².
bg = math.sqrt(a**2 + a**2)  # a·√2

# Pitágoras no triângulo ABG, retângulo em B: AG² = AB² + BG² = a² + 2a².
ag = math.sqrt(a**2 + bg**2)  # a·√3

# No triângulo retângulo ABG, AB é o cateto adjacente ao ângulo θ em A e AG é a
# hipotenusa: cos(θ) = AB / AG = 1 / √3.
angulo_ex8 = round(math.degrees(math.acos(a / ag)), 1)

print(bg, ag, angulo_ex8)  # 1.414..., 1.732..., 54.7

# Conferência com a função do utils_3d (que usa vetores, assunto do módulo 07).
print(f"Conferência: {u3d.medir_angulo_entre_retas(A, G, A, B):.4f}°")

# O triângulo ABG dentro do cubo, com o ângulo θ em A e o ângulo reto em B.
fig = u3d.criar_figura(titulo="O triângulo ABG")
fig = u3d.desenhar_cubo(fig)
for inicio, fim in [(A, B), (B, G), (G, A)]:
    fig = u3d.desenhar_segmento(fig, inicio, fim, cor=u3d.CORES["reta"], espessura=6)
fig = u3d.desenhar_angulo(fig, A, B, G, rotulo="θ")
fig = u3d.desenhar_angulo_reto(fig, B, A, G)
u3d.mostrar(fig)
```
