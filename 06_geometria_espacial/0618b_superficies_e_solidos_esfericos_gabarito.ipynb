**Exercício 2.** Numa esfera de raio 5: (a) área de uma zona de altura 2; (b) área de uma calota de altura 2; (c) volume do segmento de uma base de altura 2.

A zona e a calota têm a mesma fórmula, $S = 2\pi R h$: com a mesma altura, têm a mesma área.

```python
R, h = 5, 2
area_zona_ex2 = 2 * math.pi * R * h                       # 20π ≈ 62,83
area_calota_ex2 = 2 * math.pi * R * h                     # 20π ≈ 62,83
volume_segmento_ex2 = math.pi * h**2 / 3 * (3 * R - h)    # 52π/3 ≈ 54,45

print(area_zona_ex2, area_calota_ex2, volume_segmento_ex2)

# Conferência da calota por π(a² + h²): o raio da base é a = √(R² − (R − h)²) = 4.
a = math.sqrt(R**2 - (R - h)**2)
print(math.pi * (a**2 + h**2))
```

**Exercício 3.** O volume do setor esférico cuja calota tem altura 2, numa esfera de raio 5.

```python
R, h = 5, 2
volume_setor_ex3 = 2 / 3 * math.pi * R**2 * h   # 100π/3 ≈ 104,72

print(volume_setor_ex3)

# Conferência: o setor é o segmento de uma base (52π/3) mais o cone de vértice no
# centro, com raio 4 e altura 3 (16π = 48π/3).
print(52 * math.pi / 3 + math.pi * 4**2 * 3 / 3)
```

**Exercício 4.** O astronauta a 400 km de altitude ($R = 6371$ km): (a) altura da calota visível; (b) porcentagem visível.

Pelas relações métricas no triângulo retângulo $OTP$ (seção 2): $d = \frac{R^2}{R + H}$ e $h = R - d = \frac{RH}{R + H}$; a fração visível é $\frac{2\pi R h}{4\pi R^2} = \frac{h}{2R}$.

```python
R, H = 6371, 400
altura_calota_ex4 = R * H / (R + H)                  # ≈ 376,37 km
porcentagem_ex4 = altura_calota_ex4 / (2 * R) * 100  # ≈ 2,95%

print(altura_calota_ex4, porcentagem_ex4)
```

**Exercício 5.** A fatia de laranja de raio 5 cm entre o plano do centro e um plano a 3 cm do centro.

É um segmento esférico de duas bases: $a = 5$ (o círculo máximo), $b = \sqrt{25 - 9} = 4$ e $h = 3$.

```python
R = 5
h = 3
a = R                          # o plano passa pelo centro: círculo máximo
b = math.sqrt(R**2 - 3**2)     # 4
volume_fatia_ex5 = math.pi * h / 6 * (3 * a**2 + 3 * b**2 + h**2)   # 66π ≈ 207,35

print(volume_fatia_ex5)

# Conferência por Pappus, com a região do utils_3d:
area, centroide = u3d.medir_area_e_centroide_2d(
    u3d.regiao_meridiana_do_segmento(5, 0, 3, resolucao=2000)
)
print(area * 2 * math.pi * centroide[0])
```

**Exercício 6.** A calota de base de raio 4 e altura 2: (a) o raio da esfera; (b) a área.

Na seção meridiana, $a^2 = h(2R - h)$, ou seja, $2Rh = a^2 + h^2$.

```python
a, h = 4, 2
raio_esfera_ex6 = (a**2 + h**2) / (2 * h)      # 20 / 4 = 5
area_calota_ex6 = math.pi * (a**2 + h**2)      # 20π ≈ 62,83

print(raio_esfera_ex6, area_calota_ex6)
print(2 * math.pi * raio_esfera_ex6 * h)       # conferência por 2πRh
```

**Exercício 7.** O anel de guardanapo: esferas de raio 5 e de raio 10, furadas de modo a sobrar um anel de altura 6 (de $z = -3$ a $z = 3$).

A região que gera o anel é o segmento circular entre o arco e a corda vertical (a parede do furo). Por Pappus, volume $=$ área $\times 2\pi \times$ (distância do centroide ao eixo).

```python
regiao_raio_5 = u3d.regiao_meridiana_do_anel(5, -3, 3)
regiao_raio_10 = u3d.regiao_meridiana_do_anel(10, -3, 3)
area_5, centroide_5 = u3d.medir_area_e_centroide_2d(regiao_raio_5)
area_10, centroide_10 = u3d.medir_area_e_centroide_2d(regiao_raio_10)

volume_raio_5_ex7 = area_5 * 2 * math.pi * centroide_5[0]
volume_raio_10_ex7 = area_10 * 2 * math.pi * centroide_10[0]
volumes_iguais_ex7 = abs(volume_raio_5_ex7 - volume_raio_10_ex7) < 0.05

print(volume_raio_5_ex7, volume_raio_10_ex7, volumes_iguais_ex7)

# A fórmula do anel com a corda vertical (ℓ = h = 6): πh³/6 = 36π ≈ 113,10.
print(math.pi * 6**3 / 6)

# Por que dá igual? A região da esfera maior é bem mais fina (área menor), mas fica
# mais longe do eixo (o centroide percorre um caminho maior).
print(f"raio 5:  área {area_5:.4f}, centroide a {centroide_5[0]:.4f} do eixo")
print(f"raio 10: área {area_10:.4f}, centroide a {centroide_10[0]:.4f} do eixo")
```

**Exercício 8.** Os satélites: (a) a porcentagem vista por um satélite geoestacionário; (b) o limite da porcentagem quando a altitude cresce sem parar.

A fração visível é $\frac{h}{2R} = \frac{RH}{(R + H) \cdot 2R} = \frac{H}{2(R + H)}$. Quando $H$ é enorme, $R + H \approx H$, e a fração se aproxima de $\frac{1}{2}$, ou seja, de 50%, sem nunca chegar lá.

```python
R = 6371
H_geo = 35786
porcentagem_geo_ex8 = H_geo / (2 * (R + H_geo)) * 100   # ≈ 42,44%

for H in [1_000, 100_000, 10_000_000, 1_000_000_000]:
    porcentagem = H / (2 * (R + H)) * 100
    print(f"H = {H:>13,} km: {porcentagem:.6f}%")

limite_ex8 = 50

print(porcentagem_geo_ex8, limite_ex8)
```
