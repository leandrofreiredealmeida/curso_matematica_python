## Básico

**Exercício 1.** Verdadeiro ou falso: (a) a projeção de uma reta é sempre uma reta; (b) o segmento perpendicular é menor que qualquer oblíquo; (c) oblíquos com projeções iguais são iguais; (d) a distância entre reversas é medida na perpendicular comum; (e) o ângulo entre reta e plano pode passar de 90°.

```python
respostas_ex1 = {
    # (a) Falso: se a reta for perpendicular ao plano, a projeção é um único ponto
    #     (seção 1).
    "a": False,
    # (b) Verdadeiro: é a propriedade 1 da seção 2 (PA² = PO² + OA² > PO²).
    "b": True,
    # (c) Verdadeiro: é a propriedade 2 da seção 2 (os cabos de estai).
    "c": True,
    # (d) Verdadeiro: é o teorema da perpendicular comum (seção 3).
    "d": True,
    # (e) Falso: pela definição da seção 4, o ângulo vai de 0° a 90°.
    "e": False,
}
print(respostas_ex1)
```

**Exercício 2.** Distâncias no cubo de aresta 2: (a) $G$ ao plano $ABCD$; (b) $E$ ao plano $BCGF$; (c) entre $AC$ e $FH$; (d) $G$ à reta $AB$.

```python
aresta = 2
distancias_ex2 = {
    # (a) GC é perpendicular à base: d(G, ABCD) = GC = aresta.
    "a": aresta,
    # (b) EF é perpendicular à face BCGF: d(E, BCGF) = EF = aresta.
    "b": aresta,
    # (c) AC está na base e FH no topo, planos paralelos; a perpendicular comum liga
    #     os centros das duas faces e mede a distância entre os planos.
    "c": aresta,
    # (d) Três perpendiculares (notebook 03): GC ⊥ base e CB ⊥ AB, então GB ⊥ AB.
    #     Pitágoras no triângulo GCB: GB = √(2² + 2²) = 2√2.
    "d": math.sqrt(aresta**2 + aresta**2),
}
print(distancias_ex2)

# Conferência com o utils_3d (que usa vetores, assunto do módulo 07).
print(round(u3d.medir_distancia_ponto_plano(G, plano_pelo_nome("ABCD")), 4))
print(round(u3d.medir_distancia_ponto_plano(E, plano_pelo_nome("BCGF")), 4))
print(round(u3d.medir_distancia_entre_retas(A, C, F, H), 4))
print(round(u3d.medir_distancia(G, u3d.pe_da_perpendicular_a_reta(G, A, B)), 4))
```

**Exercício 3.** Ângulo entre a base $ABCD$ e as retas $AE$, $AF$ e $EF$.

```python
angulos_ex3 = {
    # AE é perpendicular à base: 90°.
    "AE": 90,
    # AF é oblíqua; a projeção de F na base é B, então o ângulo é FÂB. No triângulo
    # FBA, retângulo em B: tan θ = FB / AB = 2 / 2 = 1, logo θ = 45°.
    "AF": math.degrees(math.atan(2 / 2)),
    # EF é paralela à base (está no topo): 0°.
    "EF": 0,
}
print(angulos_ex3)

# Conferência com o utils_3d (que usa vetores, assunto do módulo 07).
base = plano_pelo_nome("ABCD")
for reta in angulos_ex3:
    medido = u3d.medir_angulo_reta_plano(vertices[reta[0]], vertices[reta[1]], base)
    print(f"{reta}: {medido:.1f}°")
```

## Intermediário

**Exercício 4.** Rampa de 0,8 m de altura e 10 m de horizontal: inclinação, comprimento e se atende ao limite de 8,33%.

```python
altura = 0.8
horizontal = 10

inclinacao_ex4 = altura / horizontal * 100  # tangente × 100
comprimento_ex4 = math.sqrt(horizontal**2 + altura**2)  # Pitágoras
atende_ex4 = inclinacao_ex4 <= 8.33

print(inclinacao_ex4, comprimento_ex4, atende_ex4)  # 8.0, 10.0319..., True
```

**Exercício 5.** Ângulo entre a diagonal $AG$ e o plano da base.

```python
aresta = 2

# A projeção de G na base é C, então a projeção de AG é a diagonal AC da base.
ac = math.sqrt(aresta**2 + aresta**2)  # Pitágoras no quadrado da base: 2√2
gc = aresta  # a altura de G em relação à base

# No triângulo GCA, retângulo em C: tan θ = GC / AC = 1/√2.
angulo_ex5 = math.degrees(math.atan(gc / ac))

print(angulo_ex5)  # 35.26...

# Conferência com o utils_3d (que usa vetores, assunto do módulo 07).
print(round(u3d.medir_angulo_reta_plano(A, G, plano_pelo_nome("ABCD")), 4))
```

**Exercício 6.** Torre de 12 m com 4 cabos ancorados a 5 m do pé: comprimento de cada cabo e total.

```python
altura_da_torre = 12
distancia_da_ancora = 5
numero_de_cabos = 4

# Cada cabo é um oblíquo com projeção 5: Pitágoras no triângulo POA.
cabo_ex6 = math.sqrt(altura_da_torre**2 + distancia_da_ancora**2)
# Projeções iguais, oblíquos iguais (propriedade 2): os 4 cabos são iguais.
total_ex6 = numero_de_cabos * cabo_ex6

print(cabo_ex6, total_ex6)  # 13.0 52.0
```

**Exercício 7.** Qual plano é o plano mediador da diagonal $AC$ da base: $ABFE$, $BDHF$, $ACGE$ ou $BDE$?

```python
# O plano mediador de AC passa pelo centro da base e é perpendicular a AC.
# B e D são equidistantes de A e C (lados do quadrado), e F e H também (estão acima
# de B e D). Então B, D, F e H estão no plano mediador: é o plano BDHF. (No notebook
# 03 vimos que AC ⊥ BDHF, e BDHF passa pelo ponto médio de AC.)
plano_mediador_ex7 = "BDHF"
print(plano_mediador_ex7)

# Conferência com o utils_3d (que usa vetores, assunto do módulo 07).
mediador = u3d.plano_mediador(A, C)
for nome in ["ABFE", "BDHF", "ACGE", "BDE"]:
    print(nome, u3d.classificar_posicao_planos(mediador, plano_pelo_nome(nome)))
```

**Exercício 8.** Distância entre a diagonal $AG$ e a aresta $BF$.

```python
aresta = 2

# BF é perpendicular à base. Projetando tudo na base: BF vira o ponto B e AG vira a
# diagonal AC. A distância entre AG e BF é a distância de B até a reta AC.
# No quadrado ABCD, essa distância é metade da diagonal BD (as diagonais do quadrado
# são perpendiculares e se cortam ao meio).
bd = math.sqrt(aresta**2 + aresta**2)  # Pitágoras: 2√2
distancia_ex8 = bd / 2  # √2

print(distancia_ex8)  # 1.4142...

# Conferência com o utils_3d (que usa vetores, assunto do módulo 07).
print(round(u3d.medir_distancia_entre_retas(A, G, B, F), 4))
p, q = u3d.perpendicular_comum(A, G, B, F)
print("Pé em AG:", p, "(o centro do cubo)")
print("Pé em BF:", q, "(o ponto médio de BF)")

fig = u3d.criar_figura(titulo="A perpendicular comum a AG e BF")
fig = u3d.desenhar_cubo(fig)
fig = u3d.desenhar_segmento(fig, A, G, cor=u3d.CORES["reta"], espessura=6)
fig = u3d.desenhar_segmento(fig, B, F, cor=u3d.CORES["reta_secundaria"], espessura=6)
fig = u3d.desenhar_perpendicular_comum(fig, A, G, B, F)
fig = u3d.desenhar_projecao_ortogonal(
    fig, [A, G], plano_pelo_nome("ABCD"), cor=u3d.CORES["apoio"]
)
u3d.mostrar(fig)
```

**Exercício 9.** Qual plano é bissetor da base $ABCD$ e da frente $ABFE$: $ABGH$, $ACGE$, $BDE$ ou $BCGF$?

```python
# Um plano bissetor contém a reta de interseção AB; dos quatro, só ABGH contém A e B.
# Conferindo: G está a 2 da base (GC) e a 2 da frente (GF); H também (HD e HE).
# Então A, B, G e H são equidistantes da base e da frente.
plano_bissetor_ex9 = "ABGH"
print(plano_bissetor_ex9)

# Conferência com o utils_3d (que usa vetores, assunto do módulo 07).
base, frente = plano_pelo_nome("ABCD"), plano_pelo_nome("ABFE")
for ponto, nome in [(G, "G"), (H, "H")]:
    d_base = u3d.medir_distancia_ponto_plano(ponto, base)
    d_frente = u3d.medir_distancia_ponto_plano(ponto, frente)
    print(f"{nome}: d(base) = {d_base:.1f}, d(frente) = {d_frente:.1f}")
for numero, bissetor in enumerate(u3d.planos_bissetores(base, frente), start=1):
    posicao = u3d.classificar_posicao_planos(bissetor, plano_pelo_nome("ABGH"))
    print(f"Bissetor {numero} e ABGH: {posicao}")

fig = u3d.criar_figura(titulo="ABGH é bissetor da base e da frente")
fig = u3d.desenhar_cubo(fig)
fig = u3d.desenhar_plano(fig, *base, rotulo="base", tamanho=3, opacidade=0.2)
fig = u3d.desenhar_plano(
    fig, *frente, rotulo="frente", cor=u3d.CORES["plano_terciario"], tamanho=3,
    opacidade=0.2,
)
fig = u3d.desenhar_poligono(
    fig, [A, B, G, H], rotulo="ABGH", cor=u3d.CORES["plano_secundario"]
)
u3d.mostrar(fig)
```
