**Exercício 2.** Cone reto de raio 3 e altura 4: (a) geratriz; (b) ângulo do setor; (c) área lateral; (d) área total; (e) volume.

```python
raio, altura = 3, 4

geratriz_ex2 = math.sqrt(altura**2 + raio**2)  # (a) √(16 + 9) = 5
angulo_ex2 = 360 * raio / geratriz_ex2  # (b) 360° · 3/5 = 216°
area_lateral_ex2 = math.pi * raio * geratriz_ex2  # (c) π · 3 · 5 = 15π
area_total_ex2 = math.pi * raio * (geratriz_ex2 + raio)  # (d) π · 3 · 8 = 24π
volume_ex2 = math.pi * raio**2 * altura / 3  # (e) π · 9 · 4 / 3 = 12π

print(geratriz_ex2, angulo_ex2, area_lateral_ex2, area_total_ex2, volume_ex2)

# Conferência com o utils_3d: uma pirâmide regular de 2000 lados inscrita no cone
# quase não difere dele.
solido = u3d.piramide(u3d.poligono_regular(2000, raio=raio), (0, 0, altura))
print(f"Conferência: área total {u3d.medir_area_total(*solido):.2f}")
print(f"Conferência: volume {u3d.medir_volume(*solido):.2f}")
```

**Exercício 3.** Cone equilátero de raio 2: (a) geratriz; (b) altura; (c) ângulo do setor; (d) área total; (e) volume.

```python
raio = 2

geratriz_ex3 = 2 * raio  # (a) no cone equilátero, g = 2r = 4
altura_ex3 = raio * math.sqrt(3)  # (b) h = r√3 = 2√3 (Pitágoras: 4² = h² + 2²)
angulo_ex3 = 360 * raio / geratriz_ex3  # (c) 360° · 2/4 = 180°: um semicírculo
area_total_ex3 = 3 * math.pi * raio**2  # (d) S_t = πr(2r + r) = 3πr² = 12π
volume_ex3 = math.pi * raio**2 * altura_ex3 / 3  # (e) π · 4 · 2√3 / 3 = 8√3π/3

print(geratriz_ex3, altura_ex3, angulo_ex3, area_total_ex3, volume_ex3)
```

**Exercício 4.** Chapéu feito com um setor de raio 30 cm e ângulo central de 120°: (a) raio da base; (b) altura.

```python
# (a) O arco do setor vira a circunferência da base: 2πr = 2π · 30 · 120/360.
comprimento_do_arco = 2 * math.pi * 30 * 120 / 360
raio_ex4 = comprimento_do_arco / (2 * math.pi)  # 10 cm

# (b) O raio do papel (30 cm) vira a geratriz: h² = g² − r² = 900 − 100 = 800.
altura_ex4 = math.sqrt(30**2 - raio_ex4**2)  # √800 = 20√2 ≈ 28,28 cm

print(raio_ex4, altura_ex4)
```

**Exercício 6.** Monte de areia em forma de cone, com 4 m de diâmetro e 1,5 m de altura: (a) volume; (b) viagens de um caminhão de 5 m³.

```python
raio = 4 / 2  # o raio é metade do diâmetro: 2 m

volume_ex6 = math.pi * raio**2 * 1.5 / 3  # (a) π · 4 · 1,5 / 3 = 2π ≈ 6,28 m³

# (b) 6,28 / 5 = 1,26 caminhões: uma viagem não basta, então são 2. O math.ceil
# arredonda para cima ("teto").
viagens_ex6 = math.ceil(volume_ex6 / 5)

print(volume_ex6, viagens_ex6)
```

**Exercício 7.** A função `classificar_conica(semiabertura, inclinacao_do_plano)` e a classificação das seções de um cone de semiabertura 30° por planos inclinados 0°, 40°, 60° e 80°.

```python
def classificar_conica(semiabertura: float, inclinacao_do_plano: float) -> str:
    """Classifica a seção de um cone duplo por um plano que não passa pelo vértice."""
    angulo_das_geratrizes = 90 - semiabertura  # ângulo das geratrizes com a horizontal
    if inclinacao_do_plano == 0:
        return "circunferência"  # plano perpendicular ao eixo
    if inclinacao_do_plano < angulo_das_geratrizes:
        return "elipse"  # menos inclinado que as geratrizes: corta uma folha só
    if inclinacao_do_plano == angulo_das_geratrizes:
        return "parábola"  # paralelo a uma geratriz
    return "hipérbole"  # mais inclinado: corta as duas folhas


# Com semiabertura 30°, as geratrizes formam 60° com a horizontal.
classificacoes_ex7 = [classificar_conica(30, i) for i in [0, 40, 60, 80]]

print(classificacoes_ex7)

# As quatro seções, desenhadas como na seção 1.
for inclinacao in [0, 40, 60, 80]:
    i = math.radians(inclinacao)
    plano = ((0.5, 0, 1), (math.sin(i), 0, math.cos(i)))
    titulo = f"Plano inclinado {inclinacao}°: {classificar_conica(30, inclinacao)}"
    fig = u3d.criar_figura(titulo=titulo, altura=380)
    diretriz = u3d.poligono_regular(
        80, raio=3 * math.tan(math.radians(30)), centro=(0, 0, 3)
    )
    fig = u3d.desenhar_superficie_conica(
        fig, (0, 0, 0), diretriz + diretriz[:1], opacidade=0.25, prolongar=1
    )
    for trecho in u3d.secao_plana_do_cone_duplo(30, plano):
        fig = u3d.desenhar_curva_no_espaco(
            fig, trecho, cor=u3d.CORES["destaque"], espessura=6
        )
    u3d.mostrar(fig)
```

**Exercício 8.** Disco de papel de raio 10 cm: o ângulo do setor (de 1° a 359°, de 0,1° em 0,1°) que dá o cone de maior volume.

```python
melhor_angulo = None
maior_volume = 0  # qualquer volume calculado será maior que zero

for k in range(10, 3591):
    angulo = k / 10
    raio = 10 * angulo / 360  # o arco do setor vira a circunferência da base
    altura = math.sqrt(100 - raio**2)  # o raio do papel (10) vira a geratriz
    volume = math.pi * raio**2 * altura / 3
    # Sempre que aparece um volume maior, ele e o seu ângulo passam a ser os
    # "campeões".
    if volume > maior_volume:
        maior_volume = volume
        melhor_angulo = angulo

angulo_ex8 = melhor_angulo

print(angulo_ex8)
print(f"Maior volume: {maior_volume:.2f} cm³")
```

Repare que o cone de maior volume não é nem o mais largo nem o mais alto: com o ângulo perto de 293,9°, o raio fica perto de 8,16 cm e a altura perto de 5,77 cm. Encontrar máximos e mínimos de forma exata é um dos objetivos do Cálculo (módulo 11).
