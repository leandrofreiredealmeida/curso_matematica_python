**Exercício 2.** Cubo de aresta 6: (a) $r$; (b) $\rho$; (c) $R$.

```python
a = 6
r_ex2 = a / 2  # (a) metade da aresta: 3
rho_ex2 = a * math.sqrt(2) / 2  # (b) metade da diagonal da face: 3√2 ≈ 4,243
R_ex2 = a * math.sqrt(3) / 2  # (c) metade da diagonal do cubo: 3√3 ≈ 5,196

print(r_ex2, rho_ex2, R_ex2)
```

**Exercício 3.** Tetraedro regular de aresta 6: (a) altura; (b) $r$; (c) $R$.

```python
a = 6
altura_ex3 = a * math.sqrt(6) / 3  # (a) 2√6 ≈ 4,899
r_ex3 = altura_ex3 / 4  # (b) o centro fica a um quarto da altura: √6/2 ≈ 1,225
R_ex3 = 3 * altura_ex3 / 4  # (c) o resto da altura: 3√6/2 ≈ 3,674

print(altura_ex3, r_ex3, R_ex3)
```

## Intermediário

**Exercício 4.** Bola inscrita numa caixa cúbica de aresta 20 cm: (a) volume de ar; (b) porcentagem de ar.

```python
aresta = 20
raio = aresta / 2  # a bola encosta nas seis faces: r = a/2 = 10
volume_caixa = aresta**3  # 8000 cm³
volume_bola = 4 / 3 * math.pi * raio**3  # 4000π/3 ≈ 4188,79 cm³

ar_ex4 = volume_caixa - volume_bola  # (a) ≈ 3811,21 cm³
porcentagem_ex4 = ar_ex4 / volume_caixa * 100  # (b) ≈ 47,64%

print(ar_ex4, porcentagem_ex4)
```

**Exercício 5.** Octaedro regular de aresta 4: (a) $R$; (b) volume; (c) área total; (d) $r$ pela fórmula $r = \frac{3 \cdot \text{volume}}{S}$.

```python
a = 4
R_ex5 = a * math.sqrt(2) / 2  # (a) metade da diagonal do quadrado do meio: 2√2 ≈ 2,828
# (b) Duas pirâmides de base a² e altura a√2/2: 2 · (1/3) · a² · a√2/2 = a³√2/3.
volume_ex5 = a**3 * math.sqrt(2) / 3  # 64√2/3 ≈ 30,170
# (c) Oito triângulos equiláteros de área a²√3/4: 2a²√3.
area_ex5 = 2 * a**2 * math.sqrt(3)  # 32√3 ≈ 55,426
r_ex5 = 3 * volume_ex5 / area_ex5  # (d) 2√6/3 ≈ 1,633

print(R_ex5, volume_ex5, area_ex5, r_ex5)
```

**Exercício 6.** Tetraedro formado por quatro vértices de um cubo de aresta 3: (a) aresta; (b) volume.

```python
a = 3
aresta_ex6 = a * math.sqrt(2)  # (a) diagonal de uma face do cubo: 3√2 ≈ 4,243
# (b) O cubo menos 4 "cantos", pirâmides de base a²/2 e altura a: cada uma a³/6.
volume_ex6 = a**3 - 4 * (a**3 / 6)  # a³/3 = 9

print(aresta_ex6, volume_ex6)
```

**Exercício 7.** O ângulo entre duas ligações do metano (tetraedro dentro de um cubo de aresta 2, com o carbono no centro $O$).

```python
# No triângulo AOC: OA e OC são metades da diagonal do cubo; AC é a diagonal de uma face.
a = 2
OA = OC = a * math.sqrt(3) / 2  # √3
AC = a * math.sqrt(2)  # 2√2

# Lei dos cossenos: AC² = OA² + OC² − 2 · OA · OC · cos θ.
cos_theta = (OA**2 + OC**2 - AC**2) / (2 * OA * OC)  # (3 + 3 − 8) / 6 = −1/3
angulo_ex7 = math.degrees(math.acos(cos_theta))  # ≈ 109,47°

print(f"cos θ = {cos_theta:.4f}")
print(angulo_ex7)

# Conferência com o utils_3d e o desenho da molécula.
v = u3d.vertices_do_cubo(2)
carbono = (1, 1, 1)
print(f"Ângulo medido com o utils_3d: {u3d.medir_angulo(carbono, v['A'], v['C']):.4f}°")
fig = u3d.criar_figura(titulo="A molécula de metano dentro do cubo")
fig = u3d.desenhar_cubo(fig, 2, opacidade=0.05)
fig = u3d.desenhar_poliedro(
    fig, *u3d.piramide([v["A"], v["C"], v["F"]], v["H"]), opacidade=0.15
)
for nome in "ACFH":
    fig = u3d.desenhar_segmento(fig, carbono, v[nome], cor=u3d.CORES["destaque"], espessura=6)
fig = u3d.desenhar_ponto(fig, carbono, rotulo="C (carbono)", tamanho=10)
u3d.mostrar(fig)
```

**Exercício 8.** Razão entre a esfera de Saturno e a de Mercúrio no modelo de Kepler.

```python
ordem_ex8 = ["cubo", "tetraedro", "dodecaedro", "icosaedro", "octaedro"]
razao_ex8 = 1  # um produto começa em 1
for nome in ordem_ex8:
    vertices, faces = u3d.poliedro_de_platao(nome, aresta=1)
    raios = u3d.raios_do_poliedro_regular(vertices, faces)
    # Cada poliedro "encolhe" a esfera pela razão R/r: a de fora dividida pela de dentro.
    razao_ex8 = razao_ex8 * raios["circunscrita"] / raios["inscrita"]

# √3 · 3 · 1,2584 · 1,2584 · √3 = 9 · 1,2584² ≈ 14,25
print(razao_ex8)
```

Quer dizer: no modelo de Kepler, a esfera de Saturno teria um raio cerca de **14 vezes** o da esfera de Mercúrio. Na realidade, a órbita de Saturno é cerca de **25 vezes** maior que a de Mercúrio. A beleza geométrica do modelo não resistiu às medidas: foi justamente insistindo nos dados que Kepler chegou às suas leis do movimento dos planetas.
