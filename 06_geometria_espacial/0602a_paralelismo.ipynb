## Dois aviões, um cruzamento... e nenhum perigo?

Na tela de um controlador de tráfego aéreo, as rotas de dois aviões aparecem como linhas que se cruzam. Parece que eles vão bater! Mas o controlador está tranquilo. Por quê?

Porque o radar mostra o céu **visto de cima**, achatado num plano. Na vida real, os aviões voam em **níveis de voo**: altitudes fixas, como andares invisíveis no céu. Na aviação comercial, aviões em rotas que se cruzam costumam ser separados por pelo menos 1.000 pés (cerca de 300 metros) de altura.

Cada nível é um **plano horizontal**, e todos esses planos são **paralelos** entre si. Uma rota num nível e outra rota em outro nível **nunca** se encontram, mesmo que no radar pareçam se cruzar.

🕹️ Use os controles abaixo: mude a diferença de altitude entre os aviões e alterne entre a visão do radar (de cima) e a visão 3D.

```python
# As coordenadas (x, y, z) abaixo servem só para posicionar o desenho.
VISTAS = ["Radar (de cima)", "3D"]


def desenhar_niveis_de_voo(diferenca_de_nivel: float, vista: str) -> None:
    """Desenha as rotas de dois aviões, cada uma no seu nível de voo."""
    fig = u3d.criar_figura(titulo="Dois aviões, dois níveis de voo")

    # Cada nível de voo é um plano horizontal. Com diferença 0, há um nível só.
    fig = u3d.desenhar_plano(
        fig, ponto=(0, 0, 0), normal=(0, 0, 1), rotulo="nível 1", tamanho=8,
        cor=u3d.CORES["plano"], opacidade=0.2,
    )
    if diferenca_de_nivel > 0:
        fig = u3d.desenhar_plano(
            fig, ponto=(0, 0, diferenca_de_nivel), normal=(0, 0, 1),
            rotulo="nível 2", tamanho=8, cor=u3d.CORES["plano_secundario"],
            opacidade=0.2,
        )

    # As duas rotas passam, vistas de cima, pelo mesmo ponto (0, 0).
    fig = u3d.desenhar_reta(
        fig, ponto=(0, 0, 0), direcao=(1, 0.3, 0), rotulo="rota 1",
        cor=u3d.CORES["reta"], comprimento=8, espessura=6,
    )
    fig = u3d.desenhar_reta(
        fig, ponto=(0, 0, diferenca_de_nivel), direcao=(-0.4, 1, 0), rotulo="rota 2",
        cor=u3d.CORES["reta_secundaria"], comprimento=8, espessura=6,
    )

    # Um avião sobre cada rota, ainda antes do "cruzamento".
    fig = u3d.desenhar_ponto(fig, (-2.5, -0.75, 0), cor=u3d.CORES["reta"], tamanho=10)
    fig = u3d.desenhar_ponto(
        fig, (1, -2.5, diferenca_de_nivel), cor=u3d.CORES["reta_secundaria"],
        tamanho=10,
    )

    if vista == VISTAS[0]:
        # Câmera bem em cima, olhando para baixo: é a visão do radar.
        fig = u3d.definir_camera(fig, olho=(0, 0, 2.5))
    u3d.mostrar(fig)

    if diferenca_de_nivel == 0:
        print(
            "Mesmo nível: as rotas são concorrentes. Os aviões podem se encontrar "
            "no cruzamento!"
        )
    else:
        print(
            "Níveis diferentes: as rotas são reversas. No radar parecem se cruzar, "
            "mas nunca se encontram."
        )


interact(
    desenhar_niveis_de_voo,
    diferenca_de_nivel=widgets.FloatSlider(
        min=0, max=3, step=0.5, value=1.5, description="Diferença de nível",
        continuous_update=False,
    ),
    vista=widgets.RadioButtons(options=VISTAS, value=VISTAS[0], description="Vista"),
);
```

## 1. Paralelismo de retas

Relembrando do notebook 01:

> 📌 **Definição.** Duas retas são **paralelas** quando estão contidas num mesmo plano e não têm ponto em comum (ou quando são coincidentes, caso em que dizemos que são paralelas coincidentes).

No plano, você já conhece a regra mais famosa sobre paralelas. Ela também vale no espaço:

**Postulado das paralelas (postulado de Euclides)**
> Por um ponto fora de uma reta passa **uma única** reta paralela a ela.

Por que vale no espaço? O ponto $P$ e a reta $r$ determinam um plano (é o modo 2 do notebook 01). A paralela a $r$ por $P$ precisa estar nesse plano, e dentro dele vale o que você já sabe da Geometria Plana: só existe uma.

> 🕵️ **Curiosidade histórica.** Durante mais de 2.000 anos, matemáticos tentaram **provar** o postulado das paralelas a partir dos outros postulados de Euclides, sem sucesso. No século XIX, **Gauss**, **Bolyai** e **Lobachevsky** perceberam que dava para construir geometrias perfeitamente coerentes em que ele é falso: as **geometrias não euclidianas**. Décadas depois, uma geometria desse tipo virou a linguagem da Teoria da Relatividade Geral de Einstein, que descreve a gravidade como uma curvatura do espaço-tempo.

🕹️ Mova o ponto $P$ abaixo: em qualquer posição, existe uma única paralela a $r$ passando por ele.

```python
import math


def desenhar_paralela_unica(posicao_de_P: int) -> None:
    """Desenha a reta r, o ponto P e a única paralela a r que passa por P."""
    # Dois pontos da reta r (que fica sobre o "eixo x" do desenho).
    r1, r2 = (-1, 0, 0), (1, 0, 0)

    # Seno e cosseno (módulo 05) servem só para mover o ponto P em volta da reta r,
    # sempre à mesma distância dela.
    theta = math.radians(posicao_de_P)
    P = (0, 2 * math.cos(theta), 2 * math.sin(theta))

    fig = u3d.criar_figura(titulo="r e a única paralela a r passando por P")

    # O plano determinado por r e P (modo 2): a paralela fica dentro dele.
    fig = u3d.desenhar_plano_por_tres_pontos(fig, r1, r2, P, opacidade=0.15)

    fig = u3d.desenhar_reta_por_dois_pontos(
        fig, r1, r2, rotulo="r", cor=u3d.CORES["reta"], comprimento=8, espessura=6
    )
    fig = u3d.desenhar_paralela_por_ponto(
        fig, r1, r2, P, rotulo="s", cor=u3d.CORES["reta_secundaria"],
        comprimento=8, espessura=6,
    )
    fig = u3d.desenhar_ponto(fig, P, rotulo="P", cor=u3d.CORES["destaque"], tamanho=8)
    u3d.mostrar(fig)


interact(
    desenhar_paralela_unica,
    posicao_de_P=widgets.IntSlider(
        min=0, max=360, step=30, value=60, description="posição de P (°)",
        continuous_update=False,
    ),
);
```

### Duas retas paralelas a uma terceira

No plano, se $a \parallel b$ e $b \parallel c$, então $a \parallel c$. No espaço isso continua verdade, mas **não é óbvio**: as três retas podem não estar num mesmo plano!

**Teorema (transitividade do paralelismo)**
> Se $a \parallel b$ e $b \parallel c$, então $a \parallel c$.

Pense em três arestas verticais de um cubo, como $AE$, $BF$ e $CG$. A aresta $AE$ é paralela a $BF$ (estão na face da frente), e $BF$ é paralela a $CG$ (estão na face da direita). Não existe uma face do cubo com $AE$ e $CG$ ao mesmo tempo. Mesmo assim, o teorema garante que $AE \parallel CG$: deve existir **algum** plano que contém as duas.

🕹️ Marque a caixa no widget abaixo para encontrá-lo.

<details>
<summary>Ver demonstração</summary>

Suponha $a$, $b$, $c$ distintas, com $a$ e $c$ fora de um mesmo plano com $b$ (o caso em que as três são coplanares é o da Geometria Plana).

1. Seja $\alpha$ o plano de $a$ e $b$, e $\gamma$ o plano de $b$ e $c$.
2. Escolha um ponto $P$ em $c$. A reta $a$ e o ponto $P$ determinam um plano $\delta$.
3. Os planos $\delta$ e $\gamma$ são distintos e têm o ponto $P$ em comum, logo se cortam numa reta $c'$ que passa por $P$.
4. A reta $c'$ não encontra $b$: se encontrasse num ponto $Q$, esse ponto estaria em $\delta$ e em $\alpha$, ou seja, na reta $a$ (que é a interseção de $\delta$ com $\alpha$). Mas $a$ e $b$ não têm ponto comum. Contradição.
5. Então $c'$ e $b$ estão no plano $\gamma$ sem se encontrar: são paralelas. Pelo postulado das paralelas, a única paralela a $b$ passando por $P$ é $c$. Logo $c' = c$.
6. Assim, $c$ está no plano $\delta$, junto com $a$. E $a$ e $c$ não se encontram (se se encontrassem num ponto $R$, haveria duas paralelas a $b$ passando por $R$). Portanto $a \parallel c$. ∎
</details>

```python
# Os vértices do cubo, com os nomes do módulo (A, B, C, D embaixo; E, F, G, H em cima).
vertices = u3d.vertices_do_cubo()
A, B, C, D, E, F, G, H = (vertices[nome] for nome in "ABCDEFGH")


def desenhar_transitividade(mostrar_plano: bool) -> None:
    """Desenha o cubo, as arestas AE, BF e CG e os planos que as contêm."""
    fig = u3d.criar_figura(titulo="AE ∥ BF e BF ∥ CG. E AE e CG?")
    fig = u3d.desenhar_cubo(fig)

    # O plano da face ABFE (contém AE e BF) e o da face BCGF (contém BF e CG).
    fig = u3d.desenhar_plano_por_tres_pontos(
        fig, A, B, F, cor=u3d.CORES["plano"], opacidade=0.2, tamanho=1
    )
    fig = u3d.desenhar_plano_por_tres_pontos(
        fig, B, C, G, cor=u3d.CORES["plano_secundario"], opacidade=0.2, tamanho=1
    )
    if mostrar_plano:
        # O plano diagonal ACGE: ele corta o cubo "na diagonal" e contém AE e CG.
        fig = u3d.desenhar_plano_por_tres_pontos(
            fig, A, C, G, cor=u3d.CORES["plano_terciario"], opacidade=0.3, tamanho=1
        )

    # As três arestas, redesenhadas por cima do cubo, mais grossas.
    fig = u3d.desenhar_segmento(fig, A, E, cor=u3d.CORES["reta"], espessura=7)
    fig = u3d.desenhar_segmento(fig, B, F, cor=u3d.CORES["destaque"], espessura=7)
    fig = u3d.desenhar_segmento(
        fig, C, G, cor=u3d.CORES["reta_secundaria"], espessura=7
    )
    u3d.mostrar(fig)
    print(f"Posição de AE e CG: {u3d.classificar_posicao_retas(A, E, C, G)}")


interact(
    desenhar_transitividade,
    mostrar_plano=widgets.Checkbox(
        value=False, description="Mostrar o plano de AE e CG", indent=False
    ),
);
```

## 2. Posições relativas de uma reta e um plano

Uma reta $r$ e um plano $\alpha$ podem ter:

| Pontos em comum | Posição | Notação | Exemplo do dia a dia |
|---|---|---|---|
| infinitos (a reta toda) | $r$ **está contida** em $\alpha$ | $r \subset \alpha$ | uma régua deitada sobre a mesa |
| exatamente um | $r$ e $\alpha$ são **concorrentes** (ou secantes) | $r \cap \alpha = \{P\}$ | um poste fincado no chão |
| nenhum | $r$ e $\alpha$ são **paralelos** | $r \cap \alpha = \varnothing$ | um varal esticado acima do chão |

> 📌 **Definição.** Uma reta e um plano são **paralelos** quando não têm ponto em comum.

Não existe a possibilidade de exatamente dois pontos em comum: pelo postulado da inclusão (notebook 01), se a reta tiver dois pontos no plano, ela está inteira dentro dele.

🕹️ Mexa na inclinação e na altura da reta e observe as três posições.

> 💡 **Dica.** O widget confere a posição com a função `u3d.classificar_posicao_reta_plano`. Ela faz contas com vetores, assunto do módulo 07 (Geometria Analítica); aqui ela serve só como "professor" que confere o seu raciocínio.

```python
def desenhar_reta_e_plano(inclinacao: int, altura: float) -> None:
    """Desenha o plano horizontal α e uma reta r com a inclinação e a altura dadas."""
    alfa = ((0, 0, 0), (0, 0, 1))  # o plano α, como (ponto, normal)

    # Seno e cosseno (módulo 05) servem só para inclinar o desenho da reta.
    theta = math.radians(inclinacao)
    direcao = (math.cos(theta), 0, math.sin(theta))
    P1 = (0, 0, altura)
    P2 = (direcao[0], 0, altura + direcao[2])

    posicao = u3d.classificar_posicao_reta_plano(P1, P2, alfa)
    ponto_de_encontro = u3d.intersecao_reta_plano(P1, P2, alfa)

    # Se r fura α, centramos o desenho no ponto de encontro, para que ele apareça
    # dentro do pedaço de plano desenhado. Senão, centramos embaixo de P1.
    if ponto_de_encontro is None:
        centro_do_plano, centro_da_reta = (0, 0, 0), P1
    else:
        centro_do_plano, centro_da_reta = ponto_de_encontro, ponto_de_encontro

    fig = u3d.criar_figura(titulo="Uma reta r e um plano α")
    fig = u3d.desenhar_plano(fig, ponto=centro_do_plano, normal=alfa[1], rotulo="α")
    fig = u3d.desenhar_reta(
        fig, ponto=centro_da_reta, direcao=direcao, rotulo="r", comprimento=8,
        espessura=6,
    )
    fig = u3d.desenhar_intersecao_reta_plano(fig, P1, P2, alfa, rotulo="P")
    u3d.mostrar(fig)

    mensagens = {
        "contida": "r está contida em α: todos os pontos de r estão no plano.",
        "concorrente": "r fura α no ponto P: reta e plano são concorrentes.",
        "paralela": (
            "r não tem nenhum ponto em comum com α: reta e plano são paralelos."
        ),
    }
    print(mensagens[posicao])


interact(
    desenhar_reta_e_plano,
    inclinacao=widgets.IntSlider(
        min=0, max=90, step=15, value=30, description="inclinação (°)",
        continuous_update=False,
    ),
    altura=widgets.FloatSlider(
        min=-2, max=2, step=0.5, value=1, description="altura",
        continuous_update=False,
    ),
);
```

### Como garantir que uma reta é paralela a um plano?

Verificar que uma reta "nunca" encontra um plano infinito parece impossível na prática. O teorema abaixo resolve o problema: basta encontrar **uma** reta do plano que seja paralela a ela.

**Teorema (condição de paralelismo entre reta e plano)**
> Se uma reta $a$ não está contida num plano $\alpha$ e é paralela a uma reta $b$ de $\alpha$, então $a$ é paralela a $\alpha$.

<details>
<summary>Ver demonstração</summary>

As retas paralelas $a$ e $b$ determinam um plano $\beta$, e a interseção de $\beta$ com $\alpha$ é a reta $b$. Se $a$ encontrasse $\alpha$ num ponto $P$, esse ponto estaria em $\beta$ e em $\alpha$, ou seja, em $b$. Mas $a$ e $b$ não têm ponto comum. Logo $a$ não encontra $\alpha$: são paralelos. ∎
</details>

🎯 **Aplicação.** Um eletricista quer saber se um fio esticado entre dois postes está paralelo ao chão plano. Ele não precisa de nenhum instrumento sofisticado: ao meio-dia, com o sol quase a pino, o fio projeta uma sombra no chão. Se o fio estiver paralelo à sombra (mesma distância do chão nas duas pontas), ele é paralelo a uma reta do chão e, portanto, paralelo ao chão.

Vale também a volta: **se $a \parallel \alpha$, então existem retas de $\alpha$ paralelas a $a$** (infinitas delas). Mas cuidado:

> ⚠️ **Erro comum.** Achar que uma reta paralela a um plano é paralela a **todas** as retas do plano. Ela é paralela só às retas do plano que têm a mesma direção dela; com todas as outras, ela é **reversa**.

🕹️ Gire a reta $b$ dentro do plano e compare com a reta $a$, que está acima do plano e paralela a ele.

```python
def desenhar_reta_paralela_ao_plano(angulo_de_b: int) -> None:
    """Desenha a reta a, paralela ao plano α, e uma reta b de α que gira."""
    # A reta a fica 2 unidades acima do plano α, na direção do "eixo x".
    a1, a2 = (0, 0, 2), (1, 0, 2)

    # Seno e cosseno (módulo 05) servem só para girar a reta b dentro do plano α.
    theta = math.radians(angulo_de_b)
    b1, b2 = (0, 0, 0), (math.cos(theta), math.sin(theta), 0)

    fig = u3d.criar_figura(titulo="a ∥ α e uma reta b de α")
    fig = u3d.desenhar_plano(
        fig, ponto=(0, 0, 0), normal=(0, 0, 1), rotulo="α", tamanho=8
    )
    fig = u3d.desenhar_reta_por_dois_pontos(
        fig, a1, a2, rotulo="a", cor=u3d.CORES["reta"], comprimento=8, espessura=6
    )
    fig = u3d.desenhar_reta_por_dois_pontos(
        fig, b1, b2, rotulo="b", cor=u3d.CORES["reta_secundaria"], comprimento=8,
        espessura=6,
    )
    u3d.mostrar(fig)
    print(f"Posição de a e b: {u3d.classificar_posicao_retas(a1, a2, b1, b2)}")


interact(
    desenhar_reta_paralela_ao_plano,
    angulo_de_b=widgets.IntSlider(
        min=0, max=180, step=15, value=45, description="ângulo de b (°)",
        continuous_update=False,
    ),
);
```

## 3. Duas retas reversas

Relembrando: duas retas são **reversas** quando não existe um plano que contenha as duas.

**Um jeito prático de reconhecer retas reversas**
> Se a reta $r$ está contida num plano $\alpha$ e a reta $s$ fura $\alpha$ num ponto $P$ que **não** está em $r$, então $r$ e $s$ são reversas.

Por quê? Se existisse um plano com $r$ e $s$, ele conteria $r$ e o ponto $P$, então seria o próprio $\alpha$ (modo 2 de determinação). Mas $s$ não está em $\alpha$, só fura o plano. Contradição.

Agora, um resultado importante para os próximos notebooks:

**Teorema**
> Se $r$ e $s$ são reversas, existe **um único** plano que contém $r$ e é paralelo a $s$.

**Construção, passo a passo**
1. Escolha um ponto $P$ qualquer da reta $r$.
2. Trace por $P$ a reta $s'$ paralela a $s$ (postulado das paralelas).
3. As retas $r$ e $s'$ são concorrentes em $P$; elas determinam um plano $\alpha$ (modo 3).
4. A reta $s$ é paralela a $s'$, que está em $\alpha$, e $s$ não está em $\alpha$. Pela condição de paralelismo da seção 2, $s \parallel \alpha$.

Qualquer plano que contenha $r$ e seja paralelo a $s$ precisa conter a paralela a $s$ por $P$; por isso o plano é único.

🕹️ Avance os passos no controle abaixo.

```python
# Duas retas reversas "genéricas", fora do cubo (dois pontos de cada uma).
r1, r2 = (-2, 0, 0), (2, 0, 0)
s1, s2 = (0, -2, 2), (0, 2, 3)
P = (-1, 0, 0)  # um ponto qualquer de r

TEXTOS_DOS_PASSOS = {
    1: (
        "Passo 1: as retas r e s são "
        f"{u3d.classificar_posicao_retas(r1, r2, s1, s2)}. Queremos um plano que "
        "contenha r e seja paralelo a s."
    ),
    2: "Passo 2: escolha um ponto P qualquer da reta r.",
    3: "Passo 3: trace por P a reta s' paralela a s (postulado das paralelas).",
    4: (
        "Passo 4: r e s' são concorrentes em P e determinam o plano α (modo 3). "
        "Como s ∥ s', com s' ⊂ α e s fora de α, temos s ∥ α."
    ),
}


def desenhar_construcao(passo: int) -> None:
    """Desenha a construção do plano por r paralelo a s, até o passo escolhido."""
    fig = u3d.criar_figura(titulo=f"Plano por r paralelo a s: passo {passo}")
    fig = u3d.desenhar_reta_por_dois_pontos(
        fig, r1, r2, rotulo="r", cor=u3d.CORES["reta"], comprimento=8, espessura=6
    )
    fig = u3d.desenhar_reta_por_dois_pontos(
        fig, s1, s2, rotulo="s", cor=u3d.CORES["reta_secundaria"], comprimento=8,
        espessura=6,
    )
    if passo >= 2:
        fig = u3d.desenhar_ponto(
            fig, P, rotulo="P", cor=u3d.CORES["destaque"], tamanho=8
        )
    if passo >= 3:
        fig = u3d.desenhar_paralela_por_ponto(
            fig, s1, s2, P, rotulo="s'", cor=u3d.CORES["reta_secundaria"],
            comprimento=8, espessura=5, tracejado=True,
        )
    if passo >= 4:
        fig = u3d.desenhar_plano_paralelo_a_reta(
            fig, r1, r2, s1, s2, rotulo="α", opacidade=0.25
        )
    u3d.mostrar(fig)

    print(TEXTOS_DOS_PASSOS[passo])
    if passo == 4:
        alfa = u3d.plano_paralelo_a_reta(r1, r2, s1, s2)
        print(f"Posição de s e α: {u3d.classificar_posicao_reta_plano(s1, s2, alfa)}")


interact(
    desenhar_construcao,
    passo=widgets.IntSlider(
        min=1, max=4, step=1, value=1, description="passo", continuous_update=False
    ),
);
```

## 4. Paralelismo entre planos

> 📌 **Definição.** Dois planos são **paralelos** quando não têm ponto em comum.

**Posições relativas de dois planos** (você já viu no notebook 01):

| Pontos em comum | Posição |
|---|---|
| todos | **coincidentes** ($\alpha = \beta$) |
| uma reta | **secantes** ($\alpha \cap \beta = i$) |
| nenhum | **paralelos** ($\alpha \cap \beta = \varnothing$) |

Como verificar que dois planos infinitos nunca se encontram? De novo, um teorema torna isso prático:

**Teorema (condição de paralelismo entre planos)**
> Se um plano $\alpha$ contém **duas retas concorrentes**, ambas paralelas a um plano $\beta$, então $\alpha$ e $\beta$ são paralelos.

<details>
<summary>Ver demonstração</summary>

Sejam $a$ e $b$ as retas de $\alpha$, concorrentes em $P$, ambas paralelas a $\beta$. Suponha, por absurdo, que $\alpha$ e $\beta$ se cortem numa reta $i$.

A reta $a$ está em $\alpha$, assim como $i$. Se $a$ encontrasse $i$, encontraria $\beta$, o que é impossível (pois $a \parallel \beta$). Então $a$ e $i$ são coplanares e sem ponto comum: $a \parallel i$. Pelo mesmo raciocínio, $b \parallel i$.

Assim, $a$ e $b$ seriam duas retas distintas passando por $P$, ambas paralelas a $i$. Isso contradiz o postulado das paralelas. Logo $\alpha$ e $\beta$ não se cortam: são paralelos. ∎
</details>

🎯 **Aplicação: o nível de bolha.** Um pedreiro quer saber se o piso que está construindo está na horizontal, ou seja, paralelo ao plano horizontal. Ele coloca o nível de bolha numa direção e confere; depois gira o nível e confere numa direção **que cruza a primeira**. Se a bolha ficar centralizada nas duas, o piso está nivelado. É exatamente o teorema: duas retas concorrentes do piso, ambas paralelas ao plano horizontal.

> ⚠️ **Erro comum.** Conferir uma direção só. Uma única reta do piso paralela ao plano horizontal não garante nada: o piso pode estar inclinado na outra direção. Duas direções **paralelas** entre si também não resolvem; elas precisam ser **concorrentes**.

🕹️ Teste no "piso" abaixo: incline-o nas duas direções e veja o que os dois níveis de bolha dizem.

```python
def desenhar_nivel_de_bolha(inclinacao_1: int, inclinacao_2: int) -> None:
    """Desenha o piso β, inclinado nas duas direções, e os dois níveis de bolha."""
    alfa = ((0, 0, 0), (0, 0, 1))  # o plano horizontal de referência
    P = (0, 0, 2)  # o centro do piso

    # Seno e cosseno (módulo 05) servem só para inclinar o desenho do piso:
    # u aponta para a frente e v para o lado, cada um com a sua inclinação.
    t1, t2 = math.radians(inclinacao_1), math.radians(inclinacao_2)
    P_mais_u = (math.cos(t1), 0, 2 + math.sin(t1))
    P_mais_v = (0, math.cos(t2), 2 + math.sin(t2))
    beta = u3d.plano_por_tres_pontos(P, P_mais_u, P_mais_v)

    fig = u3d.criar_figura(titulo="O piso β e os dois níveis de bolha")
    fig = u3d.desenhar_plano(
        fig, *alfa, rotulo="plano horizontal", cor=u3d.CORES["apoio"],
        opacidade=0.15,
    )
    # O piso fica centrado em P; usamos o normal do plano β construído acima.
    fig = u3d.desenhar_plano(
        fig, ponto=P, normal=beta[1], rotulo="β", tamanho=4, cor=u3d.CORES["plano"]
    )
    fig = u3d.desenhar_reta_por_dois_pontos(
        fig, P, P_mais_u, rotulo="nível 1", cor=u3d.CORES["reta"], comprimento=4,
        espessura=7,
    )
    fig = u3d.desenhar_reta_por_dois_pontos(
        fig, P, P_mais_v, rotulo="nível 2", cor=u3d.CORES["reta_secundaria"],
        comprimento=4, espessura=7,
    )
    u3d.mostrar(fig)

    # Bolha centralizada quer dizer: a direção do nível é paralela ao plano horizontal.
    bolhas = {
        "paralela": "bolha centralizada ✔",
        "concorrente": "bolha fora do centro ✘",
    }
    posicao_1 = u3d.classificar_posicao_reta_plano(P, P_mais_u, alfa)
    posicao_2 = u3d.classificar_posicao_reta_plano(P, P_mais_v, alfa)
    print(f"Nível 1: {bolhas[posicao_1]}")
    print(f"Nível 2: {bolhas[posicao_2]}")
    print(f"Piso e plano horizontal: {u3d.classificar_posicao_planos(beta, alfa)}")

    if posicao_1 == "paralela" and posicao_2 == "paralela":
        print(
            "Piso nivelado: duas retas concorrentes paralelas ao plano horizontal "
            "garantem que o piso é paralelo a ele."
        )
    else:
        print(
            "Piso inclinado: pelo menos uma direção não está paralela ao plano "
            "horizontal."
        )


interact(
    desenhar_nivel_de_bolha,
    inclinacao_1=widgets.IntSlider(
        min=-30, max=30, step=5, value=10, description="inclinação na frente (°)",
        continuous_update=False,
    ),
    inclinacao_2=widgets.IntSlider(
        min=-30, max=30, step=5, value=0, description="inclinação lateral (°)",
        continuous_update=False,
    ),
);
```

### Propriedades dos planos paralelos

A partir da definição e do teorema, demonstram-se as propriedades abaixo. Elas serão usadas o tempo todo no resto do módulo.

1. **Plano paralelo por um ponto.** Por um ponto fora de um plano passa um único plano paralelo a ele.
2. **Transitividade.** Dois planos paralelos a um terceiro são paralelos entre si (ou coincidentes).
3. **Retas de um plano.** Se $\alpha \parallel \beta$, toda reta contida em $\alpha$ é paralela a $\beta$.
4. **Uma reta de cada plano.** Se $\alpha \parallel \beta$, uma reta de $\alpha$ e uma reta de $\beta$ são **paralelas ou reversas**, nunca concorrentes. *É por isso que o controlador de voo fica tranquilo!*
5. **Plano que corta dois paralelos.** Se um plano $\gamma$ corta dois planos paralelos $\alpha$ e $\beta$, as retas de interseção são **paralelas**.
6. **Reta que fura um, fura o outro.** Se uma reta fura um de dois planos paralelos, ela fura também o outro.
7. **Reversas em planos paralelos.** Duas retas reversas sempre estão em dois planos paralelos, um contendo cada reta.

🎯 **Aplicação da propriedade 5.** Quando você corta um pão de forma com uma faca inclinada, as marcas da faca na face de cima e na face de baixo (que são planos paralelos) ficam paralelas. É também por isso que as marcas de uma rampa nos andares de um estacionamento ficam alinhadas em direções paralelas.

🕹️ Veja a propriedade 5 em ação: incline o plano $\gamma$ e observe as duas retas de interseção.

```python
def desenhar_propriedade_5(inclinacao_de_gama: int) -> None:
    """Desenha os planos paralelos α e β, o plano γ e as duas interseções."""
    alfa = ((0, 0, 0), (0, 0, 1))
    beta = ((0, 0, 2), (0, 0, 1))

    # Seno e cosseno (módulo 05) servem só para inclinar o desenho de γ, que passa
    # sempre pelo ponto (0, 0, 1), a meia altura entre α e β.
    theta = math.radians(inclinacao_de_gama)
    gama = ((0, 0, 1), (math.sin(theta), 0, math.cos(theta)))

    # Quanto mais "deitado" estiver γ, mais longe do centro ele corta α e β.
    # Por isso os pedaços desenhados crescem quando a inclinação diminui (a
    # tangente aparece aqui só para calcular o tamanho do desenho).
    tamanho_de_gama = 2 / math.sin(theta) + 1
    tamanho_de_alfa_e_beta = max(6, 2 / math.tan(theta) + 3)

    fig = u3d.criar_figura(titulo="Um plano γ cortando dois planos paralelos")
    fig = u3d.desenhar_plano(
        fig, *alfa, rotulo="α", tamanho=tamanho_de_alfa_e_beta, opacidade=0.25
    )
    fig = u3d.desenhar_plano(
        fig, *beta, rotulo="β", tamanho=tamanho_de_alfa_e_beta, opacidade=0.25,
        cor=u3d.CORES["plano_secundario"],
    )
    fig = u3d.desenhar_plano(
        fig, *gama, rotulo="γ", tamanho=tamanho_de_gama, opacidade=0.25,
        cor=u3d.CORES["plano_terciario"],
    )
    fig = u3d.desenhar_intersecao_de_planos(fig, gama, alfa, rotulo="i₁")
    fig = u3d.desenhar_intersecao_de_planos(fig, gama, beta, rotulo="i₂")
    u3d.mostrar(fig)

    # Conferência: duas retas auxiliares de γ (não desenhadas), "descendo a rampa",
    # furam α e β. Os pontos onde furam α estão em i₁; os pontos onde furam β, em i₂.
    descida = (math.cos(theta), 0, -math.sin(theta))  # uma direção dentro de γ
    auxiliar_1 = ((0, -1, 1), (descida[0], -1, 1 + descida[2]))
    auxiliar_2 = ((0, 1, 1), (descida[0], 1, 1 + descida[2]))
    i1_p = u3d.intersecao_reta_plano(*auxiliar_1, alfa)
    i1_q = u3d.intersecao_reta_plano(*auxiliar_2, alfa)
    i2_p = u3d.intersecao_reta_plano(*auxiliar_1, beta)
    i2_q = u3d.intersecao_reta_plano(*auxiliar_2, beta)
    posicao = u3d.classificar_posicao_retas(i1_p, i1_q, i2_p, i2_q)
    print(f"Posição de i₁ e i₂: {posicao}")


interact(
    desenhar_propriedade_5,
    inclinacao_de_gama=widgets.IntSlider(
        min=10, max=80, step=10, value=45, description="inclinação de γ (°)",
        continuous_update=False,
    ),
);
```

## 5. Três retas reversas duas a duas

Imagine três canos num galpão: nenhum par deles está num mesmo plano (são reversos dois a dois). Um engenheiro precisa instalar uma barra de sustentação que encoste nos dois primeiros canos e seja paralela ao terceiro. Essa barra existe? Quantas existem?

**Teorema**
> Sejam $r$, $s$ e $t$ três retas reversas duas a duas, que não sejam paralelas a um mesmo plano. Então existe **uma única** reta $x$ que é paralela a $t$ e se apoia em $r$ e em $s$ (isto é, é concorrente com as duas).

**Construção**
1. Construa o plano $\alpha$ que contém $r$ e é paralelo a $t$ (seção 3).
2. Construa o plano $\beta$ que contém $s$ e é paralelo a $t$.
3. A reta procurada é $x = \alpha \cap \beta$.

**Por que funciona?**
- $\alpha$ e $\beta$ não são o mesmo plano (senão $r$ e $s$ seriam coplanares) e se cortam porque as retas não são paralelas a um mesmo plano.
- Os dois planos são paralelos a $t$, então a interseção deles também é paralela a $t$.
- No plano $\alpha$ estão $x$ e $r$. Elas não são paralelas (senão $r$ seria paralela a $t$, mas $r$ e $t$ são reversas), então são concorrentes. Do mesmo modo, $x$ e $s$ são concorrentes no plano $\beta$.
- Qualquer reta com essas propriedades precisa estar em $\alpha$ e em $\beta$; por isso ela é única.

🕹️ No cubo abaixo, $r = AB$, $s = CG$ e $t = EH$ são reversas duas a duas. Avance os passos da construção.

```python
def desenhar_tres_reversas_no_cubo(passo: int) -> None:
    """Desenha, passo a passo, a reta x paralela a t = EH que se apoia em AB e CG."""
    alfa = u3d.plano_paralelo_a_reta(A, B, E, H)  # contém r = AB, paralelo a t
    beta = u3d.plano_paralelo_a_reta(C, G, E, H)  # contém s = CG, paralelo a t

    fig = u3d.criar_figura(titulo=f"A reta que se apoia em r e s: passo {passo}")
    fig = u3d.desenhar_cubo(fig)
    fig = u3d.desenhar_reta_por_dois_pontos(
        fig, A, B, rotulo="r", cor=u3d.CORES["reta"], comprimento=7, espessura=6
    )
    fig = u3d.desenhar_reta_por_dois_pontos(
        fig, C, G, rotulo="s", cor=u3d.CORES["reta_secundaria"], comprimento=7,
        espessura=6,
    )
    # t ganha a cor dos planos terciários, só para se diferenciar de r e s.
    fig = u3d.desenhar_reta_por_dois_pontos(
        fig, E, H, rotulo="t", cor=u3d.CORES["plano_terciario"], comprimento=7,
        espessura=6,
    )
    if passo >= 2:
        fig = u3d.desenhar_plano_paralelo_a_reta(
            fig, A, B, E, H, rotulo="α", cor=u3d.CORES["plano"], opacidade=0.25
        )
    if passo >= 3:
        fig = u3d.desenhar_plano_paralelo_a_reta(
            fig, C, G, E, H, rotulo="β", cor=u3d.CORES["plano_secundario"],
            opacidade=0.25,
        )
    if passo >= 4:
        fig = u3d.desenhar_intersecao_de_planos(
            fig, alfa, beta, rotulo="x", comprimento=7
        )
    u3d.mostrar(fig)

    if passo == 4:
        # Dois pontos de x: onde r fura β e onde s fura α (cada um deles está nos
        # dois planos, logo está na interseção x).
        x1 = u3d.intersecao_reta_plano(A, B, beta)
        x2 = u3d.intersecao_reta_plano(C, G, alfa)
        print(f"x e t: {u3d.classificar_posicao_retas(x1, x2, E, H)}")
        print(f"x e r: {u3d.classificar_posicao_retas(x1, x2, A, B)}")
        print(f"x e s: {u3d.classificar_posicao_retas(x1, x2, C, G)}")


interact(
    desenhar_tres_reversas_no_cubo,
    passo=widgets.IntSlider(
        min=1, max=4, step=1, value=1, description="passo", continuous_update=False
    ),
);
```

E se as três retas não estiverem num cubo? A construção funciona para **quaisquer** três retas reversas duas a duas (não paralelas a um mesmo plano). Cada valor do controle abaixo sorteia três retas diferentes e mostra a construção completa.

```python
import numpy as np


def sortear_reta(gerador: np.random.Generator) -> tuple[tuple, tuple]:
    """Sorteia uma reta, dada por dois pontos com coordenadas entre -2 e 2."""
    ponto_1 = tuple(gerador.uniform(-2, 2, size=3))
    ponto_2 = tuple(gerador.uniform(-2, 2, size=3))
    return ponto_1, ponto_2


def desenhar_exemplo(exemplo: int) -> None:
    """Sorteia três retas reversas duas a duas e constrói a reta x."""
    # A "semente" do sorteio é o número do exemplo: o mesmo número dá sempre as
    # mesmas retas.
    gerador = np.random.default_rng(exemplo)
    while True:
        r, s, t = sortear_reta(gerador), sortear_reta(gerador), sortear_reta(gerador)

        # Verificação 1: o teorema exige retas reversas duas a duas.
        pares = [(r, s), (r, t), (s, t)]
        if any(u3d.classificar_posicao_retas(*p, *q) != "reversas" for p, q in pares):
            continue

        # Verificação 2: as retas não podem ser paralelas a um mesmo plano. Se
        # fossem, α e β seriam paralelos e não haveria reta x.
        alfa = u3d.plano_paralelo_a_reta(*r, *t)
        beta = u3d.plano_paralelo_a_reta(*s, *t)
        if u3d.classificar_posicao_planos(alfa, beta) == "secantes":
            break

    fig = u3d.criar_figura(titulo=f"Exemplo {exemplo}: a reta x existe e é única")
    fig = u3d.desenhar_plano(fig, *alfa, rotulo="α", opacidade=0.2)
    fig = u3d.desenhar_plano(
        fig, *beta, rotulo="β", cor=u3d.CORES["plano_secundario"], opacidade=0.2
    )
    fig = u3d.desenhar_reta_por_dois_pontos(
        fig, *r, rotulo="r", cor=u3d.CORES["reta"], comprimento=8, espessura=6
    )
    fig = u3d.desenhar_reta_por_dois_pontos(
        fig, *s, rotulo="s", cor=u3d.CORES["reta_secundaria"], comprimento=8,
        espessura=6,
    )
    fig = u3d.desenhar_reta_por_dois_pontos(
        fig, *t, rotulo="t", cor=u3d.CORES["plano_terciario"], comprimento=8,
        espessura=6,
    )
    fig = u3d.desenhar_intersecao_de_planos(fig, alfa, beta, rotulo="x")

    # Os pontos onde x se apoia em r e em s.
    fig = u3d.desenhar_intersecao_reta_plano(fig, *r, beta)
    fig = u3d.desenhar_intersecao_reta_plano(fig, *s, alfa)
    u3d.mostrar(fig)


interact(
    desenhar_exemplo,
    exemplo=widgets.IntSlider(
        min=1, max=8, step=1, value=1, description="exemplo", continuous_update=False
    ),
);
```

## 6. Teorema de Tales no espaço

Na Geometria Plana você viu o Teorema de Tales: um feixe de retas paralelas cortado por duas transversais determina segmentos proporcionais. No espaço, as retas paralelas viram **planos paralelos**.

**Teorema de Tales no espaço**
> Se três planos paralelos $\alpha$, $\beta$ e $\gamma$ são cortados por duas retas transversais $r$ e $s$, nos pontos $A$, $B$, $C$ e $A'$, $B'$, $C'$, respectivamente, então
> $$\frac{AB}{BC} = \frac{A'B'}{B'C'}$$

As transversais **não precisam** estar num mesmo plano: podem até ser reversas!

<details>
<summary>Ver demonstração</summary>

Trace por $A$ a reta $s'$ paralela a $s$; ela corta $\beta$ e $\gamma$ em $B''$ e $C''$.

- **No plano de $r$ e $s'$:** os planos $\beta$ e $\gamma$ cortam esse plano em retas paralelas (propriedade 5). Pelo Tales da Geometria Plana, $\dfrac{AB}{BC} = \dfrac{AB''}{B''C''}$.
- **No plano de $s$ e $s'$:** pelo mesmo motivo, $AA'$, $B''B'$ e $C''C'$ são paralelas. Os quadriláteros $AB''B'A'$ e $B''C''C'B'$ são paralelogramos, logo $AB'' = A'B'$ e $B''C'' = B'C'$.

Juntando as duas partes: $\dfrac{AB}{BC} = \dfrac{A'B'}{B'C'}$. ∎
</details>

🎯 **Aplicação.** Num prédio, os pisos dos andares são planos paralelos. Se todos os andares têm a mesma altura, qualquer cabo esticado em linha reta do térreo até o topo (um cabo de para-raios, um fio de antena, uma escada externa reta) é dividido pelos pisos em pedaços **iguais**, seja qual for a inclinação do cabo.

🕹️ Mude as distâncias entre os planos e a inclinação da transversal $s$, e acompanhe a tabela: as duas razões são sempre iguais.

```python
def desenhar_tales(d1: float, d2: float, giro_de_s: int) -> None:
    """Desenha três planos paralelos cortados por duas transversais r e s."""
    # Os três planos horizontais: α na altura 0, β na altura d1, γ na altura d1 + d2.
    # Os pedaços ficam centrados em (0.5, 0.5) para as transversais caberem neles.
    alfa = ((0.5, 0.5, 0), (0, 0, 1))
    beta = ((0.5, 0.5, d1), (0, 0, 1))
    gama = ((0.5, 0.5, d1 + d2), (0, 0, 1))

    # Transversal r, fixa, e transversal s, que gira. Seno e cosseno (módulo 05)
    # servem só para girar o desenho de s. Em geral, r e s são reversas.
    r1, r2 = (-1.5, -1, -1), (-0.5, 0, 4)
    theta = math.radians(giro_de_s)
    s1 = (1.5, 0, 0)
    s2 = (1.5 + 0.6 * math.cos(theta), 0.6 * math.sin(theta), 1)

    # Os pontos onde cada transversal fura cada plano.
    ponto_A, ponto_B, ponto_C = (
        u3d.intersecao_reta_plano(r1, r2, plano) for plano in (alfa, beta, gama)
    )
    ponto_A_, ponto_B_, ponto_C_ = (
        u3d.intersecao_reta_plano(s1, s2, plano) for plano in (alfa, beta, gama)
    )

    fig = u3d.criar_figura(titulo="Tales no espaço", altura=550)
    cores_dos_planos = [
        u3d.CORES["plano"], u3d.CORES["plano_secundario"], u3d.CORES["plano_terciario"]
    ]
    for plano, nome, cor in zip(
        (alfa, beta, gama), "αβγ", cores_dos_planos, strict=True
    ):
        fig = u3d.desenhar_plano(
            fig, *plano, rotulo=nome, cor=cor, opacidade=0.2, tamanho=8
        )

    # As transversais inteiras (mais finas), e por cima os segmentos (mais grossos).
    fig = u3d.desenhar_reta_por_dois_pontos(
        fig, r1, r2, rotulo="r", cor=u3d.CORES["apoio"], comprimento=10, espessura=3
    )
    fig = u3d.desenhar_reta_por_dois_pontos(
        fig, ponto_A_, ponto_C_, rotulo="s", cor=u3d.CORES["apoio"],
        comprimento=u3d.medir_distancia(ponto_A_, ponto_C_) + 3, espessura=3,
    )
    for inicio, fim, cor in [
        (ponto_A, ponto_B, u3d.CORES["reta"]),
        (ponto_A_, ponto_B_, u3d.CORES["reta"]),
        (ponto_B, ponto_C, u3d.CORES["reta_secundaria"]),
        (ponto_B_, ponto_C_, u3d.CORES["reta_secundaria"]),
    ]:
        fig = u3d.desenhar_segmento(fig, inicio, fim, cor=cor, espessura=8)
    fig = u3d.desenhar_pontos(
        fig, [ponto_A, ponto_B, ponto_C, ponto_A_, ponto_B_, ponto_C_],
        rotulos=["A", "B", "C", "A'", "B'", "C'"], tamanho=5,
    )
    u3d.mostrar(fig)

    # Medimos os segmentos com u3d.medir_distancia, a nossa trena (a fórmula por
    # trás dela é assunto do módulo 07).
    AB = u3d.medir_distancia(ponto_A, ponto_B)
    BC = u3d.medir_distancia(ponto_B, ponto_C)
    A_B_ = u3d.medir_distancia(ponto_A_, ponto_B_)
    B_C_ = u3d.medir_distancia(ponto_B_, ponto_C_)

    texto_AB, texto_BC = f"AB = {AB:.3f}", f"BC = {BC:.3f}"
    texto_A_B_, texto_B_C_ = f"A'B' = {A_B_:.3f}", f"B'C' = {B_C_:.3f}"
    print(f"{'':<10}{'1º segmento':<15}{'2º segmento':<15}{'razão':>7}")
    print(f"{'Reta r':<10}{texto_AB:<15}{texto_BC:<15}{AB / BC:>7.3f}")
    print(f"{'Reta s':<10}{texto_A_B_:<15}{texto_B_C_:<15}{A_B_ / B_C_:>7.3f}")
    print(f"\nRazão entre as distâncias dos planos: d1 / d2 = {d1 / d2:.3f}")


interact(
    desenhar_tales,
    d1=widgets.FloatSlider(
        min=0.5, max=3, step=0.25, value=1, description="d1 (α até β)",
        continuous_update=False,
    ),
    d2=widgets.FloatSlider(
        min=0.5, max=3, step=0.25, value=2, description="d2 (β até γ)",
        continuous_update=False,
    ),
    giro_de_s=widgets.IntSlider(
        min=0, max=180, step=15, value=60, description="giro de s (°)",
        continuous_update=False,
    ),
);
```

## 7. Ângulo de duas retas

Quando duas retas se cruzam, elas formam quatro ângulos, dois a dois iguais (opostos pelo vértice). Vamos sempre usar o **menor** deles.

> 📌 **Definição.** O **ângulo entre duas retas** $r$ e $s$ é:
> - se forem **concorrentes**: o menor dos ângulos que elas formam ($0° < \theta \le 90°$);
> - se forem **paralelas**: $0°$;
> - se forem **reversas**: o ângulo entre $r$ e uma reta $s'$, paralela a $s$, que corta $r$.

No caso das reversas, não importa por qual ponto de $r$ você traça a paralela $s'$: todas essas paralelas têm a mesma direção e formam o mesmo ângulo com $r$.

> 📌 **Definição.**
> - Retas **perpendiculares** são retas **concorrentes** que formam ângulo de $90°$.
> - Retas **ortogonais** são retas **reversas** que formam ângulo de $90°$.

> ⚠️ **Erro comum.** Chamar duas retas reversas de "perpendiculares". Perpendicular exige que as retas se cruzem. Se elas são reversas e formam $90°$, o nome correto é **ortogonais**. Alguns livros usam "ortogonais" para os dois casos; neste curso, seguimos a distinção acima.

🎯 **Aplicação.** Em instalações industriais, tubulações de água, gás e cabos passam por cima umas das outras sem se tocar. Os projetistas preferem cruzá-las de forma **ortogonal**, porque isso facilita a montagem dos suportes e a manutenção.

## Resumo

**Retas paralelas**
- Postulado das paralelas: por um ponto fora de uma reta passa uma única paralela a ela.
- Transitividade: $a \parallel b$ e $b \parallel c$ ⟹ $a \parallel c$ (vale no espaço).

**Reta e plano**

| Pontos comuns | Posição |
|---|---|
| infinitos | contida ($r \subset \alpha$) |
| um | concorrentes |
| nenhum | paralelos |

- Condição: $a \not\subset \alpha$ e $a \parallel b$ com $b \subset \alpha$ ⟹ $a \parallel \alpha$.
- Uma reta paralela a um plano é paralela a algumas retas dele e reversa a outras.

**Retas reversas**
- $r \subset \alpha$ e $s$ fura $\alpha$ fora de $r$ ⟹ $r$ e $s$ são reversas.
- Dadas $r$ e $s$ reversas, existe um único plano que contém $r$ e é paralelo a $s$.

**Planos paralelos**
- Condição: $\alpha$ contém duas retas **concorrentes** paralelas a $\beta$ ⟹ $\alpha \parallel \beta$.
- Uma reta de cada plano: paralelas ou reversas, nunca concorrentes.
- Um plano que corta dois planos paralelos determina interseções paralelas.

**Três retas reversas duas a duas** (não paralelas a um mesmo plano): existe uma única reta paralela à terceira que se apoia nas outras duas. Ela é a interseção dos planos $\alpha$ (por $r$, paralelo a $t$) e $\beta$ (por $s$, paralelo a $t$).

**Tales no espaço:** $\alpha \parallel \beta \parallel \gamma$ cortados por transversais ⟹ $\dfrac{AB}{BC} = \dfrac{A'B'}{B'C'}$.

**Ângulo entre retas:** sempre entre $0°$ e $90°$. Perpendiculares = concorrentes a $90°$. Ortogonais = reversas a $90°$.
