As funções abaixo foram programadas no notebook principal. Elas são reconstruídas aqui para que as soluções possam usá-las.

```python
Ponto = tuple[float, float]   # um ponto do plano é uma tupla (x, y)


def determinante_dos_pontos(A: Ponto, B: Ponto, C: Ponto) -> float:
    """D = xA(yB − yC) + xB(yC − yA) + xC(yA − yB) (seção 2)."""
    xA, yA = A
    xB, yB = B
    xC, yC = C
    return xA * (yB - yC) + xB * (yC - yA) + xC * (yA - yB)


def area_orientada(A: Ponto, B: Ponto, C: Ponto) -> float:
    """Área orientada do triângulo ABC: D/2 (seção 3)."""
    return determinante_dos_pontos(A, B, C) / 2


def area_poligono(vertices: list) -> float:
    """Área de um polígono pela fórmula do cadarço (seção 5)."""
    n = len(vertices)
    soma = 0
    for i in range(n):
        xi, yi = vertices[i]
        xj, yj = vertices[(i + 1) % n]
        soma = soma + (xi * yj - xj * yi)
    return abs(soma) / 2


# As cores do sinal, como no notebook principal
COR_ANTI_HORARIO = uga.CORES["plano"]
COR_HORARIO = uga.CORES["reta_secundaria"]
```

**Exercício 8.** A função `ponto_no_triangulo(P, A, B, C)`.

```python
def ponto_no_triangulo(P: Ponto, A: Ponto, B: Ponto, C: Ponto) -> bool:
    # De que lado de cada lado (andando A → B → C → A) o ponto P está
    d1 = determinante_dos_pontos(A, B, P)
    d2 = determinante_dos_pontos(B, C, P)
    d3 = determinante_dos_pontos(C, A, P)
    # Dentro: P fica sempre do MESMO lado (todos ≥ 0 ou todos ≤ 0).
    # Um zero quer dizer que P está sobre a reta de um lado, e conta como borda.
    tem_positivo = d1 > 0 or d2 > 0 or d3 > 0
    tem_negativo = d1 < 0 or d2 < 0 or d3 < 0
    return not (tem_positivo and tem_negativo)


for vertices in [[(0, 0), (4, 0), (1, 3)], [(0, 0), (1, 3), (4, 0)]]:
    print(vertices, [ponto_no_triangulo(P, *vertices) for P in [(2, 1), (4, 3), (1, 0), (-1, 0)]])
```

**Por que funciona nos dois sentidos?** Se os vértices estão em ordem anti-horária, o interior fica à **esquerda** de cada lado e os três determinantes de um ponto interno são positivos. Em ordem horária, o interior fica à **direita** e os três são negativos. Em vez de descobrir antes o sentido, o teste só exige que **não haja sinais contrários**. Um ponto de fora fica do lado de fora de pelo menos um lado, e então aparece um sinal diferente dos outros.

Para ver a função funcionando, a figura abaixo pinta uma grade de pontos: verde dentro (ou na borda), vermelho fora.

**Exercício 10.** O Teorema de Pick conferido só com determinantes.

```python
def pick_por_determinantes(vertices: list) -> tuple:
    n = len(vertices)
    # 1. A área, pelo cadarço
    S = area_poligono(vertices)

    # Os lados do polígono, cada um como o par (Vi, Vi+1)
    lados = [(vertices[i], vertices[(i + 1) % n]) for i in range(n)]

    def na_borda(P: Ponto) -> bool:
        # P está alinhado com algum lado E dentro do retângulo envolvente desse lado
        for V, W in lados:
            alinhado = determinante_dos_pontos(V, W, P) == 0   # inteiros: conta exata
            no_retangulo = (min(V[0], W[0]) <= P[0] <= max(V[0], W[0])
                            and min(V[1], W[1]) <= P[1] <= max(V[1], W[1]))
            if alinhado and no_retangulo:
                return True   # para no primeiro lado: os vértices não são contados duas vezes
        return False

    def dentro(P: Ponto) -> bool:
        # Anti-horário: o interior fica à esquerda de todos os lados (D > 0 em todos)
        return all(determinante_dos_pontos(V, W, P) > 0 for V, W in lados)

    # 2. Percorre os pontos inteiros do retângulo envolvente do polígono
    xs = [x for x, _ in vertices]
    ys = [y for _, y in vertices]
    B = I = 0
    for x in range(min(xs), max(xs) + 1):
        for y in range(min(ys), max(ys) + 1):
            if na_borda((x, y)):
                B += 1
            elif dentro((x, y)):
                I += 1
    return S, B, I


# 3. Conferindo S = I + B/2 − 1
for vertices in [[(0, 0), (4, 1), (1, 3)], [(1, 0), (7, 2), (4, 6), (0, 4)]]:
    S, B, I = pick_por_determinantes(vertices)
    print(f"{vertices}: S = {S}, B = {B}, I = {I} → I + B/2 − 1 = {I + B / 2 - 1}  "
          f"{'✓' if math.isclose(S, I + B / 2 - 1) else '✗'}")
```

O triângulo $(0, 0)$, $(4, 1)$, $(1, 3)$ é o mesmo do módulo 04: só os vértices estão na borda ($B = 3$) e há $I = 5$ pontos dentro, e $5 + \frac{3}{2} - 1 = 5{,}5$. No quadrilátero, os lados $(1, 0) \to (7, 2)$ e $(4, 6) \to (0, 4)$ passam, cada um, por um ponto inteiro no meio, $(4, 1)$ e $(2, 5)$; com os 4 vértices, $B = 6$. E $22 + 3 - 1 = 24$.

**Por que o teste da borda usa o retângulo envolvente do lado?** O determinante zero só diz que $P$ está na **reta** do lado, que é infinita. O retângulo envolvente do lado corta essa reta exatamente no segmento.
