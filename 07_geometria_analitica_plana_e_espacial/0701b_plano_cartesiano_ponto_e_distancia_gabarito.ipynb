As funções abaixo foram programadas no notebook principal. Elas são reconstruídas aqui para que as soluções possam usá-las.

```python
Ponto = tuple[float, float]   # um ponto do plano é uma tupla (x, y)


def distancia(A: Ponto, B: Ponto) -> float:
    """Distância entre os pontos A = (xA, yA) e B = (xB, yB) (seção 1)."""
    xA, yA = A
    xB, yB = B
    return math.sqrt((xB - xA)**2 + (yB - yA)**2)


def distancia_ao_quadrado(A: Ponto, B: Ponto) -> float:
    """O quadrado da distância entre A e B (seção 2.1)."""
    xA, yA = A
    xB, yB = B
    return (xB - xA)**2 + (yB - yA)**2


def ponto_medio(A: Ponto, B: Ponto) -> Ponto:
    """Ponto médio do segmento AB (seção 3)."""
    xA, yA = A
    xB, yB = B
    return ((xA + xB) / 2, (yA + yB) / 2)


def ponto_do_segmento(A: Ponto, B: Ponto, t: float) -> Ponto:
    """Ponto que está na fração t do caminho de A até B (seção 4)."""
    xA, yA = A
    xB, yB = B
    return (xA + t * (xB - xA), yA + t * (yB - yA))


def baricentro(A: Ponto, B: Ponto, C: Ponto) -> Ponto:
    """Baricentro do triângulo ABC: a média dos três vértices (seção 5)."""
    xA, yA = A
    xB, yB = B
    xC, yC = C
    return ((xA + xB + xC) / 3, (yA + yB + yC) / 3)
```

## Exercício guiado da seção 2.2

**`foi_visto(jogador, guarda, raio)`.** Devolver `True` se o jogador estiver a uma distância menor ou igual ao raio.

Duas soluções. A primeira compara a distância com o raio; a segunda, do jeito dos programadores de jogos, compara os **quadrados** e dispensa a raiz.

```python
def foi_visto(jogador: Ponto, guarda: Ponto, raio: float) -> bool:
    # Com raiz: a distância comparada com o raio
    return distancia(jogador, guarda) <= raio


def foi_visto_sem_raiz(jogador: Ponto, guarda: Ponto, raio: float) -> bool:
    # Sem raiz: como d e raio não são negativos, d <= raio equivale a d² <= raio²
    return distancia_ao_quadrado(jogador, guarda) <= raio**2


for jogador in [(3, 2), (4, 4), (6, 7), (9, 7)]:
    print(jogador, foi_visto(jogador, (6, 4), 3), foi_visto_sem_raiz(jogador, (6, 4), 3))
```

$J(6, 7)$ está exatamente na borda ($d = 3$): por isso a comparação é **menor ou igual**. Na versão sem raiz, $9 \le 9$ é uma conta exata, sem nenhum arredondamento.

**Exercício 9.** (a) A função `testar_base_media` e o teste em 1000 triângulos.

```python
import random


def testar_base_media(A: Ponto, B: Ponto, C: Ponto) -> bool:
    # Os pontos médios dos lados AC e BC
    M_AC = ponto_medio(A, C)
    M_BC = ponto_medio(B, C)
    # O segmento que liga os dois mede metade de AB?
    return math.isclose(distancia(M_AC, M_BC), distancia(A, B) / 2)


random.seed(9)
resultados = []
for _ in range(1000):
    A, B, C = [(random.uniform(-10, 10), random.uniform(-10, 10)) for _ in range(3)]
    resultados.append(testar_base_media(A, B, C))
todos_ok_ex9 = all(resultados)

print(todos_ok_ex9, sum(resultados), "de", len(resultados))
```

(b) **A prova analítica.** Tome $A(0, 0)$, $B(2b, 0)$ e $C(2c, 2d)$, que representa **qualquer** triângulo (com $AB$ sobre o eixo $x$ e $A$ na origem). Os pontos médios são

$$M_{AC} = \left(\frac{0 + 2c}{2},\ \frac{0 + 2d}{2}\right) = (c, d) \qquad M_{BC} = \left(\frac{2b + 2c}{2},\ \frac{0 + 2d}{2}\right) = (b + c, d).$$

A distância entre eles é

$$d(M_{AC}, M_{BC}) = \sqrt{(b + c - c)^2 + (d - d)^2} = \sqrt{b^2} = |b|,$$

que é metade de $AB = |2b|$. ∎

**Por que os "2"?** Com $2b$, $2c$ e $2d$, as divisões por 2 dos pontos médios dão contas exatas, sem frações. Com $b$, $c$ e $d$, a prova funcionaria do mesmo jeito, mas cheia de $\frac{b}{2}$, $\frac{c}{2}$ e $\frac{d}{2}$.

**Bônus.** Os dois pontos médios têm a mesma ordenada $d$, então o segmento $M_{AC}M_{BC}$ é horizontal, paralelo a $AB$ (que está sobre o eixo $x$). Isso prova também a outra metade do teorema da base média: ela é **paralela** ao terceiro lado.

**Exercício 10.** Os pontos de coordenadas inteiras a uma distância exatamente 5 da origem.

```python
pontos_ex10 = []
for x in range(-5, 6):          # x de −5 a 5
    for y in range(-5, 6):      # para cada x, y de −5 a 5
        # Compara x² + y² com 25 = 5²: só contas com inteiros, sem raiz
        if x**2 + y**2 == 25:
            pontos_ex10.append((x, y))
quantidade_ex10 = len(pontos_ex10)

print(quantidade_ex10, pontos_ex10)

fig = uga.criar_plano_cartesiano(titulo="Pontos inteiros a uma distância 5 da origem")
fig = uga.desenhar_circulo_de_alcance(fig, (0, 0), 5, cor=uga.CORES["apoio"], opacidade=0.05)
fig = uga.desenhar_ponto(fig, (0, 0), rotulo="O", cor=uga.CORES["apoio"])
fig = uga.desenhar_pontos(fig, pontos_ex10, cor=uga.CORES["destaque"])
uga.mostrar(fig)
```

São **12** pontos: $(\pm 5, 0)$, $(0, \pm 5)$, $(\pm 3, \pm 4)$ e $(\pm 4, \pm 3)$. Os oito últimos vêm do terno pitagórico 3-4-5 ($3^2 + 4^2 = 5^2$), com todas as combinações de sinais e de ordem.
