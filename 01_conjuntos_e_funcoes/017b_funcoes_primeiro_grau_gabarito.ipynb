```python
# Import auxiliar necessário para os exercícios abaixo.
import sympy as sp

x_sym = sp.symbols("x")
```

## Básico

**Exercício 1.** Dada f(x) = 4x - 20, identifique os coeficientes, o zero da função e classifique-a em crescente, decrescente ou constante.

```python
# f(x) = 4x - 20 comparado com f(x) = ax + b
coef_angular_ex = 4
coef_linear_ex = -20

# zero: 4x - 20 = 0  ->  x = 20/4 = 5
zero_ex = -coef_linear_ex / coef_angular_ex

# a = 4 > 0 -> crescente
classificacao_ex = "crescente" if coef_angular_ex > 0 else (
    "decrescente" if coef_angular_ex < 0 else "constante"
)

print(coef_angular_ex, coef_linear_ex, zero_ex, classificacao_ex)
```

**Exercício 2.** Classifique cada função em "constante", "linear", "identidade" ou "nenhuma das anteriores".

```python
def q1(x):
    return 7


def q2(x):
    return -3 * x


def q3(x):
    return x


def q4(x):
    return 2 * x + 1


# q1 não depende de x -> constante (a=0, b=7)
classificacao_q1 = "constante"
# q2 não tem termo independente -> linear (a=-3, b=0)
classificacao_q2 = "linear"
# q3 tem a=1 e b=0 -> identidade
classificacao_q3 = "identidade"
# q4 tem a=2 (!= 0 e != 1) e b=1 (!= 0) -> nenhuma das anteriores
classificacao_q4 = "nenhuma das anteriores"

print(classificacao_q1, classificacao_q2, classificacao_q3, classificacao_q4)
```

## Intermediário

**Exercício 1.** Uma planta cresce a taxa constante. Depois de 2 semanas, mede 14 cm; depois de 5 semanas, mede 26 cm. Determine a lei h(x) = a·x + b.

```python
a_sym_ex, b_sym_ex = sp.symbols("a b")
eq1_ex = sp.Eq(a_sym_ex * 2 + b_sym_ex, 14)
eq2_ex = sp.Eq(a_sym_ex * 5 + b_sym_ex, 26)

# taxa de crescimento: (26-14)/(5-2) = 4 cm/semana; altura em x=0: 14 - 4*2 = 6 cm
solucao_planta_ex = sp.solve([eq1_ex, eq2_ex], [a_sym_ex, b_sym_ex])
a_planta_ex = solucao_planta_ex[a_sym_ex]
b_planta_ex = solucao_planta_ex[b_sym_ex]

print(a_planta_ex, b_planta_ex)
```

**Exercício 2.** Estude o sinal de f(x) = -5x + 35. Informe o zero e os intervalos onde f é positiva e onde é negativa.

```python
# zero: -5x + 35 = 0  ->  x = 7
zero_sinal_ex = 7

# a = -5 < 0 -> f é positiva antes do zero e negativa depois
intervalo_positivo_ex = (float("-inf"), zero_sinal_ex)
intervalo_negativo_ex = (zero_sinal_ex, float("inf"))

print(zero_sinal_ex, intervalo_positivo_ex, intervalo_negativo_ex)
```

**Exercício 3.** Resolva a inequação 3x - 12 < 0.

```python
# dividimos por 3 (positivo) -- sem inversão do sentido da desigualdade
solucao_inequacao_ex = sp.solveset(3 * x_sym - 12 < 0, x_sym, domain=sp.S.Reals)

print(solucao_inequacao_ex)
```

## Desafio

**Exercício 1.** O lucro (em milhares de reais) de uma empresa, em função das x centenas de unidades vendidas, é dado por L(x) = (x - 10)(50 - x). Para quais valores de x o lucro é positivo?

```python
lucro_ex = (x_sym - 10) * (50 - x_sym)

# produto de dois fatores afins com zeros em x=10 e x=50: o quadro de sinais
# combinado mostra que o produto é positivo entre as duas raízes
solucao_lucro_ex = sp.solve_univariate_inequality(lucro_ex > 0, x_sym)

print(solucao_lucro_ex)
```

**Exercício 2.** Uma métrica de eficiência é dada por E(x) = (x - 8) / (x - 3), com x ≠ 3. Para quais valores de x temos E(x) ≥ 0?

```python
eficiencia_ex = (x_sym - 8) / (x_sym - 3)

# quadro de sinais combinado de (x-8) e (x-3), excluindo x=3 (zero do
# denominador, fora do domínio mesmo satisfazendo a desigualdade)
solucao_eficiencia_ex = sp.solve_univariate_inequality(eficiencia_ex >= 0, x_sym)

print(solucao_eficiencia_ex)
```
