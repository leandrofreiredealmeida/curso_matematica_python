```python
# Funções e imports auxiliares definidos no notebook principal, necessários
# para os exercícios abaixo.
from fractions import Fraction
from typing import Literal

IntervaloTipo = Literal["aberto", "fechado", "semiaberto_direita", "semiaberto_esquerda"]

FECHADO_NAS_PONTAS = {
    "aberto": (False, False),
    "fechado": (True, True),
    "semiaberto_direita": (True, False),
    "semiaberto_esquerda": (False, True),
}


def eh_natural(x) -> bool:
    '''Verifica se x é um número natural (inteiro maior ou igual a zero).'''
    return isinstance(x, int) and x >= 0


def eh_inteiro(x) -> bool:
    '''Verifica se x é um número inteiro (positivo, negativo ou zero).'''
    return isinstance(x, int)


def pertence_ao_intervalo(x: float, a: float, b: float, tipo: IntervaloTipo) -> bool:
    '''Verifica se x pertence ao intervalo de extremos a e b, do tipo dado.'''
    if tipo == "aberto":
        return a < x < b
    if tipo == "fechado":
        return a <= x <= b
    if tipo == "semiaberto_direita":  # [a, b)
        return a <= x < b
    if tipo == "semiaberto_esquerda":  # (a, b]
        return a < x <= b
    raise ValueError(f"Tipo de intervalo desconhecido: {tipo}")
```

## Intermediário

**Exercício 1.** Determine se cada fração gera decimal finito ou dízima periódica.

```python
fracoes_teste = [(1, 4), (1, 3), (5, 8), (2, 7), (7, 20)]


def eh_decimal_finita(numerador: int, denominador: int) -> bool:
    '''Retorna True se numerador/denominador tem expansão decimal finita.

    Uma fração simplificada a/b tem decimal finito se e só se os únicos
    fatores primos do denominador simplificado são 2 e/ou 5.
    '''
    fracao = Fraction(numerador, denominador)
    b = fracao.denominator
    while b % 2 == 0:
        b //= 2
    while b % 5 == 0:
        b //= 5
    return b == 1


# 1/4 = 0,25 (finita) | 1/3 = 0,333... (periódica) | 5/8 = 0,625 (finita)
# 2/7 = 0,285714285714... (periódica) | 7/20 = 0,35 (finita)
resultados_fracoes = [eh_decimal_finita(n, d) for n, d in fracoes_teste]

print(resultados_fracoes)
```

**Exercício 1.** Teste numericamente que a soma dos n primeiros ímpares é sempre n², e reconheça a base e o passo indutivo de uma prova formal dessa fórmula.

```python
def soma_impares(n: int) -> int:
    '''Soma dos n primeiros números ímpares positivos (1, 3, 5, ...).'''
    return sum(2 * i - 1 for i in range(1, n + 1))


valores_de_n = range(1, 21)

# Testamos todos os n de 1 a 20 e comparamos com n**2
vale_para_todos = all(soma_impares(n) == n ** 2 for n in valores_de_n)

print(vale_para_todos)
```

A prova formal por indução (a evidência numérica acima não substitui, mas aponta o caminho):

**Afirmação:** para todo n natural (n ≥ 1), 1 + 3 + 5 + ... + (2n - 1) = n².

**Base (n = 1):** o lado esquerdo tem um único termo, 2(1) - 1 = 1. O lado direito é 1² = 1. Batem. ✅

**Passo indutivo:** suponha que vale para um k qualquer (hipótese de indução):

> 1 + 3 + ... + (2k - 1) = k²

Queremos provar que vale para k + 1, isto é, que a soma até o próximo ímpar, (2(k+1) - 1) = (2k + 1), dá (k + 1)². Partindo do lado esquerdo e usando a hipótese:

> 1 + 3 + ... + (2k - 1) + (2k + 1) = k² + (2k + 1) = (k + 1)²

que é exatamente o que queríamos. ✅ Pelo Princípio da Indução Finita, a fórmula vale para todo n natural.

**Exercício 2.** Evidência computacional de que √3 é irracional, adaptando a prova de √2.

```python
pares_testados = [(p, q) for p in range(1, 30) for q in range(1, 30)]

# Nenhum par de inteiros pequenos satisfaz p^2 = 3 * q^2
existe_par_racional = any(p ** 2 == 3 * q ** 2 for p, q in pares_testados)

print(existe_par_racional)
```

A prova formal por contradição, adaptando exatamente os mesmos passos usados para √2 (trocando "par"/divisível por 2 por divisível por 3):

1. Suponha, por contradição, que √3 **é** racional: √3 = p/q, com p, q inteiros, q ≠ 0, e mdc(p, q) = 1 (fração já simplificada).
2. Elevando ao quadrado: 3 = p² / q², logo **p² = 3q²**.
3. Isso significa que p² é múltiplo de 3. Como 3 é primo, se p² é múltiplo de 3, então **p também é múltiplo de 3** (um argumento análogo ao "quadrado de ímpar é ímpar", agora com divisibilidade por 3). Logo, p = 3k para algum inteiro k.
4. Substituindo: (3k)² = 3q² → 9k² = 3q² → **q² = 3k²**. Pelo mesmo argumento do passo 3, **q também é múltiplo de 3**.
5. Mas se p e q são ambos múltiplos de 3, eles têm o fator 3 em comum — contradizendo mdc(p, q) = 1.
6. Contradição encontrada, logo a suposição inicial está errada: **√3 é irracional**.
