```python
# Import auxiliar necessário para os exercícios abaixo.
import sympy as sp

x_sym = sp.symbols("x")
```

## Básico

**Exercício 1.** Dada f(x) = x² - 5x + 6, calcule Δ, os zeros e as coordenadas do vértice.

```python
a_ex1, b_ex1, c_ex1 = 1, -5, 6

# Δ = b² - 4ac
delta_ex1 = b_ex1**2 - 4 * a_ex1 * c_ex1

# Bhaskara: x = (-b ± √Δ) / 2a  ->  Δ = 1 > 0, duas raízes reais
zero_menor_ex1 = (-b_ex1 - sp.sqrt(delta_ex1)) / (2 * a_ex1)
zero_maior_ex1 = (-b_ex1 + sp.sqrt(delta_ex1)) / (2 * a_ex1)

# vértice: xv = -b/2a, yv = -Δ/4a
vertice_x_ex1 = sp.Rational(-b_ex1, 2 * a_ex1)
vertice_y_ex1 = sp.Rational(-delta_ex1, 4 * a_ex1)

print(delta_ex1, zero_menor_ex1, zero_maior_ex1, vertice_x_ex1, vertice_y_ex1)
```

**Exercício 2.** Classifique a concavidade de f(x) = -3x² + x - 1 e diga se ela tem máximo ou mínimo.

```python
a_ex2 = -3

# a < 0 -> concavidade para baixo -> tem máximo
concavidade_ex2 = "para baixo"
extremo_ex2 = "maximo"

print(concavidade_ex2, extremo_ex2)
```

## Intermediário

**Exercício 3.** Escreva f(x) = 2x² - 8x + 3 na forma canônica e identifique a translação em relação a f(x) = 2x².

```python
a_ex3, b_ex3, c_ex3 = 2, -8, 3

# h = -b/2a
h_ex3 = sp.Rational(-b_ex3, 2 * a_ex3)
# k = f(h)
k_ex3 = a_ex3 * h_ex3**2 + b_ex3 * h_ex3 + c_ex3

# a translação em relação a f(x) = ax² é exatamente (h, k)
translacao_ex3 = (h_ex3, k_ex3)

print(h_ex3, k_ex3, translacao_ex3)
# conferindo: f(x) = 2(x-2)² - 5
print(sp.expand(a_ex3 * (x_sym - h_ex3) ** 2 + k_ex3))
```

**Exercício 4.** Determine a imagem de f(x) = -x² + 4x - 1.

```python
a_ex4, b_ex4, c_ex4 = -1, 4, -1
delta_ex4 = b_ex4**2 - 4 * a_ex4 * c_ex4

# yv = -Δ/4a limita a imagem
limite_ex4 = sp.Rational(-delta_ex4, 4 * a_ex4)

# a = -1 < 0 -> concavidade para baixo -> Im(f) = (-∞, yv]
tipo_imagem_ex4 = "menor_igual"

print(limite_ex4, tipo_imagem_ex4)
print(f"Im(f) = (-∞, {limite_ex4}]")
```

## Desafio

**Exercício 5.** Um terreno retangular deve ter perímetro de 60 m. Expresse a área como função de um dos lados e encontre a área máxima possível.

```python
# perímetro: 2x + 2y = 60  ->  y = 30 - x
# área: A(x) = x*y = x*(30-x) = -x² + 30x
a_ex5, b_ex5, c_ex5 = -1, 30, 0

# a < 0 -> tem máximo, dado por yv = -Δ/4a
delta_ex5 = b_ex5**2 - 4 * a_ex5 * c_ex5
area_maxima_ex5 = sp.Rational(-delta_ex5, 4 * a_ex5)

print(a_ex5, b_ex5, c_ex5, area_maxima_ex5)
# ocorre em x = -b/2a = 15 m -> terreno quadrado 15m x 15m, área 225 m²
print("lado ótimo x =", sp.Rational(-b_ex5, 2 * a_ex5), "m")
```

**Exercício 6.** Dado o gráfico de uma parábola com vértice em (2, -3) passando por (0, 5), determine a lei da função.

```python
a_sym_ex6 = sp.symbols("a")

# forma canônica com o vértice conhecido: f(x) = a(x-2)² - 3
# usamos o ponto (0, 5): f(0) = a(0-2)² - 3 = 4a - 3 = 5
a_ex6 = sp.solve(sp.Eq(4 * a_sym_ex6 - 3, 5), a_sym_ex6)[0]

# expandindo a(x-2)² - 3 com a=2: f(x) = 2x² - 8x + 5
forma_geral_ex6 = sp.expand(a_ex6 * (x_sym - 2) ** 2 - 3)
b_ex6 = forma_geral_ex6.coeff(x_sym, 1)
c_ex6 = forma_geral_ex6.coeff(x_sym, 0)

print(a_ex6, b_ex6, c_ex6)
print(f"f(x) = {forma_geral_ex6}")
```
