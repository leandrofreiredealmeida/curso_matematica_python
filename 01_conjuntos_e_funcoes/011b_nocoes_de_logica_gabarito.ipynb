```python
# Funções auxiliares definidas no notebook principal, necessárias para os
# exercícios abaixo.
import itertools
from collections.abc import Callable

import pandas as pd


def condicional(p: bool, q: bool) -> bool:
    return (not p) or q


def tabela_verdade(
    proposicao: Callable[..., bool], nomes_variaveis: list[str]
) -> pd.DataFrame:
    linhas = []
    for valores in itertools.product([True, False], repeat=len(nomes_variaveis)):
        linha = dict(zip(nomes_variaveis, valores))
        linha["resultado"] = proposicao(*valores)
        linhas.append(linha)
    return pd.DataFrame(linhas)


def classificar(proposicao: Callable[..., bool], n_variaveis: int) -> str:
    resultados = [
        proposicao(*valores)
        for valores in itertools.product([True, False], repeat=n_variaveis)
    ]
    if all(resultados):
        return "tautologia"
    if not any(resultados):
        return "contradição"
    return "contingência"


def sao_equivalentes(
    prop_a: Callable[..., bool], prop_b: Callable[..., bool], n_variaveis: int
) -> bool:
    for valores in itertools.product([True, False], repeat=n_variaveis):
        if prop_a(*valores) != prop_b(*valores):
            return False
    return True
```

## Intermediário

**Exercício 1.** Classifique as proposições como tautologia, contradição ou contingência.

```python
classe_1 = classificar(lambda p: p or not p, 1)
classe_2 = classificar(lambda p: p and not p, 1)
classe_3 = classificar(lambda p, q: p and q, 2)

print(classe_1, classe_2, classe_3)
```

**Exercício 2.** Determine se os pares de proposições são logicamente equivalentes.

```python
# (P -> Q) e (not P) or Q são a mesma proposição por definição do condicional
equivalente_1 = sao_equivalentes(
    lambda p, q: condicional(p, q), lambda p, q: (not p) or q, 2
)

# P or Q basta uma ser V; P and Q exige as duas V ao mesmo tempo -> não coincidem
equivalente_2 = sao_equivalentes(lambda p, q: p or q, lambda p, q: p and q, 2)

print(equivalente_1, equivalente_2)
```

**Exercício 3.** Escreva a recíproca, a contrária e a contrapositiva de "se estuda, então passa" (P: estuda, Q: passa).

```python
original_ex = lambda p, q: condicional(p, q)
reciproca_ex = lambda p, q: condicional(q, p)
contraria_ex = lambda p, q: condicional(not p, not q)
contrapositiva_ex = lambda p, q: condicional(not q, not p)

# A contrapositiva é a única das três sempre equivalente ao condicional original
print("Original == Contrapositiva:", sao_equivalentes(original_ex, contrapositiva_ex, 2))
```

**Exercício 1.** Aplique De Morgan para negar "para todo x, (P(x) → Q(x))".

```python
P = [True, True, False, True]
Q = [True, False, False, True]

proposicao_original = all(condicional(p, q) for p, q in zip(P, Q))

# not(para-todo x, P(x) -> Q(x))  ==  existe x, (P(x) and not Q(x))
negacao = any(p and not q for p, q in zip(P, Q))

print("Proposição original:", proposicao_original)
print("Negação:", negacao)
```

**Exercício 2.** Resolva o enigma do bolo do início do notebook.

```python
suspeitos = ["Ana", "Bruno", "Caio"]


def ana_fala(quem: str) -> bool:
    return quem == "Bruno"


def bruno_fala(quem: str) -> bool:
    return quem != "Bruno"


def caio_fala(quem: str) -> bool:
    return not ana_fala(quem)


def quantas_verdadeiras(quem: str) -> int:
    return sum([ana_fala(quem), bruno_fala(quem), caio_fala(quem)])


# Testamos os três suspeitos: só um deles deixa exatamente uma fala verdadeira
culpado = next(s for s in suspeitos if quantas_verdadeiras(s) == 1)

print(f"O culpado e: {culpado}")
```
