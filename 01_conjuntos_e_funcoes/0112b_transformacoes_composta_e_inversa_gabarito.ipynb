```python
# Imports e constantes necessários para os exercícios abaixo.
import math

# Constantes da cadeia do jogo importado (definidas no notebook principal)
cotacao_dolar = 5.0     # R$ por dólar
taxa_imposto = 0.20     # 20% de imposto de importação
cupom_desconto = 20.0   # R$20 de desconto fixo (cupom)


def desfazer_desconto(preco_reais):
    return preco_reais + cupom_desconto


def desfazer_imposto(preco_reais):
    return preco_reais / (1 + taxa_imposto)


def desfazer_conversao(preco_reais):
    return preco_reais / cotacao_dolar
```

## Básico

**Exercício 1.** Dado f(x) = x², escreva g(x) que representa "f deslocada 3 unidades para a direita e 2 para cima" e calcule g(5).

```python
# Deslocar 3 para a direita: trocar x por (x - 3).
# Deslocar 2 para cima: somar 2 no final.
def g_ex1(x):
    return (x - 3) ** 2 + 2


valor_g5_ex1 = g_ex1(5)

print(valor_g5_ex1)
```

**Exercício 2.** Dado f(x) = 2x + 1 e g(x) = x² − 3, calcule (g∘f)(2) e (f∘g)(2).

```python
def f_ex2(x):
    return 2 * x + 1


def g_ex2(x):
    return x ** 2 - 3


# (g∘f)(2) = g(f(2)): primeiro f(2) = 5, depois g(5) = 25-3 = 22
gf_2_ex2 = g_ex2(f_ex2(2))

# (f∘g)(2) = f(g(2)): primeiro g(2) = 1, depois f(1) = 2*1+1 = 3
fg_2_ex2 = f_ex2(g_ex2(2))

print(gf_2_ex2, fg_2_ex2)
print("g∘f e f∘g dão resultados diferentes -> composição não é comutativa.")
```

## Intermediário

**Exercício 3.** Encontre a inversa de f(x) = 3x + 4 e verifique que f(f⁻¹(10)) = 10.

```python
def f_ex3(x):
    return 3 * x + 4


# y = 3x+4  ->  troca x por y, y por x: x = 3y+4  ->  isola y: y = (x-4)/3
def f_inversa_ex3(x):
    return (x - 4) / 3


resultado_ex3 = f_ex3(f_inversa_ex3(10))

print(resultado_ex3)
```

**Exercício 4.** Para f(x) = sen(x), determine g(x) = −2·f(x − 1) + 0.5 e calcule g(1).

```python
def g_ex4(x):
    return -2 * math.sin(x - 1) + 0.5


# em x=1: sin(1-1) = sin(0) = 0, então g(1) = -2*0 + 0.5 = 0.5
valor_g1_ex4 = g_ex4(1)

print(valor_g1_ex4)
```

## Desafio

**Exercício 5.** Refaça a cadeia do jogo importado com cotacao_dolar = 5.5, taxa_imposto = 0.20 e cupom_desconto = 20.0, e calcule o preço final (em real) para d = 45.

```python
def preco_final_ex5(preco_dolar, cotacao, taxa, cupom):
    # mesma ordem do notebook principal: converter -> imposto -> desconto
    preco_reais = preco_dolar * cotacao
    preco_com_imposto = preco_reais * (1 + taxa)
    preco_com_desconto = preco_com_imposto - cupom
    return preco_com_desconto


preco_final_45_ex5 = preco_final_ex5(45, 5.5, 0.20, 20.0)

print(preco_final_45_ex5)
```

**Exercício 6.** Um amigo pagou R$ 250,00 pelo jogo, com a cotação original (5.0), imposto de 20% e cupom de R$20. Descubra o preço original em dólar.

```python
# Desfaz a cadeia na ordem contrária: primeiro o cupom (última etapa
# aplicada), depois o imposto, depois a conversão.
dolar_original_ex6 = desfazer_conversao(desfazer_imposto(desfazer_desconto(250.0)))

print(dolar_original_ex6)
```
