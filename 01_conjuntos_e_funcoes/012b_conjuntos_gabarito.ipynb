```python
# Funções auxiliares definidas no notebook principal, necessárias para os
# exercícios abaixo.
import itertools


def conjunto_das_partes(conjunto: set) -> list[set]:
    elementos = list(conjunto)
    partes = []
    for tamanho in range(len(elementos) + 1):
        for combinacao in itertools.combinations(elementos, tamanho):
            partes.append(set(combinacao))
    return partes
```

**Exercício 4.** Aplique as Leis de De Morgan.

```python
universo_ex2 = set(range(1, 9))
P_ex = {1, 2, 3, 4}
Q_ex = {3, 4, 5, 6}

# ∁(P ∪ Q)
lado_esquerdo_ex = universo_ex2 - (P_ex | Q_ex)

# ∁P ∩ ∁Q
lado_direito_ex = (universo_ex2 - P_ex) & (universo_ex2 - Q_ex)

print(lado_esquerdo_ex == lado_direito_ex)
```

**Exercício 2.** Prove por casos que (A − B) ∪ (A ∩ B) = A sempre vale.

```python
universo_prova = {1, 2, 3}
subconjuntos_prova = conjunto_das_partes(universo_prova)

resultados = [
    ((A_teste - B_teste) | (A_teste & B_teste)) == A_teste
    for A_teste in subconjuntos_prova
    for B_teste in subconjuntos_prova
]

# Testando TODOS os pares de subconjuntos do universo, a igualdade nunca falha
sempre_verdadeira = all(resultados)

print(sempre_verdadeira)
```
