```python
# Import auxiliar necessário para os exercícios abaixo.
import sympy as sp

x_sym = sp.symbols("x", real=True)


def modulo(x):
    '''Calcula o módulo (valor absoluto) de x: a distância de x até 0.'''
    return x if x >= 0 else -x
```

## Básico

**Exercício 1.** Calcule o módulo de cada número da lista e resolva a equação modular |x − 4| = 6.

```python
numeros_ex1 = [9, -9, -2.5, 0]

# módulo de cada número: distância até zero
modulos_ex1 = [modulo(n) for n in numeros_ex1]

# |x-4|=6  ->  x-4=6  ou  x-4=-6
raiz_menor_ex1 = 4 - 6
raiz_maior_ex1 = 4 + 6

print(modulos_ex1, raiz_menor_ex1, raiz_maior_ex1)
```

**Exercício 2.** A partir da fórmula, identifique para onde o gráfico de |x + 5| está deslocado em relação a |x|.

```python
# |x + 5| = |x - (-5)|  ->  a = -5
# a < 0  ->  desloca para a ESQUERDA
deslocamento_ex2 = "esquerda"

print(deslocamento_ex2)
```

## Intermediário

**Exercício 3.** Resolva a inequação |2x + 3| > 7.

```python
# |2x+3|>7  ->  2x+3 < -7  ou  2x+3 > 7
menor_ex3 = sp.Rational(-7 - 3, 2)
maior_ex3 = sp.Rational(7 - 3, 2)

print(menor_ex3, maior_ex3)
print(f"solução: x < {menor_ex3}  ou  x > {maior_ex3}")
```

**Exercício 4 (aplicado).** Um sensor só aceita temperaturas T que satisfaçam |T − 36,5| ≤ 0,5. Determine a faixa de temperaturas aceitas.

```python
# |T-36.5|<=0.5  ->  -0.5 <= T-36.5 <= 0.5
temp_minima_ex4 = 36.5 - 0.5
temp_maxima_ex4 = 36.5 + 0.5

print(temp_minima_ex4, temp_maxima_ex4)
```

## Desafio

**Exercício 5.** Resolva a equação com módulo dos dois lados: |x − 1| = |2x − 4|.

```python
# |x-1|=|2x-4|: elevar ao quadrado (|a|^2=a^2) elimina os dois módulos de uma vez
solucoes_ex5 = sp.solve(sp.Eq((x_sym - 1) ** 2, (2 * x_sym - 4) ** 2), x_sym)
sol_menor_ex5, sol_maior_ex5 = sorted(solucoes_ex5)

print(sol_menor_ex5, sol_maior_ex5)
print(f"solução: x = {sol_menor_ex5}  ou  x = {sol_maior_ex5}")
```
