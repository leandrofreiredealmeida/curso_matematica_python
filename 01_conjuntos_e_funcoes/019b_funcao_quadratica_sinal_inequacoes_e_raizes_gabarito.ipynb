```python
# Import auxiliar necessário para os exercícios abaixo.
import sympy as sp

x_sym = sp.symbols("x")
```

## Básico

**Exercício 1.** Estude o sinal de f(x) = x² - x - 6 (identifique os intervalos onde é positiva, negativa e zero).

```python
a_ex1, b_ex1, c_ex1 = 1, -1, -6

# delta = b² - 4ac
delta_ex1 = b_ex1**2 - 4 * a_ex1 * c_ex1

# Bhaskara: Δ = 25 > 0, duas raízes reais
raiz_menor_ex1 = (-b_ex1 - sp.sqrt(delta_ex1)) / (2 * a_ex1)
raiz_maior_ex1 = (-b_ex1 + sp.sqrt(delta_ex1)) / (2 * a_ex1)

# a = 1 > 0 e Δ > 0 -> positiva FORA das raízes, negativa ENTRE as raízes
sinal_fora_ex1 = "positivo"
sinal_entre_ex1 = "negativo"

print(delta_ex1, raiz_menor_ex1, raiz_maior_ex1, sinal_fora_ex1, sinal_entre_ex1)
```

**Exercício 2.** Resolva a inequação -x² + 4x - 3 ≥ 0.

```python
a_ex2, b_ex2, c_ex2 = -1, 4, -3
delta_ex2 = b_ex2**2 - 4 * a_ex2 * c_ex2

# raízes 1 e 3; a = -1 < 0 -> positiva ENTRE as raízes
sol_menor_ex2 = (-b_ex2 - sp.sqrt(delta_ex2)) / (2 * a_ex2)
sol_maior_ex2 = (-b_ex2 + sp.sqrt(delta_ex2)) / (2 * a_ex2)
sol_menor_ex2, sol_maior_ex2 = sorted((sol_menor_ex2, sol_maior_ex2))

print(sol_menor_ex2, sol_maior_ex2)
print(f"solução: [{sol_menor_ex2}, {sol_maior_ex2}]  (inequação não-estrita -> extremos incluídos)")
```

## Intermediário

**Exercício 3.** Sem resolver a equação, determine o sinal das raízes de 2x² - 7x + 3 = 0 usando S e P.

```python
a_ex3, b_ex3, c_ex3 = 2, -7, 3

# relações de Girard
soma_ex3 = sp.Rational(-b_ex3, a_ex3)
produto_ex3 = sp.Rational(c_ex3, a_ex3)

# S > 0 e P > 0 -> as duas raízes são positivas
sinal_raizes_ex3 = "positivas"

print(soma_ex3, produto_ex3, sinal_raizes_ex3)
```

**Exercício 4.** Determine para quais valores de m a equação x² - mx + (m+3) = 0 tem duas raízes reais e distintas.

```python
m_sym = sp.symbols("m")

# Δ(m) = b² - 4ac com a=1, b=-m, c=(m+3)
delta_m_ex4 = m_sym**2 - 4 * (m_sym + 3)
print("Δ(m) =", sp.expand(delta_m_ex4), "=", sp.factor(delta_m_ex4))

# Δ(m) = (m-6)(m+2) -> raízes de Δ(m)=0
raizes_delta_m = sp.solve(sp.Eq(delta_m_ex4, 0), m_sym)
m_menor_ex4, m_maior_ex4 = sorted(raizes_delta_m)

# Δ(m) é uma parábola em m com concavidade para cima (coef. de m² é 1 > 0)
# -> Δ(m) > 0 (duas raízes reais distintas) FORA do intervalo [m_menor_ex4, m_maior_ex4]
tipo_solucao_ex4 = "fora"

print(m_menor_ex4, m_maior_ex4, tipo_solucao_ex4)
print(f"solução: m < {m_menor_ex4} ou m > {m_maior_ex4}")
```

## Desafio

**Exercício 5.** Sem calcular as raízes de f(x) = x² - 5x + 2, determine se k=1 está entre as raízes, e se k=10 está entre as raízes.

```python
a_ex5, b_ex5, c_ex5 = 1, -5, 2


def f_ex5(k):
    return a_ex5 * k**2 + b_ex5 * k + c_ex5


# teorema da comparação: sinal de a*f(k)
afk1_ex5 = a_ex5 * f_ex5(1)
afk2_ex5 = a_ex5 * f_ex5(10)

# a*f(1) < 0 -> k=1 está ENTRE as raízes
posicao_k1_ex5 = "entre"
# a*f(10) > 0 -> k=10 está FORA do intervalo das raízes
posicao_k2_ex5 = "fora"

print(afk1_ex5, afk2_ex5, posicao_k1_ex5, posicao_k2_ex5)
```

**Exercício 6 (Contextualizado).** Retome o problema do lucro L(x) = -x² + 10x - 16: para quais quantidades vendidas a empresa tem lucro de pelo menos 5 (resolva L(x) ≥ 5)?

```python
a_ex6, b_ex6, c_ex6 = -1, 10, -16

# L(x) >= 5  ->  -x²+10x-16-5 >= 0  ->  -x²+10x-21 >= 0
# multiplicando por -1 (inverte a desigualdade): x² - 10x + 21 <= 0
a_aux, b_aux, c_aux = 1, -10, 21
delta_aux = b_aux**2 - 4 * a_aux * c_aux

sol_menor_ex6 = (-b_aux - sp.sqrt(delta_aux)) / (2 * a_aux)
sol_maior_ex6 = (-b_aux + sp.sqrt(delta_aux)) / (2 * a_aux)
sol_menor_ex6, sol_maior_ex6 = sorted((sol_menor_ex6, sol_maior_ex6))

print(sol_menor_ex6, sol_maior_ex6)
print(f"solução: {sol_menor_ex6} <= x <= {sol_maior_ex6}  (centenas de unidades)")
print(f"em unidades: entre {sol_menor_ex6*100} e {sol_maior_ex6*100} unidades vendidas")
```
