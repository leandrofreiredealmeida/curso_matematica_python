## Básico

**Exercício 1.** Calcule sen(x), cos(x) e tg(x) para x = π/3 (60°), arredondados a 4 casas decimais.

```python
# aplicando diretamente as funções de math ao ângulo pi/3 e arredondando cada resultado
angulo_exercicio1 = math.pi / 3
respostas_exercicio1 = (
    round(math.sin(angulo_exercicio1), 4),
    round(math.cos(angulo_exercicio1), 4),
    round(math.tan(angulo_exercicio1), 4),
)

print(f"sen(pi/3) = {respostas_exercicio1[0]}")
print(f"cos(pi/3) = {respostas_exercicio1[1]}")
print(f"tg(pi/3)  = {respostas_exercicio1[2]}")
```

**Exercício 2.** Para x = 90°, quais das quatro funções tg, cotg, sec e cossec não estão definidas?

```python
# em x = 90 graus, cos(x) = 0 e sen(x) = 1
angulo_exercicio2 = math.radians(90)
seno_90, cosseno_90 = math.sin(angulo_exercicio2), math.cos(angulo_exercicio2)

print(f"sen(90°) = {seno_90:.4f}   cos(90°) = {cosseno_90:.4f}")

# tg e sec tem cos(x) no denominador -> indefinidas quando cos(x) = 0
# cotg e cossec tem sen(x) no denominador -> definidas, pois sen(90°) = 1
respostas_exercicio2 = sorted(["sec", "tg"])
print(f"indefinidas em x = 90°: {respostas_exercicio2}")
```

## Intermediário

**Exercício 4.** Sem calcular o valor exato, determine o sinal de tg(200°) a partir do quadrante do arco.

```python
# 200 graus esta entre 180 e 270, ou seja, no terceiro quadrante
# pela regra T-S-T-C, o terceiro quadrante e o "So Tangente": tg (e cotg) sao positivas ali
resposta_exercicio4 = "positivo"

# conferindo com o valor exato
print(f"tg(200°) = {math.tan(math.radians(200)):.4f}  ->  {resposta_exercicio4}")
```

## Desafio

**Exercício 5.** A maré de um porto varia entre 0,5 m e 2,5 m, com ciclo de 12 horas, começando na altura média e subindo. Modele como y = a·sen(bx + c) + d.

```python
mare_baixa, mare_alta = 0.5, 2.5
periodo_mare = 12  # horas

# a amplitude e metade da distância entre o máximo e o mínimo
a = (mare_alta - mare_baixa) / 2

# o deslocamento vertical e a média entre o máximo e o mínimo
d = (mare_alta + mare_baixa) / 2

# b vem diretamente do período: T = 2*pi / b
b = round(2 * math.pi / periodo_mare, 4)

# como a maré começa exatamente na média e subindo, sen(c) = 0 com a curva subindo -> c = 0
c = 0.0

respostas_exercicio5 = (a, b, c, d)
print(f"a = {a}   b = {b}   c = {c}   d = {d}")
print(f"altura(t) = {a}*sen({b}*t + {c}) + {d}")
```

**Exercício 6.** Compare os períodos das seis funções circulares e verifique que tangente e cotangente têm o menor período entre elas.

```python
# sen, cos, sec e cossec sao construídas diretamente sobre uma volta completa do ciclo -> período 2*pi
# tg e cotg se repetem depois de meia volta, porque tg(x + pi) = tg(x) (o ponto oposto no
# ciclo tem sen e cos com sinais trocados, e a razão sen/cos permanece igual) -> período pi
respostas_exercicio6 = {
    "sen": 2 * math.pi,
    "cos": 2 * math.pi,
    "tg": math.pi,
    "cotg": math.pi,
    "sec": 2 * math.pi,
    "cossec": 2 * math.pi,
}

for nome, periodo in respostas_exercicio6.items():
    print(f"{nome}: período = {periodo:.4f}")

assert respostas_exercicio6["tg"] < respostas_exercicio6["sen"]
print("\ntangente e cotangente têm, de fato, o menor período entre as seis funções.")
```
