# Gabarito — Triângulos Retângulos

Este notebook traz a resolução comentada dos exercícios de [`0509a_triangulos_retangulos.ipynb`](0509a_triangulos_retangulos.ipynb).

```python
import math


def resolver_triangulo_retangulo(
    *,
    cateto_oposto: float | None = None,
    cateto_adjacente: float | None = None,
    hipotenusa: float | None = None,
    angulo_graus: float | None = None,
) -> dict[str, float]:
    """Resolve um triângulo retângulo a partir de exatamente dois dados válidos (reaproveitada
    de 0509a_triangulos_retangulos)."""
    dados = [cateto_oposto, cateto_adjacente, hipotenusa, angulo_graus]
    if sum(dado is not None for dado in dados) != 2:
        raise ValueError("Informe exatamente dois dados: dois lados, ou um lado e um ângulo.")

    if angulo_graus is not None:
        theta = math.radians(angulo_graus)
        if cateto_adjacente is not None:
            cateto_oposto = cateto_adjacente * math.tan(theta)
            hipotenusa = cateto_adjacente / math.cos(theta)
        elif cateto_oposto is not None:
            cateto_adjacente = cateto_oposto / math.tan(theta)
            hipotenusa = cateto_oposto / math.sin(theta)
        else:
            cateto_oposto = hipotenusa * math.sin(theta)
            cateto_adjacente = hipotenusa * math.cos(theta)
    else:
        if hipotenusa is None:
            hipotenusa = math.hypot(cateto_oposto, cateto_adjacente)
        elif cateto_adjacente is None:
            cateto_adjacente = math.sqrt(hipotenusa ** 2 - cateto_oposto ** 2)
        else:
            cateto_oposto = math.sqrt(hipotenusa ** 2 - cateto_adjacente ** 2)
        angulo_graus = math.degrees(math.atan2(cateto_oposto, cateto_adjacente))

    return {
        "cateto_oposto": cateto_oposto,
        "cateto_adjacente": cateto_adjacente,
        "hipotenusa": hipotenusa,
        "angulo_graus": angulo_graus,
        "angulo_complementar_graus": 90 - angulo_graus,
    }


def distancia_por_dois_angulos(
    distancia_entre_pontos: float, angulo_perto_graus: float, angulo_longe_graus: float,
) -> float:
    """Calcula a altura (ou distância perpendicular) de um alvo inacessível a partir de dois
    pontos e dois ângulos (reaproveitada de 0509a_triangulos_retangulos)."""
    cot_perto = 1 / math.tan(math.radians(angulo_perto_graus))
    cot_longe = 1 / math.tan(math.radians(angulo_longe_graus))
    return distancia_entre_pontos / (cot_longe - cot_perto)
```

## Básico

**Exercício 1.** Um triângulo retângulo tem catetos 5 e 12. Calcule a hipotenusa.

```python
# a² = b² + c², com b=5 e c=12
hipotenusa_ex1 = math.sqrt(5 ** 2 + 12 ** 2)
print(hipotenusa_ex1)
```

**Exercício 2.** Um triângulo retângulo tem cateto oposto 9, cateto adjacente 12 e hipotenusa 15 (em relação ao ângulo α). Calcule sen(α), cos(α) e tg(α).

```python
# sen = oposto/hipotenusa, cos = adjacente/hipotenusa, tg = oposto/adjacente
sen_ex2 = 9 / 15
cos_ex2 = 12 / 15
tg_ex2 = 9 / 12
print(sen_ex2, cos_ex2, tg_ex2)
```

## Intermediário

**Exercício 3.** Resolva o triângulo retângulo com cateto adjacente 10 e ângulo agudo de referência 40°.

```python
# um lado (cateto adjacente) + um ângulo agudo já resolvem o triângulo
resultado_ex3 = resolver_triangulo_retangulo(cateto_adjacente=10, angulo_graus=40)
print(resultado_ex3)
```

**Exercício 4.** Um triângulo retângulo tem catetos 6 e 8 (logo, hipotenusa 10). Encontre a altura relativa à hipotenusa.

```python
# da relação métrica bc = ah, temos h = bc/a
altura_ex4 = (6 * 8) / 10
print(altura_ex4)
```

## Desafio

**Exercício 5.** Dois pontos alinhados com a margem oposta de um rio, separados por 40 m, deram ângulos de elevação de 62° (ponto mais próximo) e 35° (ponto mais distante). Calcule a largura do rio.

```python
# método de dois pontos e dois ângulos, arredondado a 2 casas decimais
largura_ex5 = round(distancia_por_dois_angulos(40, 62, 35), 2)
print(largura_ex5)
```

**Exercício 6.** Uma rampa precisa vencer 1,0 m de altura, com inclinação máxima permitida de 5°, num corredor com 12 m disponíveis na horizontal. A rampa cabe?

```python
# o espaço horizontal necessário é o cateto adjacente do triângulo com
# cateto oposto = altura e ângulo = inclinação máxima
espaco_necessario_ex6 = resolver_triangulo_retangulo(cateto_oposto=1.0, angulo_graus=5.0)["cateto_adjacente"]
cabe_ex6 = espaco_necessario_ex6 <= 12
print(espaco_necessario_ex6, cabe_ex6)
```
