```python
import math


def sen_soma(a: float, b: float) -> float:
    """Calcula sen(a + b) usando a fórmula de adição (função reaproveitada de 0506a_transformacoes)."""
    return math.sin(a) * math.cos(b) + math.cos(a) * math.sin(b)


def cos_soma(a: float, b: float) -> float:
    """Calcula cos(a + b) usando a fórmula de adição (função reaproveitada de 0506a_transformacoes)."""
    return math.cos(a) * math.cos(b) - math.sin(a) * math.sin(b)
```

## Básico

**Exercício 1.** Calcule sen, cos e tg de 75° usando as fórmulas de adição de arcos (75° = 45° + 30°).

```python
# 75° = 45° + 30°; usamos sen_soma e cos_soma, e dividimos sen por cos para
# obter a tangente
a75, b75 = math.radians(45), math.radians(30)
sen_75 = sen_soma(a75, b75)
cos_75 = cos_soma(a75, b75)

resultado_75 = {
    "sen": round(sen_75, 4),
    "cos": round(cos_75, 4),
    "tg": round(sen_75 / cos_75, 4),
}
print(resultado_75)
```

**Exercício 2.** Dado sen(a) = 0.6 e cos(a) = 0.8, calcule sen(2a) e cos(2a) usando as fórmulas de arco duplo.

```python
# sen(2a) = 2*sen(a)*cos(a); cos(2a) = cos²(a) - sen²(a)
sen_a_dado = 0.6
cos_a_dado = 0.8

sen_2a_calculado = round(2 * sen_a_dado * cos_a_dado, 4)
cos_2a_calculado = round(cos_a_dado ** 2 - sen_a_dado ** 2, 4)

print(sen_2a_calculado, cos_2a_calculado)
```

## Intermediário

**Exercício 3.** Dado cos(a) = −0,6, com a no 3º quadrante, calcule sen(a/2) e cos(a/2), decidindo o sinal a partir do quadrante de a/2.

```python
# a está entre 180° e 270° (3º quadrante), logo a/2 está entre 90° e 135°
# (2º quadrante): sen(a/2) é positivo e cos(a/2) é negativo
cos_a_ex3 = -0.6

sen_meio_calculado = round(math.sqrt((1 - cos_a_ex3) / 2), 4)
cos_meio_calculado = round(-math.sqrt((1 + cos_a_ex3) / 2), 4)

print(sen_meio_calculado, cos_meio_calculado)
```

**Exercício 4.** Transforme a soma sen(50°) + sen(10°) em um produto e calcule seu valor numérico.

```python
# sen p + sen q = 2*sen((p+q)/2)*cos((p-q)/2), com p=50° e q=10°
p4, q4 = math.radians(50), math.radians(10)
soma_como_produto = round(2 * math.sin((p4 + q4) / 2) * math.cos((p4 - q4) / 2), 4)

print(soma_como_produto)
```

## Desafio

**Exercício 5.** Deduza a fórmula de sen(3a) escrevendo 3a como 2a + a e aplicando a fórmula de adição.

```python
# sen(3a) = sen(2a + a) = sen(2a)*cos(a) + cos(2a)*sen(a), com sen(2a) e
# cos(2a) dados pelas fórmulas de arco duplo
def sen_3a(a_rad: float) -> float:
    sen_2a = 2 * math.sin(a_rad) * math.cos(a_rad)
    cos_2a = math.cos(a_rad) ** 2 - math.sin(a_rad) ** 2
    return sen_2a * math.cos(a_rad) + cos_2a * math.sin(a_rad)


angulos_teste = [math.radians(g) for g in (20, 50, 100)]
resultados_sen_3a = [round(sen_3a(a), 4) for a in angulos_teste]

print(resultados_sen_3a)
```

**Exercício 6.** Duas notas musicais têm frequências de 442 Hz e 437 Hz. Calcule a frequência do batimento percebido e o período desse "pulsar".

```python
# a frequência do batimento percebido é |f1 - f2|, e o período é o inverso
# dessa frequência
freq_nota1 = 442
freq_nota2 = 437

freq_batimento_calculada = abs(freq_nota1 - freq_nota2)
periodo_batimento_calculado = round(1 / freq_batimento_calculada, 4)

print(freq_batimento_calculada, periodo_batimento_calculado)
```
