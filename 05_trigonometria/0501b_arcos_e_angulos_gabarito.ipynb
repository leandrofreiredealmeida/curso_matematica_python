# Gabarito — Arcos e Ângulos

Este notebook traz a resolução comentada dos exercícios de [`0501a_arcos_e_angulos.ipynb`](0501a_arcos_e_angulos.ipynb).

⚠️ **Antes de olhar aqui:** tente resolver os exercícios sozinho no notebook original e use as células de verificação (`assert`) para conferir suas respostas. Volte para cá só se travar ou quiser comparar sua solução.

```python
import math

# funções reconstruídas do notebook principal, necessárias para as resoluções abaixo
def graus_para_radianos(angulo_graus: float) -> float:
    """Converte um ângulo de graus para radianos."""
    return angulo_graus * math.pi / 180


def radianos_para_graus(angulo_radianos: float) -> float:
    """Converte um ângulo de radianos para graus."""
    return angulo_radianos * 180 / math.pi


def comprimento_arco(raio: float, angulo_rad: float) -> float:
    """Calcula o comprimento de um arco, dado o raio e o ângulo central em radianos."""
    return raio * angulo_rad


def quadrante(angulo_graus: float) -> str:
    """Determina o quadrante de um ângulo qualquer, tratando ângulos negativos e maiores que 360°."""
    angulo_normalizado = angulo_graus % 360

    if angulo_normalizado in (0, 90, 180, 270):
        return "sobre um eixo (não pertence a nenhum quadrante)"
    if angulo_normalizado < 90:
        return "1º quadrante"
    if angulo_normalizado < 180:
        return "2º quadrante"
    if angulo_normalizado < 270:
        return "3º quadrante"
    return "4º quadrante"
```

## Básico

**Exercício 1.** Converta os ângulos [15°, 100°, 225°, 300°, 330°] de grau para radiano, arredondando a 4 casas decimais.

```python
# aplicando a fórmula theta_rad = theta_graus * pi / 180 em cada ângulo
angulos_exercicio1 = [15, 100, 225, 300, 330]
respostas_exercicio1 = [round(graus_para_radianos(a), 4) for a in angulos_exercicio1]

for angulo, resposta in zip(angulos_exercicio1, respostas_exercicio1):
    print(f"{angulo}°  ->  {resposta} rad")
```

**Exercício 2.** Identifique o quadrante dos ângulos [200°, −30°, 450°, 90°, 15°].

```python
# normalizando cada ângulo com "% 360" antes de comparar com os limites de cada quadrante
angulos_exercicio2 = [200, -30, 450, 90, 15]
respostas_exercicio2 = [quadrante(a) for a in angulos_exercicio2]

for angulo, resposta in zip(angulos_exercicio2, respostas_exercicio2):
    print(f"{angulo}°  (normalizado: {angulo % 360}°)  ->  {resposta}")
```

## Intermediário

**Exercício 3.** Uma bicicleta tem rodas de 33 cm de raio. Quantos metros ela percorre em uma volta e meia de pedalada (ângulo de 540°)?

```python
# convertendo o raio para metros e o ângulo para radianos antes de aplicar s = r*theta
raio_roda_bicicleta_cm = 33
angulo_pedalada_graus = 540

raio_roda_bicicleta_m = raio_roda_bicicleta_cm / 100
angulo_pedalada_rad = graus_para_radianos(angulo_pedalada_graus)
distancia_exercicio3 = round(comprimento_arco(raio_roda_bicicleta_m, angulo_pedalada_rad), 2)

print(f"raio = {raio_roda_bicicleta_m} m, ângulo = {angulo_pedalada_graus}° = {angulo_pedalada_rad:.4f} rad")
print(f"distância percorrida = {distancia_exercicio3} metros")
```

**Exercício 4.** Os arcos de 1000° e 280° são côngruos?

```python
# dois arcos são côngruos quando têm o mesmo resto na divisão por 360 (mesma extremidade no ciclo)
resto_1000 = 1000 % 360
resto_280 = 280 % 360
sao_congruos_exercicio4 = resto_1000 == resto_280

print(f"1000° % 360 = {resto_1000}°")
print(f"280° % 360 = {resto_280}°")
print(f"são côngruos? {sao_congruos_exercicio4}")
```

## Desafio

**Exercício 5.** Uma pista de atletismo circular tem 40 metros de raio. Um corredor percorre um arco de 250 metros ao longo da pista. Quantos graus ele percorreu?

```python
# isolando theta em s = r*theta (theta = s/r, em radianos) e só depois convertendo para graus
raio_pista_m = 40
arco_percorrido_m = 250

angulo_exercicio5_rad = arco_percorrido_m / raio_pista_m
angulo_exercicio5_graus = round(radianos_para_graus(angulo_exercicio5_rad), 2)

print(f"theta = s / r = {arco_percorrido_m} / {raio_pista_m} = {angulo_exercicio5_rad:.4f} rad")
print(f"em graus: {angulo_exercicio5_graus}°")
```

**Exercício 6.** Escreva a expressão geral dos arcos côngruos a 130° e encontre, por código, todos os que caem no intervalo [0°, 1000°].

```python
# expressão geral: 130 + k*360, k inteiro. Geramos vários candidatos e filtramos o intervalo.
angulo_base_exercicio6 = 130
candidatos_exercicio6 = [angulo_base_exercicio6 + k * 360 for k in range(-5, 5)]
arcos_no_intervalo_exercicio6 = sorted(a for a in candidatos_exercicio6 if 0 <= a <= 1000)

print(f"expressão geral: 130° + k x 360°, k inteiro")
print(f"arcos no intervalo [0°, 1000°]: {arcos_no_intervalo_exercicio6}")
```
