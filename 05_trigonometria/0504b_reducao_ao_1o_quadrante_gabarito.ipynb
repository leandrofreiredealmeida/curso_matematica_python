```python
import math


def quadrante(angulo_graus: float) -> str:
    '''Determina o quadrante de um ângulo qualquer, tratando ângulos negativos e maiores que 360°.'''
    angulo_normalizado = angulo_graus % 360

    if angulo_normalizado in (0, 90, 180, 270):
        return "sobre um eixo (não pertence a nenhum quadrante)"
    if angulo_normalizado < 90:
        return "1º quadrante"
    if angulo_normalizado < 180:
        return "2º quadrante"
    if angulo_normalizado < 270:
        return "3º quadrante"
    return "4º quadrante"


def reduzir_2o_quadrante(angulo_graus: float) -> float:
    '''Reduz um ângulo do 2º quadrante ao ângulo equivalente no 1º quadrante.'''
    return 180 - angulo_graus


def reduzir_3o_quadrante(angulo_graus: float) -> float:
    '''Reduz um ângulo do 3º quadrante ao ângulo equivalente no 1º quadrante.'''
    return angulo_graus - 180


def calculadora_universal(angulo_graus: float) -> dict[str, float]:
    '''Calcula sen, cos e tg de qualquer ângulo, reduzindo-o ao 1º quadrante.'''
    angulo_no_ciclo = angulo_graus % 360

    if angulo_no_ciclo in (0, 180):
        sen_x, cos_x = (0.0, 1.0) if angulo_no_ciclo == 0 else (0.0, -1.0)
    elif angulo_no_ciclo == 90:
        sen_x, cos_x = 1.0, 0.0
    elif angulo_no_ciclo == 270:
        sen_x, cos_x = -1.0, 0.0
    else:
        q = quadrante(angulo_no_ciclo)
        if q == "1º quadrante":
            angulo_1q, sinal_sen, sinal_cos = angulo_no_ciclo, 1, 1
        elif q == "2º quadrante":
            angulo_1q, sinal_sen, sinal_cos = reduzir_2o_quadrante(angulo_no_ciclo), 1, -1
        elif q == "3º quadrante":
            angulo_1q, sinal_sen, sinal_cos = reduzir_3o_quadrante(angulo_no_ciclo), -1, -1
        else:  # 4º quadrante
            angulo_1q, sinal_sen, sinal_cos = 360 - angulo_no_ciclo, -1, 1

        angulo_1q_rad = math.radians(angulo_1q)
        sen_x = sinal_sen * math.sin(angulo_1q_rad)
        cos_x = sinal_cos * math.cos(angulo_1q_rad)

    tg_x = sen_x / cos_x if abs(cos_x) > 1e-12 else math.nan
    return {"sen": sen_x, "cos": cos_x, "tg": tg_x}
```

## Básico

**Exercício 1.** Reduza 160° (2º quadrante), 200° (3º quadrante) e 300° (4º quadrante) ao 1º quadrante e calcule sen(160°), cos(200°) e sen(300°).

```python
# 160° está no 2º quadrante: x' = 180 - 160 = 20°, e sen(x) = sen(x')
sen_160 = round(math.sin(math.radians(180 - 160)), 4)

# 200° está no 3º quadrante: x' = 200 - 180 = 20°, e cos(x) = -cos(x')
cos_200 = round(-math.cos(math.radians(200 - 180)), 4)

# 300° está no 4º quadrante: x' = 360 - 300 = 60°, e sen(x) = -sen(x')
sen_300 = round(-math.sin(math.radians(360 - 300)), 4)

resultados_exercicio1 = [sen_160, cos_200, sen_300]
print(resultados_exercicio1)
```

**Exercício 2.** Sabendo que sen(35°) = 0,5736 e cos(35°) = 0,8192, use a paridade para calcular sen(−35°) e cos(−35°).

```python
# pela paridade, sen(-x) = -sen(x) e cos(-x) = cos(x) -- não é preciso nenhum
# cálculo trigonométrico novo, só usar os valores já dados
sen_35 = 0.5736
cos_35 = 0.8192

sen_menos_35 = -sen_35
cos_menos_35 = cos_35

respostas_exercicio2 = (sen_menos_35, cos_menos_35)
print(respostas_exercicio2)
```

## Desafio

**Exercício 5.** Implemente a lógica de redução ao 1º quadrante para calcular sen, cos e tg dos ângulos [130°, 250°, 310°, −60°, 800°], e compare com math.sin/cos/tan.

```python
# reimplementando a mesma lógica de calculadora_universal "do zero": primeiro
# reduzimos o ângulo a [0, 360) com %, depois aplicamos a fórmula do
# quadrante correspondente usando apenas seno e cosseno de ângulos do 1º
# quadrante (math.sin/math.cos restritos a 0-90°)
def calculadora_do_zero(angulo_graus: float) -> dict[str, float]:
    angulo_no_ciclo = angulo_graus % 360
    q = quadrante(angulo_no_ciclo)

    if q == "1º quadrante":
        angulo_1q, sinal_sen, sinal_cos = angulo_no_ciclo, 1, 1
    elif q == "2º quadrante":
        angulo_1q, sinal_sen, sinal_cos = 180 - angulo_no_ciclo, 1, -1
    elif q == "3º quadrante":
        angulo_1q, sinal_sen, sinal_cos = angulo_no_ciclo - 180, -1, -1
    else:
        angulo_1q, sinal_sen, sinal_cos = 360 - angulo_no_ciclo, -1, 1

    angulo_1q_rad = math.radians(angulo_1q)
    sen_x = sinal_sen * math.sin(angulo_1q_rad)
    cos_x = sinal_cos * math.cos(angulo_1q_rad)
    return {"sen": sen_x, "cos": cos_x, "tg": sen_x / cos_x}


tolerancia_exercicio5 = 1e-9
angulos_teste_exercicio5 = [130, 250, 310, -60, 800]
resultados_exercicio5 = []

for a in angulos_teste_exercicio5:
    resultado = calculadora_do_zero(a)
    bate = (
        abs(resultado["sen"] - math.sin(math.radians(a))) < tolerancia_exercicio5
        and abs(resultado["cos"] - math.cos(math.radians(a))) < tolerancia_exercicio5
        and abs(resultado["tg"] - math.tan(math.radians(a))) < tolerancia_exercicio5
    )
    resultados_exercicio5.append(bate)
    print(f"{a:>4}°: sen={resultado['sen']:.4f} cos={resultado['cos']:.4f} tg={resultado['tg']:.4f}   bate com math? {bate}")

print(f"\n{resultados_exercicio5}")
```

**Exercício 6.** Calcule sen(−750°) do início ao fim, mostrando cada passo.

```python
# passo (a): reduzir -750° ao intervalo [0°, 360°)
angulo_no_intervalo = -750 % 360  # = 330°

# passo (b): identificar o quadrante do ângulo reduzido
quadrante_do_angulo = quadrante(angulo_no_intervalo)  # "4º quadrante"

# passo (c): reduzir ao 1º quadrante e calcular sen -- 330° está no 4º
# quadrante, então x' = 360 - 330 = 30°, e sen(x) = -sen(x')
angulo_1o_quadrante = 360 - angulo_no_intervalo
sen_final = round(-math.sin(math.radians(angulo_1o_quadrante)), 4)

respostas_exercicio6 = (angulo_no_intervalo, quadrante_do_angulo, sen_final)

print(f"(a) -750° reduzido a [0°, 360°): {angulo_no_intervalo}°")
print(f"(b) quadrante: {quadrante_do_angulo}")
print(f"(c) 1º quadrante equivalente: {angulo_1o_quadrante}°, sen(-750°) = {sen_final}")
print(f"\nresposta final: {respostas_exercicio6}")
```
