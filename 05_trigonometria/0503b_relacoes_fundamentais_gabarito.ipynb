## Básico

**Exercício 1.** Verifique numericamente que sen²(x) + cos²(x) = 1 para os ângulos [17°, 63°, 128°, 194°, 275°, 340°].

```python
# para cada ângulo, calculamos sen(x)^2 + cos(x)^2 e comparamos com 1 usando uma
# tolerância pequena (1e-9), ja que operações com ponto flutuante raramente dão
# um resultado exatamente igual a 1
angulos_exercicio1 = [17, 63, 128, 194, 275, 340]
resultados_exercicio1 = [
    abs(math.sin(math.radians(a)) ** 2 + math.cos(math.radians(a)) ** 2 - 1) < 1e-9
    for a in angulos_exercicio1
]

for angulo, resultado in zip(angulos_exercicio1, resultados_exercicio1):
    print(f"{angulo}°: sen²+cos² = 1 ? {resultado}")
```

**Exercício 2.** Sabendo que sen(x) = 0,6 e cos(x) = 0,8, calcule tg(x), cotg(x), sec(x) e cossec(x).

```python
# aplicando diretamente as relações de quociente e as relações derivadas
sen_exercicio2 = 0.6
cos_exercicio2 = 0.8

tg_x = round(sen_exercicio2 / cos_exercicio2, 4)
cotg_x = round(cos_exercicio2 / sen_exercicio2, 4)
sec_x = round(1 / cos_exercicio2, 4)
cossec_x = round(1 / sen_exercicio2, 4)

respostas_exercicio2 = (tg_x, cotg_x, sec_x, cossec_x)
print(f"tg(x) = {tg_x}   cotg(x) = {cotg_x}   sec(x) = {sec_x}   cossec(x) = {cossec_x}")
```

## Intermediário

**Exercício 3.** Simplifique as expressões tg(x)·cos(x), sec²(x) − tg²(x), sen²(x)·(1 + cotg²(x)) e (1 − cos²(x))/sen(x).

```python
# (a) tg(x)*cos(x) = (sen(x)/cos(x)) * cos(x) = sen(x)  [o cos(x) se cancela]
resposta_a = "sen(x)"

# (b) sec^2(x) - tg^2(x): pela relação derivada sec^2(x) = 1 + tg^2(x), logo
#     sec^2(x) - tg^2(x) = (1 + tg^2(x)) - tg^2(x) = 1
resposta_b = "1"

# (c) sen^2(x)*(1 + cotg^2(x)): pela relação derivada, 1 + cotg^2(x) = cossec^2(x) = 1/sen^2(x)
#     logo sen^2(x) * (1/sen^2(x)) = 1
resposta_c = "1"

# (d) (1 - cos^2(x))/sen(x): pela relação fundamental, 1 - cos^2(x) = sen^2(x)
#     logo sen^2(x)/sen(x) = sen(x)
resposta_d = "sen(x)"

respostas_exercicio3 = [resposta_a, resposta_b, resposta_c, resposta_d]
print(respostas_exercicio3)
```

**Exercício 4.** Demonstre numericamente que tg(x)·cossec(x) = sec(x).

```python
# tg(x)*cossec(x) = (sen(x)/cos(x)) * (1/sen(x)) = 1/cos(x) = sec(x) -- o sen(x) se
# cancela algebricamente; conferimos numericamente para vários valores de x
x_testes_exercicio4 = [0.2, 0.8, 1.3, 2.1, 2.9, 4.0]
diferencas_exercicio4 = [
    abs(math.tan(x) * (1 / math.sin(x)) - 1 / math.cos(x)) for x in x_testes_exercicio4
]

identidade_valida_exercicio4 = max(diferencas_exercicio4) < 1e-9
print(f"maior diferença entre os dois lados: {max(diferencas_exercicio4):.2e}")
print(f"identidade válida? {identidade_valida_exercicio4}")
```

## Desafio

**Exercício 5.** Demonstre numericamente que sen⁴(x) − cos⁴(x) = sen²(x) − cos²(x).

```python
# fatorando o lado esquerdo como diferença de quadrados:
# sen^4(x) - cos^4(x) = (sen^2(x) - cos^2(x)) * (sen^2(x) + cos^2(x))
#                     = (sen^2(x) - cos^2(x)) * 1        [relação fundamental]
#                     = sen^2(x) - cos^2(x)               [exatamente o lado direito]
x_testes_exercicio5 = [0.1, 0.7, 1.4, 2.2, 3.5, 5.0]
diferencas_exercicio5 = [
    abs((math.sin(x) ** 4 - math.cos(x) ** 4) - (math.sin(x) ** 2 - math.cos(x) ** 2))
    for x in x_testes_exercicio5
]

identidade_valida_exercicio5 = max(diferencas_exercicio5) < 1e-9
print(f"maior diferença entre os dois lados: {max(diferencas_exercicio5):.2e}")
print(f"identidade válida? {identidade_valida_exercicio5}")
```

**Exercício 6.** Um bloco está sobre um plano inclinado de 35°; uma força de 120 N atua na direção da inclinação. Decomponha a força e verifique o módulo com a relação fundamental.

```python
# decompomos a força em uma componente paralela ao plano (F*cos(theta)) e uma
# componente perpendicular ao plano (F*sen(theta)); pela relação fundamental, a
# soma dos quadrados dessas componentes recupera o módulo original ao quadrado
forca_N = 120
angulo_plano_graus = 35
angulo_plano_rad = math.radians(angulo_plano_graus)

componente_paralela = forca_N * math.cos(angulo_plano_rad)
componente_perpendicular = forca_N * math.sin(angulo_plano_rad)

modulo_recuperado_exercicio6 = round(
    math.sqrt(componente_paralela**2 + componente_perpendicular**2), 2
)

print(f"componente paralela:      {componente_paralela:.2f} N")
print(f"componente perpendicular: {componente_perpendicular:.2f} N")
print(f"módulo recuperado:        {modulo_recuperado_exercicio6} N (original: {forca_N} N)")
```
