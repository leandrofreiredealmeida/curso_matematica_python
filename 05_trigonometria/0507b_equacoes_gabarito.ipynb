```python
import math

import sympy as sp


def familia_1_sen(beta: float, k: int) -> float:
    """alpha = beta + 2k*pi (função reaproveitada de 0507a_equacoes)."""
    return beta + 2 * k * math.pi


def familia_2_sen(beta: float, k: int) -> float:
    """alpha = pi - beta + 2k*pi (função reaproveitada de 0507a_equacoes)."""
    return math.pi - beta + 2 * k * math.pi


def familia_1_cos(beta: float, k: int) -> float:
    """alpha = beta + 2k*pi (função reaproveitada de 0507a_equacoes)."""
    return beta + 2 * k * math.pi


def familia_2_cos(beta: float, k: int) -> float:
    """alpha = -beta + 2k*pi (função reaproveitada de 0507a_equacoes)."""
    return -beta + 2 * k * math.pi


def resolver_e_restringir(familias: list, intervalo: tuple[float, float], k_max: int = 10) -> list[float]:
    """Gera as soluções de uma lista de famílias e devolve as que caem em [início, fim) (reaproveitada de 0507a_equacoes)."""
    inicio, fim = intervalo
    solucoes = set()
    for k in range(-k_max, k_max + 1):
        for familia in familias:
            valor = familia(k)
            if inicio <= valor < fim:
                solucoes.add(round(valor, 6))
    return sorted(solucoes)


def bhaskara(a: float, b: float, c: float) -> tuple[float, float]:
    """Resolve a*x^2 + b*x + c = 0 pela fórmula de Bhaskara (função reaproveitada de 0507a_equacoes)."""
    delta = b ** 2 - 4 * a * c
    raiz_delta = math.sqrt(delta)
    x1 = (-b + raiz_delta) / (2 * a)
    x2 = (-b - raiz_delta) / (2 * a)
    return x1, x2
```

## Básico

**Exercício 1.** Resolva sen(x) = √3/2 e encontre os dois primeiros valores de cada família de solução, em graus.

```python
# beta = pi/3 (60°), já que sen(60°) = √3/2; usamos familia_1_sen e
# familia_2_sen com k=0
beta_ex1 = math.pi / 3
resultado_ex1 = (
    round(math.degrees(familia_1_sen(beta_ex1, 0)), 1),
    round(math.degrees(familia_2_sen(beta_ex1, 0)), 1),
)
print(resultado_ex1)
```

**Exercício 2.** Restrinja a solução geral de cos(x) = −1/2 ao intervalo [0, 4π) (duas voltas completas).

```python
# beta = 2*pi/3 (120°), já que cos(120°) = -1/2
beta_ex2 = 2 * math.pi / 3
solucoes_ex2 = resolver_e_restringir(
    [lambda k: familia_1_cos(beta_ex2, k), lambda k: familia_2_cos(beta_ex2, k)],
    (0, 4 * math.pi),
)
print(solucoes_ex2)
```

## Intermediário

**Exercício 3.** Resolva cos(3x) = √2/2 no intervalo [0, 2π), tratando 3x como a incógnita.

```python
# beta = pi/4 (45°); resolvemos para u = 3x normalmente e só depois
# dividimos TODA a solução geral por 3 (inclusive o termo 2*k*pi)
beta_ex3 = math.pi / 4
solucoes_ex3 = resolver_e_restringir(
    [
        lambda k: familia_1_cos(beta_ex3, k) / 3,
        lambda k: familia_2_cos(beta_ex3, k) / 3,
    ],
    (0, 2 * math.pi),
)
print(solucoes_ex3)
```

**Exercício 4.** Resolva a equação redutível ao 2º grau 2·cos²x + cos x − 1 = 0, conferindo se as raízes são válidas.

```python
# 2*cos²x + cos x - 1 = 0 vira 2u² + u - 1 = 0, com u = cos x
u1_ex4, u2_ex4 = bhaskara(2, 1, -1)
raizes_ex4 = tuple(sorted([round(u1_ex4, 4), round(u2_ex4, 4)]))

# ambas as raízes precisam estar em [-1, 1] para serem válidas
raizes_validas_ex4 = [u for u in raizes_ex4 if abs(u) <= 1]

print(raizes_ex4, raizes_validas_ex4)
```

## Desafio

**Exercício 5.** Resolva √3·sen x + cos x = 1 usando o método do ângulo auxiliar, do início ao fim.

```python
# R = sqrt(a² + b²), phi tal que cos(phi) = a/R e sen(phi) = b/R; a equação
# vira R*sen(x + phi) = c, ou seja, sen(x + phi) = c/R
a_ex5, b_ex5, c_ex5 = math.sqrt(3), 1, 1
R_ex5 = math.sqrt(a_ex5 ** 2 + b_ex5 ** 2)
phi_ex5 = math.atan2(b_ex5, a_ex5)
beta_ex5 = math.asin(c_ex5 / R_ex5)

# duas famílias em (x + phi), depois isolamos x; guardamos as duas primeiras
# soluções não negativas
candidatos_ex5 = sorted(
    v for k in range(-2, 3)
    for v in (
        beta_ex5 - phi_ex5 + 2 * k * math.pi,
        math.pi - beta_ex5 - phi_ex5 + 2 * k * math.pi,
    )
    if v >= -1e-9
)
solucao_ex5 = tuple(round(v, 4) for v in candidatos_ex5[:2])

print(solucao_ex5)
```

**Exercício 6.** Um modelo de temperatura ao longo do dia é dado por T(h) = 6·sen(π/12·(h − 9)) + 20. Encontre os horários em que T(h) = 23, usando `sympy`.

```python
# montamos a equação e resolvemos com sympy.solveset, restringindo o
# domínio diretamente ao intervalo de um dia [0, 24)
h_sym = sp.symbols("h", real=True)
equacao_ex6 = 6 * sp.sin(sp.pi / 12 * (h_sym - 9)) + 20 - 23
solucao_ex6 = sp.solveset(equacao_ex6, h_sym, domain=sp.Interval(0, 24, False, True))
horarios_ex6 = tuple(sorted(round(float(s), 2) for s in solucao_ex6))

print(horarios_ex6)
```
