```python
import math

import sympy as sp

BASES_NOTAVEIS = {
    30: {"sen": sp.Rational(1, 2), "cos": sp.sqrt(3) / 2, "tg": sp.sqrt(3) / 3},
    45: {"sen": sp.sqrt(2) / 2, "cos": sp.sqrt(2) / 2, "tg": sp.Integer(1)},
    60: {"sen": sp.sqrt(3) / 2, "cos": sp.Rational(1, 2), "tg": sp.sqrt(3)},
}


def quadrante(angulo_graus: int) -> str:
    '''Determina o quadrante de um ângulo entre 0° e 360° (função reaproveitada de 0505a_arcos_notaveis).'''
    if angulo_graus in (0, 90, 180, 270):
        return "sobre um eixo"
    if angulo_graus < 90:
        return "1º quadrante"
    if angulo_graus < 180:
        return "2º quadrante"
    if angulo_graus < 270:
        return "3º quadrante"
    return "4º quadrante"


def valor_exato_notavel(angulo_graus: int) -> dict[str, sp.Expr | None]:
    '''Calcula sen, cos e tg exatos de um ângulo múltiplo de 30° ou 45°, em
    qualquer quadrante (função reaproveitada de 0505a_arcos_notaveis).'''
    angulo_no_ciclo = angulo_graus % 360

    if angulo_no_ciclo == 0:
        return {"sen": sp.Integer(0), "cos": sp.Integer(1), "tg": sp.Integer(0)}
    if angulo_no_ciclo == 90:
        return {"sen": sp.Integer(1), "cos": sp.Integer(0), "tg": None}
    if angulo_no_ciclo == 180:
        return {"sen": sp.Integer(0), "cos": sp.Integer(-1), "tg": sp.Integer(0)}
    if angulo_no_ciclo == 270:
        return {"sen": sp.Integer(-1), "cos": sp.Integer(0), "tg": None}

    q = quadrante(angulo_no_ciclo)
    if q == "1º quadrante":
        angulo_1q, sinal_sen, sinal_cos = angulo_no_ciclo, 1, 1
    elif q == "2º quadrante":
        angulo_1q, sinal_sen, sinal_cos = 180 - angulo_no_ciclo, 1, -1
    elif q == "3º quadrante":
        angulo_1q, sinal_sen, sinal_cos = angulo_no_ciclo - 180, -1, -1
    else:
        angulo_1q, sinal_sen, sinal_cos = 360 - angulo_no_ciclo, -1, 1

    base = BASES_NOTAVEIS[angulo_1q]
    sen_x = sinal_sen * base["sen"]
    cos_x = sinal_cos * base["cos"]
    tg_x = sen_x / cos_x
    return {"sen": sen_x, "cos": cos_x, "tg": tg_x}
```

## Básico

**Exercício 1.** Calcule sen, cos e tg de 30°, 45°, 60° e 90° usando os valores exatos deduzidos nos Blocos 2, 3 e 4.

```python
# usamos os valores exatos deduzidos nos Blocos 2, 3 e 4: sen(30)=1/2,
# cos(30)=√3/2, tg(30)=√3/3; sen(45)=cos(45)=√2/2, tg(45)=1; sen(60)=√3/2,
# cos(60)=1/2, tg(60)=√3; e sen(90)=1, cos(90)=0, tg(90) não existe
tabela_exercicio1 = {
    30: {"sen": round(1 / 2, 4), "cos": round(math.sqrt(3) / 2, 4), "tg": round(math.sqrt(3) / 3, 4)},
    45: {"sen": round(math.sqrt(2) / 2, 4), "cos": round(math.sqrt(2) / 2, 4), "tg": 1.0},
    60: {"sen": round(math.sqrt(3) / 2, 4), "cos": round(1 / 2, 4), "tg": round(math.sqrt(3), 4)},
    90: {"sen": 1.0, "cos": 0.0, "tg": None},
}
print(tabela_exercicio1)
```

**Exercício 2.** Sem consultar a tabela pronta, deduza de novo os valores de sen(30°) e sen(45°) a partir dos triângulos dos Blocos 2 e 3.

```python
# triângulo equilátero de lado 2: sen(30°) = cateto oposto (1) / hipotenusa (2)
# triângulo isósceles de catetos 1: sen(45°) = cateto (1) / hipotenusa (√2)
sen_30_deduzido = round(1 / 2, 4)
sen_45_deduzido = round(1 / math.sqrt(2), 4)

print(sen_30_deduzido, sen_45_deduzido)
```

## Intermediário

**Exercício 3.** Calcule os valores exatos de sen(120°), cos(135°), tg(210°) e sen(315°), combinando a tabela de arcos notáveis com a redução ao 1º quadrante.

```python
resultados_exercicio3 = [
    round(float(valor_exato_notavel(120)["sen"]), 4),
    round(float(valor_exato_notavel(135)["cos"]), 4),
    round(float(valor_exato_notavel(210)["tg"]), 4),
    round(float(valor_exato_notavel(315)["sen"]), 4),
]
print(resultados_exercicio3)
```

**Exercício 4.** Um projétil precisa de um alcance de aproximadamente 40,8 m, com v0 = 20 m/s e g = 9,8 m/s². Descubra qual ângulo notável de lançamento o produz.

```python
# calculamos o alcance para os três ângulos notáveis e escolhemos o mais
# próximo do alcance desejado
alcance_desejado = 40.8
candidatos = {angulo: (20 ** 2) * math.sin(math.radians(2 * angulo)) / 9.8 for angulo in (30, 45, 60)}
angulo_alcance_desejado = min(candidatos, key=lambda a: abs(candidatos[a] - alcance_desejado))

print(candidatos, "->", angulo_alcance_desejado)
```

## Desafio

**Exercício 5.** Sem usar a fórmula do alcance, explique numericamente por que um lançamento a 30° e outro a 60° alcançam a mesma distância.

```python
# sen(2*30°) = sen(60°) e sen(2*60°) = sen(120°); como 120° = 180° - 60°, a
# simetria do 2º quadrante garante que sen(120°) = sen(60°) -- por isso os
# dois ângulos de lançamento dão o mesmo alcance
sen_60 = round(math.sin(math.radians(60)), 4)
sen_120 = round(math.sin(math.radians(120)), 4)
angulos_sao_simetricos = (180 - 60) == 120

print(sen_60, sen_120, angulos_sao_simetricos)
```

**Exercício 6 (problema em aberto).** Tente encontrar uma combinação de sen(30°), cos(30°), sen(45°) e cos(45°) que se aproxime do valor real de sen(15°).

```python
# não existe uma única resposta certa aqui -- o objetivo é só tentar
# aproximar sen(15°) por tentativa. Nenhuma combinação simples de soma ou
# subtração de sen/cos acerta o valor exato -- a fórmula certa só aparece em
# 📌 Transformações (fórmula de sen(a - b) = sen(a)cos(b) - cos(a)sen(b))
tentativa_sen_15 = round((math.sin(math.radians(30)) + math.sin(math.radians(45))) / 4, 4)

print(tentativa_sen_15, "vs sen(15°) real:", round(math.sin(math.radians(15)), 4))
```
