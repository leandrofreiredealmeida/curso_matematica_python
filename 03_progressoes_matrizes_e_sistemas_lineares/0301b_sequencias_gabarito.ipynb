```python
# Imports e funções auxiliares necessários para os exercícios abaixo,
# reconstruindo o que foi definido no notebook principal.
import numpy as np


def fibonacci(n: int) -> int:
    """Calcula o n-ésimo termo de Fibonacci por recursão (F_1 = F_2 = 1)."""
    if n in (1, 2):
        return 1
    return fibonacci(n - 1) + fibonacci(n - 2)
```

## Intermediário

**Exercício 3.** Calcule os termos das duas leis abaixo (n de 1 a 6) e diga se as sequências são iguais.

```python
indices_ex3 = np.arange(1, 7)

# a_n = (n+1)^2 - (2n+1)
termos_a_ex3 = (indices_ex3 + 1) ** 2 - (2 * indices_ex3 + 1)

# b_n = n^2
termos_b_ex3 = indices_ex3**2

# (n+1)^2 - (2n+1) = n^2 + 2n + 1 - 2n - 1 = n^2, para todo n -> são iguais
sequencias_iguais_ex3 = bool(np.array_equal(termos_a_ex3, termos_b_ex3))

print(termos_a_ex3, termos_b_ex3, sequencias_iguais_ex3)
```

**Exercício 5.** Implemente a sequência de Lucas (L₁=2, L₂=1, Lₙ=Lₙ₋₁+Lₙ₋₂) e calcule os 10 primeiros termos.

```python
def lucas(n: int) -> int:
    if n == 1:
        return 2
    if n == 2:
        return 1
    return lucas(n - 1) + lucas(n - 2)


termos_lucas_ex5 = [lucas(n) for n in range(1, 11)]

print(termos_lucas_ex5)
```

**Exercício 6.** Implemente uma versão iterativa de Fibonacci e compare com a versão recursiva para o 30º termo.

```python
def fibonacci_iterativo(n: int) -> int:
    anterior, atual = 1, 1
    for _ in range(n - 2):
        anterior, atual = atual, anterior + atual
    return atual if n >= 2 else anterior


# a versão recursiva de fibonacci(30) exigiria mais de 2 milhões de
# chamadas de função (sem memoização); a versão iterativa faz o mesmo
# cálculo em 28 passos de laço
fib30_iterativo_ex6 = fibonacci_iterativo(30)

print(fib30_iterativo_ex6)
print("confere com a versão recursiva para n=10:", fibonacci_iterativo(10) == fibonacci(10))
```
