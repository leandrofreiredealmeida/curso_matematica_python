```python
# Imports e funções auxiliares necessários para os exercícios abaixo,
# reconstruindo o que foi definido no notebook principal.
import math

import numpy as np


def gerar_pg(a1: float, q: float, n: int) -> np.ndarray:
    """Gera os n primeiros termos de uma PG de primeiro termo a1 e razao q."""
    indices = np.arange(1, n + 1)
    return a1 * q ** (indices - 1)
```

## Básico

**Exercício 1.** Dada a PG com a1 = 3 e q = 2, calcule os 6 primeiros termos e classifique-a quanto à razão.

```python
# a_n = a1 * q^(n-1), para n = 1,...,6
termos_ex1 = gerar_pg(a1=3, q=2, n=6)

# a1 = 3 > 0 e q = 2 > 1 -> PG crescente
classificacao_ex1 = "crescente"

print(termos_ex1, classificacao_ex1)
```

**Exercício 2.** Uma PG tem a1 = 5, q = 3 e 6 termos. Calcule o 6º termo e a soma dos 6 termos.

```python
# a_6 = a1 * q^(6-1)
a6_ex2 = 5 * 3**5

# S_6 = a1*(q^6 - 1)/(q - 1)
soma_ex2 = 5 * (3**6 - 1) / (3 - 1)

print(a6_ex2, soma_ex2)
```

## Intermediário

**Exercício 3.** Interpole 2 meios geométricos entre 5 e 135.

```python
# com 2 meios entre os extremos, a PG final tem n = 2 + 2 = 4 termos
n_ex3 = 4
q_ex3 = (135 / 5) ** (1 / (n_ex3 - 1))
pg_ex3 = gerar_pg(a1=5, q=q_ex3, n=n_ex3)

print(pg_ex3)
```

**Exercício 4.** Uma PG infinita tem a1 = 12 e q = 1/3. Calcule a soma de todos os infinitos termos.

```python
# S_infinito = a1 / (1 - q)
soma_infinita_ex4 = 12 / (1 - 1 / 3)

print(soma_infinita_ex4)
```

## Desafio

**Exercício 5.** Um carro de R$ 80000 se deprecia 20% ao ano. Calcule o valor depois de 6 anos e o primeiro ano em que o valor fica abaixo de R$ 20000.

```python
# a_6 = a1 * q^(6-1), com q = 0.8 (deprecia 20% ao ano)
valor_ano6_ex5 = 80000 * 0.8**5

# procurando o primeiro n em que a_n = 80000 * 0.8^(n-1) fica abaixo de 20000
n = 1
while 80000 * 0.8 ** (n - 1) >= 20000:
    n += 1
ano_abaixo_20000_ex5 = n

print(valor_ano6_ex5, ano_abaixo_20000_ex5)
```

**Exercício 6.** Converta a dízima periódica 0,777... em fração usando a soma de uma PG infinita.

```python
# 0,777... = 0,7 + 0,07 + 0,007 + ... , PG infinita com a1 = 0.7 e q = 0.1
fracao_ex6 = 0.7 / (1 - 0.1)

print(fracao_ex6)  # 7/9 = 0.7777...
```
