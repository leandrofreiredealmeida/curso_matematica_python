```python
# Imports e funções auxiliares necessários para os exercícios abaixo,
# reconstruindo o que foi definido no notebook principal.
import numpy as np


def gerar_pa(a1: float, r: float, n: int) -> np.ndarray:
    """Gera os n primeiros termos de uma PA de primeiro termo a1 e razão r."""
    indices = np.arange(1, n + 1)
    return a1 + (indices - 1) * r
```

## Básico

**Exercício 1.** Dada a PA com a1 = 4 e r = 5, calcule os 6 primeiros termos e classifique-a quanto à razão.

```python
# a_n = a1 + (n-1)*r, para n = 1,...,6
termos_ex1 = gerar_pa(a1=4, r=5, n=6)

# r = 5 > 0 -> PA crescente
classificacao_ex1 = "crescente"

print(termos_ex1, classificacao_ex1)
```

**Exercício 2.** Uma PA tem a1 = 3, r = 4 e 15 termos. Calcule o 15º termo e a soma dos 15 termos.

```python
# a_15 = a1 + (15-1)*r
a15_ex2 = 3 + 14 * 4

# S_15 = (a1 + a15)*15/2
soma_ex2 = (3 + a15_ex2) * 15 / 2

print(a15_ex2, soma_ex2)
```

## Intermediário

**Exercício 3.** Interpole 4 meios aritméticos entre 6 e 26.

```python
# com 4 meios entre os extremos, a PA final tem n = 4 + 2 = 6 termos
n_ex3 = 6
r_ex3 = (26 - 6) / (n_ex3 - 1)
pa_ex3 = gerar_pa(a1=6, r=r_ex3, n=n_ex3)

print(pa_ex3)
```

**Exercício 4.** Três números em PA têm soma 24 e produto 384. Encontre os três números.

```python
# Com a notação (x - r, x, x + r): soma = 3x
x_ex4 = 24 / 3

# produto = x * (x^2 - r^2) = 384  ->  r^2 = x^2 - produto/x
r_ex4 = (x_ex4**2 - 384 / x_ex4) ** 0.5

numeros_ex4 = sorted([x_ex4 - r_ex4, x_ex4, x_ex4 + r_ex4])

print(numeros_ex4)
```

## Desafio

**Exercício 5.** Um teatro tem 25 fileiras de assentos, começando com 20 assentos na primeira fileira e aumentando 4 assentos por fileira. Calcule os assentos da fileira 25 e o total do setor.

```python
# a_25 = a1 + (25-1)*r
assentos_fileira25_ex5 = 20 + 24 * 4

# S_25 = (a1 + a25)*25/2
total_assentos_ex5 = (20 + assentos_fileira25_ex5) * 25 / 2

print(assentos_fileira25_ex5, total_assentos_ex5)
```

**Exercício 6.** Financiamento de R$ 12000 em 10 parcelas pelo SAC simplificado (amortização constante de R$ 1200, juros de 2% sobre o saldo devedor antes de cada parcela). Calcule o total de juros pago e o valor da última parcela.

```python
# saldo antes da parcela n: 12000 - (n-1)*1200
# juros da parcela n: 2% desse saldo -> forma uma PA, pois o saldo
# decresce em progressão aritmética (razão -1200) e os juros são
# proporcionais a ele
juros_1_ex6 = 0.02 * 12000  # 240
razao_juros_ex6 = 0.02 * (-1200)  # -24 (variação do saldo entre parcelas)

juros_ex6 = gerar_pa(a1=juros_1_ex6, r=razao_juros_ex6, n=10)

# S_10 = (juros_1 + juros_10)*10/2
total_juros_ex6 = (juros_ex6[0] + juros_ex6[-1]) * 10 / 2

# última parcela = amortização constante + juros da parcela 10
ultima_parcela_ex6 = 1200 + juros_ex6[-1]

print(total_juros_ex6, ultima_parcela_ex6)
```
