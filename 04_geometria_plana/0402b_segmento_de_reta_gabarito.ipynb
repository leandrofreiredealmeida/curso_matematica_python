```python
# Imports necessários para os exercícios abaixo.
import math

import numpy as np

# Pares de segmentos da figura do Exercício Básico 2 (reconstruídos do notebook principal)
pares_ex2 = {
    "a": ({"A": (-4, -1), "B": (-0.5, 0.75), "C": (4, 3)}, [("A", "B"), ("B", "C")]),
    "b": ({"D": (-4, 2), "E": (0, -2), "F": (4, 2)}, [("D", "E"), ("E", "F")]),
    "c": ({"G": (-4, 0), "H": (-1, 0), "I": (1, 0), "J": (4, 0)}, [("G", "H"), ("I", "J")]),
    "d": ({"K": (-4, -2), "L": (1, 0.5), "M": (4, 2)}, [("K", "L"), ("K", "M")]),
    "e": ({"N": (-4, 2), "O": (-1, 2), "Q": (1, -2), "R": (4, 0)}, [("N", "O"), ("Q", "R")]),
}
```

## Básico

**Exercício 1.** Medidas de segmentos numa reta numerada com A em −3, B em 2, C em 5,5 e D em −7.

```python
posicao = {"A": -3, "B": 2, "C": 5.5, "D": -7}

# Medida = módulo da diferença das posições (a ordem não importa)
medida_ab = abs(posicao["B"] - posicao["A"])  # |2 - (-3)| = 5
medida_bc = abs(posicao["C"] - posicao["B"])  # |5.5 - 2| = 3.5
medida_ad = abs(posicao["D"] - posicao["A"])  # |-7 - (-3)| = |-4| = 4
medida_cd = abs(posicao["D"] - posicao["C"])  # |-7 - 5.5| = |-12.5| = 12.5

print(medida_ab, medida_bc, medida_ad, medida_cd)
```

**Exercício 2.** Classificar os pares (a) a (e) como (consecutivos, colineares, adjacentes).

```python
# (a) AB e BC: extremidade comum B; A, B, C estão na reta y = x/2 + 1 e B fica entre A e C.
# (b) DE e EF: extremidade comum E, mas formam uma quina ("V"): não são colineares.
# (c) GH e IJ: os dois no eixo horizontal (colineares), mas sem extremidade em comum.
# (d) KL e KM: extremidade comum K, mesma reta, mas KL está DENTRO de KM (sobrepostos).
# (e) NO e QR: nenhuma extremidade comum e retas diferentes.
classificacao_ex2 = {
    "a": (True, True, True),
    "b": (True, False, False),
    "c": (False, True, False),
    "d": (True, True, False),
    "e": (False, False, False),
}

print(classificacao_ex2)
```

```python
# Conferindo por código, com as mesmas ideias da seção 4 do notebook principal
def sao_colineares(*pontos: tuple[float, float]) -> bool:
    """Testa se todos os pontos estão numa mesma reta (os dois primeiros devem ser distintos)."""
    (xa, ya), (xb, yb) = pontos[0], pontos[1]
    return all(
        math.isclose((xb - xa) * (yc - ya) - (yb - ya) * (xc - xa), 0, abs_tol=1e-9)
        for xc, yc in pontos[2:]
    )


def classificar_par(seg1: tuple, seg2: tuple) -> tuple[bool, bool, bool]:
    """Devolve (consecutivos, colineares, adjacentes) para dois segmentos."""
    comuns = set(seg1) & set(seg2)
    consecutivos = len(comuns) == 1
    colineares = sao_colineares(*seg1, *seg2)
    adjacentes = False
    if consecutivos and colineares:
        comum = comuns.pop()
        outra_1 = next(p for p in seg1 if p != comum)
        outra_2 = next(p for p in seg2 if p != comum)
        # adjacentes: as outras extremidades ficam em lados opostos do ponto comum
        adjacentes = bool(np.dot(np.subtract(outra_1, comum), np.subtract(outra_2, comum)) < 0)
    return consecutivos, colineares, adjacentes


for letra, (pontos, ((p1, p2), (q1, q2))) in pares_ex2.items():
    calculado = classificar_par((pontos[p1], pontos[p2]), (pontos[q1], pontos[q2]))
    print(f"({letra}) {calculado}  confere: {calculado == classificacao_ex2[letra]}")
```

## Intermediário

**Exercício 3.** Função de ponto médio e ponto médio do segmento de extremidades A = (−4, 7) e B = (10, −3).

```python
def ponto_medio_ex3(p1: tuple[float, float], p2: tuple[float, float]) -> tuple[float, float]:
    """Devolve o ponto médio do segmento de extremidades p1 e p2."""
    # média das abscissas e média das ordenadas, separadamente
    return ((p1[0] + p2[0]) / 2, (p1[1] + p2[1]) / 2)


a_ex3, b_ex3 = (-4, 7), (10, -3)
m_ex3 = ponto_medio_ex3(a_ex3, b_ex3)  # ((-4 + 10)/2, (7 - 3)/2) = (3, 2)

print(m_ex3)
# As duas metades são congruentes: AM = MB
print(math.dist(a_ex3, m_ex3), math.dist(m_ex3, b_ex3))
```

**Exercício 4.** (a) Distância BC na estrada; (b) função que testa se B está entre A e C.

```python
# (a) A, B, C nessa ordem -> AB e BC são adjacentes -> AC = AB + BC
#     Logo BC = AC - AB = 350 - 128 = 222 km
medida_bc_ex4 = 350 - 128


# (b) B está no segmento AC exatamente quando AB + BC = AC.
#     Se B estiver fora do segmento (na reta ou fora dela), a soma passa de AC.
def b_entre_a_e_c(a: tuple[float, float], b: tuple[float, float], c: tuple[float, float]) -> bool:
    """Devolve True se B pertence ao segmento AC."""
    return math.isclose(math.dist(a, b) + math.dist(b, c), math.dist(a, c))


print(medida_bc_ex4)
print(b_entre_a_e_c((0, 0), (2, 1), (8, 4)))   # entre A e C: True
print(b_entre_a_e_c((0, 0), (10, 5), (8, 4)))  # na reta, depois de C: False
print(b_entre_a_e_c((0, 0), (2, 4), (8, 4)))   # fora da reta: False
```

## Desafio

**Exercício 5.** Razão AP/PB para P no segmento AB, e o valor dessa razão no ponto médio.

```python
def razao_ap_pb(a: tuple[float, float], b: tuple[float, float], p: tuple[float, float]) -> float:
    """Devolve a razão AP / PB (P no segmento AB, diferente de B)."""
    return math.dist(a, p) / math.dist(p, b)


# No ponto médio, AM = MB, então AM / MB = 1
razao_no_ponto_medio = 1

print(razao_ap_pb((0, 0), (10, 0), (4, 0)))  # AP = 4, PB = 6 -> 4/6 ≈ 0.667
print(razao_ap_pb((0, 0), (10, 0), (5, 0)))  # ponto médio -> 1.0

# Observação: se P anda a fração t do caminho de A até B, então AP = t·AB e
# PB = (1 - t)·AB, e a razão fica t / (1 - t), sem depender do tamanho de AB.
for t in [0.25, 0.5, 0.75]:
    p = (10 * t, 0)
    print(f"t = {t}: AP/PB = {razao_ap_pb((0, 0), (10, 0), p):.4f}  |  t/(1 - t) = {t / (1 - t):.4f}")
```

**Exercício 6.** Ponto P do segmento AB tal que AP/PB = k.

```python
# Se AP/PB = k, então AP tem k partes e PB tem 1 parte: o segmento todo tem k + 1 partes.
# Logo P anda a fração t = k / (k + 1) do caminho de A até B, em cada coordenada:
#     P = A + t·(B - A)
# Com k = 1, t = 1/2 e P é o ponto médio.
def ponto_que_divide(a: tuple[float, float], b: tuple[float, float], k: float) -> tuple[float, float]:
    """Devolve o ponto P de AB tal que AP / PB = k (k > 0)."""
    t = k / (k + 1)
    return (a[0] + t * (b[0] - a[0]), a[1] + t * (b[1] - a[1]))


p = ponto_que_divide((0, 0), (9, 0), 2)
print(p, math.dist((0, 0), p) / math.dist(p, (9, 0)))  # (6, 0): AP = 6, PB = 3, razão 2
print(ponto_que_divide((-3, -1), (5, 3), 1))           # k = 1: ponto médio (1, 1)
```
