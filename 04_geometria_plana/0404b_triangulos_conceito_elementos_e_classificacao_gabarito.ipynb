```python
# Imports necessários para os exercícios abaixo.
import math

import numpy as np


# Ferramentas das seções 1, 2 e 4 do notebook principal (reconstruídas aqui).
def distancia(P, Q) -> float:
    """Distância entre os pontos P e Q do plano."""
    return math.dist(P, Q)


def medir_angulo(vertice, P, Q) -> float:
    """Transferidor digital: medida, em graus, do ângulo com vértice em `vertice` entre P e Q."""
    u = np.subtract(P, vertice)
    v = np.subtract(Q, vertice)
    cosseno = np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))
    return math.degrees(math.acos(np.clip(cosseno, -1, 1)))


def forma_triangulo(a: float, b: float, c: float) -> bool:
    """True se as medidas a, b e c formam um triângulo (desigualdade triangular)."""
    if min(a, b, c) <= 0:
        return False
    return a < b + c and b < a + c and c < a + b
```

**Exercício 1.** Classificar triângulos quanto aos lados (dados os lados) e quanto aos ângulos (dados os ângulos).

```python
# (a) Quanto aos lados: contar quantos lados são congruentes.
# (5, 5, 5)       três iguais             -> equilátero
# (3, 4, 5)       todos diferentes        -> escaleno
# (7, 7, 10)      dois iguais (7 e 7)     -> isósceles (base 10)
# (6, 8, 6)       dois iguais (6 e 6)     -> isósceles (a ordem não importa!)
# (2.5, 2.5, 2.5) três iguais             -> equilátero
# (4, 6, 9)       todos diferentes        -> escaleno
classificacoes_ex1a = ["equilátero", "escaleno", "isósceles", "isósceles", "equilátero", "escaleno"]

# (b) Quanto aos ângulos: basta olhar o MAIOR ângulo e comparar com 90°.
# (60, 60, 60)  maior 60°  -> acutângulo
# (90, 45, 45)  maior 90°  -> retângulo
# (100, 40, 40) maior 100° -> obtusângulo
# (30, 60, 90)  maior 90°  -> retângulo
# (80, 70, 30)  maior 80°  -> acutângulo
# (20, 35, 125) maior 125° -> obtusângulo
classificacoes_ex1b = ["acutângulo", "retângulo", "obtusângulo", "retângulo", "acutângulo", "obtusângulo"]

print(classificacoes_ex1a)
print(classificacoes_ex1b)
```

**Exercício 2.** Terceiro ângulo interno, ângulo externo e a função `terceiro_angulo_ex2`.

```python
# (a) Soma dos internos = 180°: 180 − 48 − 67 = 65
terceiro_ex2a = 180 - 48 - 67

# (b) Duas formas de chegar lá:
#     suplemento do interno adjacente: 180 − 65 = 115
#     teorema do ângulo externo: soma dos dois internos não adjacentes: 48 + 67 = 115
externo_ex2b = 180 - terceiro_ex2a
print(terceiro_ex2a, externo_ex2b, 48 + 67)


# (c) A mesma conta do item (a), agora para quaisquer dois ângulos.
def terceiro_angulo_ex2(alfa: float, beta: float) -> float:
    """Terceiro ângulo interno de um triângulo, dados os outros dois."""
    return 180 - alfa - beta


print(terceiro_angulo_ex2(90, 35))  # 55: num retângulo, os agudos são complementares
```

**Exercício 3.** Decidir se trios de medidas formam triângulo, à mão e com uma função.

```python
# (a) Teste rápido: soma dos dois menores > maior?
# (3, 4, 5)      3 + 4 = 7 > 5        -> True
# (1, 2, 3)      1 + 2 = 3 = 3        -> False (degenerado: os lados ficam deitados)
# (6, 6, 11)     6 + 6 = 12 > 11      -> True
# (2, 9, 5)      2 + 5 = 7 < 9        -> False
# (10, 10, 10)   20 > 10              -> True
# (4, 4, 8.5)    4 + 4 = 8 < 8,5      -> False
# (0.5, 0.7, 1)  0,5 + 0,7 = 1,2 > 1  -> True
respostas_ex3a = [True, False, True, False, True, False, True]


# (b) Duas versões equivalentes: as três desigualdades, ou o maior contra os outros dois.
def forma_triangulo_ex3(a: float, b: float, c: float) -> bool:
    """True se a, b e c (positivos) formam um triângulo."""
    if min(a, b, c) <= 0:  # lado de medida zero ou negativa não existe
        return False
    maior = max(a, b, c)
    return maior < (a + b + c) - maior  # soma dos outros dois = total − maior


trios_ex3 = [(3, 4, 5), (1, 2, 3), (6, 6, 11), (2, 9, 5), (10, 10, 10), (4, 4, 8.5), (0.5, 0.7, 1)]
for trio, esperado in zip(trios_ex3, respostas_ex3a):
    print(trio, forma_triangulo_ex3(*trio), "confere:", forma_triangulo_ex3(*trio) == esperado)
print((10, 2, 1), forma_triangulo_ex3(10, 2, 1))  # a pegadinha: 10 < 2 + 1 é falso
```

## Intermediário

**Exercício 4.** Classificar triângulos dados por coordenadas, com `distancia` e `medir_angulo`.

```python
def classificar_ex4(A, B, C) -> tuple[str, str]:
    """(classificação quanto aos lados, quanto aos ângulos) do triângulo de vértices A, B, C."""
    # Lados: cada um com a letra do vértice oposto
    a, b, c = distancia(B, C), distancia(C, A), distancia(A, B)
    iguais_ab, iguais_bc, iguais_ac = math.isclose(a, b), math.isclose(b, c), math.isclose(a, c)
    if iguais_ab and iguais_bc:
        lados = "equilátero"
    elif iguais_ab or iguais_bc or iguais_ac:
        lados = "isósceles"
    else:
        lados = "escaleno"

    # Ângulos: só o maior importa (no máximo um é reto ou obtuso)
    maior = max(medir_angulo(A, B, C), medir_angulo(B, C, A), medir_angulo(C, A, B))
    if math.isclose(maior, 90):
        angulos = "retângulo"
    elif maior > 90:
        angulos = "obtusângulo"
    else:
        angulos = "acutângulo"
    return lados, angulos


casos_ex4 = [
    ((0, 0), (2, 0), (1, math.sqrt(3))),  # altura sqrt(3): os três lados medem 2
    ((0, 0), (4, 0), (0, 3)),              # o clássico 3, 4, 5, reto em A
    ((0, 0), (4, 0), (2, 0.9)),            # C no meio e bem baixo: isósceles "achatado"
    ((1, 1), (5, 1), (3, 3)),              # C no meio, altura = metade da base: reto em C
    ((0, 0), (5, 0), (2, 3)),
    ((0, 0), (6, 0), (-2, 1)),             # C "para trás" de A: ângulo obtuso em A
    ((-1, 0), (1, 0), (0, 4)),
]
for vertices in casos_ex4:
    print(vertices, "->", classificar_ex4(*vertices))
```

**Exercício 5.** Ângulos de triângulos isósceles.

```python
# (a) Os dois ângulos da base são congruentes (x cada): 40 + x + x = 180 -> 2x = 140 -> x = 70
angulo_base_ex5a = (180 - 40) / 2

# (b) Dois ângulos da base de 35°: vértice = 180 − 35 − 35 = 110
angulo_vertice_ex5b = 180 - 2 * 35

# (c) O maior ângulo é 110° > 90°
classe_ex5c = "obtusângulo"


# (d) Tira o ângulo do vértice de 180° e divide o resto igualmente entre os dois da base
def angulo_base_ex5(vertice: float) -> float:
    """Medida de cada ângulo da base de um isósceles, dado o ângulo do vértice."""
    return (180 - vertice) / 2


print(angulo_base_ex5a, angulo_vertice_ex5b, classe_ex5c)
print(angulo_base_ex5(100), angulo_base_ex5(60))  # com vértice 60°, a base também tem 60°: equilátero!
```

**Exercício 6.** Ângulos com a ajuda do teorema do ângulo externo.

```python
# (a) O externo AĈD = 125° é a soma dos internos não adjacentes a ele (Â e B̂):
#     125 = 70 + B̂  ->  B̂ = 55
angulo_B_ex6a = 125 - 70
#     O interno Ĉ é suplementar do externo: 180 − 125 = 55
angulo_C_ex6a = 180 - 125
#     Conferindo: 70 + 55 + 55 = 180 ✓
#     B̂ ≡ Ĉ, então os lados opostos a eles (b = CA e c = AB) são congruentes -> isósceles
lados_ex6a = "isósceles"
print(angulo_B_ex6a, angulo_C_ex6a, 70 + angulo_B_ex6a + angulo_C_ex6a, lados_ex6a)

# (b) O externo em A é a soma dos internos B̂ e Ĉ:
#     3x = (x + 20) + 50  ->  2x = 70  ->  x = 35
x_ex6b = 70 / 2
#     Externo em A = 3 · 35 = 105; o interno Â é o suplemento: 180 − 105 = 75
angulo_A_ex6b = 180 - 3 * x_ex6b
#     Conferindo: Â + B̂ + Ĉ = 75 + 55 + 50 = 180 ✓
print(x_ex6b, angulo_A_ex6b, angulo_A_ex6b + (x_ex6b + 20) + 50)
```

**Exercício 7.** Valores inteiros possíveis para o terceiro lado de um triângulo com lados 5 e 9.

```python
# (a) Força bruta: testa cada inteiro candidato com a desigualdade triangular
valores_ex7 = [x for x in range(1, 31) if forma_triangulo(5, 9, x)]

# (b) Quantos são
quantidade_ex7 = len(valores_ex7)

# (c) Pela desigualdade triangular: |9 − 5| < x < 9 + 5, ou seja, 4 < x < 14.
#     As pontas 4 e 14 não servem (dariam triângulos degenerados).
intervalo_ex7 = (abs(9 - 5), 9 + 5)

print(valores_ex7, quantidade_ex7, intervalo_ex7)
print("4 forma?", forma_triangulo(5, 9, 4), "| 14 forma?", forma_triangulo(5, 9, 14))
```

**Exercício 8.** Ângulos dados por expressões com uma incógnita, e a classificação do triângulo.

```python
# (a) x + 2x + 3x = 180  ->  6x = 180  ->  x = 30
x_ex8a = 180 / 6
angulos_ex8a = (x_ex8a, 2 * x_ex8a, 3 * x_ex8a)  # (30, 60, 90)
#     Tem um ângulo de 90° -> retângulo.
#     Os três ângulos são diferentes, então os lados opostos também são (ao maior ângulo
#     opõe-se o maior lado, ao menor o menor) -> escaleno.
classificacao_ex8a = ("escaleno", "retângulo")

# (b) x + x + 4x = 180  ->  6x = 180  ->  x = 30
x_ex8b = 180 / 6
angulos_ex8b = (x_ex8b, x_ex8b, 4 * x_ex8b)  # (30, 30, 120)
#     120° > 90° -> obtusângulo; dois ângulos congruentes -> dois lados congruentes -> isósceles
classificacao_ex8b = ("isósceles", "obtusângulo")

print(x_ex8a, angulos_ex8a, sum(angulos_ex8a), classificacao_ex8a)
print(x_ex8b, angulos_ex8b, sum(angulos_ex8b), classificacao_ex8b)
```
