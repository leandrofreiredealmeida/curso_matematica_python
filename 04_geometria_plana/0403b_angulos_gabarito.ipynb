## Básico

**Exercício 1.** Classificar os ângulos de 72°, 90°, 135,5°, 180°, 360°, 89,9°, 90,25° e 5°.

```python
medidas_ex1 = [72, 90, 135.5, 180, 360, 89.9, 90.25, 5]

# 72° < 90°             -> agudo
# 90° exatamente        -> reto
# 90° < 135,5° < 180°   -> obtuso
# 180° exatamente       -> raso
# 360° exatamente       -> giro
# 89,9° < 90° (por pouco!)  -> agudo
# 90,25° > 90° (por pouco!) -> obtuso
# 5° < 90°              -> agudo
classificacoes_ex1 = ["agudo", "reto", "obtuso", "raso", "giro", "agudo", "obtuso", "agudo"]

for medida, classe in zip(medidas_ex1, classificacoes_ex1):
    print(f"{medida:>6g}°  ->  {classe}")
```

**Exercício 2.** Classificar os pares (a) a (f) como (consecutivos, adjacentes, complementares, suplementares).

```python
# (a) 35° e 55°: mesmo vértice, lado comum em 35°, lado a lado; 35 + 55 = 90.
# (b) 110° e 70°: mesmo vértice e lado comum em 0°, mas o de 70° está DENTRO do de 110°
#     (não adjacentes); 110 + 70 = 180 -> suplementares.
# (c) 25° e 65°: vértices diferentes (não consecutivos nem adjacentes); 25 + 65 = 90.
# (d) 120° e 60°: lado a lado, formando um ângulo raso; 120 + 60 = 180.
# (e) 40° e 60°: lado a lado (lado comum em 50°); 40 + 60 = 100, nem 90 nem 180.
# (f) 40° e 60°: mesmo vértice, mas nenhum lado em comum; soma 100.
classificacao_ex2 = {
    "a": (True, True, True, False),
    "b": (True, False, False, True),
    "c": (False, False, True, False),
    "d": (True, True, False, True),
    "e": (True, True, False, False),
    "f": (False, False, False, False),
}

print(classificacao_ex2)
```

```python
# Conferindo por código, com a mesma ideia da seção 4 do notebook principal
def classificar_par_de_angulos(angulo_1: tuple, angulo_2: tuple) -> tuple[bool, bool, bool, bool]:
    """Devolve (consecutivos, adjacentes, complementares, suplementares) para dois ângulos."""
    (v1, ini1, fim1), (v2, ini2, fim2) = angulo_1, angulo_2
    consecutivos = v1 == v2 and len({ini1, fim1} & {ini2, fim2}) == 1
    # adjacentes: os intervalos de direções só se tocam no lado comum (sem sobreposição)
    adjacentes = consecutivos and min(fim1, fim2) - max(ini1, ini2) <= 0
    soma = (fim1 - ini1) + (fim2 - ini2)
    return consecutivos, adjacentes, math.isclose(soma, 90), math.isclose(soma, 180)


for letra, (ang_1, ang_2) in pares_ex2.items():
    calculado = classificar_par_de_angulos(ang_1, ang_2)
    print(f"({letra}) {calculado}  confere: {calculado == classificacao_ex2[letra]}")
```

```python
def complemento_ex3(medida: float) -> float | str:
    """Devolve o complemento (90° - medida) ou uma mensagem se ele não existir."""
    # Só ângulos agudos (0° < medida < 90°) têm complemento
    if 0 < medida < 90:
        return 90 - medida
    return "não existe complemento"


def suplemento_ex3(medida: float) -> float | str:
    """Devolve o suplemento (180° - medida) ou uma mensagem se ele não existir."""
    # Só ângulos entre 0° e 180° (sem incluir os extremos) têm suplemento
    if 0 < medida < 180:
        return 180 - medida
    return "não existe suplemento"


for medida in (30, 89.5, 90, 120, 179, 180, 200):
    print(f"{medida:>5g}°  complemento: {complemento_ex3(medida)!s:<24} suplemento: {suplemento_ex3(medida)}")
```

**Exercício 4.** Os quatro ângulos formados por duas retas concorrentes, a partir de um deles.

```python
# (a) Com um ângulo de 63°:
#     - o vizinho está sobre a mesma reta que ele -> suplementar: 180 - 63 = 117
#     - o oposto pelo vértice é congruente: 63
#     - o outro vizinho também é suplementar: 117
angulos_ex4a = (117, 63, 117)


# (b) Em volta do vértice: alfa, seu suplemento, alfa (o.p.v.), o suplemento (o.p.v.)
def quatro_angulos_ex4(alfa: float) -> tuple[float, float, float, float]:
    """Devolve os quatro ângulos formados por duas retas concorrentes, começando por alfa."""
    beta = 180 - alfa
    return alfa, beta, alfa, beta


print(angulos_ex4a)
for alfa in (63, 40, 90, 150):
    angulos = quatro_angulos_ex4(alfa)
    print(f"alfa = {alfa:>3}°: {angulos}  soma = {sum(angulos)}°")
```

**Exercício 5.** Direção da bissetriz, os dois ângulos que ela forma e a metade de 47° 30′.

```python
# (a) A bissetriz fica "no meio" dos dois lados: a direção é a média das direções.
def direcao_bissetriz_ex5(dir_a: float, dir_b: float) -> float:
    """Direção da bissetriz do ângulo com lados nas direções dir_a < dir_b."""
    return (dir_a + dir_b) / 2


# (b) Cada parte vai de um lado até a bissetriz.
def angulos_formados_ex5(dir_a: float, dir_b: float) -> tuple[float, float]:
    """Medidas dos dois ângulos formados pela bissetriz: (dir_a até a bissetriz, bissetriz até dir_b)."""
    bissetriz = direcao_bissetriz_ex5(dir_a, dir_b)
    return bissetriz - dir_a, dir_b - bissetriz


# (c) 47° 30′ / 2: metade de 47° é 23° com 1° de resto; 1° = 60′, então
#     sobram 60′ + 30′ = 90′ para dividir por 2 -> 45′. Resultado: 23° 45′ 0″.
#     Conferindo em graus decimais: 47,5° / 2 = 23,75° = 23° + 0,75 · 60′ = 23° 45′.
metade_ex5_gms = (23, 45, 0)

print(direcao_bissetriz_ex5(20, 110), angulos_formados_ex5(20, 110))
for dir_a, dir_b in [(0, 180), (45, 60), (100, 250)]:
    primeiro, segundo = angulos_formados_ex5(dir_a, dir_b)
    print(f"lados em {dir_a}° e {dir_b}°: partes de {primeiro:g}° e {segundo:g}°  congruentes: {math.isclose(primeiro, segundo)}")
print(metade_ex5_gms)
```

**Exercício 6.** Ângulo que falta para completar uma volta, e o problema com x, 2x, 3x e 4x.

```python
# (a) Todos os ângulos em volta do ponto somam 360°, então o que falta é 360 menos a soma dos conhecidos.
def angulo_faltante_ex6(conhecidos: list[float]) -> float:
    """Medida do ângulo que falta para os ângulos em volta de um ponto somarem 360°."""
    return 360 - sum(conhecidos)


# (b) x + 2x + 3x + 4x = 360  ->  10x = 360  ->  x = 36
x_ex6 = 360 / 10
maior_angulo_ex6 = 4 * x_ex6  # 4 · 36 = 144

print(angulo_faltante_ex6([100, 80, 95]))  # 360 - 275 = 85
print(x_ex6, [k * x_ex6 for k in (1, 2, 3, 4)], maior_angulo_ex6)
```
