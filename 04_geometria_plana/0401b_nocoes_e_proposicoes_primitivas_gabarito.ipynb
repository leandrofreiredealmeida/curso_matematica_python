## Intermediário

**Exercício 3.** Função que verifica se uma lista de pontos é colinear.

```python
def sao_colineares(pontos: list[tuple[float, float]]) -> bool:
    """Devolve True se todos os pontos estão numa mesma reta."""
    # os dois primeiros pontos (distintos) determinam uma única reta
    (xa, ya), (xb, yb) = pontos[0], pontos[1]
    for xc, yc in pontos[2:]:
        # multiplicação cruzada das inclinações de A->B e A->C (sem dividir por zero)
        diferenca = (xb - xa) * (yc - ya) - (yb - ya) * (xc - xa)
        if not math.isclose(diferenca, 0, abs_tol=1e-9):
            return False  # basta um ponto fora da reta AB
    return True  # com só 2 pontos o laço nem roda: dois pontos são sempre colineares


print(sao_colineares([(0, 0), (1, 1), (2, 2)]))
print(sao_colineares([(1, 0), (1, 5), (1, -3)]))
print(sao_colineares([(0, 0), (2, 1), (4, 2), (6, 3), (8, 5)]))
```

**Exercício 4.** Contar as retas distintas determinadas pelos vértices de um quadrado e por três pontos alinhados mais um ponto fora.

```python
# Conjunto 1 (quadrado): nenhum trio é colinear, então cada par dá uma reta nova.
# São 4 pontos -> 4·3/2 = 6 pares -> 6 retas (4 lados + 2 diagonais).
retas_conjunto_1 = 6

# Conjunto 2: (0,0), (1,1) e (2,2) estão em y = x.
# Os 3 pares formados por eles determinam a MESMA reta -> conta 1.
# (0,2) liga-se a cada um dos outros 3 por retas diferentes -> mais 3.
retas_conjunto_2 = 1 + 3

print(retas_conjunto_1, retas_conjunto_2)
```

```python
# Conferindo por código: agrupa os pares pela reta que determinam
def contar_retas(pontos: list[tuple[int, int]]) -> int:
    """Conta as retas distintas determinadas por pontos de coordenadas inteiras."""
    retas = set()
    for (x1, y1), (x2, y2) in combinations(pontos, 2):
        a, b = y2 - y1, x1 - x2  # reta a·x + b·y = c
        c = a * x1 + b * y1
        divisor = math.gcd(math.gcd(a, b), c)
        a, b, c = a // divisor, b // divisor, c // divisor
        if a < 0 or (a == 0 and b < 0):  # escolhe um sinal padrão
            a, b, c = -a, -b, -c
        retas.add((a, b, c))
    return len(retas)


print(contar_retas([(0, 0), (2, 0), (2, 2), (0, 2)]))
print(contar_retas([(0, 0), (1, 1), (2, 2), (0, 2)]))
```

## Desafio

**Exercício 5.** Número de retas determinadas por n pontos, sem três colineares.

```python
# Contando à mão: n=2 -> 1 reta, n=3 -> 3 retas (triângulo), n=4 -> 6 retas.
# Raciocínio geral: cada ponto se liga aos outros n - 1, o que dá n·(n - 1)
# ligações. Mas a reta AB foi contada duas vezes (saindo de A e saindo de B),
# então dividimos por 2.
def retas_por_n_pontos(n: int) -> int:
    return n * (n - 1) // 2


print([retas_por_n_pontos(n) for n in range(2, 7)])
# Essa contagem de pares vai reaparecer em Análise Combinatória como C(n, 2)
print(retas_por_n_pontos(10), math.comb(10, 2))
```

**Exercício 6.** Número de planos determinados por n pontos no espaço, sem três colineares e sem quatro coplanares.

```python
# Escolhendo os pontos em ordem: n opções para o 1º, n - 1 para o 2º, n - 2 para o 3º.
# Mas o trio {A, B, C} aparece em 3·2·1 = 6 ordens diferentes (ABC, ACB, BAC, ...),
# e todas dão o mesmo plano. Por isso dividimos por 6.
def planos_por_n_pontos(n: int) -> int:
    return n * (n - 1) * (n - 2) // 6


print([planos_por_n_pontos(n) for n in range(3, 8)])
# Conferindo contando os trios diretamente
print(planos_por_n_pontos(10), len(list(combinations(range(10), 3))))
```
