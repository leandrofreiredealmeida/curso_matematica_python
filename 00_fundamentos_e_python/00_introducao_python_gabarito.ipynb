**Exercício 1.** Crie variáveis com nome, idade e cidade e imprima uma frase usando as três.

## Intermediário

**Exercício 1.** Dado um valor em minutos, calcule quantas horas e quantos minutos inteiros ele representa.

```python
minutos_totais = 135

# // (divisão inteira) dá quantas horas completas cabem nos minutos totais
horas = minutos_totais // 60
# % (resto da divisão) dá o que sobra depois de tirar essas horas completas
minutos = minutos_totais % 60

print(f"{minutos_totais} minutos = {horas}h{minutos}min")
```

**Exercício 2.** Troque os valores de duas variáveis sem usar uma terceira variável auxiliar.

```python
a = 10
b = 25

# Python permite atribuição múltipla: o lado direito é calculado por inteiro
# antes de qualquer atribuição acontecer, então a troca funciona sem variável auxiliar
a, b = b, a

print(f"a = {a}, b = {b}")
```

## Desafio

**Exercício.** Dado um número de segundos, calcule horas, minutos e segundos restantes, formatando a saída como `HH:MM:SS`.

```python
segundos_totais = 3725

# Quantas horas inteiras cabem no total de segundos
horas = segundos_totais // 3600
# Dos segundos que sobram depois das horas, quantos minutos inteiros cabem
minutos = (segundos_totais % 3600) // 60
# O que sobra depois de tirar horas e minutos e o resto em segundos
segundos = segundos_totais % 60

# :02d preenche com zero a esquerda quando o numero tem so um dígito
tempo_formatado = f"{horas:02d}:{minutos:02d}:{segundos:02d}"
print(tempo_formatado)
```
