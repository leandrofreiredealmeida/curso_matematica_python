## Variáveis

- Uma **variável** é como uma "caixinha" com nome que guarda um valor para você usar depois.
- Para colocar um valor dentro da caixinha, usamos o sinal de igual `=`.
- Regras para nomear variáveis em Python:
    - não pode começar com número (`2x` é inválido, `x2` é válido);
    - não pode ter espaços (use `_` no lugar, como em `nome_completo`);
    - maiúsculas e minúsculas são diferentes (`idade` e `Idade` são duas variáveis distintas — isso se chama *case-sensitive*).

📌 **Definição formal:**
- Wikipédia (PT): ["Variável (programação)"](https://pt.wikipedia.org/wiki/Vari%C3%A1vel_(programa%C3%A7%C3%A3o))

⚠️ **Erro comum:** confundir o `=` da programação com o `=` da matemática. Em matemática, `x = x + 1` não faz sentido (nenhum número é igual a ele mesmo mais 1). Em Python, essa linha significa "pegue o valor que está em `x`, some 1, e guarde o resultado de volta em `x`" — o sinal de igual é uma atribuição, não uma igualdade.

**Pergunta de checagem:**
1. O que acontece se você usar o mesmo nome de variável duas vezes seguidas, com valores diferentes?

```python
# Criando variaveis numéricas
idade = 16
preco_lanche = 12.5
desconto = 2.0

# Mostrando os valores guardados em cada variável
print(idade)
print(preco_lanche)
print(desconto)

# Reatribuindo uma variavel: o valor antigo e substituido pelo novo
preco_lanche = preco_lanche - desconto
print("Preco com desconto:", preco_lanche)
```

## Operadores aritméticos

Os operadores abaixo funcionam exatamente como na matemática que você já conhece:

| Operador | Significado | Exemplo |
|---|---|---|
| `+` | soma | `3 + 2` → `5` |
| `-` | subtração | `3 - 2` → `1` |
| `*` | multiplicação | `3 * 2` → `6` |
| `/` | divisão | `3 / 2` → `1.5` |
| `**` | potenciação | `3 ** 2` → `9` |

E dois operadores que podem ser novidades para você:

| Operador | Significado | Exemplo |
|---|---|---|
| `//` | divisão inteira (quantas vezes cabe) | `7 // 2` → `3` |
| `%` | resto da divisão (o que sobra) | `7 // 2 = 3`, sobrando `1`, então `7 % 2` → `1` |

A **ordem de precedência** é a mesma que você já usa na matemática: parênteses primeiro, depois potência, depois multiplicação/divisão, depois soma/subtração.

📌 **Definição formal:**
- Wikipédia (PT): ["Ordem das operações"](https://pt.wikipedia.org/wiki/Ordem_das_opera%C3%A7%C3%B5es)

**Isso será usado em:** praticamente todos os notebooks seguintes, especialmente em `01_conjuntos_e_funcoes/`.

**Perguntas de checagem:**
1. Quanto é `7 // 2` e quanto é `7 % 2`? Tente calcular antes de rodar o código.
2. Por que `2 + 3 * 4` não dá `20`?

```python
import ipywidgets as widgets

# Este é o seu primeiro contato com interatividade no curso:
# mexendo nos sliders, o resultado das contas muda na hora.

slider_a = widgets.IntSlider(value=8, min=-20, max=20, description="a:")
slider_b = widgets.IntSlider(value=3, min=-20, max=20, description="b:")


def mostrar_operacoes(a: int, b: int) -> None:
    '''Imprime o resultado das operacoes aritmeticas basicas entre a e b.'''
    print(f"a + b  = {a + b}")
    print(f"a - b  = {a - b}")
    print(f"a * b  = {a * b}")
    if b != 0:
        print(f"a / b  = {a / b}")
        print(f"a // b = {a // b}")
        print(f"a % b  = {a % b}")
    else:
        print("a / b, a // b e a % b nao existem quando b = 0")
    print(f"a ** b = {a ** b}")


widgets.interact(mostrar_operacoes, a=slider_a, b=slider_b)
```

## `print()` e comentários

- A função `print()` não serve só para mostrar números: ela também mostra texto e pode misturar texto com variáveis usando **f-strings** (`f"..."`).
- **Comentários** começam com `#` são ignorados pelo Python na hora de executar o código. Servem para explicar o código para você mesmo no futuro (e para outras pessoas que forem ler).

💡 **Dica:** comente o **porquê** do código, não o **o quê** — o código já mostra o que está sendo feito; o comentário deve explicar a razão por trás da escolha.

```python
nome = "Ana"
idade = 16

# Usamos f-string para misturar texto com variáveis dentro do print
print(f"Olá, {nome}! Voce tem {idade} anos.")

# Multiplicamos por 12 aqui porque queremos converter anos em meses
idade_em_meses = idade * 12
print(f"Isso equivale a {idade_em_meses} meses de vida.")
```
