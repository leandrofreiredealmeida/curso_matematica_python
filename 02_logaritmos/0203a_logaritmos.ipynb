## 1. Conceito de Logaritmo

📌 **Definição formal:**

$$\log_a(b) = x \iff a^x = b, \quad \text{com } a > 0,\ a \neq 1,\ b > 0$$

Essa expressão tem três elementos com nomes próprios:
- **base** (a): o número que é elevado ao expoente.
- **logaritmando** (b): o resultado que queremos atingir.
- **logaritmo** (x): o expoente procurado — é o que o logaritmo calcula.

Voltando ao gancho: a equação 2ˣ = 10 agora tem nome e solução formal — **x = log₂(10)**.


```python
# Logaritmos exatos "de cabeça", convertendo para uma pergunta de potência
print("log2(8)   -> 2 elevado a que dá 8?   -> 2^3 = 8   -> log2(8) = 3")
print("log10(100) -> 10 elevado a que dá 100? -> 10^2 = 100 -> log10(100) = 2")

# Conferindo com math.log(logaritmando, base)
print("\nconferindo com Python:")
print("math.log(8, 2)   =", math.log(8, 2))
print("math.log(100, 10) =", math.log(100, 10))

# Agora o logaritmo do gancho, que não tem resultado "redondo"
x = math.log(10, 2)
print(f"\nlog2(10) = {x:.4f}")
print(f"conferindo: 2^{x:.4f} = {2**x:.4f}  (bate com 10, e o valor está entre 3 e 4, como estimamos)")
```

🕹️ **Widget interativo:** escolha a base a e o logaritmando b. O widget mostra a "pergunta" (aˣ = b) e revela o valor de x.


```python
def explorar_logaritmo(a: float, b: float) -> None:
    '''Mostra a pergunta por trás de log_a(b) e revela o valor de x.'''
    print(f"pergunta: {a:g} elevado a que expoente dá {b:g}?")
    if abs(a - 1) < 1e-9:
        print("base a = 1 não é permitida -- 1 elevado a qualquer expoente é sempre 1.")
        return
    x = math.log(b, a)
    print(f"{a:g}^x = {b:g}  ->  x = log_{a:g}({b:g}) = {x:.4f}")
    print(f"conferindo: {a:g}^{x:.4f} = {a**x:.4f}")


widgets.interact(
    explorar_logaritmo,
    a=widgets.FloatSlider(value=2.0, min=0.1, max=10.0, step=0.1, description="a:"),
    b=widgets.FloatSlider(value=10.0, min=0.1, max=100.0, step=0.1, description="b:"),
)
```

**Pergunta de checagem:** o logaritmo é, na essência, a resposta a que pergunta?

🎯 **Aplicação prática:** o logaritmo aparece sempre que queremos descobrir "quantas vezes" ou "quanto tempo" algo precisa crescer/multiplicar para atingir um valor alvo.

## 2. Condições de Existência

⚠️ **Erro comum:** tentar calcular log_a(b) sem verificar se a e b respeitam as condições abaixo — e se surpreender com o erro da calculadora.

📌 **Definição formal:** log_a(b) só existe quando:
- **a > 0 e a ≠ 1** (a base) — pelas mesmas razões que restringiam a base da potência de expoente real (notebook `0201a_potencias_e_raizes`).
- **b > 0** (o logaritmando) — porque a imagem da função exponencial de mesma base (notebook `0202a_funcao_exponencial`) é só de números positivos, e é exatamente esse conjunto de valores que o logaritmo "recebe".


```python
# Testando em Python casos que violam as condições de existência
print("tentando log(-5, 2)  -- logaritmando negativo:")
try:
    math.log(-5, 2)
except ValueError as erro:
    print("  ValueError:", erro)

print("\ntentando log(8, 1)  -- base igual a 1:")
try:
    math.log(8, 1)
except ZeroDivisionError as erro:
    print("  ZeroDivisionError:", erro)

print("\nos dois erros fazem sentido: não existe expoente que transforme")
print("uma base positiva em um resultado negativo, e 1 elevado a qualquer")
print("expoente é sempre 1 -- nunca chegaria a 8.")
```

🕹️ **Widget interativo:** escolha valores de base a e logaritmando b e veja se log_a(b) é válido — e por quê.


```python
def checar_existencia(a: float, b: float) -> None:
    '''Verifica se log_a(b) existe, explicando qual condição falha, se for o caso.'''
    problemas = []
    if a <= 0:
        problemas.append("a base precisa ser positiva (a > 0)")
    elif abs(a - 1) < 1e-9:
        problemas.append("a base não pode ser 1 (a != 1)")
    if b <= 0:
        problemas.append("o logaritmando precisa ser positivo (b > 0)")

    if problemas:
        print(f"log_{a:g}({b:g}) NÃO existe:")
        for problema in problemas:
            print("  -", problema)
    else:
        print(f"log_{a:g}({b:g}) existe e vale {math.log(b, a):.4f}")


widgets.interact(
    checar_existencia,
    a=widgets.FloatSlider(value=2.0, min=-3.0, max=5.0, step=0.5, description="a:"),
    b=widgets.FloatSlider(value=8.0, min=-10.0, max=10.0, step=0.5, description="b:"),
)
```

**Pergunta de checagem:** por que log_a(b) só existe para b > 0, se a função exponencial de mesma base tem domínio igual a todo ℝ?

🎯 **Aplicação prática:** entender por que uma calculadora científica retorna erro ao calcular o log de um número negativo — situação comum em cálculos de engenharia e ciências.


## 3. Antilogaritmo

📌 **Definição formal:** dado o logaritmo x, o **antilogaritmo** é o número b tal que log_a(b) = x — ou seja, "desfazer" o logaritmo calculando aˣ. Logaritmo e antilogaritmo são operações **inversas** uma da outra.


```python
# Se x = 3 na base 10, qual é o antilogaritmo?
base = 10
x = 3
antilogaritmo = base**x
print(f"antilog de x={x} na base {base} = {base}^{x} = {antilogaritmo}")

# Percorrendo o caminho de ida (log) e volta (antilog) para conferir
numero_original = 1000
log_do_numero = math.log(numero_original, base)
volta_ao_original = base**log_do_numero
print(f"\nnúmero original: {numero_original}")
print(f"log_{base}({numero_original}) = {log_do_numero}")
print(f"antilog de {log_do_numero} na base {base} = {base}^{log_do_numero} = {volta_ao_original:g}  (voltou ao original!)")
```

**Pergunta de checagem:** se log_a(b) = x, o que representa o antilogaritmo de x na base a?

🎯 **Aplicação prática:** em tábuas de logaritmos antigas, o antilogaritmo era usado para "voltar" de um resultado logarítmico ao número original — prévia do notebook `0207a_logaritmos_decimais`.

## 4. Consequências da Definição

📌 **Definição formal:** direto da definição de logaritmo, seguem quatro consequências imediatas:

$$\log_a(1) = 0 \qquad \log_a(a) = 1 \qquad \log_a(a^m) = m \qquad a^{\log_a(b)} = b$$

A última é a **propriedade fundamental**: logaritmo e exponencial "se cancelam" quando compostas.


```python
# Verificando as quatro consequências numericamente, para várias bases
bases_teste = [2, 5, 10, 0.5]

for a in bases_teste:
    consequencia_1 = math.log(1, a)
    consequencia_2 = math.log(a, a)
    m = 4
    consequencia_3 = math.log(a**m, a)
    b = 7
    consequencia_4 = a ** math.log(b, a)
    print(
        f"a={a:>4g}:  log_a(1)={consequencia_1:g}  log_a(a)={consequencia_2:g}  "
        f"log_a(a^{m})={consequencia_3:g}  a^(log_a {b})={consequencia_4:.4f}"
    )
```

**Pergunta de checagem:** por que log_a(1) é sempre 0, não importa qual seja a base a?

🎯 **Aplicação prática:** essas consequências permitem simplificar expressões logarítmicas rapidamente, sem calculadora — muito útil em provas e cálculo mental.


## 5. Sistemas de Logaritmos

📌 **Definição formal:** dois sistemas de logaritmo têm nome e notação próprios, por serem usados o tempo todo:
- **Logaritmo decimal** (base 10): notação `log`, sem indicar a base.
- **Logaritmo natural** (base e ≈ 2,71828): notação `ln`.

Como curiosidade, existe ainda o **logaritmo binário** (base 2), muito usado em ciência da computação.


```python
# Calculando o logaritmo do mesmo número nos três sistemas
numero = 50

tabela_sistemas = pd.DataFrame(
    {
        "sistema": ["decimal (log)", "natural (ln)", "binário (log2)"],
        "base": [10, math.e, 2],
        f"log({numero})": [math.log10(numero), math.log(numero), math.log2(numero)],
    }
)
display(tabela_sistemas)
```

🕹️ **Widget interativo:** digite um número e veja seu logaritmo nas três bases simultaneamente.


```python
def comparar_sistemas(n: float) -> None:
    '''Mostra o logaritmo decimal, natural e binário do mesmo número n.'''
    if n <= 0:
        print("log só existe para números positivos (n > 0).")
        return
    tabela = pd.DataFrame(
        {
            "sistema": ["decimal (log)", "natural (ln)", "binário (log2)"],
            "base": [10, math.e, 2],
            f"log({n:g})": [math.log10(n), math.log(n), math.log2(n)],
        }
    )
    display(tabela)


widgets.interact(
    comparar_sistemas,
    n=widgets.FloatSlider(value=50.0, min=0.1, max=1000.0, step=0.1, description="n:"),
)
```

**Pergunta de checagem:** por que existem "sistemas" de logaritmo com nomes próprios, em vez de sempre usarmos uma base qualquer?

🎯 **Aplicação prática:** o logaritmo binário em ciência da computação — medir quantos bits são necessários para representar um número.

## 6. Propriedades Operatórias dos Logaritmos

📌 **Definição formal:** essas propriedades vêm diretamente das propriedades operatórias de potência (notebook `0201a_potencias_e_raizes`):

$$\log(a \cdot b) = \log(a) + \log(b) \qquad \log\left(\frac{a}{b}\right) = \log(a) - \log(b)$$

$$\log(a^m) = m \cdot \log(a) \qquad \log\left(\sqrt[n]{a}\right) = \frac{\log(a)}{n}$$


```python
# Simplificando expressões logarítmicas com as propriedades operatórias,
# usando valores conhecidos de log10(2) e log10(3)
log10_2 = math.log10(2)
log10_3 = math.log10(3)

# a) log10(6) = log10(2*3) = log10(2) + log10(3)  -- produto vira soma
log10_6 = log10_2 + log10_3
print(f"log10(6) = log10(2)+log10(3) = {log10_6:.4f}  (conferindo: {math.log10(6):.4f})")

# b) log10(9) = log10(3^2) = 2*log10(3)  -- potência vira multiplicação
log10_9 = 2 * log10_3
print(f"log10(9) = 2*log10(3) = {log10_9:.4f}  (conferindo: {math.log10(9):.4f})")

# c) log10(5) = log10(10/2) = log10(10)-log10(2)  -- quociente vira subtração
log10_5 = 1 - log10_2
print(f"log10(5) = 1-log10(2) = {log10_5:.4f}  (conferindo: {math.log10(5):.4f})")

# Um caso de uso real: transformar uma multiplicação trabalhosa numa soma fácil
print("\nexemplo histórico: 128 * 256, usando log2 (ambos são potências de 2)")
resultado_via_log = 2 ** (math.log2(128) + math.log2(256))
print(f"log2(128)+log2(256) = {math.log2(128):g}+{math.log2(256):g} = {math.log2(128)+math.log2(256):g}")
print(f"2^{math.log2(128)+math.log2(256):g} = {resultado_via_log:g}  (conferindo: 128*256 = {128*256})")
```

**Pergunta de checagem:** por que log(a·b) = log(a) + log(b), e não log(a)·log(b)?

🎯 **Aplicação prática:** essa é exatamente a ideia que tornou os logaritmos tão importantes historicamente — transformar multiplicações difíceis em somas fáceis, antes das calculadoras (prévia do bloco 8).


## 7. Mudança de Base

📌 **Definição formal:**

$$\log_a(b) = \frac{\log_c(b)}{\log_c(a)}, \quad \text{para qualquer base } c > 0,\ c \neq 1$$

Essa fórmula é útil na prática: calculadoras comuns só têm log decimal e natural, então precisamos converter para calcular logaritmos em qualquer outra base.


```python
# Calculando log2(10) por mudança de base, primeiro para base 10 e depois para base e
log2_10_via_base10 = math.log10(10) / math.log10(2)
log2_10_via_basee = math.log(10) / math.log(2)

print(f"log2(10) via base 10 = log10(10)/log10(2) = {log2_10_via_base10:.4f}")
print(f"log2(10) via base e  = ln(10)/ln(2)       = {log2_10_via_basee:.4f}")
print(f"conferindo diretamente: math.log(10, 2) = {math.log(10, 2):.4f}")
```

🕹️ **Widget interativo:** escolha a base original a, o logaritmando b e a base de conversão c, e veja o cálculo passo a passo.


```python
def mudanca_de_base(a: float, b: float, c: float) -> None:
    '''Calcula log_a(b) diretamente e por mudança de base via c, passo a passo.'''
    if a <= 0 or abs(a - 1) < 1e-9 or b <= 0 or c <= 0 or abs(c - 1) < 1e-9:
        print("valores inválidos: a e c precisam ser > 0 e != 1; b precisa ser > 0.")
        return

    resultado_direto = math.log(b, a)
    log_c_b = math.log(b, c)
    log_c_a = math.log(a, c)

    print(f"log_{a:g}({b:g}) direto = {resultado_direto:.4f}")
    print(f"\npasso a passo, convertendo para base {c:g}:")
    print(f"  log_{c:g}({b:g}) = {log_c_b:.4f}")
    print(f"  log_{c:g}({a:g}) = {log_c_a:.4f}")
    print(f"  log_{a:g}({b:g}) = {log_c_b:.4f} / {log_c_a:.4f} = {log_c_b / log_c_a:.4f}")


widgets.interact(
    mudanca_de_base,
    a=widgets.FloatSlider(value=2.0, min=0.1, max=10.0, step=0.1, description="a:"),
    b=widgets.FloatSlider(value=10.0, min=0.1, max=100.0, step=0.1, description="b:"),
    c=widgets.FloatSlider(value=10.0, min=0.1, max=10.0, step=0.1, description="c:"),
)
```

**Pergunta de checagem:** por que o resultado de log₂(10) é o mesmo, seja convertendo para base 10 ou para base e?

🎯 **Aplicação prática:** como calcular, na prática, o logaritmo em qualquer base usando Python ou uma calculadora comum (que normalmente só tem log decimal e natural).


## 8. Napier, o Cálculo com Python e o Próximo Passo

🕵️ **Curiosidade histórica:** os logaritmos foram criados em 1614 pelo matemático escocês **John Napier**, com um único objetivo: simplificar cálculos manuais enormes, transformando multiplicações trabalhosas em somas simples — exatamente a propriedade que vimos no bloco 6. Antes das calculadoras, tábuas de logaritmos eram ferramentas essenciais para astrônomos, engenheiros e navegadores.

📎 **Conexão:** lembra da lenda do tabuleiro de xadrez (notebook `0201a_potencias_e_raizes.ipynb`) e do desafio da folha de papel até a Lua (notebook `0202a_funcao_exponencial`)? Lá, perguntávamos "quanto vale depois de N dobras/casas?". Agora podemos perguntar o contrário: **"quantas dobras/casas são necessárias para atingir um valor X?"** — e o logaritmo responde essa pergunta diretamente, sem tentativa e erro.


```python
# Resolvendo com logaritmo, de forma EXATA, o desafio da folha de papel:
# com quantas dobras uma folha de 0,1mm alcançaria a Lua (384.000 km)?
espessura_inicial_mm = 0.1
distancia_lua_mm = 384_000 * 1_000 * 1_000  # km -> m -> mm

dobras_exato = math.log(distancia_lua_mm / espessura_inicial_mm, 2)
dobras_necessarias = math.ceil(dobras_exato)

print(f"dobras (valor exato e contínuo) = {dobras_exato:.4f}")
print(f"dobras necessárias (inteiro, arredondando para cima) = {dobras_necessarias}")
print(
    f"conferindo: 0,1mm * 2^{dobras_necessarias} = "
    f"{espessura_inicial_mm * 2**dobras_necessarias:,.0f} mm"
)
```

🕹️ **Widget interativo:** calculadora "quantas vezes preciso multiplicar por um fator para sair de um valor inicial e atingir (ou passar) um valor alvo?" — usando logaritmo.


```python
def calculadora_dobras(valor_inicial: float, valor_alvo: float, fator: float) -> None:
    '''Calcula quantas multiplicações por "fator" são necessárias para ir de
    valor_inicial até valor_alvo, usando logaritmo.'''
    if valor_inicial <= 0 or valor_alvo <= valor_inicial or fator <= 1:
        print("valores inválidos: valor_inicial > 0, valor_alvo > valor_inicial, fator > 1.")
        return

    n_exato = math.log(valor_alvo / valor_inicial) / math.log(fator)
    n_inteiro = math.ceil(n_exato)

    print(f"de {valor_inicial:g} até {valor_alvo:g}, multiplicando por {fator:g} a cada passo:")
    print(f"  n exato = log({valor_alvo:g}/{valor_inicial:g}) / log({fator:g}) = {n_exato:.4f}")
    print(f"  são necessários {n_inteiro} passos inteiros (arredondando para cima)")


widgets.interact(
    calculadora_dobras,
    valor_inicial=widgets.FloatText(value=0.1, description="valor inicial:"),
    valor_alvo=widgets.FloatText(value=384_000_000_000.0, description="valor alvo:"),
    fator=widgets.FloatSlider(value=2.0, min=1.1, max=10.0, step=0.1, description="fator:"),
)
```

## Resumo — cheat sheet

| Conceito | Ideia-chave |
|---|---|
| Definição | log_a(b) = x ⟺ aˣ = b (a>0, a≠1, b>0) |
| Elementos | base (a), logaritmando (b), logaritmo (x) |
| Existência | base>0, base≠1, logaritmando>0 |
| Antilogaritmo | dado x, o antilog é b tal que log_a(b)=x, ou seja, b = aˣ |
| Consequências | log_a(1)=0 \| log_a(a)=1 \| log_a(aᵐ)=m \| a^(log_a b) = b |
| Sistemas | decimal: log (base 10) \| natural: ln (base e) \| binário: log2 (base 2) |
| Propriedades | log(a·b)=log(a)+log(b) \| log(a/b)=log(a)-log(b) \| log(aᵐ)=m·log(a) \| log(ⁿ√a)=log(a)/n |
| Mudança de base | log_a(b) = log_c(b) / log_c(a), para qualquer c>0, c≠1 |

## Glossário do bloco

| Termo | Definição |
|---|---|
| Logaritmo | expoente x tal que aˣ = b, denotado log_a(b) |
| Base | o número a em log_a(b), com a>0 e a≠1 |
| Logaritmando | o número b em log_a(b), com b>0 |
| Antilogaritmo | o número b tal que log_a(b)=x, ou seja, b = aˣ |
| Logaritmo decimal | logaritmo na base 10, notação log |
| Logaritmo natural | logaritmo na base e, notação ln |
| Logaritmo binário | logaritmo na base 2, notação log2, usado em ciência da computação |
| Mudança de base | fórmula log_a(b) = log_c(b)/log_c(a), válida para qualquer base c>0, c≠1 |
