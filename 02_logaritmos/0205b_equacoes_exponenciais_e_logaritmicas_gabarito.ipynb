```python
import math

from scipy.optimize import brentq
```

## Básico

**Exercício 1.** Resolva 5ˣ = 40 usando logaritmo.

```python
# 5^x = 40  ->  aplicando log dos dois lados:  x*log(5) = log(40)
x_exercicio1 = math.log(40) / math.log(5)
print(f"x = log(40)/log(5) = {x_exercicio1:.4f}")

# conferindo
print(f"conferindo: 5^{x_exercicio1:.4f} = {5**x_exercicio1:.4f}")
```

**Exercício 2.** Resolva log₄(x) = 3 e verifique a condição de existência.

```python
# log_4(x) = 3  ->  x = 4^3, pela definição de logaritmo
x_exercicio2 = 4**3
existe_exercicio2 = x_exercicio2 > 0
print("x =", x_exercicio2)
print("condição de existência (x>0)?", existe_exercicio2)
```

## Intermediário

**Exercício 3.** Resolva 9ˣ - 4·3ˣ + 3 = 0 por substituição y = 3ˣ.

```python
# 9^x = (3^2)^x = (3^x)^2 = y^2, com y = 3^x
# a equação vira y^2 - 4y + 3 = 0
# fatorando: (y-1)(y-3) = 0  ->  y = 1 ou y = 3 (ambas positivas, nenhuma descartada)
raizes_y = [1, 3]
solucoes_exercicio3 = sorted(round(math.log(y, 3)) for y in raizes_y)
print("raízes em y:", raizes_y)
print("y=1 -> 3^x=1 -> x=0   |   y=3 -> 3^x=3 -> x=1")
print("soluções em x:", solucoes_exercicio3)
```

**Exercício 4.** Resolva log(x²-1) = log(3x-3), descartando a raiz inválida.

```python
# log(x^2-1) = log(3x-3)  ->  x^2-1 = 3x-3  ->  x^2-3x+2 = 0  ->  (x-1)(x-2)=0
raizes_algebricas = [1, 2]
print("raízes algébricas:", raizes_algebricas)

print("\nverificando as condições de existência (x^2-1>0 e 3x-3>0):")
for x in raizes_algebricas:
    condicao_1 = (x**2 - 1) > 0
    condicao_2 = (3 * x - 3) > 0
    print(f"  x={x}: x^2-1={x**2-1} ({'OK' if condicao_1 else 'FALHA'})   "
          f"3x-3={3*x-3} ({'OK' if condicao_2 else 'FALHA'})")

x_exercicio4 = 2
print(f"\nsolução válida: x = {x_exercicio4}  (x=1 é descartado, pois gera log(0), que não existe)")
```

## Desafio

**Exercício 5.** A concentração de um poluente decai pela metade a cada 3 anos. Partindo de 80 mg/L, em quantos anos ela atinge o limite legal de 5 mg/L?

```python
# concentracao_inicial*(1/2)^(t/meia_vida) = limite_legal
concentracao_inicial = 80
limite_legal = 5
meia_vida_poluente = 3  # anos

# isolando a potência: (1/2)^(t/meia_vida) = limite_legal/concentracao_inicial
# aplicando log: (t/meia_vida)*log(0.5) = log(limite_legal/concentracao_inicial)
anos_ate_limite = meia_vida_poluente * math.log(limite_legal / concentracao_inicial) / math.log(0.5)
print(f"t = {meia_vida_poluente} * log({limite_legal}/{concentracao_inicial}) / log(0.5) = {anos_ate_limite:.4f} anos")

# conferindo com método numérico, como sanity check
def diferenca_poluente(t: float) -> float:
    return concentracao_inicial * (0.5) ** (t / meia_vida_poluente) - limite_legal


anos_numerico = brentq(diferenca_poluente, 0, 200)
print(f"conferindo com método numérico: {anos_numerico:.4f} anos")
```
