## Básico

**Exercício 1.** Para f(x) = 5ˣ e g(x) = (1/3)ˣ, determine o domínio, a imagem e classifique cada uma como crescente ou decrescente.

```python
# domínio: qualquer função exponencial tem domínio R
dominio_f = "R"
dominio_g = "R"

# imagem: qualquer função exponencial tem imagem R*+ (reais positivos)
imagem_f = "R*+"
imagem_g = "R*+"

# comportamento: base 5 > 1 -> crescente | base 1/3 entre 0 e 1 -> decrescente
comportamento_f = "crescente"
comportamento_g = "decrescente"

print(dominio_f, imagem_f, comportamento_f)
print(dominio_g, imagem_g, comportamento_g)
```

**Exercício 2.** Para f(x) = 3ˣ, calcule f(0), identifique o ponto fixo por onde passa o gráfico de qualquer exponencial e a assíntota horizontal.

```python
# a^0 = 1 para qualquer base válida
f_de_zero = 3**0

# o gráfico de aˣ sempre passa por (0, 1), pois a⁰=1
ponto_fixo = (0, 1)

# a assíntota horizontal de aˣ é sempre y=0 (a curva se aproxima, mas nunca toca)
assintota_y = 0

print(f_de_zero, ponto_fixo, assintota_y)
```

## Intermediário

**Exercício 3.** Resolva as equações exponenciais reduzindo a uma base comum.
- a) 5ˣ = 125
- b) 8ˣ = 1/2

```python
# a) 5^x = 125 = 5^3  ->  x = 3
solucao_3a = 3

# b) 8^x = 1/2  ->  (2^3)^x = 2^-1  ->  2^(3x) = 2^-1  ->  3x = -1  ->  x = -1/3
solucao_3b = -1 / 3

print(solucao_3a, solucao_3b)
print("conferindo: 5^3 =", 5**3, " | 8^(-1/3) =", 8 ** (-1 / 3))
```

**Exercício 4.** Resolva as inequações exponenciais reduzindo a uma base comum.
- a) 3ˣ > 81
- b) (1/4)ˣ < 16

```python
# a) 3^x > 81 = 3^4, base 3 > 1 mantém o sentido da desigualdade  ->  x > 4
solucao_4a = "x > 4"

# b) (1/4)^x < 16 = (1/4)^-2, base 1/4 < 1 INVERTE o sentido  ->  x > -2
solucao_4b = "x > -2"

print(solucao_4a)
print(solucao_4b)
```

## Desafio

**Exercício 5.** Um investimento de R$1000, com juros compostos de 25% ao período, deve dobrar de valor. Calcule o menor número inteiro de períodos necessário para que o montante ultrapasse R$2000, testando período a período — repare que 2 não é uma potência exata de 1,25, então esse problema não é redutível a uma base comum.

```python
capital_ex5 = 1000
taxa_ex5 = 0.25

# testa período a período até o montante dobrar (não há solução por base comum aqui)
periodos_para_dobrar = 0
while capital_ex5 * (1 + taxa_ex5) ** periodos_para_dobrar < 2 * capital_ex5:
    periodos_para_dobrar += 1

print(periodos_para_dobrar, capital_ex5 * (1 + taxa_ex5) ** periodos_para_dobrar)
```
