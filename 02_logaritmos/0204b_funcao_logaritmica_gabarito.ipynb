## Básico


**Exercício 1.** Para f(x) = log_3(x) e g(x) = log_(1/4)(x), determine o domínio, a imagem e se cada uma é crescente ou decrescente.


```python
# domínio e imagem são os mesmos para qualquer base válida
dominio_f = "R*+"
imagem_f = "R"
# base 3 > 1 -> crescente
comportamento_f = "crescente"

dominio_g = "R*+"
imagem_g = "R"
# base 1/4 está entre 0 e 1 -> decrescente
comportamento_g = "decrescente"

print(dominio_f, imagem_f, comportamento_f)
print(dominio_g, imagem_g, comportamento_g)
```

**Exercício 2.** Para f(x) = log_5(x), calcule f(1), o ponto fixo por onde passa o gráfico de qualquer função logarítmica, e a equação da assíntota vertical.


```python
# log_a(1) = 0 para qualquer base
f_de_um = 0

# o gráfico de qualquer log_a(x) passa por (1, 0), pois log_a(1) = 0
ponto_fixo = (1, 0)

# a assíntota vertical de log_a(x) é sempre x = 0
assintota_x = "x = 0"

print(f_de_um, ponto_fixo, assintota_x)
```

## Intermediário


**Exercício 3.** Use a simetria entre aˣ e log_a(x) (reflexão em torno de y=x) para encontrar os pontos pedidos, sem calcular log diretamente.


```python
# a simetria em y=x troca as coordenadas (x, y) -> (y, x)

# (3, 8) está em 2^x (pois 2^3=8) -> (8, 3) está em log_2(x)
ponto_log2 = (8, 3)

# (2, 100) está em 10^x (pois 10^2=100) -> (100, 2) está em log_10(x)
ponto_log10 = (100, 2)

print(ponto_log2, ponto_log10)
print("conferindo:", math.log(ponto_log2[0], 2), math.log(ponto_log10[0], 10))
```

**Exercício 4.** Para a transformação h(x) = log_2(x + 5), determine o novo domínio e a nova assíntota vertical.


```python
# log_2(x+5) exige x+5 > 0, ou seja x > -5
novo_dominio = "x > -5"

# deslocar log_2(x) por x+5 desloca a assíntota de x=0 para x=-5
nova_assintota = "x = -5"

print(novo_dominio)
print(nova_assintota)
```

## Desafio


**Exercício 5.** Dois terremotos têm magnitudes 6,5 e 4,0 na escala Richter. Calcule quantas vezes mais forte (em amplitude) é o terremoto mais forte, usando M = log10(A/A0).


```python
magnitude_forte = 6.5
magnitude_fraco = 4.0

# a razão de amplitude é 10^(diferença de magnitude)
razao_amplitude = 10 ** (magnitude_forte - magnitude_fraco)

print(f"{razao_amplitude:g}x mais forte")
```
