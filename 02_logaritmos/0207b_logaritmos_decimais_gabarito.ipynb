# Gabarito — Logaritmos Decimais

```python
import math

import pandas as pd

# Tábua simplificada, reconstruída aqui para as resoluções que dependem dela
linhas = list(range(30, 40))
colunas = list(range(10))
tabua_simplificada = pd.DataFrame(
    {coluna: [round(math.log10(linha / 10 + coluna / 100), 4) for linha in linhas] for coluna in colunas},
    index=linhas,
)
tabua_simplificada.index.name = "linha"
tabua_simplificada.columns.name = "coluna (3º algarismo)"
```

## Básico

**Exercício 1.** Decomponha log(4780) em característica e mantissa.

```python
# log(4780): a característica é a parte inteira, a mantissa é a parte decimal
log_4780 = math.log10(4780)
caracteristica_exercicio1 = math.floor(log_4780)
mantissa_exercicio1 = round(log_4780 - caracteristica_exercicio1, 4)
decomposicao_exercicio1 = (caracteristica_exercicio1, mantissa_exercicio1)

print(f"log(4780) = {log_4780:.4f}")
print(f"característica = {caracteristica_exercicio1}   mantissa = {mantissa_exercicio1}")
print(f"\nconferindo: {caracteristica_exercicio1} + {mantissa_exercicio1} = {caracteristica_exercicio1 + mantissa_exercicio1}")
```

**Exercício 2.** Determine a característica de 27300 e de 0,0058, apenas observando a estrutura de cada número.

```python
# 27300 tem 5 dígitos na parte inteira -> característica = 5 - 1 = 4
caracteristica_27300 = 5 - 1

# 0,0058 = 5,8 x 10^-3: o 1º algarismo significativo está na 3ª casa decimal -> característica = -3
caracteristica_0058 = -3

caracteristicas_exercicio2 = (caracteristica_27300, caracteristica_0058)

print(f"27300 -> característica = {caracteristica_27300}")
print(f"0,0058 -> característica = {caracteristica_0058}")
print(f"\nconferindo com Python: {math.floor(math.log10(27300))}, {math.floor(math.log10(0.0058))}")
```

## Intermediário

**Exercício 3.** Usando a tábua simplificada, calcule log(347) por consulta direta (sem interpolação).

```python
# 347 tem 3 dígitos -> característica = 2
caracteristica_347 = 2

# normalizado: 3,47 -> linha 34 (dois primeiros algarismos), coluna 7 (terceiro algarismo)
mantissa_347 = tabua_simplificada.loc[34, 7]

log_exercicio3 = caracteristica_347 + mantissa_347

print(f"347 normalizado = 3,47  ->  linha 34, coluna 7  ->  mantissa = {mantissa_347}")
print(f"log(347) = {caracteristica_347} + {mantissa_347} = {log_exercicio3}")
print(f"conferindo com Python: math.log10(347) = {math.log10(347):.4f}")
```

**Exercício 4.** Calcule o antilogaritmo de 3,5378, decompondo em característica e mantissa e revertendo com a tábua.

```python
log_alvo = 3.5378

# separando característica e mantissa
caracteristica_alvo = math.floor(log_alvo)
mantissa_alvo = round(log_alvo - caracteristica_alvo, 4)
print(f"característica = {caracteristica_alvo}, mantissa = {mantissa_alvo}")

# consulta reversa: qual célula da tábua tem mantissa mais próxima de 0,5378?
diferencas = (tabua_simplificada - mantissa_alvo).abs()
linha_encontrada, coluna_encontrada = diferencas.stack().idxmin()
numero_normalizado = linha_encontrada / 10 + coluna_encontrada / 100
print(f"mantissa {mantissa_alvo} está na linha {linha_encontrada}, coluna {coluna_encontrada}  ->  número normalizado {numero_normalizado:.2f}")

# reposicionando a vírgula com a característica
antilog_exercicio4 = numero_normalizado * 10**caracteristica_alvo
print(f"\nantilog({log_alvo}) = {numero_normalizado:.2f} x 10^{caracteristica_alvo} = {antilog_exercicio4:.0f}")
print(f"conferindo com Python: 10^{log_alvo} = {10**log_alvo:.1f}")
```

## Desafio

**Exercício 5.** Reproduza um cálculo histórico, multiplicando 452 × 6,8 usando logaritmo e antilogaritmo.

```python
# transformando a multiplicação numa soma de logaritmos (propriedade do produto)
log_452 = math.log10(452)
log_68 = math.log10(6.8)
log_produto = log_452 + log_68

print(f"log(452) = {log_452:.4f}")
print(f"log(6,8) = {log_68:.4f}")
print(f"log(452 x 6,8) = log(452) + log(6,8) = {log_produto:.4f}")

# desfazendo com o antilogaritmo (10 elevado à soma)
produto_exercicio5 = round(10**log_produto, 1)
print(f"\nantilog({log_produto:.4f}) = 10^{log_produto:.4f} = {produto_exercicio5}")

# conferindo com a multiplicação direta
print(f"conferindo: 452 x 6,8 = {452 * 6.8}")
```
