## Básico


**Exercício 1.** Aplique a definição de logaritmo para calcular, de cabeça, log_3(81) e log_5(1/25).


```python
# 3^4 = 81, logo log_3(81) = 4
log_3_81 = 4

# 5^-2 = 1/25, logo log_5(1/25) = -2
log_5_1_25 = -2

print(log_3_81, log_5_1_25)
print("conferindo:", 3**log_3_81, 5**log_5_1_25)
```

## Intermediário


**Exercício 3.** Usando log10(2) ≈ 0,30103 e log10(3) ≈ 0,47712, calcule log10(6), log10(9) e log10(5) com as propriedades operatórias.


```python
log10_2 = 0.30103
log10_3 = 0.47712

# log10(6) = log10(2*3) = log10(2)+log10(3)  -- propriedade do produto
log10_6 = log10_2 + log10_3

# log10(9) = log10(3^2) = 2*log10(3)  -- propriedade da potência
log10_9 = 2 * log10_3

# log10(5) = log10(10/2) = log10(10)-log10(2) = 1-log10(2)  -- propriedade do quociente
log10_5 = 1 - log10_2

print(log10_6, log10_9, log10_5)
print("conferindo:", math.log10(6), math.log10(9), math.log10(5))
```

**Exercício 4.** Calcule log_5(20) por mudança de base, convertendo para base 10 e para base e, e confira que os resultados batem.


```python
# convertendo log_5(20) para base 10: log10(20)/log10(5)
log_5_20_base10 = math.log10(20) / math.log10(5)

# convertendo log_5(20) para base e (natural): ln(20)/ln(5)
log_5_20_basee = math.log(20) / math.log(5)

print(log_5_20_base10, log_5_20_basee)
print("conferindo com math.log(20, 5):", math.log(20, 5))
```

## Desafio


**Exercício 5.** Uma população cresce a 8% ao ano. Usando logaritmo, calcule (com casas decimais) em quantos anos ela dobra, e compare com uma estimativa por tentativa e erro.


```python
taxa_crescimento = 0.08

# se (1+taxa)^t = 2, então t = log(2)/log(1+taxa)
anos_para_dobrar = math.log(2) / math.log(1 + taxa_crescimento)
print(f"logaritmo (exato): {anos_para_dobrar:.4f} anos")

# comparando com uma estimativa por tentativa e erro, ano a ano
anos_estimativa = 0
populacao = 1.0
while populacao < 2.0:
    populacao *= 1 + taxa_crescimento
    anos_estimativa += 1

print(f"tentativa e erro (anos inteiros completos): {anos_estimativa} anos")
```
