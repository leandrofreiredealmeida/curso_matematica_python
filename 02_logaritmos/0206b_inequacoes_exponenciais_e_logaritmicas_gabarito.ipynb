```python
import math

import sympy as sp
```

## Básico

**Exercício 1.** Resolva (0,2)ˣ < 25 usando logaritmo.

```python
# (0,2)^x < 25  ->  aplicando log dos dois lados:  x*log(0,2) < log(25)
# como log(0,2) < 0, ao dividir os dois lados por ele o sentido se INVERTE
x_fronteira_exercicio1 = math.log(25) / math.log(0.2)
inverteu_exercicio1 = math.log(0.2) < 0

print(f"x*log(0,2) < log(25)  ->  x > log(25)/log(0,2) = {x_fronteira_exercicio1:.4f}")
print(f"log(0,2) = {math.log(0.2):.4f} < 0  ->  inverteu = {inverteu_exercicio1}")
print(f"\nconferindo: (0,2)^{x_fronteira_exercicio1:.4f} = {0.2**x_fronteira_exercicio1:.4f} (deve ser ~25)")
```

**Exercício 2.** Resolva log₃(x) < 4, intersectando com o domínio.

```python
# log_3(x) < 4  ->  x < 3^4 = 81  (base 3 > 1, mantém o sentido)
# domínio do logaritmo: x > 0
limite_superior = 3**4
solucao_exercicio2 = (0, limite_superior)

print(f"log_3(x) < 4  ->  x < {limite_superior}")
print("domínio: x > 0")
print("solução final (interseção):", solucao_exercicio2)
```

## Intermediário

**Exercício 3.** Resolva 9ˣ - 10·3ˣ + 9 < 0 por substituição y = 3ˣ.

```python
# 9^x = (3^2)^x = (3^x)^2 = y^2, com y = 3^x
# a inequação vira y^2 - 10y + 9 < 0
y = sp.symbols("y", real=True)
solucao_y = sp.solve_univariate_inequality(sp.Lt(y**2 - 10 * y + 9, 0), y, relational=False)
print("substituindo y = 3^x:  y^2 - 10y + 9 < 0  ->  y em", solucao_y)

# y=1 e y=9 são as raízes de y^2-10y+9=0; ambas positivas, nenhuma descartada
solucao_y_valida = solucao_y.intersect(sp.Interval.open(0, sp.oo))
print("intersectando com y > 0:  y em", solucao_y_valida)

limite_inferior_x = math.log(float(solucao_y_valida.start), 3)
limite_superior_x = math.log(float(solucao_y_valida.end), 3)
solucao_exercicio3 = (round(limite_inferior_x), round(limite_superior_x))
print("voltando para x (y = 3^x):", solucao_exercicio3)
```

**Exercício 4.** Resolva log_(0,5)(2x-1) > log_(0,5)(x+3), intersectando com o domínio.

```python
# base 0,5 está entre 0 e 1  ->  a comparação entre os logaritmandos INVERTE
# log_0,5(2x-1) > log_0,5(x+3)  ->  2x-1 < x+3  ->  x < 4
limite_superior = 4
print("base 0,5 (0<a<1) -> inverte:  2x-1 < x+3  ->  x < 4")

# domínio: 2x-1>0 (x>0,5) e x+3>0 (x>-3)
print("domínio: 2x-1>0 -> x>0,5   e   x+3>0 -> x>-3")

limite_inferior = 0.5
solucao_exercicio4 = (limite_inferior, limite_superior)
print("interseção final:", solucao_exercicio4)
```

## Desafio

**Exercício 5.** Resolva log(x-1) + log(x+1) < log(8), corrigindo pela interseção com o domínio.

```python
# log(x-1)+log(x+1) = log((x-1)(x+1)), pela propriedade do produto
# log((x-1)(x+1)) < log(8)  ->  (x-1)(x+1) < 8  ->  x^2-1 < 8  ->  x^2 < 9  ->  -3<x<3
print("log((x-1)(x+1)) < log(8)  ->  x^2-1 < 8  ->  x^2 < 9  ->  solução algébrica: -3 < x < 3")

# domínio: x-1>0 E x+1>0  ->  x>1  (x-1>0 é a condição mais forte)
print("domínio: x-1>0 e x+1>0  ->  x>1")

solucao_exercicio5 = (1, 3)
print("interseção entre (-3,3) e (1,+inf):", solucao_exercicio5)
print("\nrepare: sem checar o domínio, alguém poderia (erradamente) aceitar valores como x=0,")
print("que satisfazem a álgebra mas não fazem log(x-1) existir.")
```
