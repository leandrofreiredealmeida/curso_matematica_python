## Básico

**Exercício 1.** Aplique as propriedades de potência para simplificar as três expressões abaixo, sem calcular tudo "na mão".

```python
# a) 2^3 * 2^4 / 2^5 = 2^(3+4-5) = 2^2
resultado_1a = 2 ** (3 + 4 - 5)

# b) (3^2)^3 = 3^(2*3) = 3^6
resultado_1b = 3 ** (2 * 3)

# c) 5^0 + 5^(-2) = 1 + 1/25
resultado_1c = 5**0 + 5**-2

print(resultado_1a, resultado_1b, resultado_1c)
```

**Exercício 2.** Para cada par (índice, radicando), diga se a raiz enésima existe em ℝ.

```python
pares_ex2 = [(2, 16), (3, -27), (4, -16), (5, -32)]

# índice PAR exige radicando >= 0; índice ÍMPAR aceita qualquer radicando real
existe_ex2 = [not (indice % 2 == 0 and radicando < 0) for indice, radicando in pares_ex2]

print(existe_ex2)
```

## Intermediário

**Exercício 3.** Converta as potências de expoente racional para a notação de radical e calcule o valor numérico.

```python
# 8^(2/3) = (∛8)^2 = 2^2
valor_ex3a = sp.root(8, 3) ** 2

# 16^(3/4) = (⁴√16)^3 = 2^3
valor_ex3b = sp.root(16, 4) ** 3

print(valor_ex3a, valor_ex3b)
```

**Exercício 4.** Racionalize os dois denominadores, o segundo usando o conjugado.

```python
# 3/√2 * (√2/√2) = 3√2/2 = 1,5·√2
c_ex4a = sp.Rational(3, 2)

# 1/(√5-2) * (√5+2)/(√5+2): o denominador vira (√5)^2 - 2^2 = 5 - 4 = 1
d_ex4b = 2

print(c_ex4a, d_ex4b)
print(f"3/√2 = {c_ex4a}·√2   |   1/(√5-2) = √5 + {d_ex4b}")
```
